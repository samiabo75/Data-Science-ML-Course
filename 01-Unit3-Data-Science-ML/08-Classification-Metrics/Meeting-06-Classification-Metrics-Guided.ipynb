**מטרות:** לקרוא Confusion Matrix, להבחין בין TP, FP, FN, TN, לחשב Accuracy, Precision, Recall ו־F1, ולהסביר מדוע Accuracy לבדו אינו מספיק.

**ידע קודם:** סיווג באמצעות KNN, רשימות ופעולות חשבון בסיסיות.

## 1. בדיקת קורונה — מזהים אמת מול תחזית
לפי דף העבודה: מתוך 500 חולים, 450 זוהו כחולים ו־50 הוחמצו; מתוך 500 בריאים, 20 סווגו כחולים ו־480 זוהו כבריאים.

נסמן את 'חולה' כמחלקה החיובית. השורות הן **אמת**, העמודות **חיזוי**.

```python
import pandas as pd
cm = pd.DataFrame([[450,50],[20,480]],index=['אמת: חולה','אמת: בריא'],columns=['חזוי: חולה','חזוי: בריא'])
display(cm)
```

## 2. ארבעת המצבים
**TP=450** זיהוי נכון של חולים. **FN=50** חולים שהוחמצו. **FP=20** בריאים שסווגו כחולים. **TN=480** זיהוי נכון של בריאים.

חשבו קודם ביד את שיעור הסיווג הנכון.

```python
TP,FN,FP,TN = 450,50,20,480
accuracy=(TP+TN)/(TP+TN+FP+FN)
precision=TP/(TP+FP)
recall=TP/(TP+FN)
f1=2*precision*recall/(precision+recall)
print('Accuracy:',round(accuracy,3))
print('Precision:',round(precision,3))
print('Recall:',round(recall,3))
print('F1:',round(f1,3))
```

## 3. מה כל מדד מספר?
- **Accuracy:** כמה סווגו נכון מכל הדוגמאות.
- **Precision:** מתוך אלו שסווגו כחולים, כמה באמת חולים?
- **Recall:** מתוך החולים בפועל, כמה זוהו?
- **F1:** ממוצע הרמוני של Precision ו־Recall.

**דיון:** כשחוששים מהחמצת מחלה, מדוע Recall חשוב? זוהי המחשה לימודית, לא הנחיה רפואית.

## 4. מקרה נוסף — סיווג חתולים
בדף העבודה השני: נבדקו 800 חיות, מהן 100 חתולים; 90 חתולים זוהו נכון, ו־90 חיות שאינן חתולים זוהו בטעות כחתולים. בדקו אם Accuracy לבדו נותן תמונה מלאה.

```python
TP,FN,FP,TN=90,10,90,610
print('Accuracy:',round((TP+TN)/800,3))
print('Precision:',round(TP/(TP+FP),3))
print('Recall:',round(TP/(TP+FN),3))
```

## 5. מטריצה רב־מחלקתית — שלושה חתולים
בדף העבודה יש שלושה חתולים: סיימון, מאיה וסאני. לכל חתול אפשר לחשב Precision/Recall/F1 בהתייחס אליו כמחלקה החיובית, ולשניים האחרים יחד כשלילית.

```python
from sklearn.metrics import ConfusionMatrixDisplay, classification_report, confusion_matrix
import matplotlib.pyplot as plt
# שורות: אמת; עמודות: חיזוי
matrix=[[12,2,6],[13,7,5],[8,10,4]]
truth=[]
pred=[]
for i,row in enumerate(matrix):
    for j,count in enumerate(row):
        truth.extend([i]*count)
        pred.extend([j]*count)
ConfusionMatrixDisplay.from_predictions(truth,pred,display_labels=['Simon','Maya','Sunny'],cmap='Blues')
plt.show()
print(classification_report(truth,pred,target_names=['Simon','Maya','Sunny'],zero_division=0))
```
