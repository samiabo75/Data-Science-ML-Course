## פתיחה: מה שואלים לפני גרף?
`species` הוא משתנה קטגוריאלי; `body_mass_g` הוא כמותי רציף. ה־Dataset מכיל נתונים חסרים. בחרו גרף לפי השאלה ולא לפי יופיו.

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
url = "https://raw.githubusercontent.com/samiabo75/Data-Science-ML-Course/main/Datasets/penguins.csv"
df = pd.read_csv(url)
display(df.head())
df.info()
```

## Countplot: כמה תצפיות יש בכל קטגוריה?
מספר הרשומות במדגם אינו מספר הפינגווינים בטבע.

```python
plt.figure(figsize=(7,4))
sns.countplot(data=df, x="species", color="steelblue")
plt.title("Penguin observations by species")
plt.xlabel("Species")
plt.ylabel("Count")
plt.show()
```

## Histogram: התפלגות משתנה רציף
העמודות הן טווחי ערכים (bins).

```python
plt.figure(figsize=(7,4))
sns.histplot(data=df,x="body_mass_g",bins=12)
plt.title("Body mass distribution")
plt.xlabel("Body mass (g)")
plt.show()
```

## Barplot: ממוצע לפי קבוצה
זה אינו גרף לספירת תצפיות.

```python
plt.figure(figsize=(7,4))
sns.barplot(data=df,x="species",y="body_mass_g",errorbar=None)
plt.title("Mean body mass by species")
plt.ylabel("Mean mass (g)")
plt.show()
```

## Boxplot: חציון, רבעונים וחריגים
ערך חריג אינו בהכרח שגיאה.

```python
plt.figure(figsize=(7,4))
sns.boxplot(data=df,x="species",y="body_mass_g")
plt.title("Body mass by species")
plt.show()
```

## Scatterplot: קשר בין שני משתנים כמותיים
מתאם אינו סיבתיות.

```python
plt.figure(figsize=(7,4))
sns.scatterplot(data=df,x="flipper_length_mm",y="body_mass_g",hue="species")
plt.title("Flipper length and mass")
plt.show()
```

## מרכוז, פיזור וחסרים
פרשו את הממוצע, החציון וסטיית התקן; אל תניחו התפלגות נורמלית.

```python
print("ממוצע:",df["body_mass_g"].mean())
print("חציון:",df["body_mass_g"].median())
print("סטיית תקן:",df["body_mass_g"].std())
print("חסרים:",df["body_mass_g"].isna().sum())
```
