### P1 — רשימת Python ואינדקסים (10 נק׳)
כתבו ידנית טבלת אינדקסים 0–5, נבאו את שתי המדידות שיודפסו ואת אורך הרשימה, ואז השלימו את הקוד.
</div>

```python
measurements_list = [37.4, 42.1, 44.0, 39.7, 48.3, 41.5]

print(measurements_list[?])   # המדידה הראשונה
print(measurements_list[?])   # המדידה הרביעית
print(?(measurements_list))   # מספר איברים
```

<div dir="rtl" style="text-align:right; font-family:Arial; line-height:1.7">
**פירוש:** האם מדידה מספר 4 נמצאת באינדקס 4? מדוע?

### P2 — יצירת מערך NumPy (10 נק׳)
כעת ניצור מערך מאותן מדידות. השלימו את שם הספרייה ואת פונקציית היצירה. לפני הרצה: האם ערכי המדידה צפויים להשתנות?
</div>

```python
import numpy as np

a = np.?(measurements_list)
print(a)
print(?(a))                # מה סוג העצם?
```

<div dir="rtl" style="text-align:right; font-family:Arial; line-height:1.7">
**תובנה:** השוו את `type(a)` לסוג הרשימה. מה נשמר ומה השתנה?

### P3 — גישה חיובית ושלילית (10 נק׳)
הציגו את המדידה השנייה ואת המדידה האחרונה. הסבירו למה לא בטוח להשתמש ב־`a[6]` עבור שש מדידות.
</div>

```python
print(a[?])    # המדידה השנייה
print(a[?])    # המדידה האחרונה בעזרת אינדקס שלילי
```

<div dir="rtl" style="text-align:right; font-family:Arial; line-height:1.7">
**בדיקת גבולות:** מהי הסיבה העקרונית לשגיאה הצפויה ב־`a[6]`? אל תריצו את הפקודה השגויה.

### P4 — חיתוך מערך Slicing (15 נק׳)
סמנו בטבלת האינדקסים אילו ערכים ייבחרו בכל שורה. זכרו: הקצה stop אינו נכלל. השלימו.
</div>

```python
print(a[?:?])    # המדידות השנייה, השלישית והרביעית
print(a[:?])     # שתי המדידות הראשונות
print(a[?:])     # שתי המדידות האחרונות
```

<div dir="rtl" style="text-align:right; font-family:Arial; line-height:1.7">

### P5 — צורת מערך ותכונותיו (15 נק׳)
נבאו ידנית מהו מספר הממדים, צורת המערך ומספר האיברים. השלימו שמות תכונות; שימו לב ש־`dtype` מתאר את סוג המספרים.

```python
print(a.?)    # מספר ממדים
print(a.?)    # צורת המערך
print(a.?)    # מספר איברים
print(a.?)    # טיפוס הערכים
```

### P6 — יצירת מערכים נוספים (10 נק׳)
צרו רצף של מספרים אי־זוגיים מ־1 עד 7, ומערך באורך 4 שמלא אפסים. הכינו תחזית בכתב.

בדקו מדוע חוקר טוב מתעד תיקון נתונים ולא משנה בשקט את המקור. השלימו יצירת עותק ושינוי המדידה השלישית ל־44.5.

### P8 — הצצה למטריצה (10 נק׳)
לשלושה פינגווינים יש שתי מדידות: אורך מקור ועומק מקור (מ״מ). רשמו ידנית מספר שורות ועמודות, ואחר כך השלימו את ההדפסות. מטריצות לעומק נלמד במפגש 3.
</div>

```python
X = np.array([
    [37.4, 17.1],
    [42.1, 18.5],
    [44.0, 19.0]
])
print(X.?)         # shape
print(X.?)         # ndim
print(X[?, ?])     # אורך המקור של התצפית השנייה
```

### P9 — מאתרים טעויות מושגיות (5 נק׳)
התבוננו במערך הקצר. בחרו שתי טענות שגויות והסבירו מדוע הן שגויות לפני הרצה.

א. `shape` הוא `(4,1)`; ב. `a2[1:3]` מחזיר את האינדקסים 1 ו־2; ג. `a2[4]` מחזיר את המספר האחרון; ד. `ndim` הוא 1.
