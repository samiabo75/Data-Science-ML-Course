## משימה 1 — DataFrame ו־head
הקימו את הטבלה הנתונה. השלימו את שם התכונה המראה מספר שורות ועמודות, ואת מספר השורות שיוצגו ב־`head`. **דרישה:** הציגו בדיוק שתי שורות ראשונות. האם שאר השורות נמחקות?

</div>

```python
import pandas as pd
df = pd.DataFrame({
    "name": ["דניאל", "יובל", "שחר", "הדר"],
    "hours": [3, 5, 2, 4],
    "score": [72, 88, 65, 91]
})
print(df.?)
print(df.head(?))
```

## משימה 2 — Series לעומת DataFrame
בחרו פעם אחת את ציוני התלמידים בתור **Series**, ופעם שנייה בתור **DataFrame עם עמודה אחת**. מלאו את שמות העמודות במרכאות מתאימות. מה ההבדל בין `(4,)` ובין `(4,1)`?

</div>

```python
scores_series = df[?]
scores_table = df[[?]]
print(scores_series.shape)
print(scores_table.shape)
```

</div>

```python
features = df[[?, ?]]
print(features.shape)
```

## משימה 4 — תנאי וסינון
ליה רוצה לראות מי קיבל ציון **85 ומעלה**. השלימו את הסף ואת שם המסכה. ראשית נבאו רצף של ארבעה ערכי `True/False`; אחר כך בדקו אם שתי שורות נשארו.

</div>

```python
mask = df["score"] >= ?
print(mask.tolist())
selected = df[?]
print(selected[["name", "score"]])
```

## משימה 5 — סיכום מספרי והבנה
הקוד הנתון מחשב ממוצע, מינימום ומקסימום ציונים. **חובה לחשב קודם ממוצע ביד**. כתבו איזה מידע מספק הפלט, והאם הוא מוכיח ששעות חזרה גורמות לציון גבוה.
