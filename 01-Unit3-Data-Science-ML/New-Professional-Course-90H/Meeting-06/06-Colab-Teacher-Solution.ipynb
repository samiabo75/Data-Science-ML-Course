```python
import pandas as pdscores_demo = [60, 80, 100]total = 0for i in range(len(scores_demo)):    total = total + scores_demo[i]print("With for:", total / len(scores_demo))demo_df = pd.DataFrame({"score": scores_demo})print("With Pandas:", demo_df["score"].mean())
```

<div dir="rtl" lang="he" style="text-align:right;">**בדיקת הבנה:** `pd` הוא כינוי ל־Pandas; `DataFrame` היא טבלה; `{"score": scores_demo}` מגדיר עמודה בשם score; `demo_df["score"]` בוחרת את העמודה; הנקודה והסוגריים ב־`.mean()` מפעילים פעולה. מה יתרונה של עמודה בעלת שם על פני זכירת מספר העמודה? כעת התחילו את המשימות העצמאיות על נתוני ליה.</div>

<div dir="rtl" lang="he" style="text-align:right;">

## משימה 1 — טבלה וקריאה ראשונית
`shape` היא תכונה, ללא סוגריים. הממדים הם `(4,3)` — עמודת השם נספרת. `head(2)` מציגה שתי שורות, אך שומרת את ארבע שורות המקור.

</div>

```python
import pandas as pd
df = pd.DataFrame({
    "name": ["דניאל", "יובל", "שחר", "הדר"],
    "hours": [3, 5, 2, 4],
    "score": [72, 88, 65, 91]
})
print(df.shape)
print(df.head(2))
assert df.shape == (4, 3)
```

<div dir="rtl" lang="he" style="text-align:right;">

## משימה 2 — סוג התוצאה
בבחירת שם עמודה אחד מקבלים `Series` חד־ממדית `(4,)`. כאשר שם העמודה נמצא בתוך רשימה, מתקבלת `DataFrame` של 4 שורות ועמודה אחת, `(4,1)`.

</div>

```python
scores_series = df["score"]
scores_table = df[["score"]]
print(scores_series.shape)
print(scores_table.shape)
assert scores_series.shape == (4,) and scores_table.shape == (4, 1)
```

<div dir="rtl" lang="he" style="text-align:right;">

## משימה 3 — שתי עמודות
סוגריים כפולים מעבירים **רשימה של שמות עמודות**. מתקבלת `(4,2)`. הטבלה המקורית נשארת `(4,3)`. לשם חיזוי ציון לפי שעות, אין לבחור `score` כמאפיין — התרגיל כאן מלמד בחירה בלבד.

</div>

```python
features = df[["hours", "score"]]
print(features.shape)
print(df.shape)
assert features.shape == (4, 2) and df.shape == (4, 3)
```

<div dir="rtl" lang="he" style="text-align:right;">

## משימה 4 — מסכה וסינון
לפני הרצה: `[False, True, False, True]`. `mask` היא סדרה של אמת/שקר, **לא הטבלה המסוננת**. `df[mask]` מחזירה את יובל (88) ואת הדר (91). הסף `>=` כולל גם ערך השווה ל־85.

</div>

```python
mask = df["score"] >= 85
print(mask.tolist())
selected = df[mask]
print(selected[["name", "score"]])
assert mask.tolist() == [False, True, False, True]
assert selected["name"].tolist() == ["יובל", "הדר"]
```

<div dir="rtl" lang="he" style="text-align:right;">

## משימה 5 — מספרים והסקה
חישוב ידני: `(72+88+65+91)/4 = 316/4 = 79`. מינימום 65, מקסימום 91. ארבע תצפיות אינן מוכיחות ששעות תרגול **גורמות** לציון. החישוב הוא תיאור של הנתונים.

</div>

```python
print(df["score"].mean())
print(df["score"].min())
print(df["score"].max())
assert df["score"].mean() == 79.0
assert df["score"].min() == 65 and df["score"].max() == 91
```
