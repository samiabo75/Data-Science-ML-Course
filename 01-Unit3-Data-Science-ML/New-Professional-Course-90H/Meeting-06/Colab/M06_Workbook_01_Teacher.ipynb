# חוברת 1 — מבנה טבלת Pandas — פתרון מורה
הקמת טבלה עם שמות עמודות, בדיקת shape, head ו-columns

**הנחיות:** בכל משימה קראו תחילה את הדרישה, נבאו את הפלט, ואז השלימו את סימני השאלה. אל תריצו קוד עם ? לפני שהשלמתם אותו. כתבו הסבר במילים משלכם.

```python
import pandas as pd
records = {'name':['נועה','איתי','שיר','אור'], 'hours':[2,4,3,5], 'score':[65,82,74,91]}
df = pd.DataFrame(records)
print(df)
```

## משימה 1
צרו DataFrame ממילון של רשימות. מהן השורות והעמודות?

**לפני ההרצה:** מהו סוג התוצאה? מהו הפלט הצפוי? מדוע?

```python
df = pd.DataFrame(records)
```

**הסבר הפתרון למורה:** כל מפתח במילון הוא שם עמודה; כל רשימה היא הערכים בעמודה. מתקבלות ארבע שורות ושלוש עמודות.

**שאלת המשך בכיתה:** מה ישתנה אם נשנה ערך אחד בקלט?

**לפני ההרצה:** מהו סוג התוצאה? מהו הפלט הצפוי? מדוע?

```python
print(df.shape)
```

**הסבר הפתרון למורה:** התוצאה היא (4, 3): ארבע תצפיות ושלושה שדות.
