## משימה 3
הצג את הערך בשורה הראשונה ובעמודת species לפי מיקום, ואחר כך לפי תווית אינדקס רגילה.

**לפני ההרצה:** נבא את סוג התוצאה. אילו עמודות משתתפות בפעולה?

```python
print(df.iloc[0, df.columns.get_loc('species')])
```

**פירוק למורה:** iloc משתמשת במיקום. הפתרון מדגים בחירה לפי מיקום העמודה ולא מניח ש-species היא בהכרח עמודה 0.

## משימה 4
השלם בחירה לפי תווית בעמודת species עבור השורה בעלת התווית הראשונה.

**לפני ההרצה:** נבא את סוג התוצאה. אילו עמודות משתתפות בפעולה?

## משימה 5
בחר רק פינגווינים שאורך הסנפיר שלהם לפחות 200 מ"מ; הצג את מספר הנבחרים.

**לפני ההרצה:** נבא את סוג התוצאה. אילו עמודות משתתפות בפעולה?

```python
mask = df['flipper_length_mm'] >= 200
filtered = df[mask]
print(len(filtered))
```

**פירוק למורה:** השוואה מייצרת סדרה בוליאנית; df[mask] מסננת שורות. ערכי NaN אינם נבחרים.

**שאלת הבנה:** מה יקרה אם נשנה את ערך הסף או מספר השורות המוצגות?

## משימה 6
בחר שורות ש-flipper_length_mm לפחות 200 וגם species הוא Adelie; הסבר למה משתמשים ב-& ולא ב-and.

**לפני ההרצה:** נבא את סוג התוצאה. אילו עמודות משתתפות בפעולה?

```python
mask2 = (df['flipper_length_mm'] >= 200) & (df['species'] == 'Adelie')
print(df[mask2].shape)
```

**פירוק למורה:** & מבצע AND איבר-איבר בין שתי מסכות. חובה לשים כל השוואה בסוגריים.

**שאלת הבנה:** מה יקרה אם נשנה את ערך הסף או מספר השורות המוצגות?
