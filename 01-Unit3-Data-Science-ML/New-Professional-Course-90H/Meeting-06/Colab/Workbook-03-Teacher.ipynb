## משימה 1

**פירוק למורה:** isna מסמנת חסרים True; sum על ערכים בוליאניים סופרת True בכל עמודה.

## משימה 3
חשב ממוצע של מסת הגוף בקילוגרמים מתוך body_mass_g (גרמים). הצג גם את היחידות.

**לפני ההרצה:** נבא את סוג התוצאה. אילו עמודות משתתפות בפעולה?

```python
mean_kg = df['body_mass_g'].mean() / 1000
print(mean_kg)
```

**פירוק למורה:** mean מתעלמת מחסרים כברירת מחדל; חילוק ב-1000 ממיר גרמים לקילוגרמים. אין להסיק סיבתיות מהממוצע.
