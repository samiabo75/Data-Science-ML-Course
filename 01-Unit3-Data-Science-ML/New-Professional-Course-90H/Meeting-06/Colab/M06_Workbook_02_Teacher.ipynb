# חוברת 2 — בחירה וסינון — פתרון מורה
עמודה אחת מול טבלה, מסכה בוליאנית והבחנה בין שורה לעמודה

**הנחיות:** בכל משימה קראו תחילה את הדרישה, נבאו את הפלט, ואז השלימו את סימני השאלה. אל תריצו קוד עם ? לפני שהשלמתם אותו. כתבו הסבר במילים משלכם.

```python
import pandas as pd
records = {'name':['נועה','איתי','שיר','אור'], 'hours':[2,4,3,5], 'score':[65,82,74,91]}
df = pd.DataFrame(records)
print(df)
```

## משימה 1
בחרו את עמודת הציונים כסדרה חד־ממדית.

**לפני ההרצה:** מהו סוג התוצאה? מהו הפלט הצפוי? מדוע?

```python
print(df['score'])
print(df['score'].shape)
```

**הסבר הפתרון למורה:** בחירה בשם מחרוזת יחיד יוצרת Series בגודל (4,).

**שאלת המשך בכיתה:** מה ישתנה אם נשנה ערך אחד בקלט?

## משימה 2
בחרו את עמודת הציונים כטבלה.

**לפני ההרצה:** מהו סוג התוצאה? מהו הפלט הצפוי? מדוע?

```python
print(df[['score']])
print(df[['score']].shape)
```

**הסבר הפתרון למורה:** רשימת שמות בתוך הסוגריים יוצרת DataFrame בגודל (4,1).

**שאלת המשך בכיתה:** מה ישתנה אם נשנה ערך אחד בקלט?

## משימה 3
בחרו את שתי העמודות hours ו-score.

**לפני ההרצה:** מהו סוג התוצאה? מהו הפלט הצפוי? מדוע?

```python
print(df[['hours','score']])
```

**הסבר הפתרון למורה:** נשמרות ארבע שורות ושתי עמודות.

**שאלת המשך בכיתה:** מה ישתנה אם נשנה ערך אחד בקלט?

## משימה 4
הפיקו מסכה: ציון גדול או שווה 80; סננו את הטבלה.

**לפני ההרצה:** מהו סוג התוצאה? מהו הפלט הצפוי? מדוע?

```python
mask = df['score'] >= 80
print(mask)
print(df[mask])
```

**הסבר הפתרון למורה:** המסכה היא Series בוליאנית; שתי שורות מתקבלות בסינון.

**שאלת המשך בכיתה:** מה ישתנה אם נשנה ערך אחד בקלט?

## משימה 5
בחרו רק את שמות התלמידים ואת הציונים שעמדו בתנאי.

**לפני ההרצה:** מהו סוג התוצאה? מהו הפלט הצפוי? מדוע?

```python
print(df.loc[mask, ['name','score']])
```

**הסבר הפתרון למורה:** loc משלב בחירת שורות באמצעות mask ובחירת עמודות באמצעות רשימת שמות.

**שאלת המשך בכיתה:** מה ישתנה אם נשנה ערך אחד בקלט?
