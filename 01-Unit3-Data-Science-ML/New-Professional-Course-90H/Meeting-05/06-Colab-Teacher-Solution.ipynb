## משימה 1 — מדוע להמיר ל־NumPy?
`np.array` יוצרת מערך מספרי. הוספת סקלר מתבצעת איבר־איבר. לעומת זאת `hours_list + [1]` משרשרת איבר בסוף הרשימה ולא מוסיפה 1 לכל ערך.

</div>

```python
import numpy as np
hours_list = [2, 5, 4, 8]
hours_array = np.array(hours_list)
print(hours_array)
print(hours_array + 1)
assert (hours_array + 1).tolist() == [3, 6, 5, 9]
```

<div dir="rtl" lang="he" style="text-align:right;">

## משימה 2 — ממדים ואינדקסים
נועם: שתי תכונות ולכן `(2,)`. ארבעה תלמידים × שתי תכונות: `(4,2)`. שירה היא שורה 1; ציון העניין הוא עמודה 1. `students[2,0]` הוא מספר שעות האימון של תמר: 4.

</div>

```python
noam = np.array([2, 4])
students = np.array([[2, 4], [5, 8], [4, 4], [8, 7]])
print(noam.shape)
print(students.shape)
print(students[1, 1])
assert noam.shape == (2,) and students.shape == (4, 2)
assert students[1, 1] == 8 and students[2, 0] == 4
```

<div dir="rtl" lang="he" style="text-align:right;">

## משימה 3 — DataFrame ו־to_numpy
Pandas מקבלת רשימות ישירות. משתמשים ב־`to_numpy()` כאן כדי להדגים מעבר למטריצה מספרית, לא כשלב חובה בכל טבלה.

</div>

```python
import pandas as pd
df = pd.DataFrame({"hours": [2, 5, 4, 8], "interest": [4, 8, 4, 7]})
print(df.shape)
numbers = df[["hours", "interest"]].to_numpy()
print(numbers.shape)
assert df.shape == (4, 2) and numbers.shape == (4, 2)
```

## משימות 4–5 — מרחק אוקלידי ידני ובקוד
התלמיד החדש `(3,5)` ונועם `(2,4)`. הפרשים: `1,1`; ריבועים: `1,1`; סכום: `2`; מרחק: `√2≈1.414`. המרחק אינו הסתברות.

</div>

```python
new_student = np.array([3, 5])
noam = np.array([2, 4])
distance = np.linalg.norm(new_student - noam)
print(round(distance, 3))
assert round(distance, 3) == 1.414
```

<div dir="rtl" lang="he" style="text-align:right;">

## משימה 6 — broadcasting ומרחק לכל שורה
ב־`students - new_student` הווקטור החדש מופחת מכל שורה. `axis=1` מחשבת נורמה לכל שורה. הריבועים לפני השורש הם `[2,13,2,29]`; לכן המרחקים בקירוב `[1.414,3.606,1.414,5.385]`. יש **תיקו** בין נועם לתמר; `argmin` מחזירה את מיקום המינימום הראשון, 0. אין לומר שהשכן הראשון הוא הקרוב היחיד.

</div>

```python
diff = students - new_student
distances = np.linalg.norm(diff, axis=1)
closest_index = np.argmin(distances)
print(diff)
print(np.round(distances, 3))
print(closest_index)
print(students[closest_index])
assert np.array_equal(diff, np.array([[-1, -1], [2, 3], [1, -1], [5, 2]]))
assert np.allclose(distances**2, [2, 13, 2, 29])
assert closest_index == 0
```
