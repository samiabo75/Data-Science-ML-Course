## בעיה: האם אורך עלה אירוס מסייע לנבא את רוחבו?
רגרסיה ליניארית מנבאת **ערך מספרי רציף** ולא שם של קטגוריה.

```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score

df=pd.read_csv("https://raw.githubusercontent.com/arielb30/datasets/main/Iris.csv")
display(df.head())
```

## מגדירים מאפיין ומטרה
X כולל את אורך עלה הכותרת; y הוא רוחבו. הסוגריים הכפולים שומרים את X כטבלה דו־ממדית.

```python
X=df[["PetalLengthCm"]]
y=df["PetalWidthCm"]
plt.scatter(X["PetalLengthCm"],y,alpha=0.5)
plt.xlabel("Length (cm)")
plt.ylabel("Width (cm)")
plt.show()
```

## מאמנים ומנבאים
המודל מתאים ישר מן הדוגמאות: ŷ = wx + b.

```python
model=LinearRegression()
model.fit(X,y)
print("שיפוע:",model.coef_[0])
print("חיתוך:",model.intercept_)
new_flowers=pd.DataFrame({"PetalLengthCm":[2.5,4.0,5.5]})
print("תחזיות:",model.predict(new_flowers))
```

## האם התחזיות טובות?
נחשב R² על דוגמאות האימון, כדי להכיר את המדד. זה אינו מבחן על נתונים חדשים.

```python
pred=model.predict(X)
print("R² on training data:",r2_score(y,pred))
```
