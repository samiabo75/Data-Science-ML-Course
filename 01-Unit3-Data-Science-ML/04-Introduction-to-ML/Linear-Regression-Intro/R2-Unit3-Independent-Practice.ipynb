## תרגיל 1 — הבנה בלי מחשב
מודל א׳ קיבל R²=0.82 ומודל ב׳ קיבל R²=−0.35, כאשר שניהם נבדקו על אותן דוגמאות ועל אותה מטרה.

1. איזה מודל הצליח יותר לפי המדד?
2. מה פירוש ערך שלילי?
3. האם ערך 0.82 אומר ש־82% מהתחזיות היו נכונות בדיוק? נמקו.
4. האם אפשר להבטיח שהמודל יצליח באותה מידה על נתונים אחרים?

## תרגיל 2 — תחזית לעומת אמת
בטבלה הבאה מופיעים ציוני מבחן אמיתיים ותחזיות של מודל. הריצו את התא.

```python
import pandas as pd

results = pd.DataFrame({
    "actual": [60, 70, 80, 90],
    "predicted": [65, 68, 82, 87]
})
display(results)
```

1. באיזו שורה ההפרש בין התחזית לציון האמיתי הוא הגדול ביותר?
2. חשבו את ממוצע הציונים האמיתיים. מה יהיה הניחוש הקבוע של מודל שתמיד מנבא את הממוצע?
3. האם כל תחזית היא בדיוק הערך האמיתי?

```python
# השלימו את החסר:
mean_actual = results["actual"].mean()
print("הממוצע:", mean_actual)
# כתבו כאן כיצד תציגו את ההפרש המוחלט לכל שורה
```

## תרגיל 3 — מדידת R² בספריית sklearn
הריצו והשלימו את התשובות אחרי שתקבלו את התוצאה.

```python
from sklearn.metrics import r2_score

score = r2_score(results["actual"], results["predicted"])
print("R² =", round(score, 3))
```

1. האם הערך גדול מאפס?
2. האם המודל טוב יותר מחיזוי קבוע לפי ממוצע ערכי האמת, מבחינת R²?
3. כתבו במילים מה אפשר להסיק ומה עדיין אי אפשר להסיק.

## תרגיל 4 — משנים תחזיות ובודקים
שנו **רק** את התחזיות בתא הבא לערכים רחוקים מאוד מהציונים האמיתיים, ונסו להגיע ל־R² שלילי.

```python
bad_predictions = [65, 68, 82, 87]  # שנו את הערכים
print("R² חדש:", r2_score(results["actual"], bad_predictions))
```

1. אילו ערכים בחרתם?
2. האם קיבלתם תוצאה שלילית?
3. הסבירו למה המודל החדש גרוע יותר מניחוש הממוצע.

## תרגיל 5 — מסקנה מקצועית
חוקר מציג R²=0.95 ומכריז: 'המודל שלי חוזה מצוין כל תוצאה עתידית'.

כתבו שלושה משפטים: מה המדד תומך בו, איזו בדיקה חסרה (Train/Test), ומה הייתם ממליצים לחוקר לבצע.
