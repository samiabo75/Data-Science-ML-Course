## מושגים וידע קודם
אינדקסים מתחילים באפס: pixels[row,column,channel]. ערכי uint8 הם 0–255; פעולות חיסור על uint8 עלולות לגלוש ולכן לפני חישובי מרחק משתמשים ב־float. שינוי עושים על copy. reshape משנה פריסת מערך ולא משנה לבדו את התוכן החזותי. אין צורך לקרוא לתמונה מערך 'עמודות ושורות' בסדר הפוך.

## הסבר מתמטי
תמונת RGB בגובה H וברוחב W מיוצגת במערך H×W×3. ממוצע ערוץ אדום הוא סכום ערכי R חלקי H×W. וקטור של ממוצעי R,G,B הוא ייצוג בן שלושה מאפיינים; הוא אינו שומר את מיקום העצמים. לצורך גווני אפור כאן משתמשים בממוצע פשוט של הערוצים,לא בהמרת luminance משוקללת מקצועית.

## מערך ותצוגה
כל הפיקסלים כאן מומצאים במפורש; אין צילום אישי ואין Dataset חיצוני.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from PIL import Image
# מערך RGB מומצא 2x2, ערוצים R,G,B בסדר זה
pixels = np.array([
 [[200,20,10],[180,40,30]],
 [[160,60,50],[140,80,70]]
], dtype=np.uint8)
print("Shape:",pixels.shape,"Dtype:",pixels.dtype)
print("Top-left RGB:",pixels[0,0])
plt.figure();plt.imshow(pixels);plt.title("Synthetic RGB image");plt.axis("off");plt.show()
```

## קובץ והעתקה
יוצרים קובץ דוגמה מתוך הנתונים המומצאים,טוענים אותו ושומרים עותק עבודה. בפרויקט אמיתי בודקים זכויות שימוש בתמונה.

```python
Image.fromarray(pixels).save("synthetic_rgb.png")
loaded = np.asarray(Image.open("synthetic_rgb.png").convert("RGB"))
edited = loaded.copy()
edited[0,0] = [0,255,0]
display(pd.DataFrame(loaded[:,:,0]))
plt.figure();plt.imshow(edited);plt.title("Edited copy — one green pixel");plt.axis("off");plt.show()
```

## מאפיינים
ממוצעים מחושבים על שני צירי הפיקסלים,לא על ציר הערוצים. מציגים גם חישוב סכום/מספר פיקסלים לערוץ אחד.

```python
float_pixels = loaded.astype(float)
mean_r_manual = float_pixels[:,:,0].sum() / (loaded.shape[0]*loaded.shape[1])
features = float_pixels.mean(axis=(0,1))
print("Mean R manual:",mean_r_manual,"RGB features:",features)
gray = float_pixels.mean(axis=2)
plt.figure();plt.imshow(gray,cmap="gray",vmin=0,vmax=255)
plt.title("Simple channel-average grayscale");plt.axis("off");plt.show()
# היפוך מיקומי שורות שומר את הממוצעים אך משנה מיקום
reordered = loaded[::-1].copy()
print("Same mean, different layout:",np.allclose(features,reordered.astype(float).mean(axis=(0,1))))
```
