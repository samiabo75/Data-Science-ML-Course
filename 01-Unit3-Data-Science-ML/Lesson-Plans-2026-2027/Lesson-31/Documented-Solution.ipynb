המחברת היא התאמה דידקטית חדשה. אין קובץ מקור חדש מהמרצה ולכן אין Exact-Original מלאכותי. אין לייחס לה אישור כיסוי מחוון עדכני.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from PIL import Image
# מערך RGB מומצא 2x2, ערוצים R,G,B בסדר זה
pixels = np.array([
 [[20,180,40],[40,160,60]],
 [[60,140,80],[80,120,100]]
], dtype=np.uint8)
```

## תצוגה וקובץ
הערוץ הירוק הוא אינדקס 1. המקור loaded נפרד מן העותק edited.

```python
print(pixels.shape,pixels.dtype,pixels[1,0])
Image.fromarray(pixels).save("synthetic_practice.png")
loaded=np.asarray(Image.open("synthetic_practice.png").convert("RGB"))
edited=loaded.copy();edited[1,1]=[0,0,255]
plt.figure();plt.imshow(edited);plt.title("Practice edited copy");plt.axis("off");plt.show()
```

## חילוץ
ממוצעים אינם שומרים סידור מרחבי. כל גרסאות אותה תמונת מקור צריכות להישאר באותו צד החלוקה כדי למנוע זליגה; מספר תמונות צעצוע אינו מבחן הכללה.

```python
float_pixels=loaded.astype(float)
mean_g_manual=float_pixels[:,:,1].sum()/4
features=float_pixels.mean(axis=(0,1))
print("Green manual:",mean_g_manual,"RGB means:",features)
gray=float_pixels.mean(axis=2)
reordered=loaded[::-1].copy()
print(np.allclose(features,reordered.astype(float).mean(axis=(0,1))))
plt.figure();plt.imshow(gray,cmap="gray",vmin=0,vmax=255)
plt.title("Practice channel average");plt.axis("off");plt.show()
```
