## למה לומדים את זה עכשיו?
אותה הזמנה הועתקה פעמיים; שתי הזמנות שונות יכולות להיראות זהות.
**בתהליך הפרויקט:** איכות והכנה.
**מסגרת פעילה:** 2 שעות לימוד — 90 דקות, כולל דוגמה ותרגול עצמאי. המורה בוחר את משימות הליבה שבמדריך; שאר הפעילויות לחזרה או לעיון, ללא תוספת שעות.

```python
import pandas as pd
import numpy as np
raw = pd.DataFrame({
 "order_id":[201,202,203,203,204,205,206],
 "ordered_at":["2026-02-01","2026-02-02","2026-02-03","2026-02-03","2026-02-04","2026-02-05","2026-02-06"],
 "items":[1,2,3,3,2,1,2],"unit_price":[20.,10.,5.,5.,10.,20.,10.],
 "duration_min":[5.,9.,12.,12.,0.,-1.,None]
})
```

## ניקוי לפי חוזה
בחוזה התרגול 0 אינו תקין, בניגוד למקרים שבהם משך אפס אפשרי. לא מוחקים הזמנות שונות בגלל שוויון מדידות.

```python
work = raw.copy(deep=True)
display(work[work.duplicated("order_id",keep=False)])
work = work.drop_duplicates("order_id").copy()
work.loc[work["duration_min"]<=0,"duration_min"] = np.nan
eligible = work.dropna(subset=["duration_min"]).copy()
print(len(raw),len(work),len(eligible))
```

## מאפיינים
המאפיינים מחושבים ממידע זמין; אין תכונת יעד בתוך הקלט. אם רוצים לחזות הזמנות עתידיות יש לתכנן Train מוקדם ו־Test מאוחר,גם בתוך האימות.

```python
work["ordered_at"] = pd.to_datetime(work["ordered_at"],errors="raise")
work["order_month"] = work["ordered_at"].dt.month
work["order_weekday"] = work["ordered_at"].dt.dayofweek
work["order_value"] = work["items"]*work["unit_price"]
display(work)
```
