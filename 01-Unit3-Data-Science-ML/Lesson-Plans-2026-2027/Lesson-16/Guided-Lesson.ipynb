## פתיחה — מה ייחשב חיזוי בסיסי?
דודו רוצה לחזות מסה. אפשר בעתיד להשוות מודל לדרך פשוטה שמנחשת אותו ערך לכל תצפית, למשל ממוצע המסות שנלמד מנתוני אימון. היום נבין מה משמעות ממוצע ומה הוא מסתיר. לא בונים מודל ולא מודדים הצלחה.

בפרויקט אפשר לתאר את כל מאגר החקר, אבל כשמשתמשים במדד כחלק ממודל, השלמת חסרים או קו בסיס — לומדים אותו מנתוני האימון בלבד. נתוני הבדיקה מיועדים להערכה.

## 1. דוגמה קטנה לחישוב ביד
חמישה מחירים מומצאים בשקלים: 2000,2200,2200,2400,2600.

- ממוצע: סכום 11400 חלקי 5 = 2280 ₪.
- חציון: מסדרים מהנמוך לגבוה ובוחרים את הערך האמצעי — 2200 ₪.
- שכיח: הערך המופיע הכי הרבה — 2200 ₪, פעמיים.

כל מדד עונה לשאלה אחרת. הממוצע אינו חייב להופיע כערך ברשימה. אין להחליט שמדד אחד תמיד 'הנכון'.

```python
import pandas as pd
demo_df = pd.DataFrame({"price_nis": [2000, 2200, 2200, 2400, 2600]})
print("ממוצע בשקלים:", demo_df["price_nis"].mean())
print("חציון בשקלים:", demo_df["price_nis"].median())
print("ערכים שכיחים בשקלים:")
print(demo_df["price_nis"].mode())
```

mean, median ו־mode הן פעולות על העמודה. mode מחזירה Series של ערכים שכיחים; המספר 0 שמוצג משמאל הוא אינדקס בתוצאה, לא השכיח עצמו ולא שכיחותו. קוראים את הערך בצד ימין. אם יש שוויון בשכיחות המרבית עשויים לחזור כמה ערכים.

## 2. חציון למספר זוגי
מחירים מומצאים: 2000,2200,2400,2600. שני האמצעיים הם 2200 ו־2400, ולכן החציון (2200+2400)/2=2300 ₪. מיון הכרחי לפתרון ידני, אך median מבצעת את החישוב בלי שנצטרך לשנות סדר טבלה.

```python
even_df = pd.DataFrame({"price_nis": [2600, 2000, 2400, 2200]})
print("חציון ארבעה מחירים בשקלים:", even_df["price_nis"].median())
```

## 3. ניסוי מבוקר — ערך גדול אחד
נשנה בדוגמה נפרדת את 2600 ל־10000. זו הדמיית הוראה, לא תיקון נתוני אמת. לפני קוד: איזה מדד ישתנה יותר?

```python
trial_df = pd.DataFrame({"price_nis": [2000, 2200, 2200, 2400, 10000]})
print("ממוצע ניסוי:", trial_df["price_nis"].mean())
print("חציון ניסוי:", trial_df["price_nis"].median())
print("שכיחי ניסוי:")
print(trial_df["price_nis"].mode())
print("ערכי המקור:")
print(demo_df)
```

הממוצע עולה ל־3760, החציון והשכיח נשארים 2200. שינוי 7400 בערך אחד שינה את הממוצע ב־7400/5=1480. החציון עמיד יותר לשינוי קיצוני מסוג זה, אך אינו חסין לכל שינוי בנתונים. אין למחוק ערך רק כי שינה ממוצע.

## 4. נתונים אמיתיים — מסת פינגווינים

```python
url = "https://raw.githubusercontent.com/samiabo75/Data-Science-ML-Course/main/Datasets/penguins.csv"
df = pd.read_csv(url)
print("מספר הרשומות:", len(df))
print("מספר מסות ידועות:", df["body_mass_g"].count())
print("ממוצע מסה בגרם:", df["body_mass_g"].mean())
print("חציון מסה בגרם:", df["body_mass_g"].median())
print("מסות שכיחות בגרם:")
print(df["body_mass_g"].mode())
```

count סופרת ערכים שאינם חסרים בעמודה. len(df) סופרת רשומות. בגרסה זו 344 רשומות, אך רק 342 מסות ידועות. כברירת מחדל הפקודות כאן מדלגות על חסרים; אין חלוקה ב־344 לצורך ממוצע המסות הידועות, ואין מילוי חסרים באפס.

השכיח של מדידה מעוגלת תלוי ברזולוציית הרישום. אין להתייחס למסה השכיחה כאילו היא המסה היחידה או ה'נכונה'.
