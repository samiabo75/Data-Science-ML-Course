## למה לומדים את זה עכשיו?
מוצר יקר במיוחד משנה את המחיר הממוצע: האם החציון משתנה באותה מידה?
**בתהליך הפרויקט:** חקר נתונים.
**מסגרת פעילה:** 1.5 שעות לימוד — 67.5 דקות, כולל דוגמה ותרגול עצמאי. המורה בוחר את משימות הליבה שבמדריך; שאר הפעילויות לחזרה או לעיון, ללא תוספת שעות.

### 1–2. חישוב והשפעת קצה
סכום המקור 65 ולכן ממוצע 13, חציון 12 ושכיח 12. סכום הניסוי 90 ולכן ממוצע 18, חציון ושכיח 12. התוספת 25 בערך אחד העלתה ממוצע ב־25/5=5; זו דוגמה לעמידות יחסית של החציון, לא הבטחה שאינו משתנה לעולם.

```python
import pandas as pd
original_df = pd.DataFrame({"value": [10, 12, 12, 14, 17]})
experiment_df = pd.DataFrame({"value": [10, 12, 12, 14, 42]})
print("מקור ממוצע:", original_df["value"].mean())
print("מקור חציון:", original_df["value"].median())
print("מקור שכיחים:")
print(original_df["value"].mode())
print("ניסוי ממוצע:", experiment_df["value"].mean())
print("ניסוי חציון:", experiment_df["value"].median())
print("ניסוי שכיחים:")
print(experiment_df["value"].mode())
display(original_df)
```

### 3. זוגי
לאחר מיון 10,12,16,18; חציון (12+16)/2=14. median מחשבת גם בלי מיון מפורש של המקור.

```python
even_df = pd.DataFrame({"value": [18, 10, 16, 12]})
print("חציון:", even_df["value"].median())
```

### 4. אורכי סנפיר
344 רשומות, 342 ערכים ידועים. יחידה מ״מ; החסרים אינם אפס והמדדים כאן מדלגים עליהם כברירת מחדל.

```python
url = "https://raw.githubusercontent.com/samiabo75/Data-Science-ML-Course/main/Datasets/penguins.csv"
penguins_df = pd.read_csv(url)
print("רשומות:", len(penguins_df))
print("אורכי סנפיר ידועים:", penguins_df["flipper_length_mm"].count())
print("ממוצע בממ:", penguins_df["flipper_length_mm"].mean())
print("חציון בממ:", penguins_df["flipper_length_mm"].median())
print("שכיחים בממ:")
print(penguins_df["flipper_length_mm"].mode())
```

### 5. Adelie
151 אורכי סנפיר ידועים מתוך 152 רשומות. זו תת־קבוצה; השוואת ממוצעים תיאורית אינה מבחן מובהקות, הסקה סיבתית או הבטחה לגבי אוכלוסייה.

```python
adelie_mask = penguins_df["species"] == "Adelie"
adelie_df = penguins_df[adelie_mask]
print("ידועים בקבוצה:", adelie_df["flipper_length_mm"].count())
print("ממוצע בקבוצה בממ:", adelie_df["flipper_length_mm"].mean())
print("חציון בקבוצה בממ:", adelie_df["flipper_length_mm"].median())
```

### 6. אי שכיח
Biscoe היא הקטגוריה השכיחה בקובץ. mode מתאימה לשמות; לממוצע שמות אין משמעות. המקור נשמר.

```python
print("אי שכיח:")
print(penguins_df["island"].mode())
print("ממדי המקור:", penguins_df.shape)
```

### 7. דוגמת תיעוד
קבוצה: כל הרשומות עם אורך סנפיר ידוע בגרסת הקורס; מדידה כמותית רציפה במ״מ. הממוצע והחציון מופיעים בפלט, ומתארים מרכז של 342 מדידות, לא את הפיזור שלהן. במאגר יש כמה מינים ו־2 חסרי מדידה, ולכן אין לתאר מדד יחיד כערך טיפוסי מחייב לכל מין או לכל פינגווין.

דמיון או הבדל בין ממוצע וחציון אינם הוכחת נורמליות. אם ממוצע משמש פרמטר הכנה או קו בסיס במודל, מחשבים אותו רק בנתוני האימון כדי למנוע דליפת מידע.
