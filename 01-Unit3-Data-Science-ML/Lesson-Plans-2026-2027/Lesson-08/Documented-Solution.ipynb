## פתרון מתועד — מסת גוף

### 1. טעינה ומבנה
בגרסה שנבדקה 344 שורות ו־7 עמודות.

```python
import pandas as pd
url = "https://raw.githubusercontent.com/samiabo75/Data-Science-ML-Course/main/Datasets/penguins.csv"
penguins_df = pd.read_csv(url)
print("ממדים:", penguins_df.shape)
print("עמודות:", penguins_df.columns)
```

### 2. Series
בחירת שם יחיד מחזירה סדרה באורך 344. היחידה גרם.

```python
masses = penguins_df["body_mass_g"]
display(masses.head(4))
print("ממדי הסדרה:", masses.shape)
```

### 3. DataFrame
רשימת שם יחיד שומרת תוצאה דו־ממדית: 344×1. הערכים זהים, המבנה שונה.

```python
mass_table = penguins_df[["body_mass_g"]]
display(mass_table.head(4))
print("ממדי הטבלה:", mass_table.shape)
```

### 4. קלט מועמד
שתי מדידות במ״מ. כל 344 השורות נשמרו, כולל חסרים, ובאותו סדר.

```python
predictor_columns = ["flipper_length_mm", "bill_length_mm"]
predictors = penguins_df[predictor_columns]
display(predictors.head(4))
print("ממדים:", predictors.shape)
```

### 5. יעד
המסה בגרמים היא יעד מספרי. אין לכלול את היעד עצמו בקלט, משום שהתשובה תיחשף. טרם טופלו החסרים, טרם חולקו נתוני אימון ובדיקה, ולא נבנה או הוערך מודל. אין חובה ללמד את שלבי ההמשך בקוד היום.

```python
target_mass = penguins_df["body_mass_g"]
print("ממדי היעד:", target_mass.shape)
```

### 6. המקור
בחירות לקריאה לא מחקו שדות מהמקור ולא טיפלו בחסרים. איננו משנים את הערכים שנבחרו בשיעור זה.

```python
print("ממדי המקור:", penguins_df.shape)
print("עמודות המקור:", penguins_df.columns)
```

### 7. תיקון
בוחרים רשימת שמות ולא זוג שמות המופרדים בפסיק בלבד. הסוגריים הפנימיים יוצרים את הרשימה.

```python
correct_selection = penguins_df[["flipper_length_mm", "body_mass_g"]]
display(correct_selection.head(2))
```

### רשות
344×3 לפי הסדר שנבחר. species עשויה להיות יעד בשאלת סיווג; body_mass_g יעד בשאלת מסה. תפקיד העמודה תלוי בשאלת המחקר, ואין להזין אוטומטית את כל עמודות טבלת העיון כמאפיינים.

```python
review_table = penguins_df[["species", "bill_depth_mm", "body_mass_g"]]
print("ממדים:", review_table.shape)
display(review_table.head(3))
```
