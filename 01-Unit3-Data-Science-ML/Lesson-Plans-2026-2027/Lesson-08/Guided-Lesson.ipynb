## פתיחה — מה נכנס למודל ומה רוצים לחזות?
אם נרצה לסווג species לפי אורך ועומק המקור, נצטרך לבחור את שתי המדידות ואת היעד בנפרד. לא כל עמודה מתאימה אוטומטית לקלט. בחירת העמודות היום היא ארגון נתונים בלבד, לא הוכחה שאלו מאפיינים מיטביים ולא אימון מודל.

```python
import pandas as pd
url = "https://raw.githubusercontent.com/samiabo75/Data-Science-ML-Course/main/Datasets/penguins.csv"
df = pd.read_csv(url)
print("ממדי המקור:", df.shape)
print("עמודות:", df.columns)
display(df.head(3))
```

## 1. בחירת עמודה אחת לפי שמה
כותבים את שם העמודה במירכאות בתוך סוגריים מרובעים: `df["bill_length_mm"]`. שם העמודה חייב להתאים בדיוק לרשימת העמודות, כולל אותיות וקווים תחתונים.

```python
bill_lengths = df["bill_length_mm"]
display(bill_lengths.head())
print("ממדי הסדרה:", bill_lengths.shape)
```

התוצאה היא Series — סדרת ערכים חד־ממדית עם אינדקס. `(344,)` פירושו אורך 344, לא 344 עמודות. הפסיק הוא חלק מהצגת tuple בעל איבר אחד; אין צורך ללמוד tuple לעומק.

Series אינה רשימת Python, אף שגם בה יש רצף ערכים. ברירת המחדל של head מציגה חמש רשומות; אינדקסי השורות נשמרים.

## 2. אותה עמודה כטבלה
אם מעבירים רשימת שמות, מקבלים DataFrame. הרשימה נמצאת בתוך פעולת הבחירה: זוג הסוגריים הפנימי יוצר רשימה, והחיצוני מבצע את הבחירה מתוך df.

```python
bill_table = df[["bill_length_mm"]]
display(bill_table.head())
print("ממדי הטבלה:", bill_table.shape)
```

כאן `(344,1)` פירושו 344 שורות ועמודה אחת. בשתי התוצאות אותם ערכי אורך מקור, אך מבנה אחר.

| קוד | תוצאה | ממדים בגרסה שנבדקה |
|---|---|---|
| df["bill_length_mm"] | Series | (344,) |
| df[["bill_length_mm"]] | DataFrame | (344,1) |

עצירת הבנה: האם יותר סוגריים הוסיפו פינגווינים? לא — השתנה סוג המבנה בלבד.

## 3. שתי עמודות
מעבירים רשימת שני שמות. כל השורות נשארות, ורק מספר העמודות מצטמצם. סדר השמות ברשימה קובע את סדר העמודות בתוצאה.

```python
measurement_columns = ["bill_length_mm", "bill_depth_mm"]
measurements = df[measurement_columns]
display(measurements.head())
print("ממדי המדידות:", measurements.shape)
```

התוצאה היא 344×2. אפשר לכתוב גם `df[["bill_length_mm", "bill_depth_mm"]]`. שתי הדרכים בוחרות את אותן עמודות; הרשימה בשם משתנה יכולה לשפר קריאות.

לא נבחרו שורות לפי תנאי; כל הרשומות, כולל חסרים, נשארו. אין להסיק שהטבלה מוכנה למודל.
