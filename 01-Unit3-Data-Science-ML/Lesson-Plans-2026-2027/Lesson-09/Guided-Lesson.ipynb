## 1. חימום קטן — מעמודה לתנאי
שלוש תצפיות מומצאות לצורך הבנת המנגנון בלבד.

```python
import pandas as pd
tiny_df = pd.DataFrame({
    "item_id": ["A", "B", "C"],
    "price_nis": [90, 120, 100]
})
display(tiny_df)
```

לפני קוד: בתקציב 100 ₪, A ו־C עומדים בתנאי, B לא. בפייתון למדנו לבדוק ערך יחיד; כאן משווים כל ערך בעמודה לאותו סף.

```python
within_budget = tiny_df["price_nis"] <= 100
display(within_budget)
```

זו Series בוליאנית, המכונה מסכה: לפי האינדקסים 0,1,2 מתקבלים True,False,True. המסכה מכילה תוצאת בדיקה לכל שורה — היא עדיין אינה טבלת השורות שנבחרו.

```python
affordable_df = tiny_df[within_budget]
display(affordable_df)
```

הבחירה שומרת את כל השדות של A ו־C יחד, ואת האינדקסים המקוריים 0 ו־2. אינדקס 2 אינו אומר שיש שלוש שורות בתוצאה. אין צורך לאפס אינדקס כעת.

שתי פעולות נפרדות: יוצרים תנאי על עמודה; משתמשים בתוצאותיו לבחירת שורות. אין צורך בלולאה או ב־if על Series שלמה.

## 2. אותו רעיון בנתונים אמיתיים

```python
url = "https://raw.githubusercontent.com/samiabo75/Data-Science-ML-Course/main/Datasets/penguins.csv"
df = pd.read_csv(url)
print("ממדי המקור:", df.shape)
display(df.head(3))
```

העמודה species מתארת מין טקסונומי. משתמשים ב־`==` לבדיקת שוויון ובמירכאות לערך הטקסט. Adelie חייב להיכתב כפי שהוא בקובץ.

```python
adelie_mask = df["species"] == "Adelie"
display(adelie_mask.head())
adelie_df = df[adelie_mask]
print("ממדי קבוצת Adelie:", adelie_df.shape)
display(adelie_df.head())
```

בגרסה שנבדקה מתקבל (152,7): מספר השורות קטן, כל שבע העמודות נשארות. head של המסכה מציג רק חמש בדיקות; הוא אינו מוכיח שכל הבדיקות במאגר True.

אפשר לכתוב לאחר הבנה גם בשורה אחת: `df[df["species"] == "Adelie"]`. זו אינה חובה לזיכרון; שתי השורות עם שם למסכה מקלות על קריאה.
