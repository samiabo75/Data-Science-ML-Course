## פתיחה — מספר הדוגמאות או המדידה הממוצעת?
לקראת חיזוי מסה נרצה לשאול גם כמה דוגמאות יש בכל מין וגם מהי המסה הממוצעת בכל מין. אלו שאלות שונות. ספירה מתארת גודל קבוצה; ממוצע מתאר מרכז של מדידות ידועות בה.

הבדלים בממוצעים עשויים לעורר שאלת מחקר, אבל אינם מבטיחים סיווג מושלם, סיבתיות או דיוק חיזוי.

## 1. דוגמה מומצאת לחישוב ידני
מחירי מוצרים דמיוניים: בקבוצה A המחירים 100,120,140 ₪; בקבוצה B המחירים 200,220 ₪. ב־A יש יותר תצפיות, אבל הממוצע נמוך יותר: 120 לעומת 210 ₪. קבוצה גדולה אינה בהכרח בעלת מדידה ממוצעת גבוהה.

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
demo_df = pd.DataFrame({
    "group": ["A", "A", "A", "B", "B"],
    "price_nis": [100, 120, 140, 200, 220]
})
mean_prices = demo_df.groupby("group")["price_nis"].mean()
known_prices = demo_df.groupby("group")["price_nis"].count()
print("מספר מחירים ידועים לפי קבוצה:")
print(known_prices)
print("מחיר ממוצע בשקלים לפי קבוצה:")
print(mean_prices)
```

### פירוק groupby
1. `groupby("group")`: מחלק לפי קטגוריית group.
2. `["price_nis"]`: בוחר את המדידה לסיכום.
3. `mean()`: מחשב ממוצע בכל קבוצה.

התוצאה היא Series עם שמות קבוצות כאינדקס. זו טבלת סיכום, לא רשומות המקור. count סופרת את המדידות הידועות, לא בהכרח כל הרשומות בקבוצה.

```python
plt.figure(figsize=(7, 4))
sns.barplot(data=demo_df, x="group", y="price_nis", estimator="mean", errorbar=None, color="steelblue")
plt.title("Mean teaching price by group")
plt.xlabel("Group")
plt.ylabel("Mean price (NIS)")
plt.tight_layout()
plt.show()
```

גובה העמודות 120 ו־210 ₪. זה אינו גרף הספירות 3 ו־2. estimator="mean" מציין במפורש את הסיכום; errorbar=None מסתיר פסי אי־ודאות כדי להתמקד היום בממוצע. אין בכך טענה שאין אי־ודאות או פיזור. הפקודה מחשבת מן הנתונים הגולמיים; אין צורך להעביר לה טבלת ממוצעים.

## 2. מסה לפי מין בנתונים האמיתיים

```python
url = "https://raw.githubusercontent.com/samiabo75/Data-Science-ML-Course/main/Datasets/penguins.csv"
df = pd.read_csv(url)
mean_mass = df.groupby("species")["body_mass_g"].mean()
known_mass_counts = df.groupby("species")["body_mass_g"].count()
print("מסות ידועות לפי מין:")
print(known_mass_counts)
print("ממוצע מסה בגרם לפי מין:")
print(mean_mass)
```

כל ממוצע חושב ממדידות המסה הידועות בקבוצתו. בגרסה זו: Adelie — 151 מסות ידועות, Chinstrap — 68, Gentoo — 123. מספר הרשומות הכולל בקבוצות הוא 152,68,124; החסרים אינם אפס.

נציג שני גרפים באותו סדר קטגוריות כדי להקל על ההשוואה. סדר הקטגוריות הוא בחירת תצוגה, לא דירוג איכות.

```python
species_order = ["Adelie", "Chinstrap", "Gentoo"]
plt.figure(figsize=(7, 4))
sns.countplot(data=df, x="species", order=species_order, color="teal")
plt.title("Number of observations by species")
plt.xlabel("Species")
plt.ylabel("Number of observations")
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(7, 4))
sns.barplot(data=df, x="species", y="body_mass_g", order=species_order, estimator="mean", errorbar=None, color="steelblue")
plt.title("Mean body mass by species")
plt.xlabel("Species")
plt.ylabel("Mean body mass (g)")
plt.tight_layout()
plt.show()
```

## 3. קוראים הבדל בין שאלות
Adelie היא הקבוצה הגדולה בקובץ, בעוד Gentoo בעלת המסה הממוצעת הגבוהה ביותר. אין סתירה: יחידת Y והשאלה שונות. Barplot אינה מראה שכל Gentoo כבדה מכל Adelie, ואינה מציגה את הפיזור בתוך המין.

אין לחשב ממוצע של שלושת ממוצעי הקבוצות ולהציגו אוטומטית כממוצע כל הרשומות. גדלי הקבוצות והמספרים הידועים שונים; נדרש משקל מתאים. היום אפשר לחשב את ממוצע העמודה המקורית ישירות.
