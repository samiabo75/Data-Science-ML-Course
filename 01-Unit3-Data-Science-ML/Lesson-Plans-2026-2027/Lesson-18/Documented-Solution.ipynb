### 1–3. טבלה
island היא קטגוריה שמית; flipper_length_mm מדידה רציפה במ״מ. ספירה ב־Y היא מספר תצפיות; ממוצע ב־Y הוא אורך במ״מ. count על מדידה סופרת רק ידועים.

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
url = "https://raw.githubusercontent.com/samiabo75/Data-Science-ML-Course/main/Datasets/penguins.csv"
penguins_df = pd.read_csv(url)
mean_flipper_by_island = penguins_df.groupby("island")["flipper_length_mm"].mean()
known_flippers_by_island = penguins_df.groupby("island")["flipper_length_mm"].count()
print("ממוצע בממ לפי אי:")
print(mean_flipper_by_island)
print("מספר אורכים ידועים לפי אי:")
print(known_flippers_by_island)
```

בגרסה זו מספר הידועים: Biscoe=167, Dream=124, Torgersen=51. מספר הרשומות בקבוצות: 168,124,52. count אינו מספר חסרים ואינו בהכרח גובה Countplot.

```python
island_order = ["Biscoe", "Dream", "Torgersen"]
plt.figure(figsize=(7, 4))
sns.countplot(data=penguins_df, x="island", order=island_order, color="teal")
plt.title("Number of observations by island")
plt.xlabel("Island")
plt.ylabel("Number of observations")
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(7, 4))
sns.barplot(data=penguins_df, x="island", y="flipper_length_mm", order=island_order, estimator="mean", errorbar=None, color="steelblue")
plt.title("Mean flipper length by island")
plt.xlabel("Island")
plt.ylabel("Mean flipper length (mm)")
plt.tight_layout()
plt.show()
```

### 6. פירוש
Biscoe הגדולה במספר הרשומות וגם בעלת אורך הסנפיר הממוצע הגבוה בנתונים האלה. אין כלל שקבוצה גדולה תמיד בעלת ממוצע גבוה — זהו ממצא כאן בלבד.

ממוצע אינו מבטיח הפרדה של כל התצפיות, אינו מציג את הפיזור ואינו מוכיח שהאי גרם להבדל. הרכב המינים והדגימה עשויים להיות קשורים. errorbar=None מסתירה פסי אי־ודאות; היא אינה מבחן מובהקות.

### 7. כרטיס לדוגמה
שאלה: מה אורך הסנפיר הממוצע בכל אי בקובץ? יחידה מ״מ, סיכום mean למדידות הידועות; קבוצות לפי island. ממוצעים וספירות ידועות מופיעים בטבלה. מגבלה: חסרים והרכב מינים; אין מסקנה סיבתית או טענה לכלל האוכלוסייה. טבלת הסיכום מייצגת איים, לא פינגווינים בודדים.

### רשות
