## פתיחה — אותו ממוצע, חיזוי קבוע שונה באיכותו
לשתי קבוצות דידקטיות מומצאות ממוצע 10: א׳ `[8,10,12]`, ב׳ `[2,10,18]`. אם ננחש תמיד 10, באיזו קבוצה יהיו מרחקי הערכים מהניחוש גדולים יותר? בקבוצה ב׳.

לפני מודל חשוב לתאר לא רק מרכז אלא גם פיזור. לא מאמנים או מעריכים כאן מודל על נתוני בדיקה; זו המחשה של הבנת הנתונים.

## 1. טווח — המרחק בין הקצוות
טווח = מקסימום − מינימום. א׳: 12−8=4; ב׳: 18−2=16. הטווח משתמש רק בשני הקצוות ואינו מספר כיצד הערכים שביניהם מפוזרים.

```python
import pandas as pd
groups_df = pd.DataFrame({
    "group_a": [8, 10, 12],
    "group_b": [2, 10, 18]
})
print("ממוצע א:", groups_df["group_a"].mean())
print("ממוצע ב:", groups_df["group_b"].mean())
print("טווח א:", groups_df["group_a"].max() - groups_df["group_a"].min())
print("טווח ב:", groups_df["group_b"].max() - groups_df["group_b"].min())
```

## 2. למה לא מסתפקים בממוצע ההפרשים?
בקבוצה א׳ הממוצע 10. ההפרשים הם −2,0,2 וסכומם 0. סימנים מנוגדים מבטלים זה את זה. נרבע כל הפרש כדי לתאר גודל סטייה בלי ביטול.

| ערך | הפרש מהממוצע | ריבוע הפרש |
|---|---|---|
| 8 | −2 | 4 |
| 10 | 0 | 0 |
| 12 | 2 | 4 |

סכום ריבועים 8. בחישוב תיאורי של כל שלושת הערכים נחלק ב־3: שונות = 8/3 ≈ 2.67. סטיית תקן = שורש השונות ≈ 1.63.

בקבוצה ב׳ הפרשים −8,0,8; סכום ריבועים 128. שונות 128/3 ≈ 42.67; סטיית תקן ≈ 6.53. סטיית התקן גדולה פי 4 למרות ממוצע זהה.

## 3. ddof=0 — תיאור כל הערכים שלפנינו
בדוגמה זו שלושת הערכים הם כל הקבוצה המוגדרת. מציינים במפורש ddof=0 כדי לחלק במספר הערכים הידועים. כך נשווה לחישוב הידני.

```python
print("שונות א, חלוקה ב-n:", groups_df["group_a"].var(ddof=0))
print("סטיית תקן א:", groups_df["group_a"].std(ddof=0))
print("שונות ב, חלוקה ב-n:", groups_df["group_b"].var(ddof=0))
print("סטיית תקן ב:", groups_df["group_b"].std(ddof=0))
```

שונות נמדדת ביחידות בריבוע; סטיית התקן ביחידה המקורית. אם מדדנו מסה בגרמים, השונות בגרם² וסטיית התקן בגרם. סטיית תקן אינה טעות תקן, רווח סמך או המרחק של כל תצפית מהממוצע.

## 4. מדגם ואוכלוסייה — לא משנים נוסחה בשקט
כאשר מתייחסים לערכים כמדגם לצורך אומדן שונות אוכלוסייה, נהוג להשתמש בסכום ריבועי ההפרשים חלקי n−1. ב־Pandas זה ddof=1, וגם ברירת המחדל של var ו־std.

לשלושת ערכי א׳: 8/(3−1)=4; שורש 4=2. זה שונה מהתוצאה בחלוקה ב־3. אין סתירה: בחרנו מחלק אחר למטרה אחרת. אומדן השונות עם n−1 חסר הטיה בתנאי הדגימה המתאימים; סטיית התקן שמתקבלת מהשורש אינה בהכרח חסרת הטיה.

```python
print("שונות א, חלוקה ב-n-1:", groups_df["group_a"].var(ddof=1))
print("סטיית תקן א, חלוקה ב-n-1:", groups_df["group_a"].std(ddof=1))
```

הוראת עבודה: ציינו באיזה ddof השתמשתם ולמה. אין חובה להוכיח היום את תיקון המדגם. עצם קיום קובץ אינו הופך אותו לכל אוכלוסיית הפינגווינים.

## 5. נתונים אמיתיים — סיכום תיאורי של המסות הידועות
כאן נשתמש ב־ddof=0 כדי לתאר את אוסף המדידות שבקובץ, ולא כדי לטעון שאמדנו ללא הטיה את פיזור כל הפינגווינים בטבע.

```python
url = "https://raw.githubusercontent.com/samiabo75/Data-Science-ML-Course/main/Datasets/penguins.csv"
df = pd.read_csv(url)
print("מסות ידועות:", df["body_mass_g"].count())
print("ממוצע בגרם:", df["body_mass_g"].mean())
print("טווח בגרם:", df["body_mass_g"].max() - df["body_mass_g"].min())
print("שונות תיאורית בגרם בריבוע:", df["body_mass_g"].var(ddof=0))
print("סטיית תקן תיאורית בגרם:", df["body_mass_g"].std(ddof=0))
```

342 מסות ידועות ו־2 חסרות; המדדים כאן מדלגים על החסרים. הטווח 6300−2700=3600 גרם. סטיית התקן מתארת פיזור סביב ממוצע המדידות הידועות; אינה אומרת שכל פינגווין מרוחק ממנה בדיוק כך.

אין להסיק שכלל 68–95–99.7 חל רק מפני שחושבה סטיית תקן. לתיאור הנתונים יש להתבונן גם בהיסטוגרמה ובקבוצות המעורבות.
