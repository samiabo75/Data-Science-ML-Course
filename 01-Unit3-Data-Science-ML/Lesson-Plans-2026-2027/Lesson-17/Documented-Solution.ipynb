## פתרון מתועד

## למה לומדים את זה עכשיו?
שתי קבוצות מגיעות בממוצע באותו זמן; באיזו קבוצה הזמנים מפוזרים יותר?
**בתהליך הפרויקט:** חקר נתונים.
**מסגרת פעילה:** 1.5 שעות לימוד — 67.5 דקות, כולל דוגמה ותרגול עצמאי. המורה בוחר את משימות הליבה שבמדריך; שאר הפעילויות לחזרה או לעיון, ללא תוספת שעות.

### 1–2. ידני ואימות
לשתי הקבוצות ממוצע 20. א׳: טווח 4, הפרשים −2,0,2, ריבועים 4,0,4; שונות 8/3 וסטיית תקן ≈1.63. ב׳: טווח 20, הפרשים −10,0,10, ריבועים 100,0,100; שונות 200/3 וסטיית תקן ≈8.16. הפיזור ב׳ גדול יותר, אף שהממוצע זהה.

```python
import pandas as pd
groups_df = pd.DataFrame({"a": [18, 20, 22], "b": [10, 20, 30]})
print("ממוצע א:", groups_df["a"].mean())
print("ממוצע ב:", groups_df["b"].mean())
print("טווח א:", groups_df["a"].max() - groups_df["a"].min())
print("טווח ב:", groups_df["b"].max() - groups_df["b"].min())
print("שונות א:", groups_df["a"].var(ddof=0))
print("סטיית תקן א:", groups_df["a"].std(ddof=0))
print("שונות ב:", groups_df["b"].var(ddof=0))
print("סטיית תקן ב:", groups_df["b"].std(ddof=0))
```

### 3. מדגם
סכום הריבועים בא׳ 8; חלוקה ב־2 נותנת שונות 4 וסטיית תקן 2. ddof=1 היא ברירת המחדל בפעולות אלה של Pandas. אין סתירה עם חלוקה ב־3; המטרה והמחלק שונים.

```python
print("שונות א עם n-1:", groups_df["a"].var(ddof=1))
print("סטיית תקן א עם n-1:", groups_df["a"].std(ddof=1))
```

### 4. סנפיר במקור
342 ערכים ידועים. היחידה מ״מ; שונות במ״מ² וסטיית תקן במ״מ. טווח 231−172=59 מ״מ. ddof=0 מתאר כאן את הקובץ, לא מבטיח אומדן של כל האוכלוסייה.

```python
url = "https://raw.githubusercontent.com/samiabo75/Data-Science-ML-Course/main/Datasets/penguins.csv"
penguins_df = pd.read_csv(url)
print("ידועים:", penguins_df["flipper_length_mm"].count())
print("ממוצע בממ:", penguins_df["flipper_length_mm"].mean())
print("טווח בממ:", penguins_df["flipper_length_mm"].max() - penguins_df["flipper_length_mm"].min())
print("שונות בממ בריבוע:", penguins_df["flipper_length_mm"].var(ddof=0))
print("סטיית תקן בממ:", penguins_df["flipper_length_mm"].std(ddof=0))
```

### 5. Adelie
151 אורכי סנפיר ידועים מתוך 152 רשומות. זו קבוצה אחת לעומת תערובת מינים במקור; אין להסיק מובהקות או סיבתיות מהשוואת מדדים תיאורית.

```python
adelie_mask = penguins_df["species"] == "Adelie"
adelie_df = penguins_df[adelie_mask]
print("ידועים בקבוצה:", adelie_df["flipper_length_mm"].count())
print("טווח קבוצה בממ:", adelie_df["flipper_length_mm"].max() - adelie_df["flipper_length_mm"].min())
print("סטיית תקן קבוצה בממ:", adelie_df["flipper_length_mm"].std(ddof=0))
```

### 6. פירוש ומגבלה
סטיית התקן מסכמת גודל פיזור סביב ממוצע, לא מרחק קבוע של כל תצפית ולא טעות תקן של הממוצע. המדדים מדלגים על חסרים, ואינם הוכחה לנורמליות. אין למחוק תצפיות בלי בדיקת מקור והקשר. פרמטרים להכנת מודל יחושבו בהמשך מנתוני אימון בלבד.

```python
print("ממדי המקור:", penguins_df.shape)
```

### רשות
כל הערכים שווים, ולכן טווח, שונות וסטיית תקן הם 0. חלוקה בסטיית תקן 0 אינה מוגדרת; נדרש טיפול ייעודי ולא חלוקה אוטומטית.

```python
equal_df = pd.DataFrame({"value": [7, 7, 7]})
print("טווח:", equal_df["value"].max() - equal_df["value"].min())
print("שונות:", equal_df["value"].var(ddof=0))
print("סטיית תקן:", equal_df["value"].std(ddof=0))
```
