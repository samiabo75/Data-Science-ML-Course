## למה צריך את זה בלמידת מכונה?
דודו רוצה לקנות רכב משומש. בעתיד ננסה לחזות מחיר בעזרת נתונים של רכבים רבים. היום נלמד לייצג רכב אחד. שנתון וקילומטראז׳ עשויים להיות מאפיינים; מחיר יהיה יעד אפשרי לחיזוי. בחומרי אימון צריך מחירים ידועים. אין היום אימון מודל.

לפני הרצה: אילו שני רכבים אפשר להשוות בהגינות? האם די בשנתון כדי לקבוע מחיר?

## 1. משתנה שומר ערך
`=` מציב את הערך מימין בשם שמשמאל; זו אינה משוואה לפתרון. טקסט מוקף במירכאות. מספר שנרצה לחשב איתו נכתב ללא מירכאות. שמות המשתנים באנגלית, ללא רווחים, ובעלי משמעות.

```python
car_model = "רכב לדוגמה"
manufacture_year = 2021
mileage_km = 60000
price_nis = 90000
print("דגם:", car_model)
print("שנת ייצור:", manufacture_year)
print("קילומטראז׳:", mileage_km)
print("מחיר בשקלים:", price_nis)
```

כתבו בתא טקסט: מה פירוש כל משתנה ומה יחידת המדידה שלו? `2021` הוא מספר ואילו `"2021"` הוא טקסט. שינוי ערך בתא לא מעדכן פלט עד שמריצים שוב.

## 2. מנתון גולמי למאפיין מחושב
נחשב גיל לפי שנות לוח: שנת ייחוס פחות שנת ייצור. זהו קירוב שאינו מתחשב בחודש העלייה לכביש. שנת הייחוס קבועה לצורך הדוגמה.

```python
reference_year = 2026
car_age_years = reference_year - manufacture_year
print("גיל הרכב בשנות לוח:", car_age_years)
```

חישוב ידני: 2026−2021=5 שנים. בהמשך גיל עשוי לשמש מאפיין במודל. עצם חישובו אינו למידת מכונה.

ניסוי: שנו שנת ייצור ל־2023, הריצו את תא הנתונים ואת תא הגיל. הסבירו את השינוי. אחר כך החזירו ל־2021 והריצו שוב.

## 3. חיבור וחלוקה — השוואה ראשונית
נחשב ממוצע מחירים של שני רכבים דמיוניים. נוסחה: סכום שני המחירים חלקי 2. זהו תיאור שתי הדוגמאות, לא תחזית אמינה לרכב חדש.

```python
price_a_nis = 90000
price_b_nis = 110000
average_price_nis = (price_a_nis + price_b_nis) / 2
price_difference_nis = price_b_nis - price_a_nis
print("ממוצע שני המחירים בשקלים:", average_price_nis)
print("הפרש המחירים בשקלים:", price_difference_nis)
```

חישוב ידני: (90000+110000)/2=100000; הפרש: 20000. הסוגריים גורמים לחיבור להתבצע לפני החלוקה. מדוע בלי סוגריים נקבל תוצאה אחרת? מדוע שתי דוגמאות אינן מספיקות להערכת מחיר בשוק?

## 4. כפל ואחוזים
לצורך תרגול בלבד, מוכר מציע הנחה של 5%. זהו כלל נתון מראש. אין להסיק שזו ירידת הערך השנתית של רכב.

```python
discount_percent = 5
discount_nis = price_nis * discount_percent / 100
final_price_nis = price_nis - discount_nis
print("גובה ההנחה בשקלים:", discount_nis)
print("מחיר לאחר הנחה בשקלים:", final_price_nis)
```

5%=5/100. לכן ההנחה היא 4500 ₪ והמחיר הסופי 85500 ₪. `0.05` הוא שבר עשרוני (`float`); שנות ייצור נכתבו כמספרים שלמים (`int`); דגם הוא מחרוזת (`str`). אין צורך בהמרות טיפוסים היום.
