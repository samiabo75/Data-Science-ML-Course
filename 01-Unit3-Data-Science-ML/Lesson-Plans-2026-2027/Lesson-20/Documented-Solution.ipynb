## פתרון מתועד — סנפיר ומסה

## למה לומדים את זה עכשיו?
אם חסר מחיר האמת של מחשב, אי אפשר לבדוק מול מחיר שהמצאנו; חסר במאפיין הוא בעיה אחרת.
**בתהליך הפרויקט:** הכנה וחלוקה.
**מסגרת פעילה:** 2 שעות לימוד — 90 דקות, כולל דוגמה ותרגול עצמאי. המורה בוחר את משימות הליבה שבמדריך; שאר הפעילויות לחזרה או לעיון, ללא תוספת שעות.

```python
import pandas as pd
from sklearn.model_selection import train_test_split
url = "https://raw.githubusercontent.com/samiabo75/Data-Science-ML-Course/main/Datasets/penguins.csv"
penguins_df = pd.read_csv(url)
print("מקור:", penguins_df.shape)
print(penguins_df.isna().sum())
```

### 2. הגדרת משימה
סיווג species לפי סנפיר במ״מ ומסה בגרם. sex אינה נבחרת ולכן חסר בה לבדו אינו סיבה למחיקת רשומה במשימה זו. אין יעד חסר בקובץ הנוכחי, אך הכלל מפורש.

```python
model_df = penguins_df.dropna(subset=["species"]).copy()
feature_columns = ["flipper_length_mm", "body_mass_g"]
X = model_df[feature_columns]
y = model_df["species"]
print("לפני ואחרי יעד ידוע:", len(penguins_df), len(model_df))
print("X:", X.shape, "y:", y.shape)
print("חסרים במאפיינים:")
print(X.isna().sum())
```

### 3. חלוקה משותפת
random_state=17 משחזר חלוקה נתונה, stratify שומר בקירוב יחס מחלקות. 258 אימון ו־86 בדיקה. בדיקה נשמרת להערכה בסיום; בחירת שיטות תיעשה באימות בתוך האימון.

```python
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.25, random_state=17, stratify=y
)
print("אימון:", X_train.shape, y_train.shape)
print("בדיקה:", X_test.shape, y_test.shape)
print("חסרי אימון:")
print(X_train.isna().sum())
print("חסרי בדיקה:")
print(X_test.isna().sum())
```

### 4. ערכי השלמה נלמדים רק באימון
השלמה בממוצע היא שיטה פשוטה להדגמה; אינה מובטחת כטובה ביותר. אם חסרים קיימים רק באחת הקבוצות בחלוקה מסוימת, עדיין משתמשים באותו רצף תקין בשתיהן.

```python
training_means = X_train.mean()
X_train_ready = X_train.fillna(training_means)
X_test_ready = X_test.fillna(training_means)
print("פרמטרי אימון:")
print(training_means)
print("אימון אחרי:")
print(X_train_ready.isna().sum())
print("בדיקה אחרי:")
print(X_test_ready.isna().sum())
```

### 5–6. טעויות ותיעוד
השלמה מכל X או מפרמטרי Test חושפת את התהליך למידע ששמור להערכה. משתמשים בפרמטרים שנלמדו מאימון בלבד. יעד חסר אינו תשובה ידועה; אין להמציאו כדי ליצור נתון אימון 'מסומן'.

יומן לדוגמה: הוגדר סיווג species, נבחרו שתי מדידות, הופרדו רשומות בעלות יעד ידוע לפי כלל זכאות. נעשתה חלוקה משותפת 75/25 עם seed=17 ויחס מחלקות בקירוב. הממוצעים המוצגים בפלט נלמדו באימון והוחלו על שתי הקבוצות. נשמרו ממדים והתאמת אינדקסים. מגבלה: השלמה מפחיתה מידע על פיזור ואינה שחזור אמת; מנגנון החסר לא נבדק לעומק. אין עדיין מודל או ציון.

```python
print("ממדי המקור:", penguins_df.shape)
print("חסרי המקור:")
print(penguins_df.isna().sum())
```
