## דף נתונים שניתן לתלמיד מראש
כל תצפית היא רשומה המתארת יהלום. הגרסה שנבדקה כוללת 53940 שורות ו־10 שדות.

| שדה | משמעות | יחידה/דוגמה |
|---|---|---|
| carat | משקל יהלום | קראט, 1 קראט = 0.2 גרם |
| cut | איכות חיתוך | Fair, Good, Very Good, Premium, Ideal |
| color | דרגת צבע | D עד J |

| price | מחיר מתועד | דולר ארה״ב, לא ₪ ולא מחיר שוק עדכני |
| x | אורך | מ״מ |
| y | רוחב | מ״מ |
| z | עומק | מ״מ |

התלמיד אינו נדרש להכיר גמולוגיה. זו פעילות על נתונים, לא המלצת רכישה או הערכת מחיר עדכנית.

### 1. טעינה
גרסת הקישור שנבדקה: 53940×10. price הוא יעד מספרי אפשרי, אך לא בונים כאן מודל.

```python
import pandas as pd
url = "https://raw.githubusercontent.com/arielb30/datasets/main/diamonds.csv"
diamonds_df = pd.read_csv(url)
print("ממדים:", diamonds_df.shape)
print("עמודות:", diamonds_df.columns)
display(diamonds_df.head())
```

### 2. מבנים
סדרה באורך 53940 לעומת טבלה בגודל 53940×1. אותם ערכי מחיר, אותו סדר ואינדקס, מבנה שונה.

```python
price_series = diamonds_df["price"]
price_table = diamonds_df[["price"]]
display(price_series.head(3))
display(price_table.head(3))
print("Series:", price_series.shape)
print("DataFrame:", price_table.shape)
```

### 3. Ideal
תת־קבוצה לפי קטגוריית חיתוך. לא ניתן להכליל ממנה אוטומטית לכל קטגוריות החיתוך.

```python
ideal_mask = diamonds_df["cut"] == "Ideal"
ideal_df = diamonds_df[ideal_mask]
print("Ideal:", ideal_df.shape)
display(ideal_df[["carat", "cut", "color", "price"]].head(4))
```

### 4. וגם ומשקל
>= כולל 1 קראט. בגרסה שנבדקה נבחרות 5870 רשומות עם כל עשרת השדות.

```python
one_carat_mask = diamonds_df["carat"] >= 1
combined_mask = ideal_mask & one_carat_mask
ideal_one_carat_df = diamonds_df[combined_mask]
print("שני התנאים:", ideal_one_carat_df.shape)
```

### 5. מיון יורד
כל רשומה נשמרת שלמה. הרשומות הן בעלות מחיר מתועד גבוה בתוך קבוצה מוגדרת; אינן מחירון עדכני או דגימה מייצגת. במקרה שוויון בגבול אין משמעות מובטחת לסדר בין השווים.

```python
ranked_df = ideal_one_carat_df.sort_values("price", ascending=False)
top_five_df = ranked_df.head(5)
display(top_five_df[["carat", "cut", "color", "clarity", "price"]])
```

### 6. תחום משקל
שאלה: מהן שלוש הרשומות בעלות המחיר המתועד הנמוך ביותר מבין יהלומים שמשקלם 0.5–0.7 קראט, כולל הקצוות?

```python
lower_mask = diamonds_df["carat"] >= 0.5
upper_mask = diamonds_df["carat"] <= 0.7
mid_weight_mask = lower_mask & upper_mask
mid_weight_df = diamonds_df[mid_weight_mask]
mid_weight_sorted = mid_weight_df.sort_values("price")
print("בתחום:", mid_weight_df.shape)
display(mid_weight_sorted[["carat", "cut", "price"]].head(3))
```

### 7. מקור ומסקנות
דוגמאות למסקנות: בקובץ שנבדק יש 5870 רשומות Ideal שמשקלן לפחות 1 קראט; זה תיאור הקובץ, לא שיעורן בכל השוק. אפשר לזהות מחירים מתועדים גבוהים בקבוצה באמצעות הפלט; זו אינה קביעה על איכות, כדאיות או מחיר עתידי.

חיפוש לפי מחיר נתון הוא סינון; חיזוי דורש ללמוד קשרים מדוגמאות ולהעריך הכללה. price כקלט לחיזוי עצמה חושפת את התשובה. גודל מאגר אינו הוכחת תקינות, מקור ורישוי, ייצוגיות או עמידה במחוון.

```python
print("ממדי המקור:", diamonds_df.shape)
display(diamonds_df.head(2))
```

### 8. רשות
מידת אורך 0 אינה נראית כמדידה פיזיקלית סבירה של יהלום. יש לברר אם זהו קידוד חסר, שגיאת מדידה או עיבוד. לא מתקנים בלי מקור ולא מוחקים בשקט.

```python
zero_length_mask = diamonds_df["x"] == 0
zero_length_df = diamonds_df[zero_length_mask]
print("אורך אפס:", zero_length_df.shape)
display(zero_length_df.head())
```
