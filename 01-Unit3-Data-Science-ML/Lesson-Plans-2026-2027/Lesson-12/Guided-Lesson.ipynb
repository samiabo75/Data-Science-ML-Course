## פתיחה — האם נוכל בעתיד לחזות מחיר?
מחיר יכול להיות יעד מספרי, ומשקל, חיתוך ומדידות יכולים להיות מאפיינים מועמדים. היום רק מוכיחים יכולת להכיר ולבחור נתונים. הרבה רשומות אינן מבטיחות איכות, ייצוגיות או התאמה למחוון.

לפני עבודה עצמאית נדגים שאלה שונה מזו של המטלה: אילו רשומות מסוג Good הן בעלות המחיר המתועד הנמוך ביותר?

```python
import pandas as pd
url = "https://raw.githubusercontent.com/arielb30/datasets/main/diamonds.csv"
demo_df = pd.read_csv(url)
print("ממדים:", demo_df.shape)
print("עמודות:", demo_df.columns)
display(demo_df.head(3))
```

נפרק את השאלה: בוחרים קבוצה לפי cut, ממיינים לפי price בסדר עולה, מציגים שלוש רשומות ושדות נחוצים. אין צורך בתחביר חדש.

```python
good_mask = demo_df["cut"] == "Good"
good_df = demo_df[good_mask]
good_sorted = good_df.sort_values("price")
good_first_three = good_sorted.head(3)
display(good_first_three[["carat", "cut", "price"]])
print("מקור:", demo_df.shape)
print("קבוצה:", good_df.shape)
```

## דוגמת תיעוד
שאלה: שלוש הרשומות בעלות המחיר הנמוך ביותר בתוך Good. תנאי: cut שווה Good. מיון עולה לפי price בדולר ארה״ב. שלוש השורות אינן מדגם מייצג, ואין להסיק מהן מחיר שוק עדכני. שוויון במחיר אינו מצביע על סדר מהותי בין רשומות.

במחברת העצמאית תפתרו שאלות אחרות. אפשר להיעזר בדפי השיעורים הקודמים; לא פותחים את פתרון המטלה. אין קוד חדש לשינון.
