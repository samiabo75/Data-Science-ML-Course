## 1. מבינים וגם בלי קוד
דוגמת ארבעה מחשבים מומצאים: מחפשים מחיר עד 3000 ₪ וגם זיכרון של לפחות 16 GB.

| מחשב | מחיר | זיכרון | מחיר מתאים | זיכרון מתאים | שני התנאים |
|---|---|---|---|---|---|
| A | 2500 | 16 | True | True | True |
| B | 3500 | 16 | False | True | False |
| C | 2000 | 8 | True | False | False |
| D | 4000 | 8 | False | False | False |

זו טבלת ארבע האפשרויות הלוגיות: וגם אמת רק כאשר שני הצדדים אמת.

```python
import pandas as pd
tiny_df = pd.DataFrame({
    "computer_id": ["A", "B", "C", "D"],
    "price_nis": [2500, 3500, 2000, 4000],
    "memory_gb": [16, 16, 8, 8]
})
price_mask = tiny_df["price_nis"] <= 3000
memory_mask = tiny_df["memory_gb"] >= 16
both_mask = price_mask & memory_mask
display(price_mask)
display(memory_mask)
display(both_mask)
display(tiny_df[both_mask])
```

`&` מחברת שתי מסכות לפי הרשומות. מחשב A בלבד נבחר. לא משתמשים ב־`and` לחיבור שתי Series; and בפייתון אינה פעולת וגם איבר־איבר על סדרות. המסכות נוצרו מאותה טבלה, באותם אינדקסים.

## 2. שתי מסכות על נתונים אמיתיים

```python
url = "https://raw.githubusercontent.com/samiabo75/Data-Science-ML-Course/main/Datasets/penguins.csv"
df = pd.read_csv(url)
is_gentoo = df["species"] == "Gentoo"
large_mass = df["body_mass_g"] >= 5000
selected_mask = is_gentoo & large_mass
selected_df = df[selected_mask]
print("מקור:", df.shape)
print("שני התנאים:", selected_df.shape)
display(selected_df[["species", "body_mass_g"]].head())
```

מספר הרשומות בשני התנאים אינו יכול להיות גדול ממספר הרשומות בכל תנאי בנפרד. בודקים זאת בלי להוסיף פקודות חדשות.

```python
gentoo_df = df[is_gentoo]
large_mass_df = df[large_mass]
print("Gentoo:", gentoo_df.shape)
print("מסה לפחות 5000:", large_mass_df.shape)
print("שניהם:", selected_df.shape)
```

כאשר מסה חסרה בקובץ זה, השוואת NaN לסף אינה אמת, ולכן הרשומה לא תיבחר בתנאי המשולב. אין להסיק מכך שמסתה קטנה מהסף. אין כאן ניקוי חסרים.

## 3. צורה מקוצרת — רק אחרי הבנת המסכות
כשכותבים את ההשוואות ישירות, מקיפים כל תנאי בסוגריים. החיבור הוא & בין התנאים, והסוגריים המרובעים החיצוניים בוחרים שורות.

```python
compact_df = df[(df["species"] == "Gentoo") & (df["body_mass_g"] >= 5000)]
print("ממדי התוצאה המקוצרת:", compact_df.shape)
```

שתי הצורות עונות לאותה שאלה. אין חובה להעדיף שורה ארוכה; מסכות בשמות משמעותיים מועילות לתיעוד ולבדיקה.

## 4. תחום מספרי הוא גם שני תנאים
נרצה מסה בין 3500 ל־4500 גרם, כולל שני הקצוות. חייבים להיות גם מעל הגבול התחתון וגם מתחת לעליון.

```python
above_lower = df["body_mass_g"] >= 3500
below_upper = df["body_mass_g"] <= 4500
in_range_mask = above_lower & below_upper
in_range_df = df[in_range_mask]
print("בתחום כולל הקצוות:", in_range_df.shape)
display(in_range_df[["species", "body_mass_g"]].head())
```

כאן אין לכתוב `3500 <= df["body_mass_g"] <= 4500` כהשוואה משורשרת רגילה. משתמשים בשתי מסכות וב־&. כל המידות בגרמים.

## 5. הרחבת רשות — או כולל
נבחר Gentoo או Chinstrap: די שלפחות אחד התנאים אמת. `|` היא פעולת או בין מסכות.

| תנאי א | תנאי ב | א & ב | א \| ב |
|---|---|---|---|
| True | True | True | True |
| True | False | False | True |
| False | True | False | True |
| False | False | False | False |

או כולל גם מצב ששניהם אמת. רשומה שעומדת בשני הצדדים אינה מוכפלת בתוצאה.

```python
is_chinstrap = df["species"] == "Chinstrap"
either_species = is_gentoo | is_chinstrap
two_species_df = df[either_species]
print("Gentoo או Chinstrap:", two_species_df.shape)
```

באותה רשומה species אינה יכולה להיות גם Gentoo וגם Chinstrap, ולכן & בין שתי הבדיקות היה נותן קבוצה ריקה. זו תוצאה לוגית תקינה, לא בהכרח תקלה.
