## למה לומדים את זה עכשיו?
מחפשים מחשב בתקציב וגם עם זיכרון מספיק: שני תנאים צריכים להתקיים יחד.
**בתהליך הפרויקט:** חקר נתונים.
**מסגרת פעילה:** 0.5 שעות לימוד — 22.5 דקות, כולל דוגמה ותרגול עצמאי. המורה בוחר את משימות הליבה שבמדריך; שאר הפעילויות לחזרה או לעיון, ללא תוספת שעות.

### 1. לוגיקה
True וגם False נותנים False. וגם אמת רק כששני התנאים מתקיימים באותה רשומה.

```python
import pandas as pd
url = "https://raw.githubusercontent.com/samiabo75/Data-Science-ML-Course/main/Datasets/penguins.csv"
penguins_df = pd.read_csv(url)
print("מקור:", penguins_df.shape)
```

### 3. Adelie וגם סנפיר לפחות 190 מ״מ
המסכות מאותה טבלה ולכן תואמות באינדקסים. הסף כלול.

```python
adelie_mask = penguins_df["species"] == "Adelie"
long_mask = penguins_df["flipper_length_mm"] >= 190
combined_mask = adelie_mask & long_mask
adelie_long_df = penguins_df[combined_mask]
print("וגם:", adelie_long_df.shape)
display(adelie_long_df[["species", "flipper_length_mm"]].head())
```

### 4. חיתוך קבוצות
מספר התצפיות העומדות בשני התנאים אינו גדול ממספרן בכל אחת משתי הקבוצות. לכל רשומה בתוצאה מתקיימים שני התנאים.

```python
adelie_df = penguins_df[adelie_mask]
long_df = penguins_df[long_mask]
print("Adelie:", adelie_df.shape)
print("סנפיר לפחות 190:", long_df.shape)
print("שניהם:", adelie_long_df.shape)
```

### 5. תחום כולל
40<=אורך מקור<=45. משתמשים בשתי מסכות על Series ובחיבור &, לא בהשוואה משורשרת רגילה. חסרים אינם עוברים כאן את ההשוואות; לא ניתן להסיק מהו ערכם.

```python
lower_mask = penguins_df["bill_length_mm"] >= 40
upper_mask = penguins_df["bill_length_mm"] <= 45
bill_range_mask = lower_mask & upper_mask
bill_range_df = penguins_df[bill_range_mask]
print("תחום אורך מקור:", bill_range_df.shape)
display(bill_range_df[["species", "bill_length_mm"]].head())
```

### 6. סוגריים
מקיפים כל השוואה בנפרד כדי להגדיר את החיבור בין תוצאותיהן.

```python
compact_df = penguins_df[(penguins_df["species"] == "Adelie") & (penguins_df["flipper_length_mm"] >= 190)]
print("צורה מקוצרת:", compact_df.shape)
```

### 7. תיעוד לדוגמה
השאלה: אילו תצפיות הן Adelie עם סנפיר ידוע באורך 190 מ״מ לפחות? שני תנאים המחוברים בוגם; הסף כלול. המספר מופיע בפלט. נתונים חסרים אינם מאפשרים לקבוע שהתצפית מחוץ לטווח באופן ביולוגי; היא רק אינה נבחרת בבדיקה הנוכחית. אין להסיק מקבוצה זו על כלל המינים או על קשר סיבתי. המקור נשמר.

```python
print("ממדי המקור:", penguins_df.shape)
```

### רשות — או
רשומה נבחרת אם האי הוא אחד מהשניים. אין כפילות בבחירה עם מסכה. & היה מחייב אי יחיד להיות גם Dream וגם Torgersen, ולכן לא יבחר אף רשומה בנתונים האלה.

```python
dream_mask = penguins_df["island"] == "Dream"
torgersen_mask = penguins_df["island"] == "Torgersen"
either_island_mask = dream_mask | torgersen_mask
either_island_df = penguins_df[either_island_mask]
print("אחד משני האיים:", either_island_df.shape)
```
