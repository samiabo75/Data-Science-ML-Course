## 1. רשימה: כמה ערכים בשם אחד
במקום ארבעה משתני מחיר, נשתמש ברשימה: סוגריים מרובעים ובין הערכים פסיקים. סדר הערכים נשמר. כל ערך כאן הוא מחיר רכב בשקלים.

```python
prices_nis = [90000, 110000, 80000, 120000]
print("מחירי הרכבים בשקלים:", prices_nis)
```

לפני הרצה כתבו כמה ערכים אתם רואים. `prices_nis` הוא שם משתנה השומר רשימה, ולא מחיר יחיד. בדוגמה זו נקפיד שכל הרשימה תייצג אותו מאפיין ובאותה יחידה.

## 2. מקום ברשימה — אינדקס
האינדקס הראשון הוא 0. ארבעת המיקומים הם 0, 1, 2, 3. יש להבחין בין מספרו הסידורי של הרכב לבין אינדקס Python.

| רכב לפי הסדר | אינדקס | מחיר בשקלים |
|---|---|---|
| ראשון | 0 | 90000 |
| שני | 1 | 110000 |
| שלישי | 2 | 80000 |
| רביעי | 3 | 120000 |

```python
print("מחיר הרכב הראשון:", prices_nis[0])
print("מחיר הרכב השלישי:", prices_nis[2])
```

חזו בלי להריץ: מה מחזיר `prices_nis[1]`? מדוע `prices_nis[4]` מחוץ לטווח? אין צורך ליצור שגיאה כדי להמשיך; הסבירו במילים.

## 3. כמה תצפיות ומה המחיר הממוצע?
`len` סופרת איברים. `sum` מחברת מספרים. אלה פונקציות מוכנות; היום אין צורך לכתוב פונקציה משלנו.

ממוצע = סכום המחירים / מספר המחירים. חשבו קודם ביד: 90000+110000+80000+120000=400000; חלקי 4.

```python
number_of_cars = len(prices_nis)
total_price_nis = sum(prices_nis)
average_price_nis = total_price_nis / number_of_cars
print("מספר הרכבים:", number_of_cars)
print("סכום המחירים בשקלים:", total_price_nis)
print("מחיר ממוצע בשקלים:", average_price_nis)
```

מתקבל ממוצע 100000 ₪. אף רכב ברשימה אינו חייב לעלות בדיוק כך. הממוצע מתאר את הדוגמאות; הוא אינו תחזית שנבדקה. ברשימה ריקה חלוקה במספר האיברים אינה אפשרית; כאן הנתונים אינם ריקים.

## 4. מהו תחום המחירים?
`min` מחזירה ערך מינימלי ו־`max` מחזירה ערך מקסימלי. טווח = מקסימום − מינימום. הטווח אינו מספר הרכבים ואינו מלמד לבדו על כל הפיזור.

```python
lowest_price_nis = min(prices_nis)
highest_price_nis = max(prices_nis)
price_range_nis = highest_price_nis - lowest_price_nis
print("מחיר מינימלי:", lowest_price_nis)
print("מחיר מקסימלי:", highest_price_nis)
print("טווח מחירים בשקלים:", price_range_nis)
```

## 5. מרשימות לתצפיות — גשר לטבלה
כל שורה בטבלה הבאה היא רכב אחד. לשם פשטות אלו רכבים דמיוניים מאותו דגם, אך הבדלי מצב עדיין חסרים.

| מזהה | שנת ייצור | קילומטראז׳ | מחיר בשקלים |
|---|---|---|---|
| A | 2021 | 60000 | 90000 |
| B | 2023 | 30000 | 110000 |
| C | 2020 | 90000 | 80000 |
| D | 2024 | 20000 | 120000 |

נייצג עמודות ברשימות מקבילות רק כשלב מעבר. אותו אינדקס חייב לתאר אותו רכב בכל הרשימות.

```python
car_ids = ["A", "B", "C", "D"]
years = [2021, 2023, 2020, 2024]
mileages_km = [60000, 30000, 90000, 20000]
print("מזהה:", car_ids[1])
print("שנת ייצור:", years[1])
print("קילומטראז׳:", mileages_km[1])
print("מחיר:", prices_nis[1])
```

האינדקס 1 מתאר את הרכב B: שנת 2023, קילומטראז׳ 30000 ומחיר 110000 ₪. אם נשנה סדר רק ברשימת המחירים, נחבר מחיר לרכב הלא נכון. אורך זהה לבדו אינו מוכיח שההתאמה נכונה. בהמשך נשתמש ב־DataFrame לשמירת רשומות בטבלה.

מאפיינים אפשריים: שנת ייצור וקילומטראז׳. יעד חיזוי אפשרי: מחיר. המזהה משמש כאן לזיהוי בלבד; אין הצדקה להשתמש בו לחיזוי.

## 6. ניסוי מבוקר: תיקון ערך ברשימה נפרדת
נניח שבבדיקה התברר שמחיר הרכב השלישי הוא 85000. נשמור רשימת ניסוי נפרדת. `copy()` יוצרת כאן עותק נפרד של רשימת המספרים; `trial_prices_nis = prices_nis` לבדו לא היה יוצר עותק.

```python
trial_prices_nis = prices_nis.copy()
trial_prices_nis[2] = 85000
trial_average_nis = sum(trial_prices_nis) / len(trial_prices_nis)
print("רשימת המקור:", prices_nis)
print("רשימת הניסוי:", trial_prices_nis)
print("ממוצע לאחר התיקון:", trial_average_nis)
```

סכום חדש 405000, ממוצע 101250 ₪. המקור נשמר. בתא טקסט תעדו: מה שונה, מדוע, כיצד השתנתה התוצאה ומה המגבלה. אין לתקן נתוני אמת ללא בדיקת מקור; כאן הסיבה הוגדרה לצורך התרגיל.
