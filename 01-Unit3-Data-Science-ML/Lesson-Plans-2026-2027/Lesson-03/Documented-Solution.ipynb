## למה לומדים את זה עכשיו?
מחיר של מחשב אחד אינו מלמד מה יקרה בכל הדגמים. מארגנים כמה תצפיות.
**בתהליך הפרויקט:** ייצוג.
**מסגרת פעילה:** 1 שעות לימוד — 45 דקות, כולל דוגמה ותרגול עצמאי. המורה בוחר את משימות הליבה שבמדריך; שאר הפעילויות לחזרה או לעיון, ללא תוספת שעות.

### 1. ייצוג
כל אינדקס מייצג מחשב אחד. הזיכרון ב־GB והמחיר בשקלים. רשימות מקבילות הן גשר לטבלה; יש לשמור על סדר ואורך תואמים.

```python
computer_ids = ["P1", "P2", "P3", "P4", "P5"]
memory_gb = [8, 16, 8, 32, 16]
prices_nis = [1800, 2600, 1500, 4200, 2900]
print("מזהים:", computer_ids)
print("זיכרון ב-GB:", memory_gb)
print("מחירים בשקלים:", prices_nis)
```

### 2. אינדקס
המחשב השלישי נמצא באינדקס 2, משום שהספירה מתחילה ב־0.

```python
print("מחשב:", computer_ids[2])
print("זיכרון ב-GB:", memory_gb[2])
print("מחיר בשקלים:", prices_nis[2])
```

### 3. ממוצע
1800+2600+1500+4200+2900=13000; 13000/5=2600 ₪. זהו ממוצע חמש הדוגמאות בלבד.

```python
number_of_computers = len(prices_nis)
total_price_nis = sum(prices_nis)
average_price_nis = total_price_nis / number_of_computers
print("מספר מחשבים:", number_of_computers)
print("סכום בשקלים:", total_price_nis)
print("ממוצע בשקלים:", average_price_nis)
```

### 4. קצוות וטווח
מינימום 1500, מקסימום 4200, טווח 2700 ₪. אין כאן מידע על מצב פיזי, מעבד או אחריות; אין להסיק כדאיות רכישה.

```python
lowest_price_nis = min(prices_nis)
highest_price_nis = max(prices_nis)
price_range_nis = highest_price_nis - lowest_price_nis
print("מינימום:", lowest_price_nis)
print("מקסימום:", highest_price_nis)
print("טווח בשקלים:", price_range_nis)
```

### 5. ניסוי בעותק
המחשב הרביעי באינדקס 3. תיקון של 200 ₪ כלפי מטה מפחית את הממוצע ב־200/5=40 ₪. הממוצע החדש 2560 ₪. copy מתאימה כאן לרשימה של מספרים.

```python
trial_prices_nis = prices_nis.copy()
trial_prices_nis[3] = 4000
trial_average_nis = sum(trial_prices_nis) / len(trial_prices_nis)
print("מקור:", prices_nis)
print("ניסוי:", trial_prices_nis)
print("ממוצע חדש בשקלים:", trial_average_nis)
```

### 6. מילון נתונים לדוגמה

### 7. טעויות
האינדקס החמישי הוא 4; אינדקס 5 מחוץ לטווח. מיון מחירים לבדם מנתק את התאמת המחיר למזהה ולזיכרון. בטבלה יש להזיז רשומה שלמה ביחד.
