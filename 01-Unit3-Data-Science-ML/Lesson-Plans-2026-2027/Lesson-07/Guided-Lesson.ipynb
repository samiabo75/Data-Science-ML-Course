הנתונים נאספו והועמדו לרשות הציבור בידי Kristen Gorman ו־Palmer Station LTER. המקור מתאר 344 פינגווינים משלושה מינים ומאיים בארכיפלג Palmer באנטארקטיקה. חבילת המקור הפשוטה מכילה גם year; גרסת הקורס שנבדקה מכילה 7 עמודות ללא year. אין להניח שכל גרסה באינטרנט זהה.

## 1. מהו CSV?
קובץ טקסט של נתונים טבלאיים: שורת כותרת ושורות נתונים; בקובץ זה פסיק מפריד בין שדות.

```text
species,island,bill_length_mm,bill_depth_mm,flipper_length_mm,body_mass_g,sex
Adelie,Torgersen,39.1,18.7,181,3750,male
```

הרשומה הראשונה מתארת פינגווין אחד. הערך 3750 הוא מסה בגרמים, לא בקילוגרמים. הכתובת של דף GitHub מציגה אתר; כתובת Raw מספקת את תוכן CSV עצמו.

````python
import pandas as pd
url = "https://raw.githubusercontent.com/samiabo75/Data-Science-ML-Course/main/Datasets/penguins.csv"
df = pd.read_csv(url)
````

`pd.read_csv(url)` קוראת את הקובץ ובונה DataFrame. אין צורך להקליד את הרשומות ידנית. `url` היא כתובת בטקסט. חייבים חיבור אינטרנט, ולהריץ את הייבוא קודם.

למקרה שהקישור חסום: הורידו את ה־CSV מקישור ההורדה, העלו אותו ללשונית Files ב־Colab, ובתא נפרד כתבו `df = pd.read_csv("penguins.csv")`. זהו מסלול חלופי, לא תא חובה נוסף להרצה. אם לא נטען קובץ, אין להמשיך כאילו df נוצרה.

## 2. בדיקה ראשונית — מה נטען?

````python
print("שורות ועמודות:", df.shape)
print("שמות העמודות:", df.columns)
display(df.head())
````

בגרסה שנבדקה מתקבל (344,7). ברירת המחדל של head היא עד חמש שורות. 344 הוא מספר הרשומות; 7 הוא מספר כל השדות, כולל היעד, לא מספר מאפייני קלט בהכרח.

אם הפלט שונה, בדקו כתובת וגרסה; אין לשנות קוד כדי לכפות 344. כתובת main עלולה להתעדכן בעתיד.

## 3. מילון שדות
| שם | פירוש | יחידה/סוג תוכני |
|---|---|---|
| species | מין טקסונומי של פינגווין | קטגוריה; יעד אפשרי |
| island | אי הדגימה | קטגוריה |
| bill_length_mm | אורך מקור | מ״מ |
| bill_depth_mm | עומק מקור | מ״מ |
| flipper_length_mm | אורך סנפיר | מ״מ |
| body_mass_g | מסת גוף | גרם |
| sex | זוויג, זכר/נקבה | קטגוריה |

species אינו sex. מדידה רציפה יכולה להישמר במספרים שלמים בגלל רזולוציית המדידה; סוג האחסון אינו לבדו סוג המשתנה התוכני.

## 4. לא מסתפקים בתחילת הקובץ
`tail(3)` מציגה את שלוש השורות האחרונות. זו תוספת דידקטית שלנו להצצה מתחילת המאגר ומסופו.

````python
display(df.tail(3))
print("ממדי המקור אחרי ההצצות:", df.shape)
````

השורות האחרונות בגרסה זו הן Chinstrap, בעוד הראשונות Adelie. ההצצות מדגימות שלא נכון להסיק על כלל המאגר מחמש רשומות. הן אינן מוכיחות שהמאגר מייצג את כלל הפינגווינים או מגלות את כל הקבוצות.

## 5. ערך חסר אינו אפס
כבר ב־head רואים NaN במדידות ברשומה באינדקס 3. CSV מסמן חלק מהחסרים כ־NA ו־pandas מזהה אותם כברירת מחדל. חסר פירושו שהערך לא קיים/לא זמין בקובץ, לא שמסת הפינגווין אפס. לא מוחקים ולא ממלאים כעת.

כתבו איזה מידע חסר ברשומה זו ואיזה מידע עדיין קיים.

### הרחבת רשות — info
רק אחרי שמבנה הטבלה ברור. info מציגה שמות שדות, מספר ערכים שאינם חסרים וסוגי אחסון. אינה מתקנת את הנתונים. אין צורך לעטוף ב־print.

````python
df.info()
````

מספר שאינו חסר קטן מ־344 מצביע על חסרים בעמודה. Non-Null אינו מבטיח שהערכים נכונים. object עשוי להכיל כאן טקסט; float64 הוא סוג אחסון מספרי. טיפול מסודר בחסרים יגיע בהמשך.
