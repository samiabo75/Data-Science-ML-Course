## משמעות מתמטית ומושגים
groupby מחלק שורות לפי מפתח;אגרגציה מחזירה סיכום לכל קבוצה. size סופר שורות,count סופר ערכים ידועים בעמודה שנבחרה,sum מסכם ערכים ידועים ו־mean מחלק את סכומם במספר הערכים הידועים. ממוצע פשוט של ממוצעי קבוצות אינו בהכרח הממוצע הכולל כשמספרי הערכים הידועים שונים.

## פעולות שונות לאותה קבוצה
חשבו קודם על נייר את צפון ודרום. שלוש שורות בחנות אינן בהכרח שלוש מדידות ידועות.

```python
import pandas as pd
import numpy as np
sales=pd.DataFrame({"branch":["North","North","North","South","South","South"],
                   "units":[2.,3.,np.nan,1.,4.,np.nan]})
grouped=sales.groupby("branch")["units"]
print("Rows per branch:");print(grouped.size())
print("Known units values:");print(grouped.count())
print("Total units:");print(grouped.sum(min_count=1))
print("Mean units in known transactions:");print(grouped.mean())
```

## טבלה מרוכזת ומסקנה
agg כאן הוא קטע עזר להציג כמה סיכומים. ברירת המחדל sum על קבוצה ללא ערך ידוע יכולה להחזיר אפס;לכן ניתן להציג sum(min_count=1) כשצריך להבחין בין חסר לאפס.

```python
summary=sales.groupby("branch")["units"].agg(["size","count","sum","mean"])
display(summary)
print("Overall mean:",sales["units"].mean())
print("Mean of group means:",summary["mean"].mean())
```
