## פתרון מתועד — אורך מקור ומסה

## למה לומדים את זה עכשיו?
האם זמן נסיעה גדל עם המרחק? תרשים קשר אינו הוכחה שהגורם היחיד הוא מרחק.
**בתהליך הפרויקט:** חקר נתונים.
**מסגרת פעילה:** 1 שעות לימוד — 45 דקות, כולל דוגמה ותרגול עצמאי. המורה בוחר את משימות הליבה שבמדריך; שאר הפעילויות לחזרה או לעיון, ללא תוספת שעות.

### 1–2. משתנים
bill_length_mm הוא אורך מקור במ״מ ו־body_mass_g מסה בגרם; שתי מדידות כמותיות רציפות. נציג אורך ב־X ומסה ב־Y. כל נקודה היא זוג מאותה רשומה.

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
url = "https://raw.githubusercontent.com/samiabo75/Data-Science-ML-Course/main/Datasets/penguins.csv"
penguins_df = pd.read_csv(url)
print("ממדי המקור:", penguins_df.shape)
```

```python
plt.figure(figsize=(7, 4))
sns.scatterplot(data=penguins_df, x="bill_length_mm", y="body_mass_g", alpha=0.6)
plt.title("Bill length and body mass")
plt.xlabel("Bill length (mm)")
plt.ylabel("Body mass (g)")
plt.tight_layout()
plt.show()
```

### 3–4. פירוש
בגרף הכולל נראית מגמה חיובית עם פיזור וקיבוצים; אין יחס מדויק אחד לאחד. נקודה בדוגמת השאלה (40,3500) הייתה מתארת תצפית עם מקור באורך 40 מ״מ ומסה 3500 גרם. לא טענו שקיימת תצפית עם הצמד המדויק.

קשר אינו הוכחה להשפעה סיבתית. הבדלים בין מינים ומאפיינים נוספים עשויים להשתלב במראה; אין כאן ניסוי, מודל מותאם או הערכה על נתונים חדשים.

```python
adelie_mask = penguins_df["species"] == "Adelie"
adelie_df = penguins_df[adelie_mask]
plt.figure(figsize=(7, 4))
sns.scatterplot(data=adelie_df, x="bill_length_mm", y="body_mass_g", alpha=0.6)
plt.title("Bill length and mass within Adelie")
plt.xlabel("Bill length (mm)")
plt.ylabel("Body mass (g)")
plt.tight_layout()
plt.show()
```

### 5–6. קבוצה וכרטיס קשר
השאלה בתוך Adelie אינה זהה לשאלה על כלל הקובץ. בקבוצה זו יש 152 רשומות אך 151 זוגות מדידות ידועים; בכל המקור 342 זוגות. חסר לא הופך לאפס, ותצפיות חופפות יכולות להסתיר זו את זו.

כרטיס לדוגמה: נבדק קשר תיאורי בין אורך מקור למסה ב־Palmer Penguins בגרסת הקורס. המשתנים במ״מ ובגרם, הגרף כולל את הזוגות הידועים. יש מגמה לצד פיזור, והבחנה בקבוצות עשויה לסייע בפרשנות. אין הוכחת סיבתיות או ציון ביצוע. צעד עתידי: הכנת נתונים, אימון והערכה לפי חלוקה תקינה; אין ליישם זאת בקוד היום.

```python
print("מקור לאחר הבחירה והציור:", penguins_df.shape)
```

### רשות — hue
הצבע אינו משתנה שלישי מספרי בציר, אלא תווית קטגוריה. בגרף נראים קיבוצים לצד חפיפה; אין הבטחת זיהוי מושלם.

```python
plt.figure(figsize=(7, 4))
sns.scatterplot(data=penguins_df, x="bill_length_mm", y="body_mass_g", hue="species", alpha=0.6)
plt.title("Bill length and mass by species")
plt.xlabel("Bill length (mm)")
plt.ylabel("Body mass (g)")
plt.tight_layout()
plt.show()
```
