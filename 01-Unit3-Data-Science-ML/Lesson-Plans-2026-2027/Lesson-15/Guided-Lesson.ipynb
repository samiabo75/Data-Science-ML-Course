## 1. מה מייצגת נקודה?
דוגמה מומצאת של שלוש תצפיות, לא מאגר אמיתי:

| תצפית | אורך סנפיר במ״מ | מסה בגרם | נקודה (X,Y) |
|---|---|---|---|
| A | 180 | 3300 | (180,3300) |
| B | 190 | 3700 | (190,3700) |
| C | 200 | 3500 | (200,3500) |

כל נקודה משלבת שתי מדידות **מאותה רשומה**. נקודה ימינה יותר מציינת סנפיר ארוך יותר; נקודה גבוהה יותר מציינת מסה גדולה יותר. אין לחבר את הנקודות בקו לפי סדר הרשומות: זה אינו ציר זמן.

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
demo_df = pd.DataFrame({
    "flipper_mm": [180, 190, 200],
    "mass_g": [3300, 3700, 3500]
})
plt.figure(figsize=(7, 4))
sns.scatterplot(data=demo_df, x="flipper_mm", y="mass_g", color="steelblue")
plt.title("Three teaching examples")
plt.xlabel("Flipper length (mm)")
plt.ylabel("Body mass (g)")
plt.tight_layout()
plt.show()
```

לפני מעבר הלאה: איזו נקודה גבוהה ביותר? B. איזו ימנית ביותר? C. האם הארוך ביותר הוא בהכרח הכבד ביותר? לא בדוגמה. זהו קשר אפשרי עם שונות, לא כלל מדויק.

## 2. מציירים את המדידות האמיתיות

```python
url = "https://raw.githubusercontent.com/samiabo75/Data-Science-ML-Course/main/Datasets/penguins.csv"
df = pd.read_csv(url)
print("ממדי המקור:", df.shape)
plt.figure(figsize=(7, 4))
sns.scatterplot(data=df, x="flipper_length_mm", y="body_mass_g", alpha=0.6)
plt.title("Flipper length and body mass")
plt.xlabel("Flipper length (mm)")
plt.ylabel("Body mass (g)")
plt.tight_layout()
plt.show()
```

data היא הטבלה; x ו־y הם שמות המדידות. alpha=0.6 מוסיפה שקיפות כדי להקל לראות חפיפות — זו בחירת תצוגה, לא שינוי נתונים.

בגרף הכולל נראית מגמה חיובית: ערכי X גבוהים נוטים להופיע עם ערכי Y גבוהים. הנקודות אינן על קו יחיד. אין לחשב היום מקדם מתאם או להתאים קו; תחילה לומדים לקרוא את הזוגות.

## 3. מה לחפש?
- כיוון: חיובי, שלילי או בלי מגמה ברורה; היעדר מגמה ישרה אינו שולל קשר אחר.
- פיזור: האם יש מסות שונות לאורכי סנפיר דומים?
- קבוצות: האם אפשר שיש כמה קבוצות עם מאפיינים שונים?
- נקודות לבדיקה: נקודה רחוקה אינה הוכחת שגיאה ואינה מחייבת מחיקה.

לא יודעים דיוק חיזוי רק ממראה. נצטרך בעתיד נתוני בדיקה ומדד מתאים.

## 4. הרחבה לפי ההתקדמות — אותה תמונה, צבע לפי מין
hue מוסיפה מקרא וצבע לפי קטגוריה. ערכי X ו־Y אינם משתנים. זו המחשה לכך שהמגמה הכוללת יכולה לשלב הבדלים בין קבוצות ומגמות בתוך כל קבוצה.

```python
plt.figure(figsize=(7, 4))
sns.scatterplot(data=df, x="flipper_length_mm", y="body_mass_g", hue="species", alpha=0.6)
plt.title("Flipper length and mass by species")
plt.xlabel("Flipper length (mm)")
plt.ylabel("Body mass (g)")
plt.tight_layout()
plt.show()
```

אין להסיק שכל זוג מדידות מספיק לזיהוי מין ללא טעות. קבוצות יכולות לחפוף. המין עשוי להיות קשור להבדלים במדידות, אך הגרף אינו ניסוי שמוכיח סיבתיות.
