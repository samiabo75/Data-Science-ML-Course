## פתרון מתועד — סנפיר

### 1. שאלה
כיצד מתפלגים אורכי הסנפיר הידועים במאגר? זהו משתנה כמותי רציף במ״מ, גם אם נשמר במספרים שלמים. X הוא אורך, Y הוא מספר ערכים בתחום.

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
url = "https://raw.githubusercontent.com/samiabo75/Data-Science-ML-Course/main/Datasets/penguins.csv"
penguins_df = pd.read_csv(url)
print("ממדים:", penguins_df.shape)
```

```python
plt.figure(figsize=(7, 4))
sns.histplot(data=penguins_df, x="flipper_length_mm", bins=8, color="steelblue")
plt.title("Flipper length distribution - 8 bins")
plt.xlabel("Flipper length (mm)")
plt.ylabel("Number of observations")
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(7, 4))
sns.histplot(data=penguins_df, x="flipper_length_mm", bins=16, color="teal")
plt.title("Flipper length distribution - 16 bins")
plt.xlabel("Flipper length (mm)")
plt.ylabel("Number of observations")
plt.tight_layout()
plt.show()
```

### 4–5. פירוש
אותם 342 אורכים ידועים חולקו ליותר תחומים צרים. הטווח הידוע בגרסה זו הוא 172–231 מ״מ. רוחב תחום ב־8 תחומים הוא 59/8=7.375 מ״מ, וב־16 הוא 3.6875 מ״מ. אין צורך שהתלמיד יחשב זאת אם אין שליטה בחלוקה.

הגרפים מראים שכיחות בתחומי אורך, לא את ערכי כל הרשומות או את מספר המינים. כדאי לתאר טווחים לפי התצוגה ולנמק. לא קובעים נורמליות ממראה, וייתכן שהתצוגה מערבת קבוצות שונות. תצוגה מפורטת יותר אינה מידע חדש שנאסף.

### 6. חסרים
סכום הגבהים בגרפים אלה 342; 2 חסרי מדידה אינם נכללים. שינוי מספר התחומים על כל אותו טווח אינו משנה מספר זה. המקור נשאר 344×7.

```python
print("ממדי המקור:", penguins_df.shape)
```

### 7. כרטיס לדוגמה
שאלה: התפלגות אורך סנפיר ידוע ב־Palmer Penguins בגרסת הקורס. כמותי רציף, מ״מ. הוצגו 8 ואחר כך 16 תחומים; Y מספר תצפיות. הטווח הידוע הוא 172–231 מ״מ, והחלוקה משנה את פירוט המראה. מגבלה: שני חסרים ותלות ב־bins, ומדגם שאינו כלל הפינגווינים. אין ניבוי או הוכחת סיבתיות.

### רשות
משווים אותה מדידה ובאותם גבולות תחומים, אך הספירות הן של אוכלוסיות שונות בגודלן.

```python
adelie_mask = penguins_df["species"] == "Adelie"
adelie_df = penguins_df[adelie_mask]
shared_bins = [170, 180, 190, 200, 210, 220, 230, 240]
plt.figure(figsize=(7, 4))
sns.histplot(data=adelie_df, x="flipper_length_mm", bins=shared_bins, color="steelblue")
plt.title("Adelie flipper lengths - shared bins")
plt.xlabel("Flipper length (mm)")
plt.ylabel("Number of observations")
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(7, 4))
sns.histplot(data=penguins_df, x="flipper_length_mm", bins=shared_bins, color="teal")
plt.title("All flipper lengths - shared bins")
plt.xlabel("Flipper length (mm)")
plt.ylabel("Number of observations")
plt.tight_layout()
plt.show()
```

Adelie כוללת 151 אורכי סנפיר ידועים בגרסה זו, לעומת 342 בכל המקור. אין להשוות גובה ספירה כאילו הוא אחוז. ניתוח מנורמל דורש הגדרה נפרדת ואינו נלמד כאן.
