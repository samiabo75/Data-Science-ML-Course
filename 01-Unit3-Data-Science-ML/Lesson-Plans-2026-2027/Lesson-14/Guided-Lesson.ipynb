## 1. טווחי ערכים במקום קטגוריות
דוגמה מומצאת של שש מסות בגרמים: 3100,3400,3600,3900,4100,4400. נחלק לתחומים שווים.

| תחום מסה | ערכים בו | ספירה |
|---|---|---|
| 3000 עד פחות מ־3500 | 3100,3400 | 2 |
| 3500 עד פחות מ־4000 | 3600,3900 | 2 |
| 4000 עד 4500 כולל הקצה האחרון | 4100,4400 | 2 |

גובה עמודה הוא מספר הערכים בתחום, לא המסה עצמה ולא ממוצע. כל תחום נקרא bin. כאן רוחב כל תחום 500 גרם.

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
demo_df = pd.DataFrame({"mass_g": [3100, 3400, 3600, 3900, 4100, 4400]})
plt.figure(figsize=(7, 4))
sns.histplot(data=demo_df, x="mass_g", bins=[3000, 3500, 4000, 4500], color="steelblue")
plt.title("Six teaching examples by mass interval")
plt.xlabel("Mass (g)")
plt.ylabel("Number of observations")
plt.tight_layout()
plt.show()
```

בדוגמה זו התחומים סגורים משמאל ופתוחים מימין, והאחרון כולל גם את הגבול הימני. מסה ששווה בדיוק 3500 תיכנס לתחום השני. אין הכפלה של ערך בשני תחומים.

Countplot ספרה קטגוריות; היסטוגרמה מקבצת מדידה לתחומים. עמודות סמוכות מתארות תחומים על ציר מספרי.

## 2. נתונים אמיתיים — מסה ב־Penguins

```python
url = "https://raw.githubusercontent.com/samiabo75/Data-Science-ML-Course/main/Datasets/penguins.csv"
df = pd.read_csv(url)
print("ממדי המקור:", df.shape)
```

```python
plt.figure(figsize=(7, 4))
sns.histplot(data=df, x="body_mass_g", bins=12, color="steelblue")
plt.title("Body mass distribution - 12 bins")
plt.xlabel("Body mass (g)")
plt.ylabel("Number of observations")
plt.tight_layout()
plt.show()
```

`bins=12` מבקשת 12 תחומים שווי רוחב על טווח הנתונים. לא מדובר ב־12 תצפיות. בגרסה זו טווח המסות הידועות 2700–6300 גרם: רוחב תחום הוא (6300−2700)/12=300 גרם.

כתבו תחום שבו נראה ריכוז יחסי של תצפיות לפי הגרף. אל תטענו מתוך הגרף לערך השכיח המדויק או לתכונות כלל האוכלוסייה.

## 3. שינוי bins — אותם נתונים, חלוקה אחרת

```python
plt.figure(figsize=(7, 4))
sns.histplot(data=df, x="body_mass_g", bins=24, color="teal")
plt.title("Body mass distribution - 24 bins")
plt.xlabel("Body mass (g)")
plt.ylabel("Number of observations")
plt.tight_layout()
plt.show()
```

כאן רוחב התחום 150 גרם. השוו לציור הקודם: מה התחדד ומה נראה תנודתי יותר? השוו את ערכי הצירים, לא רק גובה חזותי. העמודה הגבוהה עשויה להשתנות בגלל חלוקת התחומים, אף שהתצפיות לא השתנו.

אין מספר bins אחד נכון לכל שאלה. לא בוחרים חלוקה רק כדי לקבל צורת פעמון. תצוגה שנראית פעמונית אינה הוכחה לנורמליות; ערבוב קבוצות עשוי ליצור צורה מורכבת.
