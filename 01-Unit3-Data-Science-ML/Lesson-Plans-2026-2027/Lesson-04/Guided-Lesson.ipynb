## למה זה נחוץ ללמידת מכונה?
דודו רוצה להשתמש בגיל הרכב כמאפיין לחיזוי מחיר. כשיש הרבה רכבים, לא נרצה לכתוב חישוב נפרד לכל רכב. לולאה תפעיל אותה פעולה על כל שנת ייצור. זהו עיבוד נתונים לקראת מודל; אין כאן אימון מודל.

בשיעור הזה לומדים מבנה חדש אחד: `for`. משתמשים רק במשתנים, בחשבון וברשימה שכבר הכרנו.

## חימום — מה כבר יודעים?
הנתונים הם של שלושה רכבים דמיוניים. מה מחזיר `years[0]`? כמה איברים יש? חשבו ביד גיל רכב משנת 2021 בשנת ייחוס 2026.

```python
years = [2021, 2023, 2020]
reference_year = 2026
print("שנת הייצור הראשונה:", years[0])
print("מספר רכבים:", len(years))
print("גיל הרכב הראשון:", reference_year - years[0])
```

## 1. קודם קוראים בלי קוד
ניקח בתורו כל ערך ברשימת השנים ונציג אותו:

| מעבר | הערך הנוכחי | מה יודפס |
|---|---|---|
| ראשון | 2021 | 2021 |
| שני | 2023 | 2023 |
| שלישי | 2020 | 2020 |

כל ערך נלקח לפי סדר הרשימה. לא צריך לבחור אינדקס בכל פעם.

```python
for year in years:
    print(year)
```

### פירוק שורה אחר שורה
- `for`: התחלת לולאה.
- `year`: משתנה שמקבל בכל מעבר את הערך הבא. כאן הוא שנת ייצור, לא אינדקס.
- `in years`: קח את הערכים מהרשימה `years`.
- `:`: מסיימת את כותרת הלולאה.
- ארבעת הרווחים לפני `print` הם הזחה: ההוראה שייכת לגוף הלולאה.

שאלו: האם Python חישבה לבד את השנים? לא; היא עברה על הערכים הנתונים. אחרי המעבר השלישי אין עוד ערך, ולכן הלולאה מסתיימת.

## עצירה לבדיקה
חזו את הפלט עבור הרשימה `[2024, 2022]`. כמה פעמים יודפס ערך? האם `year` יקבל 0 ו־1 או 2024 ו־2022? כתבו תשובה לפני שממשיכים.

## 2. אותה פעולה לכל שנת ייצור
בתוך הלולאה נחשב גיל. המשתנה `car_age_years` מתעדכן בכל מעבר. בשלב זה לא נבנית רשימת גילים חדשה; מציגים כל תוצאה בנפרד. הגיל הוא בקירוב שנות לוח.

```python
for year in years:
    car_age_years = reference_year - year
    print("שנת ייצור:", year, "גיל בשנות לוח:", car_age_years)
```

### טבלת מעקב מלאה
| מעבר | year | החישוב | car_age_years |
|---|---|---|---|
| 1 | 2021 | 2026−2021 | 5 |
| 2 | 2023 | 2026−2023 | 3 |
| 3 | 2020 | 2026−2020 | 6 |

נוסחה: גיל בשנות לוח = שנת ייחוס פחות שנת ייצור. זהו מאפיין מחושב אפשרי. מודל יצטרך דוגמאות עם מחירים ותהליך הערכה כדי לחזות מחיר; הלולאה לבדה אינה מספקת זאת.

## 3. מה חוזר ומה מתבצע פעם אחת?
ההוראה המוזחת מתבצעת לכל ערך. ההוראה שחזרה לתחילת השורה נמצאת מחוץ ללולאה.

```python
for year in years:
    print("בתוך הלולאה:", year)
print("סיימנו לעבור על כל השנים")
```

לפני הרצה: כמה פעמים תופיע הודעת הסיום? מה ישתנה אם נזיח גם אותה? הסבירו, בלי לשנות עדיין את התא. ההזחה מגדירה את גבולות הפעולה החוזרת.

## 4. הרחבה מדורגת — צבירה
ממשיכים רק אחרי שמבינים את השלבים הקודמים. רוצים גם סכום גילים וממוצע. מתחילים מסכום 0 לפני הלולאה; בכל מעבר מוסיפים את הגיל לסכום הקודם. `total_age_years = total_age_years + car_age_years` היא השמה: קודם מחשבים את צד ימין, ואז שומרים תוצאה חדשה בצד שמאל.

```python
total_age_years = 0
for year in years:
    car_age_years = reference_year - year
    total_age_years = total_age_years + car_age_years
print("סכום הגילים:", total_age_years)
average_age_years = total_age_years / len(years)
print("גיל ממוצע בשנים:", average_age_years)
```

### מעקב צבירה
| מעבר | גיל נוכחי | סכום לפני | סכום אחרי |
|---|---|---|---|
| 1 | 5 | 0 | 5 |
| 2 | 3 | 5 | 8 |
| 3 | 6 | 8 | 14 |

הממוצע הוא 14/3 ≈ 4.67 שנים. לא מאפסים את הסכום בתוך הלולאה. מחשבים את הממוצע אחרי שאספנו את כל הערכים. הרשימה כאן אינה ריקה. אם צבירה עדיין אינה ברורה, עוצרים כאן וחוזרים במפגש הבא.
