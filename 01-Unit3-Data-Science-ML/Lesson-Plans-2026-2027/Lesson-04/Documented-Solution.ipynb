## פתרון מתועד — מחשבים

### 1–2. מעבר על ערכים
הערכים המודפסים: 2022, 2024, 2021. משתנה הלולאה מקבל שנה ולא אינדקס.

```python
years = [2022, 2024, 2021]
for year in years:
    print(year)
```

### 3. חישוב גיל
2026−2022=4; 2026−2024=2; 2026−2021=5. כל מעבר מחשב את אותו ביטוי עבור שנה אחרת.

```python
reference_year = 2026
for year in years:
    computer_age_years = reference_year - year
    print("שנת ייצור:", year, "גיל בשנות לוח:", computer_age_years)
```

| מעבר | שנה | גיל |
|---|---|---|
| 1 | 2022 | 4 |
| 2 | 2024 | 2 |
| 3 | 2021 | 5 |

זהו קירוב לפי שנים; אין תאריך ייצור מלא.

### 4. גבולות הלולאה
הודעת הסיום אינה מוזחת ולכן תודפס פעם אחת לאחר שלושת המעברים.

```python
for year in years:
    print("שנה:", year)
print("הסתיים עיבוד השנים")
```

### 5. תיקון תחביר
נדרשות נקודתיים בסוף כותרת for והזחה של גוף הלולאה.

```python
for year in years:
    print(year)
```

### 6. פירוש
גיל מחושב משנת ייחוס ושנת ייצור, ולכן הוא משתנה נגזר. לשנת ייחוס קבועה יש קשר חשבוני מלא בין שנת הייצור לגיל. לחיזוי מחיר נחוצים גם נתוני מחירים ידועים ומאפיינים רלוונטיים כגון מעבד, זיכרון ומצב. אין התאמת מודל או הערכה על דוגמאות חדשות בשיעור זה.

תיעוד לדוגמה: חישבנו גיל לכל מחשב כדי לתאר מאפיין אפשרי. הפלט הוא 4,2,5 שנות לוח. המידע מומצא ואינו מאגר לפרויקט גמר.

### 7. הרחבת צבירה
| מעבר | גיל | סכום לפני | סכום אחרי |
|---|---|---|---|
| 1 | 4 | 0 | 4 |
| 2 | 2 | 4 | 6 |
| 3 | 5 | 6 | 11 |

סכום 11; ממוצע 11/3 ≈ 3.67 שנים. האתחול קודם ללולאה כדי שהסכום לא יימחק בכל מעבר.

```python
total_age_years = 0
for year in years:
    computer_age_years = reference_year - year
    total_age_years = total_age_years + computer_age_years
average_age_years = total_age_years / len(years)
print("סכום הגילים:", total_age_years)
print("ממוצע הגילים:", average_age_years)
```

### אתגר
שינוי סדר השנים משנה את סדר ההדפסה אך לא את סכום הגילים או ממוצעם. כל עוד התצפיות עצמן לא השתנו, פעולת הסיכום זהה.
