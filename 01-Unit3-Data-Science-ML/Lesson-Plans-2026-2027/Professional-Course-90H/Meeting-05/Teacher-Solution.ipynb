## פתרון חלק 2
הפתרון הבא תואם לנתוני התרגול העצמאי, ולא לנתוני הדוגמה הפתורה.

</div>

```python
import pandas as pd
url = "https://raw.githubusercontent.com/samiabo75/Data-Science-ML-Course/main/Datasets/penguins.csv"
penguins_df = pd.read_csv(url)
print("ממדי המקור:", penguins_df.shape)
```

<div dir="rtl" lang="he" style="text-align: right;">

### 2. מסכה
השוואה טקסטואלית עם == מחזירה Series בוליאנית. כל ערך מקושר לאינדקס של תצפית.

</div>

```python
gentoo_mask = penguins_df["species"] == "Gentoo"
display(gentoo_mask.head())
```

<div dir="rtl" lang="he" style="text-align: right;">

### 3. Gentoo
בחירה במסכה משאירה את כל העמודות של הרשומות שנבחרו, ואינה מאפסת אינדקסים.

</div>

```python
gentoo_df = penguins_df[gentoo_mask]
print("ממדים:", gentoo_df.shape)
display(gentoo_df.head())
```

<div dir="rtl" lang="he" style="text-align: right;">

### 4. Dream
יוצרים מסכה חדשה על המקור; אילו היינו מפעילים את הבדיקה רק על gentoo_df, היינו עונים על שאלה אחרת.

</div>

```python
dream_mask = penguins_df["island"] == "Dream"
dream_df = penguins_df[dream_mask]
print("ממדי קבוצת Dream:", dream_df.shape)
```

<div dir="rtl" lang="he" style="text-align: right;">

### 5. אורך סנפיר
>= כולל 220. בעמודת float שבקובץ זה NaN אינו עובר את ההשוואה. אין להסיק שהערך החסר קטן מהסף.

</div>

```python
long_flipper_mask = penguins_df["flipper_length_mm"] >= 220
long_flippers_df = penguins_df[long_flipper_mask]
print("ממדים:", long_flippers_df.shape)
display(long_flippers_df.head())
```

<div dir="rtl" lang="he" style="text-align: right;">

### 6. עמודות לעומת שורות
המסכה הפחיתה שורות. רשימת השמות מפחיתה עמודות. המקור נשאר 344×7.

</div>

```python
flipper_review = long_flippers_df[["species", "flipper_length_mm"]]
print("ממדי העיון:", flipper_review.shape)
print("ממדי המקור:", penguins_df.shape)
display(flipper_review.head())
```

<div dir="rtl" lang="he" style="text-align: right;">

### 7. תיעוד לדוגמה
שאלה: אילו תצפיות במאגר כוללות אורך סנפיר ידוע של לפחות 220 מ״מ? תנאי: flipper_length_mm>=220. מספר התוצאות מופיע בפלט. נשמרו כל השדות בטבלת הקבוצה, ואחר כך נבחרו שני שדות לעיון. ערכים חסרים אינם נבחרים, אבל לא הוכח שהם מתחת לסף. אין הכללה מכאן לכלל הפינגווינים ואין אימון מודל.

מין יחיד אינו בסיס למסווג בין כמה מינים; חסרות לו דוגמאות מהמחלקות האחרות. head על מסכה מציג רק חלק מהבדיקות.

</div>

<div dir="rtl" lang="he" style="text-align: right;">

### רשות — בדיקת גבול
הפרש מספר השורות נובע מהרשומות השוות בדיוק ל־ 220 מ״מ, ולא משינוי במספר השדות.

</div>

```python
strict_mask = penguins_df["flipper_length_mm"] > 220
strict_flippers_df = penguins_df[strict_mask]
print("לפחות 220:", long_flippers_df.shape)
print("יותר מ-220:", strict_flippers_df.shape)
print("מספר הרשומות בדיוק בסף:", len(long_flippers_df) - len(strict_flippers_df))
```
