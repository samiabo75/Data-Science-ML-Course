# מפגש 5 — פתרון מורה מלא, מבוקר
נתוני מעבדה: penguins.csv בגרסת הקורס, 344×7. תשובות אלה אינן מיועדות למצגת תלמיד. בדיקות assert הן בקרת קוד ולא תחליף להסבר מחקרי.

</div>

```python
import pandas as pd
url = "https://raw.githubusercontent.com/samiabo75/Data-Science-ML-Course/main/Datasets/penguins.csv"
penguins_df = pd.read_csv(url)
print('Source dimensions:',penguins_df.shape)
assert penguins_df.shape == (344,7)
```

<div dir="rtl" lang="he" style="text-align:right">

## חלק 1 — בחירת שדה יחיד כסדרה
body_mass_g הוא יעד אפשרי לחיזוי מסה בגרמים. Series היא חד־ממדית וצורתה (344,).

</div>

```python
masses = penguins_df['body_mass_g']
display(masses.head(4))
print('Series shape:',masses.shape)
assert masses.shape == (344,)
```

<div dir="rtl" lang="he" style="text-align:right">

## אותה עמודה בטבלה
שימוש ברשימה של שם העמודה מחזיר DataFrame דו־ממדי, ולא משנה את מספר התצפיות.

</div>

```python
mass_table = penguins_df[['body_mass_g']]
display(mass_table.head(4))
assert mass_table.shape == (344,1)
```

<div dir="rtl" lang="he" style="text-align:right">

## שתי מדידות כמאפיינים אפשריים
flipper_length_mm ו־bill_length_mm נבחרות בסדר זה, שתיהן במ״מ. אין כאן אימון מודל.

</div>

```python
predictor_columns = ['flipper_length_mm','bill_length_mm']
predictors = penguins_df[predictor_columns]
display(predictors.head(4))
assert predictors.shape == (344,2)
assert list(predictors.columns) == predictor_columns
```

<div dir="rtl" lang="he" style="text-align:right">

## יעד והצלבת שורות
היעד body_mass_g נפרד מ־X. גם הקלט וגם היעד נשמרים באותו סדר ואותם אינדקסים. לפני אימון עתידי יש לטפל בחסרים תוך שמירה על גבולות קבוצת הבדיקה.

</div>

```python
target_mass = penguins_df['body_mass_g']
assert target_mass.shape == (344,)
assert 'body_mass_g' not in predictors.columns
assert (predictors.index == target_mass.index).all()
assert penguins_df.shape == (344,7)
```

<div dir="rtl" lang="he" style="text-align:right">

## שגיאת בחירה
למבחר של שתי עמודות צריך רשימת שמות בתוך הסוגריים החיצוניים. כאשר שמים שני שמות מופרדים בפסיק בביטוי אחד זה לא תחביר בחירת מספר עמודות בטבלה הרגילה שלנו.

</div>

```python
correct_review = penguins_df[['flipper_length_mm','body_mass_g']]
assert correct_review.shape == (344,2)
```

<div dir="rtl" lang="he" style="text-align:right">

## חלק 2 — מסכה לפי קטגוריה
מסכה היא Series בוליאנית באורך 344. Gentoo מסנן ל־124 שורות עם 7 עמודות.

</div>

```python
gentoo_mask = penguins_df['species'] == 'Gentoo'
display(gentoo_mask.head(5))
gentoo_df = penguins_df[gentoo_mask]
print('Gentoo dimensions:',gentoo_df.shape)
display(gentoo_df.head())
assert gentoo_mask.shape == (344,)
assert gentoo_df.shape == (124,7)
```

<div dir="rtl" lang="he" style="text-align:right">

## Dream — אותה כמות אבל לא אותה קבוצה
סינון עצמאי מן המקור נותן 124 תצפיות מאי Dream. אין להסיק זהות קבוצות מהשוואת המספר 124.

</div>

```python
dream_mask = penguins_df['island'] == 'Dream'
dream_df = penguins_df[dream_mask]
print('Dream dimensions:',dream_df.shape)
assert dream_df.shape == (124,7)
assert set(gentoo_df.index) != set(dream_df.index)
```

<div dir="rtl" lang="he" style="text-align:right">

## סינון לפי מדידת סנפיר
>=220 מחזיר 43 תצפיות; חסרים אינם עוברים את הסף, אבל אין הוכחה שהמדידה החסרה קצרה.

</div>

```python
long_flipper_mask = penguins_df['flipper_length_mm'] >= 220
long_flippers_df = penguins_df[long_flipper_mask]
print('At least 220 mm:',long_flippers_df.shape)
assert long_flippers_df.shape == (43,7)
assert penguins_df['flipper_length_mm'].isna().sum() == 2
```

<div dir="rtl" lang="he" style="text-align:right">

## שדות לעיון לאחר בחירת שורות
מספר השורות נשאר 43, אבל שתי עמודות נבחרו.

</div>

```python
flipper_review = long_flippers_df[['species','flipper_length_mm']]
display(flipper_review.head())
print('Review dimensions:',flipper_review.shape)
assert flipper_review.shape == (43,2)
assert penguins_df.shape == (344,7)
```

<div dir="rtl" lang="he" style="text-align:right">

## הרחבה — הבדל סף
>220 מחזיר 35; ההפרש 8 הוא מספר התצפיות שמדידתן בדיוק 220 מ״מ.

</div>

```python
strict_mask = penguins_df['flipper_length_mm'] > 220
strict_df = penguins_df[strict_mask]
equal_count = (penguins_df['flipper_length_mm'] == 220).sum()
print('At least 220:',len(long_flippers_df),'Over 220:',len(strict_df),'Exactly 220:',equal_count)
assert len(strict_df) == 35
assert equal_count == 8
assert len(long_flippers_df) - len(strict_df) == equal_count
```

<div dir="rtl" lang="he" style="text-align:right">

## דו״ח מסכם למורה
ממדי Series ו־DataFrame שונים אך אינם משנים ערכים כשנבחרת אותה עמודה. בחירת עמודות משאירה 344 שורות; מסכת Gentoo או Dream בוחרת כל אחת 124 (לא אותן שורות); מסכת אורך סנפיר >=220 בוחרת 43. תלמיד צריך לציין יחידות, מקור וקבוצה, ולהסביר שגם קבוצת נתונים מסוננת אינה ראיה לסיבתיות, ייצוגיות או איכות חיזוי.

**שמירה על שלמות X/y:** כל שורה חייבת להתאים ליעד של אותו פרט; לא למיין צד אחד בנפרד ולא להכניס את היעד למאפייניו.
