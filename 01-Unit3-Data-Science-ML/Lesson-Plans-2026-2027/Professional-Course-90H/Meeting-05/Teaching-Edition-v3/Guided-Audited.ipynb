# מפגש 5 — מחברת מורה מודרכת: עמודות או שורות?
**דוגמה א:** מחשבים A–D מומצאים — לא פתרון מטלת Penguins. לפני הרצת כל קוד: חיזוי פלט וממדים. לאחריה: הסבר במילים.

סיפור: דודו רוצה לבחור שדות לחיזוי מחיר, ואז לזהות מחשבים שעומדים בתקציב. אלה שני סוגי בחירה.

</div>

```python
import pandas as pd
computers_df = pd.DataFrame({
    'computer_id':['A','B','C','D'],
    'year':[2021,2023,2020,2024],
    'memory_gb':[8,16,8,32],
    'price_nis':[2100,2700,3600,2400]
})
display(computers_df)
assert computers_df.shape == (4,4)
```

<div dir="rtl" lang="he" style="text-align:right">

## חלק א — מה ההבדל בין Series ל־DataFrame?
בחירה של שם יחיד תיצור Series חד־ממדית. בחירה באמצעות רשימה של שם תיצור DataFrame. הביאו הוכחה מממדים וסוגים.

</div>

```python
price_series = computers_df['price_nis']
price_table = computers_df[['price_nis']]
print('Series:', type(price_series), price_series.shape)
print('DataFrame:', type(price_table), price_table.shape)
assert price_series.shape == (4,)
assert price_table.shape == (4,1)
```

<div dir="rtl" lang="he" style="text-align:right">

## בוחרים שתי עמודות בלבד
האם בחירה של year ו־memory_gb מסירה מחשבים? לפני הרצה חזו shape.

</div>

```python
X = computers_df[['year','memory_gb']]
y = computers_df['price_nis']
display(X)
print('X:',X.shape,'y:',y.shape)
assert X.shape == (4,2) and y.shape == (4,)
assert 'price_nis' not in X.columns
```

<div dir="rtl" lang="he" style="text-align:right">

## חלק ב — מסכה בוליאנית
תקציב 2500. חזו ארבע תוצאות לפי הסדר A,B,C,D. True אינו הסתברות: הוא תוצאת תנאי.

</div>

```python
budget_nis = 2500
mask = computers_df['price_nis'] <= budget_nis
print(mask)
assert mask.tolist() == [True,False,False,True]
```

<div dir="rtl" lang="he" style="text-align:right">

## בוחרים את השורות המתאימות
המסכה שומרת את כל שדות A ו־D, לא מוחקת את המקור. כמה שורות? כמה עמודות? מה יהיו האינדקסים?

</div>

```python
affordable_df = computers_df[mask]
display(affordable_df)
print('Filtered:',affordable_df.shape,'Original:',computers_df.shape)
assert affordable_df.shape == (2,4)
assert computers_df.shape == (4,4)
assert affordable_df['computer_id'].tolist() == ['A','D']
```

<div dir="rtl" lang="he" style="text-align:right">

## חישוב ידני: כמה אחוז מההצעות מתאימות?
מכיוון שנבחרו שתי שורות מתוך ארבע, 2/4×100=50%. זה רק אחוז של ארבע הצעות מומצאות, לא נתון על השוק.

</div>

```python
selected_count = len(affordable_df)
all_count = len(computers_df)
selection_rate = selected_count/all_count*100
print('Selection rate (%):',selection_rate)
assert (selected_count,all_count,selection_rate) == (2,4,50)
```

<div dir="rtl" lang="he" style="text-align:right">

## משלבים: בחירת שורות ואז עמודות
בחירה של שני שדות מתוך שתי השורות שנבחרו צריכה לתת (2,2). לא שינינו אף ערך במקור.

</div>

```python
review_df = affordable_df[['year','memory_gb']]
display(review_df)
assert review_df.shape == (2,2)
assert computers_df.shape == (4,4)
```

<div dir="rtl" lang="he" style="text-align:right">

## גרף מורה מוכן בלבד — אין צורך ללמוד Matplotlib
הגרף מציג את ארבעת המחירים ואת התקציב. ערכים מתחת לקו או בגובהו עומדים בתנאי.

</div>

```python
import matplotlib.pyplot as plt
plt.figure(figsize=(6.5,4))
plt.bar(computers_df['computer_id'],computers_df['price_nis'])
plt.axhline(budget_nis,linestyle='--',label='Budget: 2500 NIS')
plt.xlabel('Computer ID')
plt.ylabel('Price (NIS)')
plt.ylim(0,4200)
plt.legend()
plt.show()
```
