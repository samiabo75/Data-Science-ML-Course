## חישוב הכנסה ידני

(R=q\cdot p=12\times8=96) ש״ח הכנסה (לא רווח). שינוי קלט מ־12 ל־15 ישנה את התוצאה ל־120 ש״ח ללא אימון מודל.

</div>

```python
sandwiches_sold = 12
price_per_sandwich_nis = 8
revenue_nis = sandwiches_sold * price_per_sandwich_nis
print('הכנסה בשקלים:', revenue_nis)
assert revenue_nis == 96
print('ניסוי עם 15 כריכים:', 15 * price_per_sandwich_nis)
assert 15 * price_per_sandwich_nis == 120
```

## הגרף — הציגו אותו רק כהמחשה של נתונים מומצאים

```python
# זהו קוד הדגמה מוכן — לא נדרש לדעת לכתוב matplotlib במפגש 1
import matplotlib.pyplot as plt
sales = [12, 16, 9, 15]
days = ["Sun", "Mon", "Tue", "Wed"]
plt.figure(figsize=(6, 3))
plt.bar(days, sales)
plt.xlabel("Day")
plt.ylabel("Sandwiches sold (count)")
plt.title("Illustrative kiosk observations (invented)")
plt.ylim(0, 18)
plt.show()
```

## ממוצע ידני

## פתרון משימת הספרייה

7+5=12 השאלות ביום המתואר. אם שואלים כמה יושאלו מחר, צריך מידע היסטורי, תיעוד ומבחן ביצועים; חיבור אינו למידה.

</div>

```python
hebrew_loans = 7
arabic_loans = 5
total_loans = hebrew_loans + arabic_loans
print('מספר השאלות ביום שנבדק:', total_loans)
assert total_loans == 12
```

## פתרון המחשב המשומש

גיל בשנות לוח: 2026−2022=4. ממוצע: (2400+3000)/2=2700 ₪. הפרש: 3000−2400=600 ₪. אין להסיק מהדוגמה את שווי השוק.

</div>

```python
manufacture_year = 2022
reference_year = 2026
memory_gb = 16
price_a_nis = 2400
price_b_nis = 3000
age_years = reference_year - manufacture_year
average_price_nis = (price_a_nis + price_b_nis) / 2
price_difference_nis = price_b_nis - price_a_nis
print('גיל בשנות לוח:', age_years)
print('ממוצע מחירים בשקלים:', average_price_nis)
print('הפרש מחירים בשקלים:', price_difference_nis)
assert (age_years, average_price_nis, price_difference_nis) == (4, 2700, 600)
```

### שינוי מבוקר

שנת 2024 ומחיר 2600: גיל 2, ממוצע 2800, הפרש 400. הסבירו את סיבת כל שינוי.

</div>

```python
test_year = 2024
test_price_nis = 2600
print('בניסוי: גיל', reference_year - test_year)
print('בניסוי: ממוצע', (test_price_nis + price_b_nis) / 2)
print('בניסוי: הפרש', price_b_nis - test_price_nis)
assert reference_year - test_year == 2
assert (test_price_nis + price_b_nis) / 2 == 2800
assert price_b_nis - test_price_nis == 400
```
