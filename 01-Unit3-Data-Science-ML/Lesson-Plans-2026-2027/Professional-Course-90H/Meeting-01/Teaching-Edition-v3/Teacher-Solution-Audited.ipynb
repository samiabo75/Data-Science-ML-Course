## דוגמת הקיוסק
כמות 12, מחיר 8 ₪ לכריך. כפל: 12×8=96 ₪ הכנסה, לא רווח. בודקים שינוי מבוקר לכמות 15 → 120 ₪ בלי אימון.

</div>

```python
sandwiches_sold = 12
price_per_sandwich_nis = 8
revenue_nis = sandwiches_sold * price_per_sandwich_nis
print("Revenue (NIS):", revenue_nis)
assert revenue_nis == 96
assert 15 * price_per_sandwich_nis == 120
```

<div dir="rtl" lang="he" style="text-align:right;">

## ארבע תצפיות והגרף
נתונים מומצאים, לא מחקר. כלל ראשון: גרף אמין חייב להציג אותם ארבעה מספרים שנמצאים בטבלה.

</div>

```python
# תא הדגמה נתון: תחביר גרפים יילמד בהמשך היחידה
import matplotlib.pyplot as plt

days = ["Sun", "Mon", "Tue", "Wed"]
sold = [12, 16, 9, 15]   # נתונים מומצאים להוראה

plt.figure(figsize=(7, 4))
plt.bar(days, sold, color="#3173AD")
plt.ylim(0, 18)
plt.xlabel("Day")
plt.ylabel("Sandwiches sold (count)")
plt.title("Illustrative sales - invented example")
for day_index in range(4):
    plt.text(day_index, sold[day_index] + 0.25, str(sold[day_index]),
             ha="center")
plt.show()
```

<div dir="rtl" lang="he" style="text-align:right;">

## פתרון ידני לממוצע
12+16+9+15=52; מספר ימים 4; ממוצע 52/4=13 כריכים ליום. זו סטטיסטיקה תיאורית.

</div>

```python
total_sold = 12 + 16 + 9 + 15
number_of_days = 4
average_daily_sales = total_sold / number_of_days
print("Average daily sales:", average_daily_sales)
assert total_sold == 52 and average_daily_sales == 13
```

<div dir="rtl" lang="he" style="text-align:right;">

## משימת ספרייה
7 ספרים בעברית ועוד 5 בערבית → 12 השאלות באותו יום. זהו חישוב לפי כלל, לא מודל ML.

</div>

```python
hebrew_books = 7
arabic_books = 5
total_books = hebrew_books + arabic_books
print("Total loans:", total_books)
assert total_books == 12
```

## מחשב משומש: ייצוג נתונים
יש להקפיד על יחידות. אין ללמוד על שווי שוק של מחשב ממחירם של שני מחשבים בלבד.

</div>

```python
computer_model = "Study laptop"
manufacture_year = 2022
memory_gb = 16
price_nis = 2400
reference_year = 2026
print("Model:", computer_model)
print("Memory (GB):", memory_gb)
print("Price (NIS):", price_nis)
assert (manufacture_year, memory_gb, price_nis, reference_year) == (2022, 16, 2400, 2026)
```

<div dir="rtl" lang="he" style="text-align:right;">

## חישוב גיל
2026−2022=4 **שנות לוח**. כדי לחשב גיל מדויק דרושים גם יום וחודש.

</div>

```python
computer_age_years = reference_year - manufacture_year
print("Approx. calendar years:", computer_age_years)
assert computer_age_years == 4
```

<div dir="rtl" lang="he" style="text-align:right;">

## ממוצע והפרש מחירים
ממוצע: (2400+3000)/2=2700 ₪. הפרש: 3000−2400=600 ₪.

</div>

```python
other_price_nis = 3000
average_price_nis = (price_nis + other_price_nis) / 2
price_difference_nis = other_price_nis - price_nis
print("Average price (NIS):", average_price_nis)
print("Difference (NIS):", price_difference_nis)
assert (average_price_nis, price_difference_nis) == (2700, 600)
```

<div dir="rtl" lang="he" style="text-align:right;">

## הרחבה: הנחה
2400×10/100=240 ₪ הנחה; מחיר סופי 2160 ₪. בשינוי שנה 2024 ומחיר 2600, גיל 2, ממוצע 2800 והפרש 400.

</div>

```python
discount_percent = 10
discount_nis = price_nis * discount_percent / 100
final_price_nis = price_nis - discount_nis
print("Discount (NIS):", discount_nis)
print("After discount (NIS):", final_price_nis)
assert (discount_nis, final_price_nis) == (240, 2160)
new_year = 2024
new_price = 2600
assert reference_year - new_year == 2
assert (new_price + other_price_nis) / 2 == 2800
assert other_price_nis - new_price == 400
```
