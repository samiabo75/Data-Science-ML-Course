# מפגש 1 — מחברת מודרכת לאחר בקרת איכות
סיפור דודו והקיוסק. לפני כל הרצה: כותבים ציפייה לפלט ומנמקים. הנתונים על ארבעת הימים מומצאים ללמידה.

**מטרות:** להבין מהי תצפית, להבחין בין חישוב לחיזוי, להשתמש במשתנים ולהסביר פלט.

</div>

<div dir="rtl" lang="he" style="text-align:right;">

## שלב א — קודם מחשבים ידנית
נתון: 12 כריכים נמכרו במחיר 8 ₪ לכריך. הנוסחה היא R=q×p. הציבו q=12 ו־p=8. קבלו 96 ₪ הכנסה. **הכנסה אינה רווח** — אין מידע על עלויות.

כתבו מראש איזה מספר תדפיס התוכנית.

</div>

```python
sandwiches_sold = 12
price_per_sandwich_nis = 8
revenue_nis = sandwiches_sold * price_per_sandwich_nis
print("Revenue (NIS):", revenue_nis)
assert revenue_nis == 96
```

<div dir="rtl" lang="he" style="text-align:right;">

## שלב ב — שינוי קלט, לא למידה
נניח שנמכרו 15 כריכים באותו מחיר. קודם מחשבים ביד 15×8, ואז בודקים. שינוי קלט לא הפך את התוכנית למודל ML.

</div>

```python
new_quantity = 15
new_revenue_nis = new_quantity * price_per_sandwich_nis
print("If 15 sandwiches are sold:", new_revenue_nis)
assert new_revenue_nis == 120
```

## שלב ג — ארבע תצפיות וגרף
ראשון 12, שני 16, שלישי 9, רביעי 15. **אלה מספרי דוגמה מומצאים.** כל שורה בטבלה מייצגת יום; יעד אפשרי להמשך יהיה מכירות ביום חדש. הגרף הבא מיוצר מקוד מוכן — אין צורך ללמוד כעת רשימות, לולאות או Matplotlib.

</div>

```python
# תא הדגמה נתון: תחביר גרפים יילמד בהמשך היחידה
import matplotlib.pyplot as plt

days = ["Sun", "Mon", "Tue", "Wed"]
sold = [12, 16, 9, 15]   # נתונים מומצאים להוראה

plt.figure(figsize=(7, 4))
plt.bar(days, sold, color="#3173AD")
plt.ylim(0, 18)
plt.xlabel("Day")
plt.ylabel("Sandwiches sold (count)")
plt.title("Illustrative sales - invented example")
for day_index in range(4):
    plt.text(day_index, sold[day_index] + 0.25, str(sold[day_index]),
             ha="center")
plt.show()
```

<div dir="rtl" lang="he" style="text-align:right;">

## שלב ד — ממוצע: חישוב ידני לפני Python
סכום: 12+16+9+15=52 כריכים. מספר ימים: 4. ממוצע: 52/4=13 כריכים ליום.

כאן הממוצע מתאר את העבר. הוא עשוי לשמש בעתיד תחזית בסיס, אבל **אינו מודל מאומן ואינו מבטיח את מחר**.

</div>

```python
total_sold = 12 + 16 + 9 + 15
number_of_days = 4
average_daily_sales = total_sold / number_of_days
print("Average sandwiches per day:", average_daily_sales)
assert total_sold == 52 and average_daily_sales == 13
```
