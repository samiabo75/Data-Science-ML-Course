## פתרון חלק 2

<div dir="rtl" lang="he" style="text-align: right;">

### 1. הנתונים
שם הדגם הוא טקסט; השנה, הזיכרון והמחיר הם ערכים מספריים. יש לציין יחידות.

</div>

```python
computer_model = "מחשב לדוגמה"
manufacture_year = 2022
memory_gb = 16
price_nis = 2400
reference_year = 2026
print("דגם:", computer_model)
print("שנת ייצור:", manufacture_year)
print("זיכרון ב-GB:", memory_gb)
print("מחיר בשקלים:", price_nis)
```

<div dir="rtl" lang="he" style="text-align: right;">

### 2. גיל
2026−2022=4 שנים. החיסור הופך שני נתוני שנה למאפיין מחושב, בקירוב שנות לוח.

</div>

```python
computer_age_years = reference_year - manufacture_year
print("גיל בשנות לוח:", computer_age_years)
```

<div dir="rtl" lang="he" style="text-align: right;">

### 3. ממוצע והפרש
(2400+3000)/2=2700 ₪; 3000−2400=600 ₪. סדר הפעולות מחייב סוגריים בחישוב הממוצע.

</div>

```python
other_price_nis = 3000
average_price_nis = (price_nis + other_price_nis) / 2
price_difference_nis = other_price_nis - price_nis
print("ממוצע בשקלים:", average_price_nis)
print("הפרש בשקלים:", price_difference_nis)
```

<div dir="rtl" lang="he" style="text-align: right;">

### 4. אחוזים
2400×10/100=240 ₪ הנחה; 2400−240=2160 ₪ לתשלום. זהו כלל ידני שהוגדר מראש.

</div>

```python
discount_percent = 10
discount_nis = price_nis * discount_percent / 100
final_price_nis = price_nis - discount_nis
print("הנחה בשקלים:", discount_nis)
print("מחיר סופי בשקלים:", final_price_nis)
```

### 6. תיעוד לדוגמה

שנת ייצור וזיכרון עשויים להיות מאפיינים; מחיר הוא יעד אפשרי. למידת מכונה מחייבת דוגמאות ותהליך התאמת מודל; כאן חישבנו כלל ידוע. שתי דוגמאות אינן בסיס לתחזית אמינה. אין להסיק קשר סיבתי משנת ייצור ומחיר.

### אתגר
סכום שלושת המחירים חלקי מספרם: (2400+3000+3600)/3=3000 ₪.

</div>

```python
third_price_nis = 3600
three_price_average_nis = (price_nis + other_price_nis + third_price_nis) / 3
print("ממוצע שלושה מחירים בשקלים:", three_price_average_nis)
```
