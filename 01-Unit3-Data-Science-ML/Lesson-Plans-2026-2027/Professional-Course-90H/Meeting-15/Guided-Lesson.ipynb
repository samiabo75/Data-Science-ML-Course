## המושגים הנחוצים
KNN שומר דוגמאות מתויגות. בחיזוי מחשבים מרחק, בוחרים K שכנים ומצביעים לפי התוויות. מרחק אוקלידי הוא שורש סכום ריבועי הפרשים. היעד אינו חלק מהמרחק. מגדירים מדיניות לתיקו, ובודקים השפעת סולם ו־K בלי להבטיח שגדול תמיד טוב יותר.

</div>

<div dir="rtl" lang="he" style="text-align: right;">

## דוגמה פתורה לפני קוד
המרחק בין [0,0] ל־[3,4] הוא 5. אם שלוש התוויות הקרובות הן A,B,B, תחזית K=3 היא B. אמת הדוגמה החדשה עדיין אינה ידועה.

</div>

```python
dx = 3 - 0
dy = 4 - 0
distance = (dx ** 2 + dy ** 2) ** 0.5
print(distance)
```

<div dir="rtl" lang="he" style="text-align: right;">

**פלט צפוי:** 5.0

## מדוע זה חשוב?
מרחק, ייצוג מאפיינים ו־KNN הם גשר ישיר למסווגים מבוססי מרחק ביחידה 5. אין מלמדים את יחידת ההמשך במלואה כעת.

### חישוב ופירוש
קראו את השמות והיחידות, כתבו ציפייה לפני הרצה ובדקו אותה בפלט. הסבירו מה השתנה ומה נשאר קבוע. אין צורך לשנן פקודה בלי להבין את פעולתה.

</div>

```python
import pandas as pd
training_df = pd.DataFrame({
    "id": ["A", "B", "C", "D", "E"],
    "f1": [3, 5, 5, 7, 3],
    "f2": [2, 3, 5, 3, 8],
    "label": ["Red", "Blue", "Blue", "Red", "Red"]
})
query_f1 = 3
query_f2 = 3
```

<div dir="rtl" lang="he" style="text-align: right;">

### המחשה חזותית
לפני הרצה הגדירו שאלה, סוגי משתנים ויחידות. אחרי הרצה פרשו את הצירים ואת הממצא. אין להסיק סיבתיות או למחוק קצה רק מן הציור.

</div>

```python
import seaborn as sns
import matplotlib.pyplot as plt
plt.figure(figsize=(7, 5))
sns.scatterplot(data=training_df, x="f1", y="f2", hue="label", palette={"Red": "firebrick", "Blue": "royalblue"}, s=80)
plt.scatter([query_f1], [query_f2], marker="*", s=220, color="black", label="New observation")
plt.title("Teaching data for nearest-neighbor classification")
plt.xlabel("Feature 1 (teaching scale)")
plt.ylabel("Feature 2 (teaching scale)")
plt.axis("equal")
plt.legend()
plt.tight_layout()
plt.show()
```

<div dir="rtl" lang="he" style="text-align: right;">

### חישוב ופירוש
קראו את השמות והיחידות, כתבו ציפייה לפני הרצה ובדקו אותה בפלט. הסבירו מה השתנה ומה נשאר קבוע. אין צורך לשנן פקודה בלי להבין את פעולתה.

</div>

```python
distance_df = training_df.copy()
distance_df["distance"] = (
    (distance_df["f1"] - query_f1) ** 2 +
    (distance_df["f2"] - query_f2) ** 2
) ** 0.5
ranked_df = distance_df.sort_values("distance")
display(ranked_df)
```

<div dir="rtl" lang="he" style="text-align: right;">

### חישוב ופירוש
קראו את השמות והיחידות, כתבו ציפייה לפני הרצה ובדקו אותה בפלט. הסבירו מה השתנה ומה נשאר קבוע. אין צורך לשנן פקודה בלי להבין את פעולתה.

</div>

```python
neighbors_1 = ranked_df.head(1)
neighbors_3 = ranked_df.head(3)
neighbors_5 = ranked_df.head(5)
print("תחזית/תחזיות K=1:")
print(neighbors_1["label"].mode())
print("תחזית/תחזיות K=3:")
print(neighbors_3["label"].mode())
print("תחזית/תחזיות K=5:")
print(neighbors_5["label"].mode())
```
