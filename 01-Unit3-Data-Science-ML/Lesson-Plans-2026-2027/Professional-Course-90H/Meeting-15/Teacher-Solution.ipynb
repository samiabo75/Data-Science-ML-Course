### 2–3. קלט ומרחקים
המאפיינים f1, f2; label היא תווית, id מזהה בלבד. מרחקים: P1=1, P2=2, P3=√8≈2.83, P4=4, P5=5. כל המרחקים בדוגמה שונים, ולכן אין תיקו סביב גבול השכנות.

</div>

```python
import pandas as pd
training_df = pd.DataFrame({
    "id": ["P1", "P2", "P3", "P4", "P5"],
    "f1": [2, 4, 4, 6, 2],
    "f2": [1, 2, 4, 2, 7],
    "label": ["B", "A", "A", "B", "A"]
})
query_f1 = 2
query_f2 = 2
distance_df = training_df.copy()
distance_df["distance"] = (
    (distance_df["f1"] - query_f1) ** 2 +
    (distance_df["f2"] - query_f2) ** 2
) ** 0.5
ranked_df = distance_df.sort_values("distance")
display(ranked_df)
```

<div dir="rtl" lang="he" style="text-align: right;">

### 4. תחזיות
K=1: B. K=3: שתי A מול B אחת, לכן A. K=5: שלוש A מול שתי B, לכן A. שינוי K אינו חייב לשנות את התחזית.

</div>

```python
neighbors_1 = ranked_df.head(1)
neighbors_3 = ranked_df.head(3)
neighbors_5 = ranked_df.head(5)
print("K=1:")
print(neighbors_1["label"].mode())
print("K=3:")
print(neighbors_3["label"].mode())
print("K=5:")
print(neighbors_5["label"].mode())
display(neighbors_3[["id", "label", "distance"]])
```

### 5–6. פירוש
אין תווית אמת לקלט החדש ואין קבוצת בדיקה, ולכן אין לחשב Accuracy מהתחזיות האלה. K=7 אינו אפשרי לבקשת שבעה שכנים מתוך חמישה. K=1 עלול להיות רגיש לתווית שגויה או לתצפית חריגה קרובה. K גדול עלול לטשטש קבוצות מקומיות.

דוגמת תיעוד: סיווג A/B משני מאפיינים מספריים בעלי סקאלה שנקבעה בדוגמת הוראה. מרחק אוקלידי והצבעת קולות שווה; K=3 בחר P1, P2, P3 והחזיר A ברוב 2:1. זו תחזית, לא אמת. בנתונים אמיתיים נדרשים בדיקת איכות, חלוקה והכנה מאימון, בחירת K באימות והערכה בסיום.
