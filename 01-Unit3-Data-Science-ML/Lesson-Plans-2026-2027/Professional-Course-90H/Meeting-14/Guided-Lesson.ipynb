### תשתית הנתונים והחלוקה
קראו תחילה את היעד ואת שמות המאפיינים. החלוקה שומרת התאמה בין שורות X ל־y. הפרמטר random_state מאפשר שחזור ואינו משפר דיוק בפני עצמו.

</div>

```python
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import OneHotEncoder
url = "https://raw.githubusercontent.com/samiabo75/Data-Science-ML-Course/main/Datasets/penguins.csv"
df = pd.read_csv(url)
model_df = df.dropna(subset=["species"]).copy()
X = model_df[["island"]]
y = model_df["species"]
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.25, random_state=42, stratify=y
)
print("חסרים במאפיין האימון:")
print(X_train.isna().sum())
print("לפני קידוד:", X_train.shape, X_test.shape)
```

<div dir="rtl" lang="he" style="text-align: right;">

### התאמה ויישום
זהו מה לומד פרמטרים ובאילו שורות נעשה fit. ודאו שכל הכנה שנלמדת מנתונים משתמשת באימון המתאים. לאחר מכן פירשו את הפלט במילים.

</div>

```python
encoder = OneHotEncoder(handle_unknown="ignore", sparse_output=False)
encoder.set_output(transform="pandas")
encoder.fit(X_train)
X_train_encoded = encoder.transform(X_train)
X_test_encoded = encoder.transform(X_test)
print("קטגוריות שנלמדו באימון:", encoder.categories_)
print("אחרי קידוד:", X_train_encoded.shape, X_test_encoded.shape)
display(X_train_encoded.head(3))
display(X_test_encoded.head(3))
```

## גשר יישומי — שתי הכנות לטבלה אחת
זהו קוד הכנה בלבד, לפני מודל. numeric ו־categorical הם שמות מסלולים. תכונת הקבוצה אינה סולם סדרי.

</div>

```python
import pandas as pd
from sklearn.compose import ColumnTransformer
from sklearn.impute import SimpleImputer
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler, OneHotEncoder
example_train = pd.DataFrame({"age": [10., 20., None], "group": ["A", "B", "A"]})
prepare = ColumnTransformer([
    ("numeric", Pipeline([("fill", SimpleImputer(strategy="median")), ("scale", StandardScaler())]), ["age"]),
    ("categorical", OneHotEncoder(handle_unknown="ignore", sparse_output=False), ["group"])
])
print(prepare.fit_transform(example_train))
```
