### 1–2. הגדרה
island שמית, ללא סדר טבעי. נמוך/בינוני/גבוה סדרית, אך אין הבטחה למרווחים שווים. המשימה מדגימה מאפיין קטגוריאלי בלבד, לא בחירת מאפיין מיטבית.

</div>

```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import OneHotEncoder
url = "https://raw.githubusercontent.com/samiabo75/Data-Science-ML-Course/main/Datasets/penguins.csv"
penguins_df = pd.read_csv(url)
model_df = penguins_df.dropna(subset=["species"]).copy()
X = model_df[["island"]]
y = model_df["species"]
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.25, random_state=17, stratify=y
)
print("לפני:", X_train.shape, X_test.shape)
print("חסרי אימון:")
print(X_train.isna().sum())
```

<div dir="rtl" lang="he" style="text-align: right;">

### 3. אותו מבנה לשתי הקבוצות
encoder לומד קטגוריות באימון. אין fit ב־ Test. בקובץ ובחלוקה האלה נלמדות שלוש קטגוריות ומתקבלות שלוש עמודות.

</div>

```python
encoder = OneHotEncoder(handle_unknown="ignore", sparse_output=False)
encoder.set_output(transform="pandas")
encoder.fit(X_train)
X_train_encoded = encoder.transform(X_train)
X_test_encoded = encoder.transform(X_test)
print("קטגוריות:", encoder.categories_)
print("עמודות:", X_train_encoded.columns)
print("אחרי:", X_train_encoded.shape, X_test_encoded.shape)
display(X_train_encoded.head(3))
display(X_test_encoded.head(3))
```

## פתרון להעברת ההכנה המשולבת
החציון שנלמד הוא10. אחרי השלמה הגילים5,15,10. התקנון נלמד מאימון בלבד. שתי עמודות One-Hot נוצרות ל־Cול־D. הקטגוריהEחדשה ולכן מקבלת אפסים במדיניות הנתונה; זהו ייצוג טכני ולא הוכחה שהיא דומה לכולן.

</div>

```python
import pandas as pd
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler, OneHotEncoder
lab_train = pd.DataFrame({"age": [5., 15., None], "group": ["C", "D", "D"]})
lab_new = pd.DataFrame({"age": [10.], "group": ["E"]})
prepare = ColumnTransformer([
    ("numeric", Pipeline([("fill", SimpleImputer(strategy="median")), ("scale", StandardScaler())]), ["age"]),
    ("categorical", OneHotEncoder(handle_unknown="ignore", sparse_output=False), ["group"])
])
train_prepared = prepare.fit_transform(lab_train)
new_prepared = prepare.transform(lab_new)
print(train_prepared)
print(new_prepared)
print(train_prepared.shape, new_prepared.shape)
```
