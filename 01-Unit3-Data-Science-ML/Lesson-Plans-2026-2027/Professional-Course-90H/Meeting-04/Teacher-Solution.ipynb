## מקור וגרסה

### 1–2. קובץ וטעינה
שורת הכותרת מתחילה ב־ species; הרשומה שאחריה מתארת פינגווין Adelie באי Torgersen. לדוגמה אורך מקור 39.1 מ״מ. CSV הוא קובץ; DataFrame הוא אובייקט טבלאי בזיכרון לאחר הקריאה.

</div>

```python
import pandas as pd
dataset_url = "https://raw.githubusercontent.com/samiabo75/Data-Science-ML-Course/main/Datasets/penguins.csv"
penguins_df = pd.read_csv(dataset_url)
```

### 5. חסרים
ברשומה באינדקס 3 ידועים species=Adelie ו־ island=Torgersen; חסרים ארבע המדידות והזוויג. חסר אינו מספר אפס. החלטה על טיפול מחייבת בדיקת היקף וסיבה; לא בוצע טיפול כאן.

### 6. כרטיס מאגר לדוגמה
מקור: Palmer Penguins, גרסת CSV בעלת 7 שדות ב־ Datasets של הקורס, נבדקה ב־ 19.10.2026. תצפית: פינגווין אחד. ממדים: 344×7.
יעד אפשרי: body_mass_g, מסה בגרמים — יעד מספרי לבעיית רגרסיה. מאפיינים אפשריים: flipper_length_mm ו־ bill_length_mm. אין להזין את המסה עצמה כקלט לחיזוי אותה מסה.

שאלה: האם מדידות מקור וסנפיר מסייעות לחיזוי המסה? טרם נבדק קשר, לא הותאם מודל ולא הוערכה הכללה. הנתונים מתארים דגימה מאזור מוגדר, לא את כלל הפינגווינים. יש חסרים שמחייבים בירור.

קטלוג הקורס מתעד דרישת 1000 רשומות במסמך ההשתלמות שסופק; לכן 344 הרשומות אינן עומדות ברף זה. זו הבחנה מול תיעוד ההשתלמות, לא הכרעה לגבי כל מחוון רשמי עדכני.

</div>
