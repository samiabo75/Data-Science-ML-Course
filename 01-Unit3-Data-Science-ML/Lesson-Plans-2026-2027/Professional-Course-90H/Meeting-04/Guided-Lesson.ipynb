## המושגים הנחוצים
CSV הוא קובץ טבלאי טקסטואלי. pd.read_csv קוראת אותו לטבלה. info מתארת מבנה, טיפוסים וערכים ידועים. describe מפיקה סיכום סטטיסטי. חסר אינו אפס. כרטיס מקור כולל יוצר, גרסה, תאריך, מילון שדות, יחידות ותנאי שימוש.

## דוגמה פתורה לפני קוד
בטבלה בעלת ארבע שורות ועמודת גיל עם שלושה ערכים ידועים יש חסר אחד. count בסיכום מתייחס לערכים הידועים בעמודה, ולא בהכרח לכל השורות.

</div>

```python
import pandas as pd
df = pd.DataFrame({"age": [10, 20, None, 30]})
print(df.shape)
print(df["age"].count())
```
