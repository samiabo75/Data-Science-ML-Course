## 1. סיפור, מקור ויחידות
דודו מצא ברשת מדידות פינגווינים. האם הוא רשאי להסיק שהקובץ מייצג את כל פינגוויני העולם? לא. מקור הנתונים והיחידות חשובים לפני חישוב.

מקור: https://github.com/allisonhorst/palmerpenguins | קובץ הקורס: https://github.com/samiabo75/Data-Science-ML-Course/blob/main/Datasets/penguins.csv

תצפית: רשומת מדידה של פינגווין אחד. אורך מקור ב־mm ומסת גוף ב־g.

</div>

<div dir="rtl" lang="he" style="text-align:right;">

## 2. ארבע רשומות ראשונות מהמקור — חישוב ידני
אורכי המקור: 39.1, 39.5, 40.3, NA במ״מ. לכן יש ארבע שורות, שלוש מדידות ידועות ואחת חסרה; שיעור 1/4×100=25% **בארבעת הרשומות בלבד**. ממוצע ידועים 118.9/3≈39.63 מ״מ.

</div>

```python
import pandas as pd

tiny_df = pd.DataFrame({'bill_length_mm': [39.1, 39.5, 40.3, None]})
print('Shape:', tiny_df.shape)
print('Known measurements:', tiny_df['bill_length_mm'].count())
print('Mean known length (mm):', tiny_df['bill_length_mm'].mean())
assert tiny_df.shape == (4, 1)
assert tiny_df['bill_length_mm'].count() == 3
assert abs(tiny_df['bill_length_mm'].mean() - 118.9 / 3) < 1e-9
```

<div dir="rtl" lang="he" style="text-align:right;">

## 3. למה לא משנים חסר לאפס?
הציבו בכתב: ממוצע ללא חסר =118.9/3; עם אפס מומצא =118.9/4. השני נותן 29.725, אך אין לנו מדידה של 0 ולכן אין הצדקה לשינוי הזה.

</div>

```python
correct_mean_mm = (39.1 + 39.5 + 40.3) / 3
invalid_zero_mean_mm = (39.1 + 39.5 + 40.3 + 0) / 4
print('Observed mean (mm):', correct_mean_mm)
print('Invented-zero mean (not justified):', invalid_zero_mean_mm)
assert abs(correct_mean_mm - 39.63333333333333) < 1e-8
assert abs(invalid_zero_mean_mm - 29.725) < 1e-9
```

<div dir="rtl" lang="he" style="text-align:right;">

## 4. טעינת קובץ CSV אמיתי מקישור Raw
ה־URL הבא מצביע על קובץ הקורס. אם הרשת אינה נגישה, העלו את penguins.csv למערכת Files של Colab והחליפו את כתובת הקובץ בנתיב שהועלה. **לא משתמשים בכתובת blob של דף GitHub.**

</div>

```python
dataset_url = "https://raw.githubusercontent.com/samiabo75/Data-Science-ML-Course/main/Datasets/penguins.csv"
penguins_df = pd.read_csv(dataset_url)
print('Rows and columns:', penguins_df.shape)
print('Column names:', list(penguins_df.columns))
assert penguins_df.shape == (344, 7)
```

<div dir="rtl" lang="he" style="text-align:right;">

## 5. הצצה אינה מחיקה
הראו שש ראשונות ושתי אחרונות. בדקו שוב את ממדי המקור. האם כל המינים מופיעים בשורות הראשונות? לא חייבים.

</div>

```python
display(penguins_df.head(6))
display(penguins_df.tail(2))
print('Original shape after previews:', penguins_df.shape)
assert penguins_df.shape == (344, 7)
```

## 6. info ו־describe — מפרשים, לא רק מריצים
info מציגה סוגים ו־Non-Null Count; describe מציגה סיכום מספרי. מה פירוש count=342 כאשר יש 344 שורות? שתי מדידות חסרות בעמודה.

</div>

```python
penguins_df.info()
display(penguins_df.describe())
print('Known bill lengths:', penguins_df['bill_length_mm'].count())
assert penguins_df['bill_length_mm'].count() == 342
assert penguins_df['sex'].count() == 333
```

<div dir="rtl" lang="he" style="text-align:right;">

## 7. בדיקת מינים
שש השורות הראשונות אינן מייצגות בהכרח את מגוון המינים. בודקים את כל ערכי השדה בעזרת unique.

</div>

```python
print('Species:', penguins_df['species'].unique())
assert set(penguins_df['species'].unique()) == {'Adelie', 'Gentoo', 'Chinstrap'}
```

<div dir="rtl" lang="he" style="text-align:right;">

## 8. גרף חסרים מן הנתונים עצמם — קוד הדגמה למורה בלבד
אין צורך ללמוד כעת Matplotlib. התרשים נוצר מתוך הקובץ הנטען; הערכים אינם נבחרים רק בשביל להיראות יפה.

</div>

```python
import matplotlib.pyplot as plt
known = penguins_df.count()
missing = len(penguins_df) - known
fig, ax = plt.subplots(figsize=(10, 4.5))
ax.barh(list(known.index), known.values, label='Known')
ax.barh(list(known.index), missing.values, left=known.values, label='Missing')
ax.set_xlim(0, 365)
ax.set_xlabel('Number of records')
ax.set_title('Palmer Penguins - source completeness')
ax.legend()
for i, (ok, na) in enumerate(zip(known.values, missing.values)):
    ax.text(348, i, str(na) + ' missing', va='center', fontsize=9)
plt.tight_layout()
plt.show()
assert len(penguins_df) == 344
```
