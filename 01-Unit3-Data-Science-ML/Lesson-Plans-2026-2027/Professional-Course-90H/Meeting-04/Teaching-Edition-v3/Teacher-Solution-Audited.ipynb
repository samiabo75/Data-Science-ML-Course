מבוסס על מטלת Penguins המקורית. בדיקות assert מסייעות לבדוק שהקובץ שנטען הוא גרסת הקורס. אין להציג פתרון מורה במצגת NotebookLM.

## 1. מקור, CSV וגרסה
Palmer Penguins — המדידות נאספו בידי Kristen Gorman ו־Palmer Station LTER. חבילת palmerpenguins מיוחסת ל־Horst, Hill ו־Gorman (2020). גרסת הקורס נבדקה ומכילה 344 שורות ו־7 עמודות, ללא year.

</div>

```python
import pandas as pd
dataset_url = "https://raw.githubusercontent.com/samiabo75/Data-Science-ML-Course/main/Datasets/penguins.csv"
penguins_df = pd.read_csv(dataset_url)
assert penguins_df.shape == (344, 7)
print('Rows/columns:', penguins_df.shape)
```

<div dir="rtl" lang="he" style="text-align:right;">

## 2. שמות ומילון שדות
העמודות הן species, island, bill_length_mm, bill_depth_mm, flipper_length_mm, body_mass_g, sex. המדידות במ״מ, למעט מסת גוף בגרמים.

</div>

```python
expected_columns = ['species','island','bill_length_mm','bill_depth_mm','flipper_length_mm','body_mass_g','sex']
print('Columns:', list(penguins_df.columns))
assert list(penguins_df.columns) == expected_columns
```

<div dir="rtl" lang="he" style="text-align:right;">

## 3. הצצות ולמה אינן מייצגות דגימה
שש שורות ראשונות מתארות Adelie; שתי האחרונות מתארות Chinstrap. יש במאגר גם Gentoo; ההצצות אינן הוכחה על כל התפלגות המינים.

</div>

```python
display(penguins_df.head(6))
display(penguins_df.tail(2))
print('Source shape:', penguins_df.shape)
assert penguins_df.shape == (344,7)
assert set(penguins_df['species'].unique()) == {'Adelie','Gentoo','Chinstrap'}
```

<div dir="rtl" lang="he" style="text-align:right;">

## 4. info ו־describe
info נותנת פירוט Non-Null Counts וטיפוסים; describe סיכום סטטיסטי. אין לשכתב חסרים אוטומטית.

</div>

```python
penguins_df.info()
display(penguins_df.describe())
print('Known counts:', penguins_df.count().to_dict())
expected_known = {'species':344,'island':344,'bill_length_mm':342,'bill_depth_mm':342,'flipper_length_mm':342,'body_mass_g':342,'sex':333}
assert penguins_df.count().to_dict() == expected_known
```

<div dir="rtl" lang="he" style="text-align:right;">

## 5. חישוב חסרים בעמודת מקור וביקורת
344−342=2; שיעור 2/344×100≈0.5814%. אל תציגו זאת כ־25%: 25% חושב רק בתת־הדוגמה של ארבע שורות.

</div>

```python
total_rows = len(penguins_df)
known_bill_lengths = penguins_df['bill_length_mm'].count()
missing_bill_lengths = total_rows - known_bill_lengths
missing_percent = missing_bill_lengths / total_rows * 100
print('Missing bill lengths:', missing_bill_lengths)
print('Missing percentage:', missing_percent)
assert (total_rows,known_bill_lengths,missing_bill_lengths) == (344,342,2)
assert abs(missing_percent-200/344) < 1e-10
```

<div dir="rtl" lang="he" style="text-align:right;">

## 6. הצבת ארבע השורות הראשונות בנוסחה
אורכי המקור הם 39.1,39.5,40.3,NA. ממוצע 118.9/3≈39.63 מ״מ. השלמת החסר באפס הייתה יוצרת ממוצע 29.725 מ״מ, ללא הצדקה.

</div>

```python
first_four = penguins_df['bill_length_mm'].head(4)
print('First four known:', first_four.count())
print('First-four mean:', first_four.mean())
assert first_four.count() == 3
assert abs(first_four.mean()-118.9/3) < 1e-10
assert first_four.isna().sum() == 1
```

<div dir="rtl" lang="he" style="text-align:right;">

## 7. ערכים ייחודיים והסבר מדוע head אינו מספיק
שלושת המינים המופיעים בנתונים הם Adelie, Gentoo, Chinstrap.

</div>

```python
species_names = penguins_df['species'].unique()
print('Species:', species_names)
assert set(species_names) == {'Adelie','Gentoo','Chinstrap'}
```

<div dir="rtl" lang="he" style="text-align:right;">

## 8. גרף מורה מתוך נתוני המקור
מטרת הגרף להמחיש ידועים מול חסרים **בכל עמודה**. שדה עם ערכים חסרים אינו אומר בהכרח שיש פחות שורות.

</div>

```python
import matplotlib.pyplot as plt
known = penguins_df.count()
missing = len(penguins_df)-known
fig, ax = plt.subplots(figsize=(10,4.5))
ax.barh(list(known.index), known.values, label='Known')
ax.barh(list(known.index), missing.values, left=known.values, label='Missing')
ax.set_xlabel('Number of records')
ax.set_xlim(0,365)
ax.set_title('Penguins CSV completeness by column')
ax.legend()
for i, na in enumerate(missing.values):
    ax.text(348, i, f'{na} missing', va='center', fontsize=9)
plt.tight_layout()
plt.show()
assert missing.to_dict() == {'species':0,'island':0,'bill_length_mm':2,'bill_depth_mm':2,'flipper_length_mm':2,'body_mass_g':2,'sex':11}
```
