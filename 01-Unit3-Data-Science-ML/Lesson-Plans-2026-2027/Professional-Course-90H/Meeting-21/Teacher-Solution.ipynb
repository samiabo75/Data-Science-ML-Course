## דרך פתרון
הפערים הם 0.35,0.04,0.02. B נבחר לפי האימות. A עשוי להדגים התאמת יתר, C עשוי להדגים התאמת חסר; אלה הסברים אפשריים ולא אבחון חד־משמעי מהטבלה בלבד. Test נשמר להערכה לאחר סגירת הבחירה.

</div>

```python
import pandas as pd
results = pd.DataFrame({"candidate": ["A", "B", "C"], "train": [1.00, 0.88, 0.60], "validation": [0.65, 0.84, 0.58]})
results["gap"] = results["train"] - results["validation"]
print(results)
print("Chosen:", results.loc[results["validation"].idxmax(), "candidate"])
```
