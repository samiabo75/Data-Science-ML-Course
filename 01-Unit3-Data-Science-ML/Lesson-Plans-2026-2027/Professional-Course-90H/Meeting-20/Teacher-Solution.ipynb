## נתונים והסבר
נתונים מומצאים: מתוך 100 הודעות, 10 הן זבל. מסנן בסיס מנבא תמיד תקין. מסנן אחר נותן TP=6,FP=4,FN=4,TN=86. המחלקה החיובית היא זבל.

## דרך פתרון
למסנן האחר Accuracy=0.92 ושלושת המדדים P/R/F1 שווים 0.6. לבסיס Accuracy=0.9 ו־Recall=0. Precision של הבסיס אינו מוגדר כי אין חיזויים חיוביים; בדיווח אפשר לקבוע אפס. מסנן האחר טוב יותר בזיהוי זבל בדוגמה, אך ההשוואה אינה הוכחה לאוכלוסייה חדשה.

</div>

```python
TP = 6
FP = 4
FN = 4
TN = 86
accuracy = (TP + TN) / (TP + FP + FN + TN)
precision = TP / (TP + FP)
recall = TP / (TP + FN)
f1 = 2 * TP / (2 * TP + FP + FN)
print(accuracy, precision, recall, f1)
print("Baseline accuracy:", 90 / 100)
print("Baseline recall:", 0 / 10)
```
