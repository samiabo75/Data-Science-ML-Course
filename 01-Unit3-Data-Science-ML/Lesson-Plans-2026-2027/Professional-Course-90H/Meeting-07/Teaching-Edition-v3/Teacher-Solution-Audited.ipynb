# מפגש 7 — פתרון מורה עם בדיקות אמת
מעבדה מקורית: Countplot לאיים בקרב Adelie ובהשוואה למקור; Histogram אורך סנפיר ב־8/16 bins. תוצאות בדוקות מול קובץ הקורס. אין להעלות מחברת זו ל־NotebookLM של התלמיד.

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
source = 'https://raw.githubusercontent.com/samiabo75/Data-Science-ML-Course/main/Datasets/penguins.csv'
penguins_df = pd.read_csv(source)
print('Source:',penguins_df.shape)
assert penguins_df.shape == (344,7)
```

## 1. סיווג שדות
species/island/sex הם קטגוריאליים. body_mass_g, flipper_length_mm הן מדידות כמותיות בגרם ובמ״מ. חסר אינו קטגוריה ביולוגית ואינו מדידה אפס.

## 2. Adelie ותדירויות האיים
תוצאות הקורס: Dream=56, Torgersen=52, Biscoe=44 בתוך Adelie; סך 152. הגרף מציג ספירת תצפיות ולא אוכלוסיית טבע.

```python
adelie_mask = penguins_df['species'] == 'Adelie'
adelie_df = penguins_df[adelie_mask]
print('Adelie:',adelie_df.shape)
print(adelie_df['island'].value_counts())
assert adelie_df.shape == (152,7)
assert adelie_df['island'].value_counts().to_dict() == {'Dream':56,'Torgersen':52,'Biscoe':44}
```

```python
plt.figure(figsize=(7,4))
sns.countplot(data=adelie_df,x='island')
plt.title('Adelie observations by island')
plt.xlabel('Island (category)')
plt.ylabel('Number of observations')
plt.ylim(0,180)
plt.tight_layout()
plt.show()
```

## 3. כלל המאגר — אותה יחידה, קבוצה שונה
Biscoe=168, Dream=124, Torgersen=52. כששומרים אותה סקאלה לשני הגרפים קל לראות שינוי, אך אין להסיק סיבתיות.

```python
all_counts = penguins_df['island'].value_counts()
print(all_counts)
assert all_counts.to_dict() == {'Biscoe':168,'Dream':124,'Torgersen':52}
assert all_counts.sum() == len(penguins_df)
```

```python
plt.figure(figsize=(7,4))
sns.countplot(data=penguins_df,x='island')
plt.title('All species: observations by island')
plt.xlabel('Island (category)')
plt.ylabel('Number of observations')
plt.ylim(0,180)
plt.tight_layout()
plt.show()
```

## 4. הרחבת חסרים במשתנה קטגוריאלי
sex כולל male=168, female=165, וחסרים=11. שרטוט Countplot לא יציג NaN כברירת מחדל; סכום שתי הקטגוריות 333 ולא 344.

```python
sex_known = penguins_df['sex'].value_counts()
print('Known sex values:',sex_known.to_dict())
assert sex_known.to_dict() == {'male':168,'female':165}
assert int(sex_known.sum()) == 333
assert penguins_df['sex'].isna().sum() == 11
```

## 5. היסטוגרמה של אורך סנפיר — 8 תחומים
יש 342 מדידות ידועות בטווח 172–231 מ״מ. bins=8 משנה אופן חלוקה, לא מספר מדידות.

```python
flipper = penguins_df['flipper_length_mm']
print('Known:',flipper.count(),'Range:',flipper.min(),flipper.max())
assert flipper.count() == 342
assert flipper.min() == 172 and flipper.max() == 231
plt.figure(figsize=(7,4))
sns.histplot(data=penguins_df,x='flipper_length_mm',bins=8)
plt.title('Flipper lengths: 8 bins')
plt.xlabel('Flipper length (mm)')
plt.ylabel('Number of measurements')
plt.tight_layout()
plt.show()
```

## 6. היסטוגרמה של אותן מדידות — 16 תחומים
יש לשאול איזה מידע השתנה בפועל. התצפיות לא השתנו; המספר 16 הוא מספר תחומים על הציר.

```python
plt.figure(figsize=(7,4))
sns.histplot(data=penguins_df,x='flipper_length_mm',bins=16)
plt.title('Flipper lengths: 16 bins')
plt.xlabel('Flipper length (mm)')
plt.ylabel('Number of measurements')
plt.tight_layout()
plt.show()
assert penguins_df.shape == (344,7)
assert penguins_df['flipper_length_mm'].count() == 342
```

## 7. אימות סכומי התחומים באמצעות פלט ההיסטוגרמה
למורה בלבד: אם כל טווח המדידות הידועות מוצג בגרף count, סכום ספירות העמודות 342. ההשוואה מתייחסת לאותו קובץ.

```python
import numpy as np
values = penguins_df['flipper_length_mm'].dropna().to_numpy()
counts_8, edges_8 = np.histogram(values,bins=8)
counts_16, edges_16 = np.histogram(values,bins=16)
print('Sum of 8 intervals:',counts_8.sum())
print('Sum of 16 intervals:',counts_16.sum())
assert counts_8.sum() == 342
assert counts_16.sum() == 342
assert len(edges_8) == 9 and len(edges_16) == 17
```

## 8. רשות: Adelie עם אותם גבולות
בקבוצת Adelie יש 151 מדידות אורך סנפיר ידועות. עם אותם גבולות התחומים אפשר להשוות מספרים, אך לא להציג ספירות כאילו היו שיעורים כאשר גדלי הקבוצות שונים.

```python
shared_bins = [170,180,190,200,210,220,230,240]
assert adelie_df['flipper_length_mm'].count() == 151
plt.figure(figsize=(7,4))
sns.histplot(data=adelie_df,x='flipper_length_mm',bins=shared_bins)
plt.title('Adelie, shared bins')
plt.xlabel('Flipper length (mm)')
plt.ylabel('Number of measurements')
plt.tight_layout()
plt.show()
```

## הגנה ומסקנה
בחירה ב־countplot/histplot תלויה במשמעות המשתנה. שינוי bins אינו משנה את מקור הנתונים. גרף אינו מאשר נורמליות או סיבתיות; ספירת תצפיות במדגם אינה ספירת אוכלוסייה בטבע. פרויקט: כרטיסי גרף תכנוניים בלבד עד הסדרת שימוש בקבוצת Test.
