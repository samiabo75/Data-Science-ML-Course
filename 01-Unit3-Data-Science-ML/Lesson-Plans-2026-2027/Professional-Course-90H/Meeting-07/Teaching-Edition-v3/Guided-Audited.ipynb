# מפגש 7 — מחברת הדגמה: פעילות במועדון וזמני הגעה
נתוני ההדגמה **מומצאים במפורש** ואינם תוצאות מעבדת Penguins. לפני קוד: חיזוי ידני של ספירות וגרפים. לאחר קוד: השוואה לחישוב, פירוש הצירים ומגבלה.

## 1. סיווג לפני קוד
activity הוא משתנה קטגוריאלי שמי; arrival_minutes הוא זמן כמותי. אל תבחרו את הגרף לפי סוג המשתנה בפייתון בלבד.

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
club_df = pd.DataFrame({'activity':['Robotics','Sports','Robotics','Music','Sports','Robotics']})
print('Source:',club_df.shape)
display(club_df)
assert club_df.shape == (6,1)
```

## 2. שכיחות קטגוריות
חיזוי ידני: רובוטיקה 3, ספורט 2, מוזיקה 1. הנוסחה סופרת כמה פעמים כל קטגוריה מופיעה. הסכום צריך להיות 6.

```python
frequencies = club_df['activity'].value_counts()
print(frequencies)
assert frequencies.to_dict() == {'Robotics':3,'Sports':2,'Music':1}
assert frequencies.sum() == len(club_df)
```

## 3. Countplot ושתי שאלות על צירים
ציר X = שם פעילות, ציר Y = מספר תשובות מתוך שש. מדוע אין משמעות לאורך הקטע בין Robotics ל־Sports?

```python
plt.figure(figsize=(7,4))
sns.countplot(data=club_df,x='activity')
plt.title('Club survey: six answers')
plt.xlabel('Activity (category)')
plt.ylabel('Number of answers')
plt.tight_layout()
plt.show()
```

## 4. זמן הגעה — מחשבים ידנית לפני גרף
ערכים: 10,12,18,22 דקות. בתחום [10,20) יש 3, בתחום [20,30) יש 1. לא מדובר בשמות קבוצות אלא בתחומי מדידה מספרית.

```python
arrival_minutes = [10,12,18,22]
first_count = 0
second_count = 0
for value in arrival_minutes:
    if 10 <= value < 20:
        first_count += 1
    if 20 <= value < 30:
        second_count += 1
print('Manual counts:',first_count,second_count)
assert (first_count,second_count) == (3,1)
assert first_count + second_count == len(arrival_minutes)
```

## 5. היסטוגרמה בשני תחומים מפורשים
כל עמודה היא *תחום זמן*, ולא שם קבוצת ילדים. שני התחומים שווי רוחב — 10 דקות.

```python
plt.figure(figsize=(7,4))
plt.hist(arrival_minutes,bins=[10,20,30])
plt.title('Four arrivals: two bins')
plt.xlabel('Arrival time (minutes)')
plt.ylabel('Number of observations')
plt.tight_layout()
plt.show()
```

## 6. אותה סדרה, ארבעה תחומים
טבלת המעקב: [10,15) מכיל 2, [15,20) מכיל 1, [20,25) מכיל 1, [25,30) מכיל 0. הסכום נשאר 4.

```python
plt.figure(figsize=(7,4))
plt.hist(arrival_minutes,bins=[10,15,20,25,30])
plt.title('Same four arrivals: four bins')
plt.xlabel('Arrival time (minutes)')
plt.ylabel('Number of observations')
plt.tight_layout()
plt.show()
assert len(arrival_minutes) == 4
```

## 7. חיבור למאגר Penguins
עכשיו נזכרים בשאלות המעבדה, *בלי למסור לתלמיד פתרון מוכן*: כמה תצפיות מכל אי יש בקרב Adelie לעומת כלל המאגר, וכיצד מתפלגים אורכי הסנפיר?

```python
dataset_url = 'https://raw.githubusercontent.com/samiabo75/Data-Science-ML-Course/main/Datasets/penguins.csv'
penguins_df = pd.read_csv(dataset_url)
print('Penguins source dimensions:',penguins_df.shape)
print('Known flipper measurements:',penguins_df['flipper_length_mm'].count())
assert penguins_df.shape == (344,7)
assert penguins_df['flipper_length_mm'].count() == 342
```
