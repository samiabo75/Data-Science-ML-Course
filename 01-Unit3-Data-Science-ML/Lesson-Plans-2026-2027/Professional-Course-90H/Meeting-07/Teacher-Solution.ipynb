## 1. סוג המשתנה נקבע לפי המשמעות
| סוג | השאלה | דוגמה |
|---|---|---|
| קטגוריאלי שמי | לאיזו קבוצה שייך? ללא סדר טבעי | species, island |
| קטגוריאלי סדרי | איזו רמה? יש סדר, לא בהכרח מרווחים שווים | דירוג נמוך/בינוני/גבוה; זו דוגמה כללית שאינה עמודה בקובץ |
| כמותי בדיד | כמה? ספירה | מספר תצפיות בכל מין, נתון מסוכם שניצור בגרף |
| כמותי רציף | כמה נמדד? | אורך מקור, אורך סנפיר ומסה |

מסה שנרשמה במספר שלם נשארת מדידה רציפה מבחינת המשמעות. מספר מזהה הוא תווית ולא גודל כמותי בגלל הספרות. קידוד מספרי של קטגוריות אינו הופך אותן למדידות.

סוג אחסון בפייתון אינו סוג המשתנה התוכני. species הוא מין טקסונומי; sex הוא זוויג — לא אותה עמודה.

</div>

<div dir="rtl" lang="he" style="text-align: right;">

### 2–3. קבוצה
Adelie כוללת 152 רשומות בגרסה שנבדקה. island הוא משתנה קטגוריאלי שמי; בודקים ספירות בתוך תת־הקבוצה.

</div>

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
url = "https://raw.githubusercontent.com/samiabo75/Data-Science-ML-Course/main/Datasets/penguins.csv"
penguins_df = pd.read_csv(url)
adelie_mask = penguins_df["species"] == "Adelie"
adelie_df = penguins_df[adelie_mask]
print("מקור:", penguins_df.shape)
print("Adelie:", adelie_df.shape)
```

```python
plt.figure(figsize=(7, 4))
sns.countplot(data=adelie_df, x="island", color="steelblue")
plt.title("Adelie observations by island")
plt.xlabel("Island")
plt.ylabel("Number of observations")
plt.tight_layout()
plt.show()
```

<div dir="rtl" lang="he" style="text-align: right;">

### 4. פירוש
בקבוצת Adelie: Dream=56, Torgersen=52, Biscoe=44. אלו מספרי רשומות; Dream הגדולה בקבוצה זו. אין להסיק מספרי אוכלוסייה בטבע, ואין להסיק שכך מתחלקים כל המינים.

</div>

```python
plt.figure(figsize=(7, 4))
sns.countplot(data=penguins_df, x="island", color="teal")
plt.title("All penguin observations by island")
plt.xlabel("Island")
plt.ylabel("Number of observations")
plt.tight_layout()
plt.show()
```

<div dir="rtl" lang="he" style="text-align: right;">

### 5. מקור לעומת קבוצה
בכל המאגר Biscoe=168, Dream=124, Torgersen=52. Biscoe הגדולה במקור, אך לא בקרב Adelie. שינוי אוכלוסיית הבחירה משנה את השאלה. ההבדל אינו מוכיח קשר סיבתי בין אי למין.

### 6. בחירת גרפים
מסה היא מדידה רציפה: Histogram. מסה מול סנפיר הן שתי מדידות: Scatterplot. ספירת אי כקטגוריה: Countplot.

### 7. כרטיס גרף לדוגמה
שאלה: כמה תצפיות Adelie מכל אי? קבוצה: Adelie בלבד, 152 רשומות. X אי, משתנה שמי; Y מספר תצפיות, ספירה בדידה. Dream מונה 56 רשומות והיא הגדולה בקבוצה. מגבלה: דגימה מוגדרת, לא מפקד אוכלוסייה או הוכחת סיבתיות. מקור: Palmer Penguins בגרסת הקורס 344×7.

</div>

male=168, female=165 בגרסה שנבדקה; 11 חסרים אינם מוצגים כעמודה. לכן 333 ערכים ידועים בגרף אינם מספר כל הרשומות. אין לשנות מקור או למלא חסרים רק לשם הציור.

## פתרון חלק 2
הפתרון הבא תואם לנתוני התרגול העצמאי, ולא לנתוני הדוגמה הפתורה.

</div>

<div dir="rtl" lang="he" style="text-align: right;">

### 1. שאלה
כיצד מתפלגים אורכי הסנפיר הידועים במאגר? זהו משתנה כמותי רציף במ״מ, גם אם נשמר במספרים שלמים. X הוא אורך, Y הוא מספר ערכים בתחום.

</div>

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
url = "https://raw.githubusercontent.com/samiabo75/Data-Science-ML-Course/main/Datasets/penguins.csv"
penguins_df = pd.read_csv(url)
print("ממדים:", penguins_df.shape)
```

```python
plt.figure(figsize=(7, 4))
sns.histplot(data=penguins_df, x="flipper_length_mm", bins=8, color="steelblue")
plt.title("Flipper length distribution - 8 bins")
plt.xlabel("Flipper length (mm)")
plt.ylabel("Number of observations")
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(7, 4))
sns.histplot(data=penguins_df, x="flipper_length_mm", bins=16, color="teal")
plt.title("Flipper length distribution - 16 bins")
plt.xlabel("Flipper length (mm)")
plt.ylabel("Number of observations")
plt.tight_layout()
plt.show()
```

<div dir="rtl" lang="he" style="text-align: right;">

### 4–5. פירוש
אותם 342 אורכים ידועים חולקו ליותר תחומים צרים. הטווח הידוע בגרסה זו הוא 172–231 מ״מ. רוחב תחום ב־ 8 תחומים הוא 59/8=7.375 מ״מ, וב־ 16 הוא 3.6875 מ״מ. אין צורך שהתלמיד יחשב זאת אם אין שליטה בחלוקה.

הגרפים מראים שכיחות בתחומי אורך, לא את ערכי כל הרשומות או את מספר המינים. כדאי לתאר טווחים לפי התצוגה ולנמק. לא קובעים נורמליות ממראה, וייתכן שהתצוגה מערבת קבוצות שונות. תצוגה מפורטת יותר אינה מידע חדש שנאסף.

</div>

<div dir="rtl" lang="he" style="text-align: right;">

### 6. חסרים
סכום הגבהים בגרפים אלה 342; 2 חסרי מדידה אינם נכללים. שינוי מספר התחומים על כל אותו טווח אינו משנה מספר זה. המקור נשאר 344×7.

</div>

```python
print("ממדי המקור:", penguins_df.shape)
```

<div dir="rtl" lang="he" style="text-align: right;">

### 7. כרטיס לדוגמה
שאלה: התפלגות אורך סנפיר ידוע ב־ Palmer Penguins בגרסת הקורס. כמותי רציף, מ״מ. הוצגו 8 ואחר כך 16 תחומים; Y מספר תצפיות. הטווח הידוע הוא 172–231 מ״מ, והחלוקה משנה את פירוט המראה. מגבלה: שני חסרים ותלות ב־ bins, ומדגם שאינו כלל הפינגווינים. אין ניבוי או הוכחת סיבתיות.

### רשות
משווים אותה מדידה ובאותם גבולות תחומים, אך הספירות הן של אוכלוסיות שונות בגודלן.

</div>

```python
adelie_mask = penguins_df["species"] == "Adelie"
adelie_df = penguins_df[adelie_mask]
shared_bins = [170, 180, 190, 200, 210, 220, 230, 240]
plt.figure(figsize=(7, 4))
sns.histplot(data=adelie_df, x="flipper_length_mm", bins=shared_bins, color="steelblue")
plt.title("Adelie flipper lengths - shared bins")
plt.xlabel("Flipper length (mm)")
plt.ylabel("Number of observations")
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(7, 4))
sns.histplot(data=penguins_df, x="flipper_length_mm", bins=shared_bins, color="teal")
plt.title("All flipper lengths - shared bins")
plt.xlabel("Flipper length (mm)")
plt.ylabel("Number of observations")
plt.tight_layout()
plt.show()
```
