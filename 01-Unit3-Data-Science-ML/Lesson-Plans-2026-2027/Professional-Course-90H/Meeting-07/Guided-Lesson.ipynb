## דוגמה פתורה לפני קוד
בערכים 10, 12, 18 ו־22, התחום 10 עד פחות מ־20 מכיל שלוש תצפיות. התחום הבא מכיל אחת. אלה תחומי מדידה ולא שמות קבוצות.

</div>

```python
values = [10, 12, 18, 22]
count = 0
for value in values:
    if 10 <= value < 20:
        count = count + 1
print(count)
```

### טעינה ובדיקת קלט
טוענים את מקור הנתונים שהופיע בתרגול. בודקים שדות וממדים לפני שימוש. כתובת Raw שונה מדף התצוגה של GitHub.

</div>

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
url = "https://raw.githubusercontent.com/samiabo75/Data-Science-ML-Course/main/Datasets/penguins.csv"
df = pd.read_csv(url)
print("ממדים:", df.shape)
display(df.head(3))
```

<div dir="rtl" lang="he" style="text-align: right;">

### המחשה חזותית
לפני הרצה הגדירו שאלה, סוגי משתנים ויחידות. אחרי הרצה פרשו את הצירים ואת הממצא. אין להסיק סיבתיות או למחוק קצה רק מן הציור.

</div>

```python
plt.figure(figsize=(7, 4))
sns.countplot(data=df, x="species", color="steelblue")
plt.title("Penguin observations by species")
plt.xlabel("Species")
plt.ylabel("Number of observations")
plt.tight_layout()
plt.show()
```
