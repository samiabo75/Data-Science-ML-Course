# מפגש 19: הדגמת מורה Audited
8 מוצרים. חיובי Defect; אמת בשורות, חיזוי בעמודות. חיזוי ידני לפני קוד.

```python
import pandas as pd
from sklearn.metrics import confusion_matrix,accuracy_score,precision_score,recall_score,f1_score,classification_report
actual=["OK"]*4+["Defect"]*4
predicted=["OK","OK","Defect","OK","OK","OK","Defect","Defect"]
labels=["OK","Defect"]
matrix=confusion_matrix(actual,predicted,labels=labels)
table=pd.DataFrame(matrix,index=labels,columns=labels)
tn,fp,fn,tp=matrix.ravel()
display(table)
```

```python
assert matrix.tolist()==[[3,1],[2,2]]
assert matrix.sum()==8
assert (tn,fp,fn,tp)==(3,1,2,2)
assert tp+fp==3
assert tp+fn==4
```

## מכנים שונים
Precision: מתוך החזויים חיובי. Recall: מתוך החיוביים באמת. F1 ממוצע הרמוני. הסבירו כל מכנה.

```python
precision=tp/(tp+fp)
recall=tp/(tp+fn)
f1=2*tp/(2*tp+fp+fn)
accuracy=(tn+tp)/matrix.sum()
print(precision,recall,f1,accuracy)
assert abs(precision-precision_score(actual,predicted,pos_label="Defect"))<1e-12
assert abs(recall-recall_score(actual,predicted,pos_label="Defect"))<1e-12
assert abs(f1-f1_score(actual,predicted,pos_label="Defect"))<1e-12
assert abs(accuracy-accuracy_score(actual,predicted))<1e-12
```

```python
assert abs(precision-2/3)<1e-12
assert recall==0.5
assert abs(f1-4/7)<1e-12
assert accuracy==0.625
assert abs(f1-2*precision*recall/(precision+recall))<1e-12
```

## מכנה אפס
כאשר אין חיזויים חיוביים, Precision מתמטי הוא 0/0 לא מוגדר. מדיניות zero_division=0 היא החלטת דיווח.

```python
edge_actual=["OK"]*8+["Defect"]*2
edge_pred=["OK"]*10
assert accuracy_score(edge_actual,edge_pred)==0.8
assert recall_score(edge_actual,edge_pred,pos_label="Defect")==0
assert precision_score(edge_actual,edge_pred,pos_label="Defect",zero_division=0)==0
assert f1_score(edge_actual,edge_pred,pos_label="Defect",zero_division=0)==0
```

## שינוי תחזית
שנו מוצר Defect שהוחמץ לחיזוי Defect בעותק. מה קרה ל־Recall?

```python
changed=predicted.copy()
changed[4]="Defect"
assert recall_score(actual,changed,pos_label="Defect")>recall
assert accuracy_score(actual,changed)>accuracy
assert confusion_matrix(actual,changed,labels=labels).sum()==8
```
