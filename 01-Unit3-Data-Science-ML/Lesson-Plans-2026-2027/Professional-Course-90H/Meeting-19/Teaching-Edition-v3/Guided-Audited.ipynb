# מפגש 19: הדגמת מורה Audited
8 מוצרים. חיובי Defect; אמת בשורות, חיזוי בעמודות. חיזוי ידני לפני קוד.

```python
import pandas as pd
from sklearn.metrics import confusion_matrix,accuracy_score,precision_score,recall_score,f1_score,classification_report
actual=["OK"]*4+["Defect"]*4
predicted=["OK","OK","Defect","OK","OK","OK","Defect","Defect"]
labels=["OK","Defect"]
matrix=confusion_matrix(actual,predicted,labels=labels)
table=pd.DataFrame(matrix,index=labels,columns=labels)
tn,fp,fn,tp=matrix.ravel()
display(table)
```

```python
assert matrix.tolist()==[[3,1],[2,2]]
assert matrix.sum()==8
assert (tn,fp,fn,tp)==(3,1,2,2)
assert tp+fp==3
assert tp+fn==4
```

## מכנים שונים
Precision: מתוך החזויים חיובי. Recall: מתוך החיוביים באמת. F1 ממוצע הרמוני. הסבירו כל מכנה.

```python
precision=tp/(tp+fp)
recall=tp/(tp+fn)
f1=2*tp/(2*tp+fp+fn)
accuracy=(tn+tp)/matrix.sum()
print(precision,recall,f1,accuracy)
assert abs(precision-precision_score(actual,predicted,pos_label="Defect"))<1e-12
assert abs(recall-recall_score(actual,predicted,pos_label="Defect"))<1e-12
assert abs(f1-f1_score(actual,predicted,pos_label="Defect"))<1e-12
assert abs(accuracy-accuracy_score(actual,predicted))<1e-12
```

```python
assert abs(precision-2/3)<1e-12
assert recall==0.5
assert abs(f1-4/7)<1e-12
assert accuracy==0.625
assert abs(f1-2*precision*recall/(precision+recall))<1e-12
```

## מכנה אפס
כאשר אין חיזויים חיוביים, Precision מתמטי הוא 0/0 לא מוגדר. מדיניות zero_division=0 היא החלטת דיווח.

```python
edge_actual=["OK"]*8+["Defect"]*2
edge_pred=["OK"]*10
assert accuracy_score(edge_actual,edge_pred)==0.8
assert recall_score(edge_actual,edge_pred,pos_label="Defect")==0
assert precision_score(edge_actual,edge_pred,pos_label="Defect",zero_division=0)==0
assert f1_score(edge_actual,edge_pred,pos_label="Defect",zero_division=0)==0
```

## שינוי תחזית
שנו מוצר Defect שהוחמץ לחיזוי Defect בעותק. מה קרה ל־Recall?

```python
changed=predicted.copy()
changed[4]="Defect"
assert recall_score(actual,changed,pos_label="Defect")>recall
assert accuracy_score(actual,changed)>accuracy
assert confusion_matrix(actual,changed,labels=labels).sum()==8
```

## העברה ל־Penguins
המשיכו את הקוד הנתון במחברת המקור. מחלקת היעד בדוגמת המורה היא Chinstrap מול היתר; תרגיל התלמיד יבחר Adelie. שמרו על K=3, seed42, Train-only preprocessing ו־Test=25%.

```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score

url = "https://raw.githubusercontent.com/samiabo75/Data-Science-ML-Course/main/Datasets/penguins.csv"
df = pd.read_csv(url)
display(df.head())
display(df.isna().sum())

model_df = df.dropna(subset=["species"]).copy()
features = ["flipper_length_mm","body_mass_g"]
X = model_df[features].copy()
y = model_df["species"].copy()
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.25, random_state=42, stratify=y
)
print("Train:", len(X_train), "Test:", len(X_test))

training_means = X_train.mean()
X_train_ready = X_train.fillna(training_means)
X_test_ready = X_test.fillna(training_means)

scaler = StandardScaler().set_output(transform="pandas")
scaler.fit(X_train_ready)
X_train_scaled = scaler.transform(X_train_ready)
X_test_scaled = scaler.transform(X_test_ready)
display(X_train_scaled.head())

model = KNeighborsClassifier(n_neighbors=3, weights="uniform", metric="euclidean")
model.fit(X_train_scaled, y_train)
predictions = model.predict(X_test_scaled)
print(predictions[:5])

comparison = pd.DataFrame(
    {"actual": y_test, "predicted": predictions}, index=y_test.index
)
comparison["correct"] = comparison["actual"] == comparison["predicted"]
display(comparison.head(10))
correct_count = comparison["correct"].sum()
accuracy = accuracy_score(y_test, predictions)
print("Correct:", correct_count, "out of", len(y_test))
print("Accuracy:", accuracy, "Percent:", 100 * accuracy)
```

```python
labels_species=["Adelie","Chinstrap","Gentoo"]
cm_species=confusion_matrix(y_test,predictions,labels=labels_species)
target="Chinstrap"
j=labels_species.index(target)
tp_target=int(cm_species[j,j])
fn_target=int(cm_species[j,:].sum()-tp_target)
fp_target=int(cm_species[:,j].sum()-tp_target)
tn_target=int(cm_species.sum()-tp_target-fp_target-fn_target)
assert cm_species.shape==(3,3)
assert cm_species.sum()==len(y_test)
assert abs(cm_species.trace()/len(y_test)-accuracy_score(y_test,predictions))<1e-12
assert tp_target+fn_target==sum(y_test==target)
assert tp_target+fp_target==sum(predictions==target)
assert tp_target+fp_target+fn_target+tn_target==len(y_test)
print(target,tp_target,fp_target,fn_target,tn_target)
```
