### 1–2. חישוב והשפעת קצה
סכום המקור 65 ולכן ממוצע 13, חציון 12 ושכיח 12. סכום הניסוי 90 ולכן ממוצע 18, חציון ושכיח 12. התוספת 25 בערך אחד העלתה ממוצע ב־ 25/5=5; זו דוגמה לעמידות יחסית של החציון, לא הבטחה שאינו משתנה לעולם.

</div>

```python
import pandas as pd
original_df = pd.DataFrame({"value": [10, 12, 12, 14, 17]})
experiment_df = pd.DataFrame({"value": [10, 12, 12, 14, 42]})
print("מקור ממוצע:", original_df["value"].mean())
print("מקור חציון:", original_df["value"].median())
print("מקור שכיחים:")
print(original_df["value"].mode())
print("ניסוי ממוצע:", experiment_df["value"].mean())
print("ניסוי חציון:", experiment_df["value"].median())
print("ניסוי שכיחים:")
print(experiment_df["value"].mode())
display(original_df)
```

### 1–2. ידני ואימות
לשתי הקבוצות ממוצע 20. א׳: טווח 4, הפרשים −2,0,2, ריבועים 4,0,4; שונות 8/3 וסטיית תקן ≈1.63. ב׳: טווח 20, הפרשים −10,0,10, ריבועים 100,0,100; שונות 200/3 וסטיית תקן ≈8.16. הפיזור ב׳ גדול יותר, אף שהממוצע זהה.

</div>

```python
import pandas as pd
groups_df = pd.DataFrame({"a": [18, 20, 22], "b": [10, 20, 30]})
print("ממוצע א:", groups_df["a"].mean())
print("ממוצע ב:", groups_df["b"].mean())
print("טווח א:", groups_df["a"].max() - groups_df["a"].min())
print("טווח ב:", groups_df["b"].max() - groups_df["b"].min())
print("שונות א:", groups_df["a"].var(ddof=0))
print("סטיית תקן א:", groups_df["a"].std(ddof=0))
print("שונות ב:", groups_df["b"].var(ddof=0))
print("סטיית תקן ב:", groups_df["b"].std(ddof=0))
```

<div dir="rtl" lang="he" style="text-align: right;">

### 3. מדגם
סכום הריבועים בא׳ 8; חלוקה ב־ 2 נותנת שונות 4 וסטיית תקן 2. ddof=1 היא ברירת המחדל בפעולות אלה של Pandas. אין סתירה עם חלוקה ב־ 3; המטרה והמחלק שונים.

</div>

```python
print("שונות א עם n-1:", groups_df["a"].var(ddof=1))
print("סטיית תקן א עם n-1:", groups_df["a"].std(ddof=1))
```
