# מפגש 8 — פתרון מורה עם בדיקות
אימות קוד אינו במקום נימוק מתמטי. קובץ הקורס כולל 344 תצפיות אך 342 ערכי סנפיר ידועים.

```python
import pandas as pd
import numpy as np
original_df=pd.DataFrame({'value':[10,12,12,14,17]})
experiment_df=pd.DataFrame({'value':[10,12,12,14,42]})
assert original_df['value'].mean()==13
assert experiment_df['value'].mean()==18
assert original_df['value'].median()==12
assert experiment_df['value'].median()==12
assert original_df['value'].mode().tolist()==[12]
```

```python
even_df=pd.DataFrame({'value':[18,10,16,12]})
assert even_df['value'].median()==14
```

## חישוב ידני: סכומי ריבועי הסטיות 8 ו־200. היחידות בשונות דקות².

```python
groups_df=pd.DataFrame({'a':[18,20,22],'b':[10,20,30]})
assert groups_df.mean().tolist()==[20,20]
assert (groups_df.max()-groups_df.min()).tolist()==[4,20]
assert np.isclose(groups_df['a'].var(ddof=0),8/3)
assert np.isclose(groups_df['b'].var(ddof=0),200/3)
assert np.isclose(groups_df['a'].std(ddof=0),np.sqrt(8/3))
assert np.isclose(groups_df['b'].std(ddof=0),np.sqrt(200/3))
assert groups_df['a'].var(ddof=1)==4
```

## העברה לאורכי סנפיר: יחידות מ״מ ו־מ״מ²

```python
url='https://raw.githubusercontent.com/samiabo75/Data-Science-ML-Course/main/Datasets/penguins.csv'
penguins_df=pd.read_csv(url)
flippers=penguins_df['flipper_length_mm']
print('Known',flippers.count(),'Mean',flippers.mean(),'Variance',flippers.var(ddof=0),'SD',flippers.std(ddof=0))
assert penguins_df.shape==(344,7)
assert flippers.count()==342
adelie_df=penguins_df[penguins_df['species']=='Adelie']
assert len(adelie_df)==152
print('Adelie known:',adelie_df['flipper_length_mm'].count())
```
