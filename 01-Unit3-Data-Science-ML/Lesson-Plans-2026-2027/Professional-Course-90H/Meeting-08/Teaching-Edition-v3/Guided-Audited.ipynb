# מפגש 8 — מחברת הדגמה: אותו מרכז, פיזור שונה
לפני הרצה: ניבוי וחישוב ידני, אחריה: בדיקת יחידות ופירוש.

```python
import pandas as pd
import numpy as np
A = [18,20,22]
B = [10,20,30]
df = pd.DataFrame({'line_a':A,'line_b':B})
display(df)
assert df.shape==(3,2)
```

## מרכז: מדוע ממוצע זהה אינו מספר את הסיפור?

```python
print('Mean:',df.mean().to_dict())
print('Median:',df.median().to_dict())
print('Range:',(df.max()-df.min()).to_dict())
assert df.mean().tolist()==[20,20]
assert (df.max()-df.min()).tolist()==[4,20]
```

## שונות תיאורית: מחלקים ב־n, ולכן ddof=0

```python
print('Variance:',df.var(ddof=0).to_dict())
print('SD:',df.std(ddof=0).to_dict())
assert np.isclose(df['line_a'].var(ddof=0),8/3)
assert np.isclose(df['line_b'].std(ddof=0),np.sqrt(200/3))
```

## מדגם: מחלקים ב־n−1; אין לערבב את ההגדרות

```python
print('Sample variance:',df.var(ddof=1).to_dict())
assert df['line_a'].var(ddof=1)==4
```

## גרף על אותו ציר דקות — קוד מורה להמחשה

```python
import matplotlib.pyplot as plt
plt.figure(figsize=(8,3))
plt.scatter(A,[1]*3,label='Line A')
plt.scatter(B,[2]*3,label='Line B')
plt.axvline(20,linestyle='--',label='Mean 20 min')
plt.yticks([1,2],['A','B'])
plt.xlabel('Travel time (minutes)')
plt.legend()
plt.show()
```

## מסקנה והעברה
A ו־B בעלי אותו ממוצע אך פיזור שונה. אין כאן ראיה לנורמליות או למידת מכונה.
