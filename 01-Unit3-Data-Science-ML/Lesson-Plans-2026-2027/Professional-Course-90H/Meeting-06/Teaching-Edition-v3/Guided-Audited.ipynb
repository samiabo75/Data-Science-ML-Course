# מפגש 6 — חקר מודרך: AND ואז מיון
סיפור דודו: רוצים מחשבים בין 1500 ל־4000 ₪ כולל, ואז מיון מחיר יורד. **לפני הרצה** מלאו ביד טבלת שני תנאים ושערו את הפלט. ערכי מחשבים מומצאים להוראה.

</div>

```python
import pandas as pd
computers_df = pd.DataFrame({'computer_id':['A','B','C','D'],'memory_gb':[16,8,32,16],'price_nis':[1000,2500,4000,4500]})
display(computers_df)
assert computers_df.shape == (4,3)
```

<div dir="rtl" lang="he" style="text-align:right">

## שני תנאים נפרדים ואז AND
מחיר לפחות 1500 וגם לכל היותר 4000. חזו את המסכות: מחיר תחתון [False,True,True,True], עליון [True,True,True,False].

</div>

```python
lower = computers_df['price_nis'] >= 1500
upper = computers_df['price_nis'] <= 4000
both = lower & upper
print('Lower mask:',lower.tolist())
print('Upper mask:',upper.tolist())
print('AND mask:',both.tolist())
assert both.tolist() == [False,True,True,False]
```

<div dir="rtl" lang="he" style="text-align:right">

## בחירת שורות: מסכה אינה מיון
המקור צריך להישאר עם 4 שורות; המסוננת תהיה עם B,C בלבד ושלוש עמודות.

</div>

```python
selected_df = computers_df[both]
display(selected_df)
assert selected_df['computer_id'].tolist() == ['B','C']
assert selected_df.shape == (2,3)
assert computers_df['price_nis'].tolist() == [1000,2500,4000,4500]
```

<div dir="rtl" lang="he" style="text-align:right">

## מיון רשומות שלמות
אחרי סינון, מיון יורד לפי מחיר. יש להשאיר את מזהה המחשב ואת הזיכרון צמודים למחירו.

</div>

```python
ranked_df = selected_df.sort_values('price_nis',ascending=False)
display(ranked_df)
assert ranked_df['computer_id'].tolist() == ['C','B']
assert ranked_df['memory_gb'].tolist() == [32,8]
assert computers_df.shape == (4,3)
```

<div dir="rtl" lang="he" style="text-align:right">

## חישוב שיעור נבחרים וגבול
2 מתוך 4 = 50% בארבע הצעות אלו בלבד. אם הגבול העליון יהיה קטן ממש מ־4000, C לא ייבחר.

</div>

```python
rate = len(selected_df)/len(computers_df)*100
strict = computers_df[(computers_df['price_nis']>=1500)&(computers_df['price_nis']<4000)]
print('Selection rate (%)',rate)
print('Strict upper:',strict['computer_id'].tolist())
assert rate == 50.0
assert strict['computer_id'].tolist() == ['B']
```

<div dir="rtl" lang="he" style="text-align:right">

## גרף השוואתי למורה בלבד
ציר אנכי: מחיר בשקלים; קווים לסף תחתון ולעליון. התלמיד לא נדרש ללמוד Matplotlib בשיעור זה.

</div>

```python
import matplotlib.pyplot as plt
plt.figure(figsize=(7,4))
plt.bar(computers_df['computer_id'], computers_df['price_nis'])
plt.axhline(1500,linestyle='--',label='Lower 1500 NIS')
plt.axhline(4000,linestyle=':',label='Upper 4000 NIS')
plt.ylabel('Price (NIS)')
plt.xlabel('Computer ID')
plt.ylim(0,5000)
plt.legend()
plt.show()
```

## מעבר ל־Penguins האמיתי
לומדים להעביר את התהליך: ניצור שתי מסכות על אותו DataFrame ואז נמיין שורות שלמות. **הנתונים לא הופכים למודל לומד.**

</div>

```python
penguins_url = 'https://raw.githubusercontent.com/samiabo75/Data-Science-ML-Course/main/Datasets/penguins.csv'
penguins_df = pd.read_csv(penguins_url)
print('Penguins source:',penguins_df.shape)
assert penguins_df.shape == (344,7)
```

<div dir="rtl" lang="he" style="text-align:right">

## Adelie וגם אורך סנפיר לפחות 190 מ״מ
תנו תחילה לתלמידים להציע שתי מסכות ולנבא האם מספר תצפיות החיתוך יכול לעלות על מספר כל תת־קבוצה.

</div>

```python
adelie_mask = penguins_df['species'] == 'Adelie'
long_mask = penguins_df['flipper_length_mm'] >= 190
chosen = penguins_df[adelie_mask & long_mask]
print('Adelie:',adelie_mask.sum(),'Long:',long_mask.sum(),'Both:',len(chosen))
display(chosen[['species','flipper_length_mm']].head(4))
assert len(chosen) <= min(adelie_mask.sum(),long_mask.sum())
```

<div dir="rtl" lang="he" style="text-align:right">

## ממיינים מסה ושומרים רשומה שלמה
מיינו Adelie לפי body_mass_g בסדר יורד; ההצגה של חמש רשומות היא הצגת קצוות בלבד, **לא מדגם מייצג**.

</div>

```python
adelie_df = penguins_df[adelie_mask]
adelie_sorted = adelie_df.sort_values('body_mass_g', ascending=False)
display(adelie_sorted[['species','island','body_mass_g']].head(5))
print('Before/after sort:',adelie_df.shape,adelie_sorted.shape)
assert adelie_df.shape == adelie_sorted.shape
```

<div dir="rtl" lang="he" style="text-align:right">

## דוגמה ל־NaN בסוף המיון
בטבלת Penguins חסרות שתי מדידות אורך סנפיר. מיון עם ערכים חסרים אינו מתנהג כאילו החסר הוא אפס.

</div>

```python
flipper_desc = penguins_df.sort_values('flipper_length_mm',ascending=False)
display(flipper_desc[['species','flipper_length_mm']].tail(3))
assert flipper_desc['flipper_length_mm'].isna().sum() == 2
assert flipper_desc['flipper_length_mm'].tail(2).isna().all()
```
