# מפגש 6 — פתרון מלא למורה: סינון, AND ומיון
החישובים מיועדים למורה ואינם למצגת. המחברת משתמשת בקישורי המקור ובבדיקות assert. קובץ Diamonds הגדול דורש גישה חיצונית ל־Raw URL בזמן הרצת Colab.

</div>

```python
import pandas as pd
penguins_url = 'https://raw.githubusercontent.com/samiabo75/Data-Science-ML-Course/main/Datasets/penguins.csv'
penguins_df = pd.read_csv(penguins_url)
print('Penguins source:',penguins_df.shape)
assert penguins_df.shape == (344,7)
```

<div dir="rtl" lang="he" style="text-align:right">

## חלק 1 — מיון סנפיר ומסה
מיון שורות שלמות משאיר מזהה/מין/אי/מסה מחוברים. ערכי NaN מוצגים בסוף כברירת מחדל.

</div>

```python
flippers_ascending = penguins_df.sort_values('flipper_length_mm')
flippers_descending = penguins_df.sort_values('flipper_length_mm',ascending=False)
display(flippers_ascending[['species','flipper_length_mm']].head(5))
display(flippers_descending[['species','flipper_length_mm']].tail(3))
assert flippers_ascending.shape == flippers_descending.shape == penguins_df.shape
assert flippers_descending['flipper_length_mm'].tail(2).isna().all()
```

```python
adelie_mask = penguins_df['species'] == 'Adelie'
adelie_df = penguins_df[adelie_mask]
adelie_mass_sorted = adelie_df.sort_values('body_mass_g',ascending=False)
adelie_top_three = adelie_mass_sorted.head(3)
display(adelie_top_three[['species','island','body_mass_g']])
assert len(adelie_df) == 152
assert len(adelie_top_three) == 3
assert adelie_mass_sorted.shape == adelie_df.shape
```

<div dir="rtl" lang="he" style="text-align:right">

## חלק 2 — AND וטווח על Penguins
שני תנאים נפרדים על אותו DataFrame. המפגש אינו מצריך groupby או טכניקות שמירה מתקדמות.

</div>

```python
long_mask = penguins_df['flipper_length_mm'] >= 190
combined_mask = adelie_mask & long_mask
adelie_long_df = penguins_df[combined_mask]
print('Adelie',adelie_mask.sum(),'Length >= 190',long_mask.sum(),'Both',len(adelie_long_df))
assert len(adelie_long_df) <= min(adelie_mask.sum(),long_mask.sum())
assert all(adelie_long_df['species']=='Adelie')
assert all(adelie_long_df['flipper_length_mm']>=190)
```

```python
lower_mask = penguins_df['bill_length_mm'] >= 40
upper_mask = penguins_df['bill_length_mm'] <= 45
bill_range_df = penguins_df[lower_mask & upper_mask]
compact_df = penguins_df[(penguins_df['species']=='Adelie') & (penguins_df['flipper_length_mm']>=190)]
print('Bill range:',bill_range_df.shape,'Compact:',compact_df.shape)
assert compact_df.index.equals(adelie_long_df.index)
assert (bill_range_df['bill_length_mm']>=40).all()
assert (bill_range_df['bill_length_mm']<=45).all()
assert penguins_df.shape == (344,7)
```

<div dir="rtl" lang="he" style="text-align:right">

## חלק 3 — Diamonds: נתוני ההשתלמות
לפי תיעוד ggplot2 של diamonds, מבנה מוכר הוא 53,940×10; עם זאת יש לבצע בדיקת מקור מלאה לקובץ זה ב־Colab. לא נרשמו כאן מספרי קבוצות בלתי מאומתים מראש.

</div>

```python
diamonds_url = 'https://raw.githubusercontent.com/arielb30/datasets/main/diamonds.csv'
diamonds_df = pd.read_csv(diamonds_url)
print('Diamonds source:',diamonds_df.shape)
print('Fields:',list(diamonds_df.columns))
display(diamonds_df.head())
required = {'carat','cut','color','clarity','depth','table','price','x','y','z'}
assert required.issubset(diamonds_df.columns)
assert len(diamonds_df) > 0
```

<div dir="rtl" lang="he" style="text-align:right">

## price כסדרה וכטבלה, ובחירת Ideal

</div>

```python
price_series = diamonds_df['price']
price_table = diamonds_df[['price']]
ideal_mask = diamonds_df['cut']=='Ideal'
ideal_df = diamonds_df[ideal_mask]
print('Price series/table:',price_series.shape,price_table.shape)
print('Ideal:',ideal_df.shape)
display(ideal_df[['carat','cut','color','price']].head(4))
assert price_series.shape == (len(diamonds_df),)
assert price_table.shape == (len(diamonds_df),1)
assert (ideal_df['cut']=='Ideal').all()
```

<div dir="rtl" lang="he" style="text-align:right">

## Ideal וגם carat לפחות 1 — כל שורה עוברת שני תנאים

</div>

```python
one_carat_mask = diamonds_df['carat']>=1
ideal_one_carat_df = diamonds_df[ideal_mask & one_carat_mask]
print('Ideal and at least 1 carat:',ideal_one_carat_df.shape)
assert (ideal_one_carat_df['cut']=='Ideal').all()
assert (ideal_one_carat_df['carat']>=1).all()
assert len(ideal_one_carat_df)<=len(ideal_df)
```

<div dir="rtl" lang="he" style="text-align:right">

## מיון מחיר יורד — חמש רשומות יקרות בקבוצה
המחירים בקובץ בדולרים ומתארים מאגר נתונים, לא מחירון עדכני. כשיש מחירים שווים בגבול, אין להכריז על סדר 'נכון' בין השווים.

</div>

```python
ranked_df = ideal_one_carat_df.sort_values('price',ascending=False)
top_five_df = ranked_df.head(5)
display(top_five_df[['carat','cut','color','clarity','price']])
assert ranked_df.shape == ideal_one_carat_df.shape
assert top_five_df['price'].is_monotonic_decreasing
```

<div dir="rtl" lang="he" style="text-align:right">

## טווח משקל 0.5–0.7 ומיון עולה

</div>

```python
lower_carat = diamonds_df['carat']>=0.5
upper_carat = diamonds_df['carat']<=0.7
mid_weight_df = diamonds_df[lower_carat & upper_carat]
mid_weight_sorted = mid_weight_df.sort_values('price')
display(mid_weight_sorted[['carat','cut','price']].head(3))
assert (mid_weight_df['carat']>=0.5).all()
assert (mid_weight_df['carat']<=0.7).all()
assert diamonds_df.shape[1] >= 10
```

<div dir="rtl" lang="he" style="text-align:right">

## תיעוד, בקרת איכות ומשמעות
תלמיד יכול לומר 'מצאתי את חמש הרשומות היקרות בקבוצת Ideal שקראט לפחות 1', אבל אינו רשאי להסיק שבאלה טמון ממוצע מחיר השוק. אפשר לבדוק x=0 כסימן שדורש בירור, מבלי למחוק או להמציא סיבה. אין כאן חיזוי, התאמת מודל או שימוש ב־Test.

</div>

```python
zero_x_df = diamonds_df[diamonds_df['x']==0]
print('Zero x values - investigate:',zero_x_df.shape)
print('Original source unchanged:',diamonds_df.shape)
assert diamonds_df.shape[1]>=10
```
