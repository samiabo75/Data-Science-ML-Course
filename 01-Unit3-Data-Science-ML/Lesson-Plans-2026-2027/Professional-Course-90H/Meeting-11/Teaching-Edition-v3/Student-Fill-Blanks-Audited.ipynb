## חלק 1: [10,12,14,16,30] — הרבעונים בשיטת Pandas linear
לפני קוד: סדרו; חשבו חציון, Q1 ו־Q3 לפי `quantile(...,interpolation='linear')` כפי שנדרש בחומר המקור; מצאו IQR, הגדרות וקצות שפמים **מן התצפיות**. החליטו איזה ערך רק מסומן לבדיקה.

```python
import pandas as ?
import seaborn as sns
import matplotlib.pyplot as plt
demo_df=pd.DataFrame({'value':[10,12,14,16,30]})
q1=demo_df['value'].?(.25,interpolation='linear')
q2=demo_df['value'].?()
q3=demo_df['value'].?(.75,interpolation='linear')
iqr=q3 ? q1
lower=q1 ? 1.5*iqr
upper=q3 ? 1.5*iqr
print(q1,q2,q3,iqr,lower,upper)
```

## Boxplot של חמש המדידות
כתבו כותרת, יחידות, ואל תסמנו את הגדר במקום השפם. מה יכולה להיות סיבה אמיתית לערך חריג?

```python
plt.figure(figsize=(6,4))
sns.?(data=demo_df,y=?,whis=1.5,color='lightblue')
plt.title('Teaching observations')
plt.ylabel('Measured value')
plt.show()
```

## חלק 2: בדיקת איכות של עסקאות אחרות
זהו נתון מדומה של ההשתלמות, **לא** דוגמת המורה. כל שורה מייצגת הזמנה. order_id אמור להיות ייחודי; לפי חוזה התרגול duration_min חייב להיות חיובי ממש (>0).

```python
import numpy as np
raw=pd.DataFrame({'order_id':[201,202,203,203,204,205,206], 'ordered_at':['2026-02-01','2026-02-02','2026-02-03','2026-02-03','2026-02-04','2026-02-05','2026-02-06'], 'items':[1,2,3,3,2,1,2], 'unit_price':[20.,10.,5.,5.,10.,20.,10.], 'duration_min':[5.,9.,12.,12.,0.,-1.,np.nan]})
print('Before:',len(raw))
display(raw)
```

### משימה 1: מה באמת כפול?
הציגו את כל הרשומות בעלות order_id חוזר; השוו לתוצאות כאשר בוחרים items ו־unit_price. אילו רשומות *ייתכן* שהן עסקאות שונות למרות דמיון?

```python
same_ids=raw[raw.?(subset=[?],keep=False)]
same_measurements=raw[raw.?(subset=[?,?],keep=False)]
display(same_ids)
display(same_measurements)
```

### משימה 2: עותק עבודה ויומן לפני/אחרי
יצרו `work` עותק של raw. **בדקו** שמזהה חוזר מתייחס לשורת ייבוא זהה בכל השדות; רק אחר כך הסירו עותק כפול של הרשומה המלאה. אל תמחקו עסקאות שונות בגלל אותו מחיר/כמות. השאירו raw שלם.

```python
work=raw.?(deep=True)
print('Exact duplicate rows:',work.?().sum())
work=work.?(keep='first')
print('After exact duplicate import:',len(work))
print('Raw unchanged:',len(raw))
```

### משימה 3: משכים לא תקינים בחוזה התרגול
סמנו **רק** משכים לא חסרים שאינם חיוביים כחסרים ב־work. דווחו מספר שינויים, מספר משכים חסרים בסיום, וממדי מקור לעומת עותק. אין לשנות שדות אחרים.

```python
invalid=work['duration_min'].?() & (work['duration_min'] ? 0)
print('Newly flagged:',invalid.sum())
work.loc[invalid,?]=np.nan
print('Final missing durations:',work['duration_min'].?().sum())
display(work)
```
