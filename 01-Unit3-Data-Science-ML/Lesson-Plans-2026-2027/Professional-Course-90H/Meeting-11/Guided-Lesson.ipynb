## המושגים הנחוצים
Boxplot מסכם חציון ורבעונים. IQR=Q3−Q1. נקודה מחוץ לשפם אינה הוכחה לטעות. כפילות נבדקת לפי זהות ומשמעות התצפית. שומרים מקור ויומן שינויים, ומנמקים כל כלל סבירות. אין להסיר נתונים רק כדי לשפר ציון מודל.

</div>

<div dir="rtl" lang="he" style="text-align: right;">

## דוגמה פתורה לפני קוד
אם Q1=10 ו־Q3=20, אז IQR=10 והגבול העליון לפי כלל 1.5IQR הוא 35. ערך 40 דורש בדיקה, אך אינו בהכרח שגוי.

</div>

```python
q1 = 10
q3 = 20
iqr = q3 - q1
upper_fence = q3 + 1.5 * iqr
print(upper_fence)
```

<div dir="rtl" lang="he" style="text-align: right;">

**פלט צפוי:** 35.0

### חישוב ופירוש
קראו את השמות והיחידות, כתבו ציפייה לפני הרצה ובדקו אותה בפלט. הסבירו מה השתנה ומה נשאר קבוע. אין צורך לשנן פקודה בלי להבין את פעולתה.

</div>

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
demo_df = pd.DataFrame({"value": [2, 4, 6, 8, 20]})
q1 = demo_df["value"].quantile(0.25, interpolation="linear")
q2 = demo_df["value"].median()
q3 = demo_df["value"].quantile(0.75, interpolation="linear")
iqr = q3 - q1
print("Q1:", q1, "חציון:", q2, "Q3:", q3, "IQR:", iqr)
```

<div dir="rtl" lang="he" style="text-align: right;">

### המחשה חזותית
לפני הרצה הגדירו שאלה, סוגי משתנים ויחידות. אחרי הרצה פרשו את הצירים ואת הממצא. אין להסיק סיבתיות או למחוק קצה רק מן הציור.

</div>

```python
lower_fence = q1 - 1.5 * iqr
upper_fence = q3 + 1.5 * iqr
print("גבול תחתון:", lower_fence, "גבול עליון:", upper_fence)
plt.figure(figsize=(7, 4))
sns.boxplot(data=demo_df, y="value", whis=1.5, color="lightblue")
plt.title("Teaching example: quartiles and a flagged value")
plt.xlabel("All teaching observations")
plt.ylabel("Value")
plt.tight_layout()
plt.show()
```
