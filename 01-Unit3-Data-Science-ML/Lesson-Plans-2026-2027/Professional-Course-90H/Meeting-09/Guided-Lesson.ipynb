## המושגים הנחוצים
בתרשים פיזור כל נקודה מייצגת זוג מדידות מאותה שורה. קורלציית Pearson מתארת קשר ליניארי ונעה בין −1 ל־1 כאשר השונויות חיוביות. קורלציה אינה סיבתיות, וערך קרוב לאפס אינו שולל קשר לא ליניארי. בחירת מאפיינים בעזרת יעד נעשית בנתוני הפיתוח.

</div>

<div dir="rtl" lang="he" style="text-align: right;">

## דוגמה פתורה לפני קוד
ל־x=[−1,0,1] ול־y=[1,0,1] יש קשר ברור של ריבוע, אך קורלציית Pearson היא אפס. זו המחשה דידקטית, לא ממצא על מאגר אמיתי.

</div>

```python
import pandas as pd
x = pd.Series([-1., 0., 1.])
y = x ** 2
print(x.corr(y))
```

### תשתית הנתונים והחלוקה
קראו תחילה את היעד ואת שמות המאפיינים. החלוקה שומרת התאמה בין שורות X ל־y. הפרמטר random_state מאפשר שחזור ואינו משפר דיוק בפני עצמו.

</div>

```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
url = "https://raw.githubusercontent.com/samiabo75/Data-Science-ML-Course/main/Datasets/penguins.csv"
df = pd.read_csv(url)
development, heldout = train_test_split(df, test_size=0.25, random_state=42)
development.info()
display(development.describe())
display(development.describe(include="object"))
print("Species:", development["species"].unique())
print("Unique species:", development["species"].nunique(dropna=True))
display(development.isna().sum())
```

<div dir="rtl" lang="he" style="text-align: right;">

### המחשה חזותית
לפני הרצה הגדירו שאלה, סוגי משתנים ויחידות. אחרי הרצה פרשו את הצירים ואת הממצא. אין להסיק סיבתיות או למחוק קצה רק מן הציור.

</div>

```python
numeric = ["bill_length_mm", "bill_depth_mm", "flipper_length_mm", "body_mass_g"]
corr = development[numeric].corr()
pair_counts = development[numeric].notna().astype(int).T.dot(
    development[numeric].notna().astype(int))
display(corr)
display(pair_counts)
plt.figure(figsize=(7, 5))
sns.heatmap(corr, vmin=-1, vmax=1, cmap="coolwarm", annot=True, fmt=".2f")
plt.title("Development data — Pearson correlations")
plt.tight_layout()
plt.show()
plt.figure()
sns.scatterplot(data=development, x="flipper_length_mm", y="body_mass_g", hue="species")
plt.title("Development data — relationship and groups")
plt.tight_layout()
plt.show()
```
