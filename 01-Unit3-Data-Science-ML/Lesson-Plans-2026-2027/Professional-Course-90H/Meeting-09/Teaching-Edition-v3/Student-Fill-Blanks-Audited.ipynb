## 0. טעינת מקור וחלוקה שמורה
הקובץ ההשתלמות הוא Palmer Penguins, והחלוקה מתקבעת עם random_state=17. בדקו רק את development לאחר חלוקה.

```python
import pandas as ?
import seaborn as sns
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
url="https://raw.githubusercontent.com/samiabo75/Data-Science-ML-Course/main/Datasets/penguins.csv"
penguins_df=pd.?(url)
development,heldout=train_test_split(penguins_df,test_size=0.25,random_state=17)
```

הציגו info ו־describe מספרי וקטגוריאלי של development בלבד. כתבו במילים מה מייצגת שורה אחת ואיזה שדה נמדד בגרמים.

```python
development.?()
display(development.?())
display(development.describe(include=?))
```

## 3. מתאם בין אורך מקור למסת גוף
שמרו רק זוגות עם שתי מדידות ידועות; דווחו מספר זוגות ידועים, חשבו Pearson וציירו Scatterplot. יחידות x במ״מ, y בגרמים. הסבירו מה ניתן ומה לא ניתן להסיק.

```python
pair=development[[?,?]].?()
print('Known pairs:',?(pair))
r=pair[?].?(pair[?])
print('Pearson r:',r)
plt.figure(figsize=(6,4))
sns.scatterplot(data=pair,x=?,y=?)
plt.xlabel('Bill length (mm)')
plt.ylabel('Body mass (g)')
plt.show()
```

## 4. Heatmap של ארבע מדידות
בחרו את ארבע המדידות המספריות. הציגו מטריצה והמחשה. בחרו תא אחד וכתבו פירוש ללא טענת סיבתיות. מדוע צריך לבדוק גם Scatterplot?

```python
numeric=[?,?,?,?]
corr=development[numeric].?()
display(corr)
plt.figure(figsize=(7,5))
sns.heatmap(corr,vmin=-1,vmax=1,annot=True,fmt='.2f')
plt.title('Development correlations')
plt.tight_layout()
plt.show()
```

## 5. קשר לא־ליניארי

```python
toy_x=pd.Series([-3.,-2.,-1.,0.,1.,2.,3.])
toy_y=toy_x ? 2
print('Pearson:',toy_x.?(toy_y))
plt.figure(figsize=(6,4))
plt.scatter(toy_x,toy_y)
plt.xlabel('x')
plt.ylabel('x squared')
plt.show()
```
