## פתרון חלק 1
הפתרון הבא תואם לנתוני התרגול העצמאי, ולא לנתוני הדוגמה הפתורה.

</div>

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
url = "https://raw.githubusercontent.com/samiabo75/Data-Science-ML-Course/main/Datasets/penguins.csv"
df = pd.read_csv(url)
development, heldout = train_test_split(df, test_size=0.25, random_state=17)
```

<div dir="rtl" lang="he" style="text-align: right;">

## סקירה וספירות
אי הוא קטגוריה שמית; מסה כמותית. מספר ערכים שונים קטן ממספר תצפיות. מדגם אזורי אינו מייצג אוטומטית את כל הפינגווינים.

</div>

```python
development.info()
display(development.describe())
display(development.describe(include="object"))
print(development["island"].unique())
print(development["island"].nunique(dropna=True))
display(development["island"].value_counts(dropna=False))
```

<div dir="rtl" lang="he" style="text-align: right;">

## זוג ומטריצה
חישוב בזוגות ידועים אינו מצדיק השלמת יעד או הסקת סיבתיות. בחירת מאפיינים לא משתמשת ב־ heldout.

</div>

```python
numeric = ["bill_length_mm", "bill_depth_mm", "flipper_length_mm", "body_mass_g"]
pair = development[["bill_length_mm", "body_mass_g"]].dropna()
print("Known pairs:", len(pair), "Pearson:", pair["bill_length_mm"].corr(pair["body_mass_g"]))
corr = development[numeric].corr()
display(corr)
plt.figure(figsize=(7,5))
sns.heatmap(corr, vmin=-1, vmax=1, cmap="coolwarm", annot=True, fmt=".2f")
plt.title("Practice development correlations"); plt.tight_layout(); plt.show()
plt.figure()
sns.scatterplot(data=pair, x="bill_length_mm", y="body_mass_g")
plt.title("Practice development — bill and mass"); plt.tight_layout(); plt.show()
```
