## חלק 2: עסקות מומצאות אחרות — לא הדוגמה הפתורה
A=[1,חסר], B=[2,2,חסר], C=[5] יחידות. לפני הרצה חשבו ביד לכל חנות size/count/sum/mean. כל שורה היא עסקה, לא יחידת מוצר.

```python
import numpy as np
sales=pd.DataFrame({'branch':['A','A','B','B','B','C'],'units':[1.,np.nan,2.,2.,np.nan,5.]})
display(sales)
print('Rows:',sales.?)
```

### שלב א — ארבע פעולות בנפרד
כתבו תחילה במילים מה כל פעולה סופרת או מודדת. חסר אינו אפס.

```python
grouped=sales.?(?)[?]
print('size:',grouped.?())
print('count:',grouped.?())
print('sum:',grouped.?())
print('mean:',grouped.?())
```

### שלב ב — טבלת agg
יצרו DataFrame המשלבת ארבע פעולות. השוו לחישוב הידני שלכם.

```python
summary=grouped.?([?,?,?,?])
display(summary)
```

### שלב ג — שתי תשובות שונות לממוצע
חשבו את הממוצע של כל המדידות הידועות; אחריו ממוצע פשוט של שלושת ממוצעי החנויות. הסבירו למי ניתן משקל שווה בכל מקרה.

```python
overall_mean=sales[?].?()
mean_of_group_means=summary[?].?()
print('Overall:',overall_mean)
print('Mean of group means:',mean_of_group_means)
```
