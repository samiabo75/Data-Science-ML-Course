# מפגש 10 — מחברת הדגמה: GroupBy מן השאלה עד הגרף
## סיפור: שני דוכנים בקיוסק
נתונים מומצאים ללימוד. לפני הרצה — צפו פלט: מה מספר השורות? מהי כמות יחידות ידועה? אין להתייחס ל־NaN כאפס.

```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
sales_df=pd.DataFrame({'shop':['A','A','B','B'],'units':[1,np.nan,2,4]})
display(sales_df)
assert sales_df.shape==(4,2)
```

## מחשבים תחילה ידנית
A: size=2, count=1, sum=1, mean=1; B: size=2, count=2, sum=6, mean=3. כל שורה בתוצאה מסכמת **דוכן**, לא עסקה.

```python
by_shop=sales_df.groupby('shop')['units']
print('Size:',by_shop.size().to_dict())
print('Count:',by_shop.count().to_dict())
print('Sum:',by_shop.sum().to_dict())
print('Mean:',by_shop.mean().to_dict())
assert by_shop.size().to_dict()=={'A':2,'B':2}
assert by_shop.count().to_dict()=={'A':1,'B':2}
assert by_shop.sum().to_dict()=={'A':1.,'B':6.}
assert by_shop.mean().to_dict()=={'A':1.,'B':3.}
```

## אותה תוצאה בטבלה אחת
לימדו את agg רק אחרי שהבנתם את כל אחת מארבע הפעולות.

```python
summary=by_shop.agg(['size','count','sum','mean'])
display(summary)
assert summary.shape==(2,4)
assert sales_df['units'].isna().sum()==1
```

## שני גרפים — שתי שאלות שונות
Countplot שואלת כמה *שורות עסקאות* לכל דוכן; Barplot עם mean מראה *יחידות מדווחות בממוצע* לעסקה.

```python
plt.figure(figsize=(6,4))
sns.countplot(data=sales_df,x='shop',order=['A','B'],color='teal')
plt.title('Transactions by shop')
plt.ylabel('Number of transactions')
plt.show()
plt.figure(figsize=(6,4))
sns.barplot(data=sales_df,x='shop',y='units',order=['A','B'],estimator='mean',errorbar=None,color='steelblue')
plt.title('Mean reported units per transaction')
plt.ylabel('Mean units (known transactions)')
plt.show()
```

## מקרה גבול: אין אף כמות ידועה
כשהקבוצה כוללת רק NaN, mean היא NaN. ברירת מחדל של sum עלולה להציג 0 למרות שלא נמדדו אפס יחידות.

```python
missing_df=pd.DataFrame({'shop':['C','C'],'units':[np.nan,np.nan]})
missing_group=missing_df.groupby('shop')['units']
print('No known values:',missing_group.count().to_dict())
print('Default sum:',missing_group.sum().to_dict())
print('Conservative sum:',missing_group.sum(min_count=1).to_dict())
print('Mean:',missing_group.mean().to_dict())
assert missing_group.size()['C']==2
assert missing_group.count()['C']==0
assert np.isnan(missing_group.sum(min_count=1)['C'])
assert np.isnan(missing_group.mean()['C'])
```

## מעבר למאגר אמיתי — בלי לחשוף את תרגיל האיים
הדגמת המורה משתמשת ב־species ובמסת גוף, כדי שתלמידים יפתרו באופן עצמאי את תרגיל island ואורך הסנפיר.

```python
url='https://raw.githubusercontent.com/samiabo75/Data-Science-ML-Course/main/Datasets/penguins.csv'
penguins_df=pd.read_csv(url)
print('Source:',penguins_df.shape)
species_mass=penguins_df.groupby('species')['body_mass_g'].agg(['size','count','mean'])
display(species_mass)
assert penguins_df.shape==(344,7)
assert species_mass['size'].sum()==344
assert species_mass['count'].sum()==342
```
