### 1–3. טבלה
island היא קטגוריה שמית; flipper_length_mm מדידה רציפה במ״מ. ספירה ב־ Y היא מספר תצפיות; ממוצע ב־ Y הוא אורך במ״מ. count על מדידה סופרת רק ידועים.

</div>

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
url = "https://raw.githubusercontent.com/samiabo75/Data-Science-ML-Course/main/Datasets/penguins.csv"
penguins_df = pd.read_csv(url)
mean_flipper_by_island = penguins_df.groupby("island")["flipper_length_mm"].mean()
known_flippers_by_island = penguins_df.groupby("island")["flipper_length_mm"].count()
print("ממוצע בממ לפי אי:")
print(mean_flipper_by_island)
print("מספר אורכים ידועים לפי אי:")
print(known_flippers_by_island)
```

<div dir="rtl" lang="he" style="text-align: right;">

בגרסה זו מספר הידועים: Biscoe=167, Dream=124, Torgersen=51. מספר הרשומות בקבוצות: 168,124,52. count אינו מספר חסרים ואינו בהכרח גובה Countplot.

</div>

```python
island_order = ["Biscoe", "Dream", "Torgersen"]
plt.figure(figsize=(7, 4))
sns.countplot(data=penguins_df, x="island", order=island_order, color="teal")
plt.title("Number of observations by island")
plt.xlabel("Island")
plt.ylabel("Number of observations")
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(7, 4))
sns.barplot(data=penguins_df, x="island", y="flipper_length_mm", order=island_order, estimator="mean", errorbar=None, color="steelblue")
plt.title("Mean flipper length by island")
plt.xlabel("Island")
plt.ylabel("Mean flipper length (mm)")
plt.tight_layout()
plt.show()
```

## פתרון חלק 2
הפתרון הבא תואם לנתוני התרגול העצמאי, ולא לנתוני הדוגמה הפתורה.

</div>

```python
import pandas as pd
import numpy as np
# עסקאות דידקטיות מומצאות;NaN אינו אפס יחידות
sales=pd.DataFrame({"branch":["A","A","B","B","B","C"],
                   "units":[1.,np.nan,2.,2.,np.nan,5.]})
```

<div dir="rtl" lang="he" style="text-align: right;">

## סיכומים
A: שתי שורות, ערך ידוע אחד, סכום 1, ממוצע 1; B: שלוש שורות, שניים ידועים, סכום 4, ממוצע 2; C: שורה אחת, סכום וממוצע 5. שורת תוצאה מתארת חנות, לא עסקה.

</div>

```python
grouped=sales.groupby("branch")["units"]
print(grouped.size());print(grouped.count())
print(grouped.sum(min_count=1));print(grouped.mean())
summary=grouped.agg(["size","count","sum","mean"])
display(summary)
```

<div dir="rtl" lang="he" style="text-align: right;">

## מסקנות והסבר
הממוצע הכולל 2.5 מול 8/3≈2.6667. כל ערך ידוע מקבל משקל שווה בממוצע הכולל; בממוצע הממוצעים כל חנות מקבלת משקל שווה. ניבוי ביקוש עתידי משתמש רק בהיסטוריה הזמינה בזמן התחזית.

</div>

```python
overall=sales["units"].mean()
mean_of_means=summary["mean"].mean()
print("Overall:",overall,"Mean of means:",mean_of_means)
```

<div dir="rtl" lang="he" style="text-align: right;">

## דגשים להגנה
groupby מחלק שורות לפי מפתח; אגרגציה מחזירה סיכום לכל קבוצה. size סופר שורות, count סופר ערכים ידועים בעמודה שנבחרה, sum מסכם ערכים ידועים ו־ mean מחלק את סכומם במספר הערכים הידועים. ממוצע פשוט של ממוצעי קבוצות אינו בהכרח הממוצע הכולל כשמספרי הערכים הידועים שונים.
לא סופרים count ככמות שורות כולל חסרים; לא מפרשים סכום יחידות כמספר עסקאות; לא מחברים קטגוריות; לא מסיקים שהחנות בעלת סכום גדול יותר בעלת עסקה ממוצעת גדולה יותר. שומרים מקור; הנתונים מומצאים ואין כאן מסקנת שוק אמיתית.
