### חישוב ופירוש
קראו את השמות והיחידות, כתבו ציפייה לפני הרצה ובדקו אותה בפלט. הסבירו מה השתנה ומה נשאר קבוע. אין צורך לשנן פקודה בלי להבין את פעולתה.

</div>

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
demo_df = pd.DataFrame({
    "group": ["A", "A", "A", "B", "B"],
    "price_nis": [100, 120, 140, 200, 220]
})
mean_prices = demo_df.groupby("group")["price_nis"].mean()
known_prices = demo_df.groupby("group")["price_nis"].count()
print("מספר מחירים ידועים לפי קבוצה:")
print(known_prices)
print("מחיר ממוצע בשקלים לפי קבוצה:")
print(mean_prices)
```

<div dir="rtl" lang="he" style="text-align: right;">

### המחשה חזותית
לפני הרצה הגדירו שאלה, סוגי משתנים ויחידות. אחרי הרצה פרשו את הצירים ואת הממצא. אין להסיק סיבתיות או למחוק קצה רק מן הציור.

</div>

```python
plt.figure(figsize=(7, 4))
sns.barplot(data=demo_df, x="group", y="price_nis", estimator="mean", errorbar=None, color="steelblue")
plt.title("Mean teaching price by group")
plt.xlabel("Group")
plt.ylabel("Mean price (NIS)")
plt.tight_layout()
plt.show()
```

### חישוב ופירוש
קראו את השמות והיחידות, כתבו ציפייה לפני הרצה ובדקו אותה בפלט. הסבירו מה השתנה ומה נשאר קבוע. אין צורך לשנן פקודה בלי להבין את פעולתה.

</div>

```python
import pandas as pd
import numpy as np
sales=pd.DataFrame({"branch":["North","North","North","South","South","South"],
                   "units":[2.,3.,np.nan,1.,4.,np.nan]})
grouped=sales.groupby("branch")["units"]
print("Rows per branch:");print(grouped.size())
print("Known units values:");print(grouped.count())
print("Total units:");print(grouped.sum(min_count=1))
print("Mean units in known transactions:");print(grouped.mean())
```

<div dir="rtl" lang="he" style="text-align: right;">

### חישוב ופירוש
קראו את השמות והיחידות, כתבו ציפייה לפני הרצה ובדקו אותה בפלט. הסבירו מה השתנה ומה נשאר קבוע. אין צורך לשנן פקודה בלי להבין את פעולתה.

</div>

```python
summary=sales.groupby("branch")["units"].agg(["size","count","sum","mean"])
display(summary)
print("Overall mean:",sales["units"].mean())
print("Mean of group means:",summary["mean"].mean())
```
