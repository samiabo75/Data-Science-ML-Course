# מפגש 3 — פתרון מורה מלא ומוסבר
**לשימוש מורה בלבד; לא למצגת.** נתוני P1–P4 ורשימת המחירים הם מן המעבדה המקורית, בנפרד מנתוני ההדגמה.

</div>

<div dir="rtl" lang="he" style="text-align:right;">

## חלק א — השוואות ומקרה גבול
2400≤3000 True; ‏3200≤3000 False; ‏3000<3000 False; ‏3000≤3000 True. = משמש להשמה, == בודק שוויון.

</div>

```python
assert (2400 <= 3000, 3200 <= 3000, 3000 < 3000, 3000 <= 3000) == (True,False,False,True)
print('Truth table verified')
```

<div dir="rtl" lang="he" style="text-align:right;">

### משימה 2 — תנאי והודעת סיום
מחיר 2600 מתאים, ולכן הודעת התאמה והודעת סיום יודפסו. אם המחיר 3200, רק הודעת הסיום תופיע. להחזיר את המקור ל־2600.

</div>

```python
price_nis = 2600
budget_nis = 3000
if price_nis <= budget_nis:
    print('המחיר מתאים:', price_nis)
print('הבדיקה הסתיימה')
assert price_nis <= budget_nis
assert not (3200 <= budget_nis)
```

<div dir="rtl" lang="he" style="text-align:right;">

### משימה 3 — מעבר על כל הרשימה וסינון הדפסות
חמישה מחירים; שלושה עומדים בתקציב עד 3000 כולל. אין כאן סינון DataFrame או שינוי של הרשימה.

</div>

```python
prices_nis = [1800,3200,3000,4200,2600]
for price_nis in prices_nis:
    print('מחיר (NIS):', price_nis)
for price_nis in prices_nis:
    if price_nis <= budget_nis:
        print('עומד בתקציב:', price_nis)
assert len(prices_nis) == 5
assert [p for p in prices_nis if p <= budget_nis] == [1800,3000,2600]
```

<div dir="rtl" lang="he" style="text-align:right;">

### משימה 4 — טבלת מעקב
| מעבר | מחיר | מחיר≤3000? | יודפס? |
|---|---:|---|---|
| 1 | 1800 | True | כן |
| 2 | 3200 | False | לא |
| 3 | 3000 | True | כן |
| 4 | 4200 | False | לא |
| 5 | 2600 | True | כן |

חמישה מעברים, שלוש הדפסות. False אינו עוצר את הלולאה.

</div>

<div dir="rtl" lang="he" style="text-align:right;">

### משימה 5 — פחות ממש
התנאי <3000 מוציא את המחיר השווה בדיוק לסף. יודפסו 1800 ו־2600 בלבד.

</div>

```python
for price_nis in prices_nis:
    if price_nis < budget_nis:
        print('מתחת לתקציב:', price_nis)
assert [p for p in prices_nis if p < budget_nis] == [1800,2600]
```

<div dir="rtl" lang="he" style="text-align:right;">

### משימה 6 — שמירת מקור והזחה
פקודת print(price_nis) צריכה להיות מוזחת לאותה רמה כמו ההודעה שבתוך if, אחרת היא תרוץ בכל מעבר.

</div>

```python
print('המקור:', prices_nis)
for price_nis in prices_nis:
    if price_nis <= budget_nis:
        print('מתאים')
        print(price_nis)
assert prices_nis == [1800,3200,3000,4200,2600]
```

<div dir="rtl" lang="he" style="text-align:right;">

### משימת רשות — מונה
מונה לא מסכם שקלים אלא מספר הצעות. הוא יתקדם מ־0 ל־1, אחר כך 2, ולבסוף 3.

</div>

```python
matching_count = 0
for price_nis in prices_nis:
    if price_nis <= budget_nis:
        matching_count = matching_count + 1
print('מספר הצעות מתאימות:', matching_count)
assert matching_count == 3
```

<div dir="rtl" lang="he" style="text-align:right;">

## חלק ב — DataFrame של P1–P4
כל שורה מחשב; ארבעה שדות. העמודות שוות אורך ומתארות את אותן ארבע תצפיות.

</div>

```python
import pandas as pd
computer_data = {
    'computer_id': ['P1','P2','P3','P4'],
    'year': [2022,2024,2021,2023],
    'memory_gb': [8,16,8,32],
    'price_nis': [1800,3200,1500,4200]
}
computers_df = pd.DataFrame(computer_data)
display(computers_df)
assert computers_df.shape == (4,4)
```

<div dir="rtl" lang="he" style="text-align:right;">

### משימה 8 — ממדים ושמות
shape=(4,4): ארבע תצפיות וארבע עמודות נתונים. האינדקס ברירת המחדל אינו שדה נוסף.

</div>

```python
print('ממדים:', computers_df.shape)
print('שמות שדות:', list(computers_df.columns))
assert list(computers_df.columns) == ['computer_id','year','memory_gb','price_nis']
```

<div dir="rtl" lang="he" style="text-align:right;">

### משימה 9 — הצצה
head(3) מציגה P1,P2,P3. הטבלה המקורית עדיין מכילה גם את P4.

</div>

```python
display(computers_df.head(3))
print('ממדי המקור:', computers_df.shape)
assert computers_df.head(3).shape == (3,4)
assert computers_df.shape == (4,4)
assert computers_df.iloc[3]['computer_id'] == 'P4'
```

<div dir="rtl" lang="he" style="text-align:right;">

### הרחבה — הוספת שדה בטבלת ניסוי נפרדת
אין לשנות את המקור; מחשבים הם אותם ארבעה, והוספת storage_gb הופכת את הצורה ל־(4,5).

</div>

```python
experiment_data = {
    'computer_id': ['P1','P2','P3','P4'],
    'year': [2022,2024,2021,2023],
    'memory_gb': [8,16,8,32],
    'price_nis': [1800,3200,1500,4200],
    'storage_gb': [256,512,256,1000]
}
experiment_df = pd.DataFrame(experiment_data)
print('מקור:', computers_df.shape, 'ניסוי:', experiment_df.shape)
assert computers_df.shape == (4,4)
assert experiment_df.shape == (4,5)
```
