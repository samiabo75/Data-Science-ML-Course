# מפגש 3 — מחברת מודרכת עם מורה
**תנאי תקציב, if ו־DataFrame ראשון**. הנתונים A–D הם נתוני הוראה מומצאים ואינם פתרונות המעבדה. לפני כל תא: חזו את הפלט. אחרי כל תא: פירושו במילים וביחידות.

</div>

<div dir="rtl" lang="he" style="text-align:right;">

## 1. קודם בודקים ביד מקרה גבול
תקציב 2500 ₪. 2100 מתאים, 2700 אינו מתאים, 2500 **מתאים** כאשר התנאי קטן או שווה. לעומת זאת 2500 קטן ממש מ־2500? לא. בדקו True/False לפני הרצה.

</div>

```python
budget_nis = 2500
print('2100 <= 2500:', 2100 <= budget_nis)
print('2700 <= 2500:', 2700 <= budget_nis)
print('2500 <= 2500:', 2500 <= budget_nis)
print('2500 < 2500:', 2500 < budget_nis)
assert (2100 <= budget_nis,2700 <= budget_nis,2500 <= budget_nis,2500 < budget_nis) == (True,False,True,False)
```

<div dir="rtl" lang="he" style="text-align:right;">

## 2. השמה = לעומת בדיקת שוויון ==
שני הסימנים אינם אותו דבר. הראשון מכניס ערך למשתנה; השני שואל האם שני הערכים שווים ומחזיר True או False.

</div>

```python
price_nis = 2400
print('מחיר שווה לגבול:', price_nis == budget_nis)
print('מחיר עומד בתקציב:', price_nis <= budget_nis)
assert (price_nis == budget_nis) is False
```

<div dir="rtl" lang="he" style="text-align:right;">

## 3. תנאי if והזחה
חזו כמה הודעות יודפסו במחיר 2400, וכמה יישארו אם נחליף ל־2700. הודעת הסיום אינה מוזחת לתוך ה־if.

</div>

```python
price_nis = 2400
if price_nis <= budget_nis:
    print('Within budget')
print('Check finished')
price_nis = 2700
if price_nis <= budget_nis:
    print('Within budget')
print('Check finished')
```

<div dir="rtl" lang="he" style="text-align:right;">

## 4. משתמשים ב־for שכבר למדנו
ארבעה מחשבים: A=2100, B=2700, C=3600, D=2400. שימו לב: ארבעה מעברים, רק שתי הדפסות. תוצאת False אינה עוצרת את ה־for.

</div>

```python
prices_nis = [2100, 2700, 3600, 2400]
for price_nis in prices_nis:
    if price_nis <= budget_nis:
        print('מתאים:', price_nis)
print('סיום בדיקת הרשימה')
assert [p for p in prices_nis if p <= budget_nis] == [2100, 2400]
```

<div dir="rtl" lang="he" style="text-align:right;">

## 5. טבלת מעקב ידנית — התלמיד מסביר
| מעבר | מחיר | מחיר ≤ 2500? | הודפס |
|---|---:|---|---|
| 1 | 2100 | True | 2100 |
| 2 | 2700 | False | לא |
| 3 | 3600 | False | לא |
| 4 | 2400 | True | 2400 |

קוד ה־assert בהדגמה הוא בדיקת מורה בלבד; תלמידים אינם צריכים בשיעור זה ללמוד כתיבה של רשימת comprehension.

</div>

## 6. תרשים הדגמה מוכן למורה בלבד
התרשים מתאים לטבלה בדיוק. המורה מציג אותו, אך התלמיד אינו נדרש ללמוד כתיבת Matplotlib בשיעור זה.

</div>

```python
import matplotlib.pyplot as plt
computer_ids = ['A', 'B', 'C', 'D']
prices_nis = [2100, 2700, 3600, 2400]
plt.figure(figsize=(7,4))
plt.bar(computer_ids, prices_nis)
plt.axhline(2500, linestyle='--', label='Budget: 2500 NIS')
plt.xlabel('Computer ID')
plt.ylabel('Price (NIS)')
plt.title('Illustrative computer offers')
plt.ylim(0,4500)
plt.legend()
plt.show()
```

## 7. מרשימות לטבלה: כל שורה מתארת מחשב אחד
יצירת מילון רשימות לצורך DataFrame. שמות העמודות חשובים: מזהה, שנה, זיכרון ומחיר. אינדקס ברירת מחדל אינו עמודה חמישית.

</div>

```python
import pandas as pd
computer_data = {
    'computer_id': ['A','B','C','D'],
    'year': [2021,2023,2020,2024],
    'memory_gb': [8,16,8,32],
    'price_nis': [2100,2700,3600,2400]
}
computers_df = pd.DataFrame(computer_data)
display(computers_df)
assert computers_df.shape == (4,4)
```
