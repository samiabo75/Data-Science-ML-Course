## תחנה 1 — נזכרים במפגש 1
בדוגמה הקודמת נמכרו 12,16,9,15 כריכים. זהו נתון של ארבעה ימים. מהו מספר הימים ומהו הממוצע?

</div>

```python
sandwiches_sold = [12, 16, 9, 15]
days_count = len(sandwiches_sold)
average_sandwiches = sum(sandwiches_sold) / days_count
print('מספר ימים:', days_count)
print('ממוצע כריכים:', average_sandwiches)
assert days_count == 4 and average_sandwiches == 13
```

<div dir="rtl" lang="he" style="text-align: right;">

## תחנה 2 — מועדון המחשבים
דודו קיבל ארבע הצעות. נכתוב רשימת מחירים. האינדקס הראשון הוא 0.

| אינדקס | 0 | 1 | 2 | 3 |
|---|---:|---:|---:|---:|
| מחיר בש״ח | 2100 | 2700 | 3600 | 2400 |

**שאלת חשיבה:** מה ההבדל בין המחיר השלישי לבין price[3]?

</div>

```python
prices_nis = [2100, 2700, 3600, 2400]
print('מחיר ראשון:', prices_nis[0])
print('מחיר שלישי:', prices_nis[2])
assert prices_nis[2] == 3600
```

<div dir="rtl" lang="he" style="text-align: right;">

## תחנה 3 — מהנוסחה לפלט
ממוצע = סכום המחירים / מספר המחשבים. חישוב ידני: (2100+2700+3600+2400)/4 = 10800/4 = 2700 ₪.

טווח = מקסימום − מינימום = 3600−2100 = 1500 ₪.

</div>

```python
count_computers = len(prices_nis)
total_price_nis = sum(prices_nis)
average_price_nis = total_price_nis / count_computers
price_range_nis = max(prices_nis) - min(prices_nis)
print('מספר מחשבים:', count_computers)
print('סכום מחירים בשקלים:', total_price_nis)
print('מחיר ממוצע בשקלים:', average_price_nis)
print('טווח מחירים בשקלים:', price_range_nis)
assert (count_computers, total_price_nis, average_price_nis, price_range_nis) == (4, 10800, 2700, 1500)
```

```python
# גרף המחשה מוכן — אין צורך לדעת לכתוב matplotlib בשיעור 2
import matplotlib.pyplot as plt
computer_ids = ["A", "B", "C", "D"]
prices_nis = [2100, 2700, 3600, 2400]
plt.figure(figsize=(6, 3.5))
plt.bar(computer_ids, prices_nis)
plt.ylabel("Price (NIS)")
plt.xlabel("Computer ID")
plt.title("Illustrative computer offers — invented data")
plt.ylim(0, 4200)
plt.show()
```

## תחנה 5 — אותה פעולה על כל שנה
שנת הייחוס היא 2026. השנים הן 2021,2023,2020,2024. חשבו **ידנית** את הגילים 5,3,6,2 לפני הרצה. הבהירו ש־year מקבל *שנת ייצור*, ולא מספר מיקום.

</div>

```python
years = [2021, 2023, 2020, 2024]
reference_year = 2026
for year in years:
    computer_age_years = reference_year - year
    print('שנת ייצור:', year, 'גיל בשנות לוח:', computer_age_years)
print('סיימנו לעבור על השנים')
```
