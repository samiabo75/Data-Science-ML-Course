## חלק א — ייצוג רשימות
כל אינדקס באותן שלוש רשימות מתייחס לאותו מחשב. למשל האינדקס 2 מתייחס ל־P3.

</div>

```python
computer_ids = ['P1', 'P2', 'P3', 'P4', 'P5']
memory_gb = [8, 16, 8, 32, 16]
prices_nis = [1800, 2600, 1500, 4200, 2900]
print('מזהים:', computer_ids)
print('זיכרון (GB):', memory_gb)
print('מחירים (NIS):', prices_nis)
assert len(computer_ids) == len(memory_gb) == len(prices_nis) == 5
```

<div dir="rtl" lang="he" style="text-align: right;">

### פתרון משימה 2 — אינדקס
המחשב השלישי הוא באינדקס 2. הפלט P3, 8 GB, 1500 ₪. אין לבלבל מספר סידורי אנושי ומיקום Python.

</div>

```python
third_index = 2
print('מחשב שלישי:', computer_ids[third_index])
print('זיכרון (GB):', memory_gb[third_index])
print('מחיר (NIS):', prices_nis[third_index])
assert (computer_ids[third_index], memory_gb[third_index], prices_nis[third_index]) == ('P3', 8, 1500)
```

<div dir="rtl" lang="he" style="text-align: right;">

### פתרון משימה 3 — ממוצע
חישוב ידני: 1800+2600+1500+4200+2900=13000; מספר תצפיות 5; ממוצע 13000/5=2600 ₪. הממוצע תקף לחמש הצעות המחיר בלבד.

</div>

```python
computer_count = len(prices_nis)
total_price_nis = sum(prices_nis)
average_price_nis = total_price_nis / computer_count
print('מספר מחשבים:', computer_count)
print('סכום מחירים (NIS):', total_price_nis)
print('ממוצע מחירים (NIS):', average_price_nis)
assert (computer_count, total_price_nis, average_price_nis) == (5, 13000, 2600)
```

<div dir="rtl" lang="he" style="text-align: right;">

### פתרון משימה 4 — טווח
מינימום 1500, מקסימום 4200, טווח 2700 ₪. טווח אינו אומר מה מצב החומרה, ואינו מדד לכדאיות קנייה ללא מאפיינים נוספים.

</div>

```python
minimum_price_nis = min(prices_nis)
maximum_price_nis = max(prices_nis)
price_range_nis = maximum_price_nis - minimum_price_nis
print('מינימום:', minimum_price_nis, 'מקסימום:', maximum_price_nis, 'טווח:', price_range_nis)
assert (minimum_price_nis, maximum_price_nis, price_range_nis) == (1500, 4200, 2700)
```

<div dir="rtl" lang="he" style="text-align: right;">

### פתרון משימה 5 — העתקה ותיקון
P4 נמצא באינדקס 3; לפני התיקון 4200 ואחרי 4000. סך המחירים יורד ב־200 ולכן הממוצע ב־200/5=40: 2600−40=2560 ₪. הרשימה המקורית לא משתנה.

</div>

```python
corrected_prices_nis = prices_nis.copy()
corrected_prices_nis[3] = 4000
corrected_average_nis = sum(corrected_prices_nis) / len(corrected_prices_nis)
print('מקור:', prices_nis)
print('לאחר תיקון:', corrected_prices_nis)
print('ממוצע לאחר תיקון:', corrected_average_nis)
assert prices_nis[3] == 4200
assert corrected_average_nis == 2560
assert corrected_prices_nis[3] == 4000
```

<div dir="rtl" lang="he" style="text-align: right;">

## חלק ב — מעבר על שלוש שנות ייצור
הערכים של year יהיו 2022,2024,2021 (ולא 0,1,2). הפלט על פי הסדר ברשימה.

</div>

```python
years = [2022, 2024, 2021]
printed_years = []
for year in years:
    print(year)
    printed_years.append(year)
assert printed_years == [2022, 2024, 2021]
```

<div dir="rtl" lang="he" style="text-align: right;">

### פתרון משימה 7 — גיל וטבלת מעקב
נוסחה ידנית: גיל בשנות לוח ≈ 2026 − שנת ייצור. תוצאות: 4,2,5.

| מעבר | year | גיל | הסבר |
|---|---:|---:|---|
| 1 | 2022 | 4 | 2026−2022 |
| 2 | 2024 | 2 | 2026−2024 |
| 3 | 2021 | 5 | 2026−2021 |

זהו הבדל בין שנים ולא גיל מדויק בתאריך.

</div>

```python
reference_year = 2026
computed_ages = []
for year in years:
    age_years = reference_year - year
    computed_ages.append(age_years)
    print('שנה:', year, 'גיל בשנות לוח:', age_years)
assert computed_ages == [4, 2, 5]
```

<div dir="rtl" lang="he" style="text-align: right;">

### פתרון משימה 8 — הזחה
גוף הלולאה מודפס שלוש פעמים. הודעת הסיום נמצאת מחוץ לגוף (ללא הזחה) ולכן מופיעה פעם אחת.

</div>

```python
for year in years:
    print('מעבדים את שנת:', year)
print('העיבוד הסתיים')
```

<div dir="rtl" lang="he" style="text-align: right;">

### פתרון משימה 9 — תיקון תחביר
גרסת התלמיד חסרה נקודתיים אחרי כותרת for וגם הזחה לפני print. לאחר תיקון שתי הטעויות:

</div>

```python
for year in years:
    print(year)
```

<div dir="rtl" lang="he" style="text-align: right;">

### הרחבת מקור: סכום וממוצע גיל באמצעות צובר
אם מלמדים צבירה כמטלת רשות, מאתחלים total_age_years=0 מחוץ ללולאה, מוסיפים בכל מעבר ומחלקים במספר השנים. טבלת מעקב: לפני/אחרי 0→4, 4→6, 6→11. ממוצע 11/3≈3.67.

</div>

```python
total_age_years = 0
for year in years:
    current_age = reference_year - year
    total_age_years = total_age_years + current_age
average_age_years = total_age_years / len(years)
print('סכום גילים:', total_age_years, 'ממוצע גילים:', average_age_years)
assert total_age_years == 11
assert abs(average_age_years - 11 / 3) < 1e-12
```
