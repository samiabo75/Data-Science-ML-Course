**המטרה:** מרשימות ואינדקסים לממוצע, גרף ולולאת for. המחשבים A–D הם דוגמה **מומצאת להוראה**.

היום לא נלמד לכתוב ספריית גרפים — קוד התרשים ניתן למורה. לפני כל קטע Python נחזה פלט ונבדוק את החישוב הידני.

## שלב 1 — רשימה ומיקומים
רשימת המחירים: A=2100 ₪, B=2700 ₪, C=3600 ₪, D=2400 ₪. האינדקסים הם 0,1,2,3. מה יחזיר האיבר השלישי?

</div>

```python
prices_nis = [2100, 2700, 3600, 2400]
print('מספר איברים:', len(prices_nis))
print('מחיר ראשון:', prices_nis[0])
print('מחיר שלישי:', prices_nis[2])
assert len(prices_nis) == 4 and prices_nis[2] == 3600
```

<div dir="rtl" lang="he" style="text-align:right">

## שלב 2 — ממוצע: קודם ידני
סכום = 2100+2700+3600+2400=10800 ₪. מספר הצעות = 4. ממוצע = 10800/4=2700 ₪. טווח = 3600−2100=1500 ₪.

לפני הרצה: כתבו את יחידת התוצאה ולמה אין זו תחזית למחיר של מחשב חדש.

</div>

```python
total_price_nis = sum(prices_nis)
number_of_computers = len(prices_nis)
average_price_nis = total_price_nis / number_of_computers
price_range_nis = max(prices_nis) - min(prices_nis)
print('סכום (NIS):', total_price_nis)
print('ממוצע (NIS):', average_price_nis)
print('טווח (NIS):', price_range_nis)
assert total_price_nis == 10800
assert average_price_nis == 2700
assert price_range_nis == 1500
```

## שלב 3 — רואים את המחירים
הקוד הבא מצייר את ארבעת המחירים. זו הדגמת מורה, לא תחביר שהתלמידים צריכים לכתוב כעת. כל עמודה מתאימה לאותו ערך בטבלה.

</div>

```python
# גרף הדגמה מוכן למורה בלבד - אין צורך ללמוד matplotlib בשיעור 2
import matplotlib.pyplot as plt
computer_ids = ["A", "B", "C", "D"]
prices_nis = [2100, 2700, 3600, 2400]
plt.figure(figsize=(7, 4))
plt.bar(computer_ids, prices_nis, color="#3173ad")
plt.xlabel("Computer ID")
plt.ylabel("Price (NIS)")
plt.title("Computer offers - invented example")
plt.ylim(0, 4500)
for i in range(len(prices_nis)):
    plt.text(i, prices_nis[i] + 70, str(prices_nis[i]), ha="center")
plt.show()
```

## שלב 4 — מחשבים גיל בשנות לוח
שנת הייחוס היא 2026. השנים הן [2021,2023,2020,2024]. קודם ביד: 5,3,6,2 שנות לוח. זה אינו גיל מדויק בתאריך.

</div>

```python
years = [2021, 2023, 2020, 2024]
reference_year = 2026
for year in years:
    age = reference_year - year
    print('שנת ייצור:', year, 'גיל בשנות לוח:', age)
print('הלולאה הסתיימה')
```

## שלב 6 — הקשר בין תצפיות
חשוב לשלוף באותו אינדקס מכל הרשימות. המחשב B הוא האיבר השני, באינדקס 1. מה יקרה אם נמיין רק את המחירים?

</div>

```python
computer_ids = ['A', 'B', 'C', 'D']
print('המחשב השני:', computer_ids[1])
print('השנה שלו:', years[1])
print('המחיר שלו:', prices_nis[1])
assert (computer_ids[1], years[1], prices_nis[1]) == ('B', 2023, 2700)
```
