# מפגש 2 — פתרון מורה מאומת
**אין להעלות את המחברת הזאת למצגת התלמיד.** אלה אותם נתונים שבמעבדה העצמאית P1–P5; כל המספרים מומצאים.

</div>

<div dir="rtl" lang="he" style="text-align:right">

## חלק א — רשימות
שלושת השדות באותו סדר כדי שכל אינדקס ייצג את אותו מחשב.

</div>

```python
computer_ids = ['P1', 'P2', 'P3', 'P4', 'P5']
memory_gb = [8, 16, 8, 32, 16]
prices_nis = [1800, 2600, 1500, 4200, 2900]
print(computer_ids)
print(memory_gb)
print(prices_nis)
assert len(computer_ids) == len(memory_gb) == len(prices_nis) == 5
```

<div dir="rtl" lang="he" style="text-align:right">

### משימה 2 — מחשב שלישי
הראשון באינדקס 0, השני ב־1 והשלישי ב־2: P3, זיכרון 8GB, מחיר 1500 ₪.

</div>

```python
third_index = 2
print('מחשב:', computer_ids[third_index])
print('זיכרון (GB):', memory_gb[third_index])
print('מחיר (NIS):', prices_nis[third_index])
assert (computer_ids[third_index], memory_gb[third_index], prices_nis[third_index]) == ('P3', 8, 1500)
```

<div dir="rtl" lang="he" style="text-align:right">

### משימה 3 — סכום וממוצע
1800+2600=4400; 4400+1500=5900; 5900+4200=10100; 10100+2900=13000. מספר מחשבים 5; ממוצע 13000/5=2600 ₪.

</div>

```python
number_of_computers = len(prices_nis)
total_price_nis = sum(prices_nis)
average_price_nis = total_price_nis / number_of_computers
print('מספר מחשבים:', number_of_computers)
print('סכום (NIS):', total_price_nis)
print('ממוצע (NIS):', average_price_nis)
assert (number_of_computers, total_price_nis, average_price_nis) == (5, 13000, 2600)
```

<div dir="rtl" lang="he" style="text-align:right">

### משימה 4 — טווח
המחיר המינימלי הוא 1500 והמרבי הוא 4200, ולכן הטווח 2700 ₪.

</div>

```python
lowest_price_nis = min(prices_nis)
highest_price_nis = max(prices_nis)
price_range_nis = highest_price_nis - lowest_price_nis
print('טווח (NIS):', price_range_nis)
assert (lowest_price_nis, highest_price_nis, price_range_nis) == (1500, 4200, 2700)
```

<div dir="rtl" lang="he" style="text-align:right">

### משימה 5 — עותק ותיקון
P4 נמצא באינדקס 3. שינוי 4200 ל־4000 מפחית 200 ₪ מהסכום, ולכן הממוצע יורד 200/5=40 ₪: 2600−40=2560.

</div>

```python
trial_prices_nis = prices_nis.copy()
trial_prices_nis[3] = 4000
trial_average_nis = sum(trial_prices_nis) / len(trial_prices_nis)
print('מקור:', prices_nis)
print('ניסוי:', trial_prices_nis)
print('ממוצע ניסוי:', trial_average_nis)
assert prices_nis[3] == 4200
assert trial_prices_nis[3] == 4000
assert trial_average_nis == 2560
```

<div dir="rtl" lang="he" style="text-align:right">

## חלק ב — לולאה
התלמיד צריך לחזות את סדר הפלט ולא רק את ערכי הגיל. year מכיל ערכי שנים 2022,2024,2021.

</div>

```python
years = [2022, 2024, 2021]
printed_years = []
for year in years:
    print(year)
    printed_years.append(year)
assert printed_years == [2022, 2024, 2021]
```

<div dir="rtl" lang="he" style="text-align:right">

### משימה 7 — מחשבים גיל
שנת ייחוס 2026: 2022→4; 2024→2; 2021→5 שנות לוח.

</div>

```python
reference_year = 2026
ages = []
for year in years:
    age_years = reference_year - year
    ages.append(age_years)
    print('שנת ייצור:', year, 'גיל בשנות לוח:', age_years)
assert ages == [4, 2, 5]
```

<div dir="rtl" lang="he" style="text-align:right">

### משימה 8 — הודעת סיום
print בתוך גוף הלולאה רץ שלוש פעמים. הודעת סיום ללא הזחה רצה לאחר הלולאה פעם אחת.

</div>

```python
for year in years:
    print('בתוך הלולאה:', year)
print('העיבוד הסתיים')
```

<div dir="rtl" lang="he" style="text-align:right">

### משימה 9 — צובר כרשות
מתחילים מ־0. אחרי המעבר הראשון 4, אחרי השני 6, אחרי השלישי 11. הממוצע 11/3 בקירוב 3.67.

</div>

```python
total_age_years = 0
for year in years:
    age_years = reference_year - year
    total_age_years = total_age_years + age_years
average_age_years = total_age_years / len(years)
print('סכום גילים:', total_age_years)
print('ממוצע גילים:', average_age_years)
assert total_age_years == 11
assert abs(average_age_years - 11 / 3) < 1e-12
```
