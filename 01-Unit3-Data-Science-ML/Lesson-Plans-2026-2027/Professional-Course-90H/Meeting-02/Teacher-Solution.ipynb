## פתרון חלק 1
הפתרון הבא תואם לנתוני התרגול העצמאי, ולא לנתוני הדוגמה הפתורה.

<div dir="rtl" lang="he" style="text-align: right;">

### 1. ייצוג
כל אינדקס מייצג מחשב אחד. הזיכרון ב־ GB והמחיר בשקלים. רשימות מקבילות הן גשר לטבלה; יש לשמור על סדר ואורך תואמים.

</div>

```python
computer_ids = ["P1", "P2", "P3", "P4", "P5"]
memory_gb = [8, 16, 8, 32, 16]
prices_nis = [1800, 2600, 1500, 4200, 2900]
print("מזהים:", computer_ids)
print("זיכרון ב-GB:", memory_gb)
print("מחירים בשקלים:", prices_nis)
```

<div dir="rtl" lang="he" style="text-align: right;">

### 2. אינדקס
המחשב השלישי נמצא באינדקס 2, משום שהספירה מתחילה ב־ 0.

</div>

```python
print("מחשב:", computer_ids[2])
print("זיכרון ב-GB:", memory_gb[2])
print("מחיר בשקלים:", prices_nis[2])
```

<div dir="rtl" lang="he" style="text-align: right;">

### 3. ממוצע
1800+2600+1500+4200+2900=13000; 13000/5=2600 ₪. זהו ממוצע חמש הדוגמאות בלבד.

</div>

```python
number_of_computers = len(prices_nis)
total_price_nis = sum(prices_nis)
average_price_nis = total_price_nis / number_of_computers
print("מספר מחשבים:", number_of_computers)
print("סכום בשקלים:", total_price_nis)
print("ממוצע בשקלים:", average_price_nis)
```

<div dir="rtl" lang="he" style="text-align: right;">

### 4. קצוות וטווח
מינימום 1500, מקסימום 4200, טווח 2700 ₪. אין כאן מידע על מצב פיזי, מעבד או אחריות; אין להסיק כדאיות רכישה.

</div>

```python
lowest_price_nis = min(prices_nis)
highest_price_nis = max(prices_nis)
price_range_nis = highest_price_nis - lowest_price_nis
print("מינימום:", lowest_price_nis)
print("מקסימום:", highest_price_nis)
print("טווח בשקלים:", price_range_nis)
```

<div dir="rtl" lang="he" style="text-align: right;">

### 5. ניסוי בעותק
המחשב הרביעי באינדקס 3. תיקון של 200 ₪ כלפי מטה מפחית את הממוצע ב־ 200/5=40 ₪. הממוצע החדש 2560 ₪. copy מתאימה כאן לרשימה של מספרים.

</div>

```python
trial_prices_nis = prices_nis.copy()
trial_prices_nis[3] = 4000
trial_average_nis = sum(trial_prices_nis) / len(trial_prices_nis)
print("מקור:", prices_nis)
print("ניסוי:", trial_prices_nis)
print("ממוצע חדש בשקלים:", trial_average_nis)
```

<div dir="rtl" lang="he" style="text-align: right;">

| memory_gb | נפח זיכרון | GB | מאפיין |

מאפיינים חסרים: מעבד ומצב המחשב. חמש דוגמאות מומצאות אינן מאגר מייצג. אין התאמת מודל או בדיקת הכללה; חישובי סיכום אינם אימון.

### 7. טעויות
האינדקס החמישי הוא 4; אינדקס 5 מחוץ לטווח. מיון מחירים לבדם מנתק את התאמת המחיר למזהה ולזיכרון. בטבלה יש להזיז רשומה שלמה ביחד.

## פתרון חלק 2

<div dir="rtl" lang="he" style="text-align: right;">

### 1–2. מעבר על ערכים
הערכים המודפסים: 2022, 2024, 2021. משתנה הלולאה מקבל שנה ולא אינדקס.

</div>

```python
years = [2022, 2024, 2021]
for year in years:
    print(year)
```

<div dir="rtl" lang="he" style="text-align: right;">

### 3. חישוב גיל
2026−2022=4; 2026−2024=2; 2026−2021=5. כל מעבר מחשב את אותו ביטוי עבור שנה אחרת.

</div>

```python
reference_year = 2026
for year in years:
    computer_age_years = reference_year - year
    print("שנת ייצור:", year, "גיל בשנות לוח:", computer_age_years)
```

<div dir="rtl" lang="he" style="text-align: right;">

| מעבר | שנה | גיל |
|---|---|---|
| 1 | 2022 | 4 |
| 2 | 2024 | 2 |
| 3 | 2021 | 5 |

זהו קירוב לפי שנים; אין תאריך ייצור מלא.

</div>

<div dir="rtl" lang="he" style="text-align: right;">

### 4. גבולות הלולאה
הודעת הסיום אינה מוזחת ולכן תודפס פעם אחת לאחר שלושת המעברים.

</div>

```python
for year in years:
    print("שנה:", year)
print("הסתיים עיבוד השנים")
```

<div dir="rtl" lang="he" style="text-align: right;">

### 5. תיקון תחביר

### 6. פירוש
גיל מחושב משנת ייחוס ושנת ייצור, ולכן הוא משתנה נגזר. לשנת ייחוס קבועה יש קשר חשבוני מלא בין שנת הייצור לגיל. לחיזוי מחיר נחוצים גם נתוני מחירים ידועים ומאפיינים רלוונטיים כגון מעבד, זיכרון ומצב. אין התאמת מודל או הערכה על דוגמאות חדשות בשיעור זה.

תיעוד לדוגמה: חישבנו גיל לכל מחשב כדי לתאר מאפיין אפשרי. הפלט הוא 4,2,5 שנות לוח. המידע מומצא ואינו מאגר לפרויקט גמר.

</div>

<div dir="rtl" lang="he" style="text-align: right;">

### 7. הרחבת צבירה
| מעבר | גיל | סכום לפני | סכום אחרי |
|---|---|---|---|
| 1 | 4 | 0 | 4 |
| 2 | 2 | 4 | 6 |
| 3 | 5 | 6 | 11 |

סכום 11; ממוצע 11/3 ≈ 3.67 שנים. האתחול קודם ללולאה כדי שהסכום לא יימחק בכל מעבר.

</div>

```python
total_age_years = 0
for year in years:
    computer_age_years = reference_year - year
    total_age_years = total_age_years + computer_age_years
average_age_years = total_age_years / len(years)
print("סכום הגילים:", total_age_years)
print("ממוצע הגילים:", average_age_years)
```
