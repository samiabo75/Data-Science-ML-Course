## פתיחה — מה מודל לחיזוי מחיר צריך לראות?
דודו רוצה לחזות מחיר רכב לפי שנת ייצור וקילומטראז׳. ברשימות מקבילות צריך לזכור איזה מחיר שייך לאיזה רכב. בטבלה כל שורה מרכזת נתוני רכב אחד וכל עמודה מתארת שדה.

נלמד היום לארגן ולקרוא נתונים. אין אימון מודל. שנת ייצור וקילומטראז׳ הם מאפיינים אפשריים, מחיר הוא יעד אפשרי; מזהה הרכב מיועד לזיהוי בלבד.

## 1. קוראים טבלה לפני קוד
| מזהה | שנת ייצור | קילומטראז׳ | מחיר בשקלים |
|---|---|---|---|
| A | 2021 | 60000 | 90000 |
| B | 2023 | 30000 | 110000 |
| C | 2020 | 90000 | 80000 |
| D | 2024 | 20000 | 120000 |

שאלו: כמה רכבים? כמה שדות? איזה מידע מתארת השורה של C? **תצפית/רשומה**: רכב אחד. **שדה/עמודה**: תכונה או נתון על הרכב. לא כל עמודה בהכרח תהיה מאפיין במודל.

## 2. ספרייה היא אוסף כלים מוכנים
`pandas` היא ספרייה לעבודה עם נתונים טבלאיים. `import` טוענת אותה לסביבת העבודה. `as pd` נותנת לה קיצור שנשתמש בו. מריצים את התא לפני יצירת הטבלה.

```python
import pandas as pd
```

## 3. גשר קצר — מילון שמקשר שם עמודה לרשימה
כאן מספיק להכיר מבנה אחד: `{שם: רשימת ערכים}`. הסוגריים המסולסלים עוטפים מילון; המפתח הוא שם העמודה במירכאות; אחרי נקודתיים מופיעה רשימת הערכים. פסיק מפריד בין זוגות. זהו שימוש ממוקד במילון, לא שיעור מלא במילונים.

כל הרשימות באותו אורך ובאותו סדר של רכבים.

```python
data = {
    "car_id": ["A", "B", "C", "D"],
    "year": [2021, 2023, 2020, 2024],
    "mileage_km": [60000, 30000, 90000, 20000],
    "price_nis": [90000, 110000, 80000, 120000]
}
```

הצביעו על שם עמודה ועל הרשימה המתאימה לו. מדוע חייבים ארבעה מחירים לארבעה רכבים? אורך זהה לבדו אינו מבטיח סדר נכון; זו אחריותנו בהקלדת הדוגמאות.

## 4. יצירת DataFrame
`pd.DataFrame(data)` יוצרת אובייקט טבלאי מן המילון. `df` הוא שם המשתנה שבחרנו לשמירת הטבלה. שימו לב לאותיות הגדולות D ו־F. `display` מציגה טבלה במחברת.

```python
df = pd.DataFrame(data)
display(df)
```

כעת הטבלה כוללת אינדקס ברירת מחדל 0,1,2,3. האינדקס שמוצג משמאל אינו עמודת `car_id`, ואינו שדה נתונים חמישי. לדוגמה אינדקס 1 הוא שורת הרכב B.

DataFrame שומר ערכי שדות של רשומות ביחד. בשיעור הבא נלמד פעולות שבוחרות שורות מתוך טבלה.

## 5. כמה שורות וכמה עמודות?
`df.shape` מציגה זוג מספרים: מספר שורות ואחריו מספר עמודות. אין סוגריים אחרי shape. נשתמש ב־`df.columns` לקריאת שמות העמודות.

```python
print("ממדי הטבלה — שורות, עמודות:", df.shape)
print("שמות העמודות:", df.columns)
```

התוצאה `(4, 4)` פירושה ארבע תצפיות וארבע עמודות, לא ארבעה מאפייני חיזוי: יש גם מזהה וגם יעד. סוג המבנה של הזוג נקרא tuple; אין צורך ללמוד פעולות עליו היום.

## 6. הצצה לשתי רשומות
`head(2)` מחזירה את שתי השורות הראשונות. זוהי הצצה, לא מחיקה של שאר השורות.

```python
display(df.head(2))
print("הממדים אחרי ההצצה:", df.shape)
```

בשתי השורות המוצגות נראים הרכבים A ו־B. הטבלה המקורית נשארת בגודל 4×4. חזו: מה יוצג ב־head(3)? ואם נבקש יותר שורות ממספרן בטבלה, יוחזרו כל השורות הקיימות.
