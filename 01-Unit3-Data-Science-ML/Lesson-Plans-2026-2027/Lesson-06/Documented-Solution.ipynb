### 1–2. רשומות ועמודות
כל שורה מייצגת מחשב אחד; יש ארבע שורות וארבע עמודות. שמות העמודות הם מפתחות במילון וכל רשימה היא ערכי עמודה. רשימות שוות אורך נדרשות ליצירת הטבלה ממילון רשימות.

```python
import pandas as pd
computer_data = {
    "computer_id": ["P1", "P2", "P3", "P4"],
    "year": [2022, 2024, 2021, 2023],
    "memory_gb": [8, 16, 8, 32],
    "price_nis": [1800, 3200, 1500, 4200]
}
```

```python
computers_df = pd.DataFrame(computer_data)
display(computers_df)
```

### 3. ממדים ושמות
(4,4) פירושו ארבע תצפיות וארבע עמודות. אינדקס ברירת המחדל אינו עמודת נתונים נוספת.

```python
print("ממדים:", computers_df.shape)
print("עמודות:", computers_df.columns)
```

### 4. הצצה אינה מחיקה
head(3) מחזירה את שורות P1,P2,P3; הטבלה המקורית נשארת בת ארבע שורות.

```python
display(computers_df.head(3))
print("ממדי המקור:", computers_df.shape)
```

### 5. כרטיס לדוגמה
מקור: ארבע דוגמאות מומצאות להוראה. תצפית: מחשב אחד. ממדים: 4×4.

| שדה | משמעות | יחידה/תפקיד |
|---|---|---|
| computer_id | מזהה מחשב | מזהה בלבד |
| year | שנת ייצור | שנה, מאפיין אפשרי |
| memory_gb | נפח זיכרון | GB, מאפיין אפשרי |
| price_nis | מחיר | ₪, יעד אפשרי |

שאלה: כיצד המחיר משתנה לצד גיל וזיכרון? מגבלה: אין נתוני מעבד, מצב או מקור אמיתי; אי אפשר להסיק מכלל השוק או סיבתיות. לא הותאם מודל.

### 6. שגיאות
