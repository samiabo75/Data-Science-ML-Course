## 1. סוג המשתנה נקבע לפי המשמעות
| סוג | השאלה | דוגמה |
|---|---|---|
| קטגוריאלי שמי | לאיזו קבוצה שייך? ללא סדר טבעי | species, island |
| קטגוריאלי סדרי | איזו רמה? יש סדר, לא בהכרח מרווחים שווים | דירוג נמוך/בינוני/גבוה; זו דוגמה כללית שאינה עמודה בקובץ |
| כמותי בדיד | כמה? ספירה | מספר תצפיות בכל מין, נתון מסוכם שניצור בגרף |
| כמותי רציף | כמה נמדד? | אורך מקור, אורך סנפיר ומסה |

מסה שנרשמה במספר שלם נשארת מדידה רציפה מבחינת המשמעות. מספר מזהה הוא תווית ולא גודל כמותי בגלל הספרות. קידוד מספרי של קטגוריות אינו הופך אותן למדידות.

סוג אחסון בפייתון אינו סוג המשתנה התוכני. species הוא מין טקסונומי; sex הוא זוויג — לא אותה עמודה.

### 2–3. קבוצה
Adelie כוללת 152 רשומות בגרסה שנבדקה. island הוא משתנה קטגוריאלי שמי; בודקים ספירות בתוך תת־הקבוצה.

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
url = "https://raw.githubusercontent.com/samiabo75/Data-Science-ML-Course/main/Datasets/penguins.csv"
penguins_df = pd.read_csv(url)
adelie_mask = penguins_df["species"] == "Adelie"
adelie_df = penguins_df[adelie_mask]
print("מקור:", penguins_df.shape)
print("Adelie:", adelie_df.shape)
```

```python
plt.figure(figsize=(7, 4))
sns.countplot(data=adelie_df, x="island", color="steelblue")
plt.title("Adelie observations by island")
plt.xlabel("Island")
plt.ylabel("Number of observations")
plt.tight_layout()
plt.show()
```

### 4. פירוש
בקבוצת Adelie: Dream=56, Torgersen=52, Biscoe=44. אלו מספרי רשומות; Dream הגדולה בקבוצה זו. אין להסיק מספרי אוכלוסייה בטבע, ואין להסיק שכך מתחלקים כל המינים.

```python
plt.figure(figsize=(7, 4))
sns.countplot(data=penguins_df, x="island", color="teal")
plt.title("All penguin observations by island")
plt.xlabel("Island")
plt.ylabel("Number of observations")
plt.tight_layout()
plt.show()
```

### 5. מקור לעומת קבוצה
בכל המאגר Biscoe=168, Dream=124, Torgersen=52. Biscoe הגדולה במקור, אך לא בקרב Adelie. שינוי אוכלוסיית הבחירה משנה את השאלה. ההבדל אינו מוכיח קשר סיבתי בין אי למין.

### 6. בחירת גרפים
מסה היא מדידה רציפה: Histogram. מסה מול סנפיר הן שתי מדידות: Scatterplot. ספירת אי כקטגוריה: Countplot.

### 7. כרטיס גרף לדוגמה
שאלה: כמה תצפיות Adelie מכל אי? קבוצה: Adelie בלבד, 152 רשומות. X אי, משתנה שמי; Y מספר תצפיות, ספירה בדידה. Dream מונה 56 רשומות והיא הגדולה בקבוצה. מגבלה: דגימה מוגדרת, לא מפקד אוכלוסייה או הוכחת סיבתיות. מקור: Palmer Penguins בגרסת הקורס 344×7.
