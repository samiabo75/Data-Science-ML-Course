## פתיחה — האם כל מחלקות היעד מיוצגות באותה מידה?
אם בעתיד נסווג species לפי מדידות, נרצה לדעת כמה דוגמאות יש מכל מין. מחלקה עם מעט דוגמאות עלולה לקבל פחות הזדמנויות ללמידה ולהערכה. ספירת הדוגמאות היא תיאור מדגם, לא ציון איכות למודל ולא מפקד הפינגווינים בטבע.

## 1. סוג המשתנה נקבע לפי המשמעות
| סוג | השאלה | דוגמה |
|---|---|---|
| קטגוריאלי שמי | לאיזו קבוצה שייך? ללא סדר טבעי | species, island |
| קטגוריאלי סדרי | איזו רמה? יש סדר, לא בהכרח מרווחים שווים | דירוג נמוך/בינוני/גבוה; זו דוגמה כללית שאינה עמודה בקובץ |
| כמותי בדיד | כמה? ספירה | מספר תצפיות בכל מין, נתון מסוכם שניצור בגרף |
| כמותי רציף | כמה נמדד? | אורך מקור, אורך סנפיר ומסה |

מסה שנרשמה במספר שלם נשארת מדידה רציפה מבחינת המשמעות. מספר מזהה הוא תווית ולא גודל כמותי בגלל הספרות. קידוד מספרי של קטגוריות אינו הופך אותן למדידות.

סוג אחסון בפייתון אינו סוג המשתנה התוכני. species הוא מין טקסונומי; sex הוא זוויג — לא אותה עמודה.

### עצירה לפני קוד
סווגו את species, sex, body_mass_g ו־bill_length_mm ונמקו. האם צריך לחשב ממוצע של שמות המינים? איזה גרף יענה על 'כמה מכל מין'?

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
url = "https://raw.githubusercontent.com/samiabo75/Data-Science-ML-Course/main/Datasets/penguins.csv"
df = pd.read_csv(url)
print("ממדים:", df.shape)
display(df.head(3))
```

## 2. Countplot — ספירת תצפיות בכל קטגוריה
Seaborn מציירת גרפים מתוך טבלה; sns הוא הקיצור שלה. Matplotlib משמשת כאן לכותרות ולצירים, בקיצור plt. אין צורך ללמוד את כל הספריות.

נציג קוד ציור כתבנית: table ועמודת הקטגוריה הן ההחלטות המרכזיות. הטקסטים בתוך הגרף באנגלית לתצוגה יציבה, והתיעוד והפרשנות בעברית.

```python
plt.figure(figsize=(7, 4))
sns.countplot(data=df, x="species", color="steelblue")
plt.title("Penguin observations by species")
plt.xlabel("Species")
plt.ylabel("Number of observations")
plt.tight_layout()
plt.show()
```

### קוראים את הגרף
ציר X: קטגוריות המין. ציר Y: מספר רשומות, לא ממוצע מסה. כל תצפית מוסיפה 1 לקטגוריה שלה. ציר הספירה מתחיל ב־0.
בגרסת הקורס: Adelie=152, Gentoo=124, Chinstrap=68; הסכום 344. אלה ספירות מאומתות לצורך ההסבר, לא פקודה חדשה לשינון.

Adelie היא הקבוצה הגדולה בקובץ ו־Chinstrap הקטנה. אין להסיק ש־Adelie בהכרח הנפוצה בעולם. עצם הבדל הגדלים אינו מחייב איזון אוטומטי או מוכיח שמודל ייכשל.

## 3. שאלה אחרת — קטגוריה אחרת
כמה רשומות מכל אי? לומדים לשנות את השאלה, העמודה וכותרות הצירים; לא את תבנית הציור כולה.

```python
plt.figure(figsize=(7, 4))
sns.countplot(data=df, x="island", color="teal")
plt.title("Penguin observations by island")
plt.xlabel("Island")
plt.ylabel("Number of observations")
plt.tight_layout()
plt.show()
```

הספירות בגרסת הקורס: Biscoe=168, Dream=124, Torgersen=52. זהו פיזור הדגימה באיים המיוצגים, לא גודל אוכלוסיית האיים. אין להסיק מכאן מהו המין הנפוץ בכל אי; לכך דרושה שאלה נוספת.

## 4. בוחרים גרף לפי השאלה
| שאלה | סוג מידע | גרף מתאים להמשך |
|---|---|---|
| כמה תצפיות בכל מין? | קטגוריה וספירה | Countplot — היום |
| כיצד מסה מתפלגת? | מדידה כמותית אחת | Histogram — שיעור הבא |
| האם מסה קשורה לאורך סנפיר? | שתי מדידות כמותיות | Scatterplot — בהמשך |
| מה המסה הממוצעת בכל מין? | קטגוריה וסיכום מספרי | Barplot — אחרי לימוד מדדים |

לא מציירים Countplot על כל ערך מדידה רציפה במאגר; נרצה תחומי ערכים בהיסטוגרמה. Barplot אינו תחליף אוטומטי לספירה: כברירת מחדל רגילה של השימוש בו הוא מסכם מדידה, לרוב ממוצע.
