### 1–2. הגדרה
island שמית, ללא סדר טבעי. נמוך/בינוני/גבוה סדרית, אך אין הבטחה למרווחים שווים. המשימה מדגימה מאפיין קטגוריאלי בלבד, לא בחירת מאפיין מיטבית.

```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import OneHotEncoder
url = "https://raw.githubusercontent.com/samiabo75/Data-Science-ML-Course/main/Datasets/penguins.csv"
penguins_df = pd.read_csv(url)
model_df = penguins_df.dropna(subset=["species"]).copy()
X = model_df[["island"]]
y = model_df["species"]
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.25, random_state=17, stratify=y
)
print("לפני:", X_train.shape, X_test.shape)
print("חסרי אימון:")
print(X_train.isna().sum())
```

### 3. אותו מבנה לשתי הקבוצות
encoder לומד קטגוריות באימון. אין fit ב־Test. בקובץ ובחלוקה האלה נלמדות שלוש קטגוריות ומתקבלות שלוש עמודות.

```python
encoder = OneHotEncoder(handle_unknown="ignore", sparse_output=False)
encoder.set_output(transform="pandas")
encoder.fit(X_train)
X_train_encoded = encoder.transform(X_train)
X_test_encoded = encoder.transform(X_test)
print("קטגוריות:", encoder.categories_)
print("עמודות:", X_train_encoded.columns)
print("אחרי:", X_train_encoded.shape, X_test_encoded.shape)
display(X_train_encoded.head(3))
display(X_test_encoded.head(3))
```

### 4. קריאה
לשורת Dream, למשל, island_Dream=1 ושתי האחרות 0. לא נוספו תצפיות; מספר שורות נשאר 258 ו־86. הייצוג אינו מרחק גאוגרפי ואינו מבטיח הכללה לאיים שלא נראו.

```python
toy_train = pd.DataFrame({"shape": ["circle", "square", "circle"]})
toy_test = pd.DataFrame({"shape": ["triangle", "square"]})
toy_encoder = OneHotEncoder(handle_unknown="ignore", sparse_output=False)
toy_encoder.set_output(transform="pandas")
toy_encoder.fit(toy_train)
toy_test_encoded = toy_encoder.transform(toy_test)
display(toy_test_encoded)
```

### 5–6. מדיניות ומגבלה
triangle לא נראתה באימון ולכן שתי עמודות הצורות הידועות הן 0. square מקבלת 1 בעמודת square. זהו טיפול מוגדר בקלט חדש, לא לימוד משמעותו. לא מוסיפים עמודה מהבדיקה ולא מריצים fit מחדש.

יומן: island שמית; One-Hot עם כל הקטגוריות שנלמדו באימון וללא drop; חדשים מקבלים אפסים. אין חסרי island בקובץ, אך יש צורך במדיניות נפרדת אם יגיעו חסרים. המקור נשמר. קשר בין אי למין עשוי לשקף דגימה, ולכן בחירת island דורשת הנמקה לפי שימוש עתידי. y נשארת טקסט נפרד; אין צורך לקודד אותה כך ל־KNN כאן.
