## למה לומדים את זה עכשיו?
רשת לומדת ייצוג מורכב יותר; בכיתה י׳ בונים את שפת הנתונים, השגיאה והאימות.
**בתהליך הפרויקט:** העשרה להמשך.
**מסגרת פעילה:** העשרה בלבד — 0 שעות במסלול החובה. המורה בוחר את משימות הליבה שבמדריך; שאר הפעילויות לחזרה או לעיון, ללא תוספת שעות.

## מושגים וידע קודם
רשת עצבית אינה רק פרספטרון יחיד. פונקציית הפעלה לא ליניארית מאפשרת לשילוב שכבות לייצג מעבר לפונקציה ליניארית אחת. לא כל רשת עמוקה ואין ללמד CNN/Backprop מלא ביחידה זו בלי דרישה מאומתת. הטבלה בעלת ארבע שורות משמשת להבנת ייצוג ואימון בלבד; הצלחה עליה אינה ראיה להכללה.

## הסבר מתמטי
שכבה מחשבת z=Wx+b ואז פונקציית הפעלה. שתי יחידות סף יכולות לחשב OR ו־AND; שילוב OR−2·AND−.5 עם סף >0 נותן XOR. הספים בדוגמה הם ידניים ואינם נגזרים בתהליך אימון. ברשת מאומנת מגדירים Loss; backpropagation מחשב גרדיאנטים בעזרת כלל השרשרת,ואופטימיזציה מעדכנת משקולות. בדוגמת gradient descent: w_new=w−η·gradient; זו נוסחת העדכון של האופטימייזר,לא הגדרת backprop עצמה.

## רשת ידנית
משקולות נתונות להמחשת יכולת ייצוג; השורות הן ארבע כניסות,העמודות שתי יחידות ביניים.

```python
import numpy as np
import pandas as pd
from sklearn.neural_network import MLPClassifier
# טבלת אמת מומצאת,לא נתוני עולם אמיתי
X=np.array([[0.,0.],[0.,1.],[1.,0.],[1.,1.]])
y=np.array([0,1,1,0])

hidden_or=(X.sum(axis=1)-0.5>0).astype(int)
hidden_and=(X.sum(axis=1)-1.5>0).astype(int)
manual_predictions=(hidden_or-2*hidden_and-0.5>0).astype(int)
display(pd.DataFrame({"x1":X[:,0],"x2":X[:,1],"OR":hidden_or,"AND":hidden_and,"XOR":manual_predictions}))
```

## רשת שלומדת
MLP עם שכבה חבויה קטנה ולא ליניארית,solver lbfgs קבוע. זו הדגמת אימון; לא מחפשים ארכיטקטורה לפי ציון Test,ואין Test בדוגמת ארבעת הזוגות.

```python
model=MLPClassifier(hidden_layer_sizes=(4,),activation="tanh",
 solver="lbfgs",alpha=0.0,max_iter=5000,random_state=42)
model.fit(X,y)
predictions=model.predict(X)
print("Training predictions only:",predictions)
print("Training loss:",model.loss_,"Optimizer iterations:",model.n_iter_)
```

## מושגים ותיעוד
הסקלרים מומצאים: מראים כיוון עדכון בלבד. אין זה חישוב backprop של הרשת שהורצה.

```python
old_weight=0.5
learning_rate=0.1
gradient=0.2
new_weight=old_weight-learning_rate*gradient
print("Toy gradient descent weight:",new_weight)
```

תעדו מה נכנס,מה חושב,מה התקבל ומה אי אפשר להסיק. כתבו שני משפטים על הקשר לפרויקט.
לא מכנים משקולות ידניות 'אימון'; לא מכנים loss אפס כמעט 'הכללה מושלמת'; לא מציגים פונקציית סף כאילו היא ההפעלה של MLP tanh; לא מזהים backprop עם צעד האופטימייזר. למתקשים טבלת חישוב חבויה,למצטיינים הסבר מדוע כמה שכבות ליניאריות ללא הפעלה אינן מוסיפות ייצוג לא ליניארי.
