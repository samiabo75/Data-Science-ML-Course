## למה לומדים את זה עכשיו?
רשת לומדת ייצוג מורכב יותר; בכיתה י׳ בונים את שפת הנתונים, השגיאה והאימות.
**בתהליך הפרויקט:** העשרה להמשך.
**מסגרת פעילה:** העשרה בלבד — 0 שעות במסלול החובה. המורה בוחר את משימות הליבה שבמדריך; שאר הפעילויות לחזרה או לעיון, ללא תוספת שעות.

```python
import numpy as np
import pandas as pd
from sklearn.neural_network import MLPClassifier
# טבלת אמת מומצאת,לא נתוני עולם אמיתי
X=np.array([[0.,0.],[0.,1.],[1.,0.],[1.,1.]])
y=np.array([1,0,0,1])
```

## ייצוג ידני
שינוי הפלט אינו אימון; כל המשקולות נקבעו ביד. שתי יחידות חבויות יכולות להפריד אזורים שאינם קו יחיד.

```python
hidden_or=(X.sum(axis=1)-.5>0).astype(int)
hidden_and=(X.sum(axis=1)-1.5>0).astype(int)
manual_predictions=1-(hidden_or-2*hidden_and-.5>0).astype(int)
print("Manual XNOR:",manual_predictions)
```

## אימון
אין טענה להכללה מארבע דוגמאות אימון; מאפייני האופטימייזר והפלט עשויים להשתנות בין גרסאות. אין רשת למידה עמוקה מאומתת לשימוש כאן.

```python
model=MLPClassifier(hidden_layer_sizes=(4,),activation="tanh",
 solver="lbfgs",alpha=0.0,max_iter=5000,random_state=42)
model.fit(X,y)
predictions=model.predict(X)
print("Training only:",predictions,"Loss:",model.loss_)
```

## עדכון והסבר
משקל חדש .83. forward מחשב תחזית,Loss מודד אי־התאמה,backprop מחשב נגזרות ואופטימייזר מעדכן. הדוגמה משתמשת ב־gradient descent; lbfgs שבספרייה הוא אופטימייזר אחר,ולכן אין לטעון שזהו הצעד המדויק שביצע.

```python
old_weight=.8;learning_rate=.1;gradient=-.3
new_weight=old_weight-learning_rate*gradient
print(new_weight)
```
