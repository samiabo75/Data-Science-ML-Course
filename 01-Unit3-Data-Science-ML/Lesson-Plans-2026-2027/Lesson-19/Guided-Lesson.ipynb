## 1. דוגמה קטנה — רבעונים
חמישה ערכים מומצאים: 2,4,6,8,20. נשתמש בשיטת הרבעונים הליניארית של Pandas, התואמת כאן לחישוב שעליו מבוסס הגרף.

בדוגמה זו Q1=4, החציון Q2=6, ו־Q3=8. Q1 ו־Q3 מתארים את גבולות החלק האמצעי של ההתפלגות; ברעיון הקופסה מייצגת את כ־50% האמצעיים. במדגם קטן ועם ערכים חוזרים אין לפרש זאת כהבטחה לספירת בדיוק חצי מהנקודות.

טווח בין־רבעוני: IQR=Q3−Q1=8−4=4. יחידתו זהה לנתונים. יש שיטות אחרות לחישוב רבעונים; הן עשויות להחזיר ערכים אחרים, ולכן מתעדים שיטה ולא מתקנים מקור בשקט.

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
demo_df = pd.DataFrame({"value": [2, 4, 6, 8, 20]})
q1 = demo_df["value"].quantile(0.25, interpolation="linear")
q2 = demo_df["value"].median()
q3 = demo_df["value"].quantile(0.75, interpolation="linear")
iqr = q3 - q1
print("Q1:", q1, "חציון:", q2, "Q3:", q3, "IQR:", iqr)
```

## 2. גבולות ושפמים — לא אותו דבר
בכלל 1.5×IQR, הגבול התחתון Q1−1.5×IQR והעליון Q3+1.5×IQR. בדוגמה: −2 ו־14.

השפמים מגיעים לערכי הנתונים הקיצוניים שנמצאים בתוך הגבולות — כאן 2 ו־8. הם אינם חייבים להגיע לגבולות −2 ו־14. 20 מעבר לגבול ויסומן כנקודה נפרדת.

הסימון אומר 'מעבר לכלל הגרפי הזה', לא 'שגיאה מוכחת'. ערך בדיוק בגבול נכלל בשפם אם הוא קיים בנתונים.

```python
lower_fence = q1 - 1.5 * iqr
upper_fence = q3 + 1.5 * iqr
print("גבול תחתון:", lower_fence, "גבול עליון:", upper_fence)
plt.figure(figsize=(7, 4))
sns.boxplot(data=demo_df, y="value", whis=1.5, color="lightblue")
plt.title("Teaching example: quartiles and a flagged value")
plt.xlabel("All teaching observations")
plt.ylabel("Value")
plt.tight_layout()
plt.show()
```

### קוראים את התמונה
הקו בקופסה הוא חציון 6, לא ממוצע. תחתית הקופסה 4, ראש הקופסה 8. השפם העליון מסתיים ב־8 כי זה הערך הגבוה ביותר בתוך הגבולות; 20 מוצג בנפרד. נתון גדול אינו נמחק מהמקור.

## 3. מסה לפי מין — מרכז ופיזור יחד
ה־Boxplot מחשב רבעונים וגבולות בנפרד לכל קבוצה. איננו מחילים את רבעוני כל המאגר על כל מין.

```python
url = "https://raw.githubusercontent.com/samiabo75/Data-Science-ML-Course/main/Datasets/penguins.csv"
df = pd.read_csv(url)
species_order = ["Adelie", "Chinstrap", "Gentoo"]
plt.figure(figsize=(7, 4))
sns.boxplot(data=df, x="species", y="body_mass_g", order=species_order, whis=1.5, color="lightblue")
plt.title("Body mass distributions by species")
plt.xlabel("Species")
plt.ylabel("Body mass (g)")
plt.tight_layout()
plt.show()
```

קראו: באיזו קבוצה החציון גבוה יותר? האם הקופסאות והשפמים חופפים? האם קבוצה עם ממוצע גבוה חייבת להיות ללא חפיפה עם אחרות? לא.

גובה הקופסה כאן מייצג IQR של מסה בגרמים, לא מספר רשומות. רוחב הקופסה בהגדרה זו אינו גודל המדגם. כדי לדעת גודל מדגם צריך ספירה נפרדת.

חסרי מסה לא משתתפים בחישוב הגרף; הם נשארים ב־df. אין להסיק נורמליות מצורת הקופסה או מובהקות מעצם חפיפה/אי־חפיפה.
