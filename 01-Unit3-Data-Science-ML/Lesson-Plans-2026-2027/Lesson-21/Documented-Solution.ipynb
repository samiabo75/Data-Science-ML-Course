## פתרון מתועד

### 1. ידני
(60−50)/5=2, (45−50)/5=−1. אלו מרחקים מממוצע האימון ביחידות סטיית תקן, לא הסתברויות.

```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
url = "https://raw.githubusercontent.com/samiabo75/Data-Science-ML-Course/main/Datasets/penguins.csv"
penguins_df = pd.read_csv(url)
model_df = penguins_df.dropna(subset=["species"]).copy()
X = model_df[["bill_length_mm", "body_mass_g"]]
y = model_df["species"]
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.25, random_state=17, stratify=y
)
training_means = X_train.mean()
X_train_ready = X_train.fillna(training_means)
X_test_ready = X_test.fillna(training_means)
```

### 3. fit באימון
הסרגל נלמד מן האימון לאחר השלמה. StandardScaler משתמשת בסטיית תקן ddof=0. בקשת pandas היא עזרת תצוגה.

```python
scaler = StandardScaler()
scaler.set_output(transform="pandas")
scaler.fit(X_train_ready)
print("ממוצעים לפי סדר העמודות:", scaler.mean_)
print("קני מידה לפי סדר העמודות:", scaler.scale_)
```

### 4. אותו transform
אין fit חדש בבדיקה. האינדקסים והעמודות נשמרים כי הוגדר פלט pandas.

```python
X_train_scaled = scaler.transform(X_train_ready)
X_test_scaled = scaler.transform(X_test_ready)
display(X_train_scaled.head(3))
display(X_test_scaled.head(3))
print("אימון:", X_train_scaled.shape, "בדיקה:", X_test_scaled.shape)
```

### 5. בקרת סרגל
ממוצעי האימון בקירוב 0 וסטיות התקן בקירוב 1 לשני המאפיינים הלא־קבועים. בבדיקה אין דרישה כזו. מעבר מגרם לק״ג מכפיל גם את הערכים, הממוצע וסטיית התקן באותו גורם חיובי, ולכן בתקנון מחדש מאותו אימון התוצאות שקולות עד דיוק מספרי. זה אינו אומר שכל שינוי מאפיינים לא משפיע על מודל.

```python
print("ממוצעים אחרי:")
print(X_train_scaled.mean())
print("סטיות תקן אחרי, ddof=0:")
print(X_train_scaled.std(ddof=0))
```

### 6. תיעוד ומגבלה
הוגדר סיווג species, נבחרו שתי מדידות, בוצעה חלוקה והשלמה לפי אימון. לאחר מכן נלמד scaler מאימון בלבד והוחל על שתי הקבוצות. היחידות המקוריות מ״מ וגרם הומרו לערכים ללא יחידת מקור. מספר התצפיות והתאמת התוויות נשמרו.

fit על Test היה משנה את פרמטרי הכלי לפי הבדיקה, ופוגע בסרגל המשותף ובהפרדה. תקנון אינו ניבוי, אינו מנרמל צורה להתפלגות נורמלית ואינו מבטיח שיפור ביצועים. באימות צולב יש ללמוד הכנה בתוך הקפל.

```python
print("מקור:", penguins_df.shape)
print("עמודות X:", X.columns)
print("ממדי y:", y.shape)
```
