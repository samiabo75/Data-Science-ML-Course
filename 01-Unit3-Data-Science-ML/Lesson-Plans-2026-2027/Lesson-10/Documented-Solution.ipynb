## פתרון מתועד

```python
import pandas as pd
url = "https://raw.githubusercontent.com/samiabo75/Data-Science-ML-Course/main/Datasets/penguins.csv"
penguins_df = pd.read_csv(url)
print("מקור:", penguins_df.shape)
```

### 2. סדר עולה
שדה המיון הוא אורך סנפיר במ״מ. חמש הראשונות הן בעלות הערכים הידועים הנמוכים בראש המיון, לא בהכרח דגימה מייצגת.

```python
flippers_ascending = penguins_df.sort_values("flipper_length_mm")
display(flippers_ascending[["species", "flipper_length_mm"]].head(5))
```

### 3. סדר יורד
ascending=False משנה את כיוון הסדר. מספר השדות והרשומות בטבלה הממוינת נשאר 344×7.

```python
flippers_descending = penguins_df.sort_values("flipper_length_mm", ascending=False)
display(flippers_descending[["species", "flipper_length_mm"]].head(5))
print("ממדים:", flippers_descending.shape)
```

### 4. חסרים ואינדקס
חסרי אורך סנפיר מגיעים לסוף כברירת מחדל. הם אינם אורך 0 ואינם בהכרח קצרים. האינדקס נשמר מן המקור ולכן אינו דירוג.

```python
display(flippers_descending[["species", "flipper_length_mm"]].tail(3))
```

### 5. קבוצה, סדר, הצצה
Adelie כוללת 152 רשומות בגרסה שנבדקה. המיון משאיר 152; head(3) בוחרת שלוש. כל שדות הרשומה נשמרים יחד.

```python
adelie_mask = penguins_df["species"] == "Adelie"
adelie_df = penguins_df[adelie_mask]
adelie_mass_sorted = adelie_df.sort_values("body_mass_g", ascending=False)
adelie_top_three = adelie_mass_sorted.head(3)
display(adelie_top_three[["species", "island", "body_mass_g"]])
print("קבוצה:", adelie_df.shape)
print("ממוינת:", adelie_mass_sorted.shape)
print("שלוש:", adelie_top_three.shape)
```

### 6. מקור ותיעוד
שאלה: מהן שלוש רשומות Adelie בעלות המסה הידועה הגבוהה ביותר במאגר? מיון יורד לפי body_mass_g, בגרמים. הרשומות הן קצוות בקבוצה מוגדרת, לא מדגם אקראי או מייצג. אין כאן הוכחה לשגיאה ואין בסיס מספיק לאימון כללי. המקור נשאר ללא שינוי.

```python
print("מקור:", penguins_df.shape)
display(penguins_df.head(3))
```

### 7. התאמה בין שדות
מיון עמודה לבד והצמדתה לפי מיקום למזהים המקוריים עלול לשייך מדידה לישות שגויה. sort_values על DataFrame מזיזה כל רשומה בשלמותה.

### רשות

```python
adelie_mass_ascending = adelie_df.sort_values("body_mass_g")
display(adelie_mass_ascending[["species", "body_mass_g"]].head(3))
display(adelie_top_three[["species", "body_mass_g"]])
```

רואים את קצוות המסה הידועים בקבוצת Adelie. זהו תיאור, לא הוכחה לסיבתיות, לא קביעה שערך חריג סטטיסטית ולא אישור ששגיאה קיימת.
