## פתיחה — אילו תצפיות כדאי לבדוק לפני מודל?
לפני שננסה לחזות מסת גוף, נרצה להכיר את ערכי הקצה. מיון יאפשר לראות תצפיות בעלות מסה גבוהה ונמוכה. ערך קיצוני ביחס לאחרים אינו בהכרח שגיאה: צריך לבדוק יחידות, מקור והקשר. מיון אינו מודל, ואינו מודד לבדו קשר בין מאפיינים.

## 1. מתחילים בשלוש רשומות
דוגמת מחירים מומצאת בלבד. חזו את סדר המזהים ממחיר נמוך לגבוה. חשוב להזיז כל שורה בשלמותה כדי לשמור התאמת מזהה ומחיר.

```python
import pandas as pd
tiny_df = pd.DataFrame({
    "item_id": ["A", "B", "C"],
    "price_nis": [90, 120, 100]
})
ascending_tiny = tiny_df.sort_values("price_nis")
display(ascending_tiny)
```

סדר המזהים: A,C,B; מחירים: 90,100,120. `sort_values` מחזירה טבלה ממוינת. ברירת המחדל היא סדר עולה. האינדקסים המקוריים 0,2,1 נשמרו; הם אינם דירוג חדש.

```python
descending_tiny = tiny_df.sort_values("price_nis", ascending=False)
display(descending_tiny)
display(tiny_df)
```

`ascending=False` מבקשת סדר יורד. כאן `=` הוא מתן ערך לפרמטר בשם ascending, לא בדיקת שוויון. המקור לא השתנה כי שמרנו את התוצאה בשם אחר ולא השתמשנו ב־inplace. אין צורך ללמד inplace בשיעור זה.

## 2. נתונים אמיתיים — מסת גוף

```python
url = "https://raw.githubusercontent.com/samiabo75/Data-Science-ML-Course/main/Datasets/penguins.csv"
df = pd.read_csv(url)
print("ממדי המקור:", df.shape)
```

```python
mass_ascending = df.sort_values("body_mass_g")
display(mass_ascending[["species", "body_mass_g"]].head(5))
print("ממדי הטבלה הממוינת:", mass_ascending.shape)
```

אלה חמש הרשומות הראשונות לפי מסה ידועה בסדר עולה. המיון לא הפחית את מספר השורות או העמודות. הרשומה הקלה ביותר בגרסה שנבדקה בעלת מסה 2700 גרם. הקביעה אינה אבחון של טעות או מצב בריאותי.

## 3. חמש הרשומות בעלות המסה הגבוהה ביותר

```python
mass_descending = df.sort_values("body_mass_g", ascending=False)
top_five_mass = mass_descending.head(5)
display(top_five_mass[["species", "island", "body_mass_g"]])
```

בגרסה שנבדקה המסה הגבוהה ביותר היא 6300 גרם. head(5) היא בחירה של חמש רשומות לאחר מיון, לא דגימה מייצגת. אין להשתמש אוטומטית בחמש הקיצוניות כנתוני אימון לכלל האוכלוסייה.

אם יש שוויון בערך המיון, ייתכנו כמה רשומות באותו ערך. אין לפרש את סדרן כהבדל בערך. אם יש שוויון בגבול חמש הרשומות, ייתכנו רשומות נוספות עם אותו ערך שלא הוצגו.

## 4. היכן המדידות החסרות?
ברירת המחדל של sort_values מציבה חסרים בסוף, גם בסדר עולה וגם ביורד. לכן tail של מיון יורד אינה דרך בטוחה למציאת הערכים הנמוכים ביותר.

```python
display(mass_descending[["species", "body_mass_g"]].tail(3))
```

שתי רשומות המסה החסרה נמצאות בסוף גרסת הקורס. אין פירוש הדבר שהן הקלות ביותר. כדי לראות את הערכים הנמוכים הידועים בחרנו head של מיון עולה. החסרים נשמרו ולא טופלו; אין למחוק אותם רק מפני שהופיעו בסוף.

## 5. מחברים סינון ומיון — בשלבים נפרדים
שאלה: אילו חמש רשומות Gentoo הן בעלות המסה הידועה הגבוהה ביותר?

```python
gentoo_mask = df["species"] == "Gentoo"
gentoo_df = df[gentoo_mask]
gentoo_sorted = gentoo_df.sort_values("body_mass_g", ascending=False)
gentoo_top_five = gentoo_sorted.head(5)
display(gentoo_top_five[["species", "body_mass_g"]])
print("מקור:", df.shape)
print("קבוצת Gentoo:", gentoo_df.shape)
print("קבוצה ממוינת:", gentoo_sorted.shape)
print("חמש רשומות:", gentoo_top_five.shape)
```

הסינון בחר קבוצה; המיון סידר אותה; head הציגה עד חמש רשומות; בחירת העמודות צמצמה את שדות התצוגה. סדר הפעולות עונה לשאלה 'בתוך Gentoo', ולא 'בחר חמישה מכל המאגר ואז חפש בהם Gentoo'.
