## פתיחה — בחירת נתונים בדרך למודל
דודו בוחן רכבים שמחירם עד 100000 ₪. איך נציג רק מחירים שעומדים בתקציב? זו שאלת סינון. בהמשך נשתמש בתנאים כדי לבחור תצפיות ולחקור קבוצות במאגר.

התקציב הוא כלל ידוע, לא כלל שנלמד מנתונים. סינון לפי מחיר יכול לשרת שאלה תיאורית, אך אם בונים מודל לכל שוק הרכב, סינון כזה מצמצם את אוכלוסיית המחקר ועלול להגביל את ההכללה. אין להסיק שרכב שעומד בתקציב בהכרח מתאים או משתלם.

## חזרה קצרה
קראו את הלולאה: כמה מחירים יודפסו? איזה משתנה מקבל בכל מעבר את הערך הנוכחי?

```python
prices_nis = [90000, 110000, 100000, 80000]
for price_nis in prices_nis:
    print("מחיר בשקלים:", price_nis)
```

## 1. תנאי הוא ביטוי שמחזיר אמת או שקר
`<=` פירושו קטן או שווה. בדקו ידנית: האם 90000 קטן או שווה ל־100000? ומה לגבי 110000?

```python
budget_nis = 100000
print(90000 <= budget_nis)
print(110000 <= budget_nis)
print(100000 <= budget_nis)
```

הפלט הוא `True`, `False`, `True`. אלו ערכים בוליאניים (`bool`), לא מילים במירכאות. הסף נכלל כי כתוב קטן **או שווה**.

| ביטוי | משמעות |
|---|---|
| x < limit | קטן מהסף, בלי שוויון |
| x <= limit | קטן מהסף או שווה לו |
| x > limit | גדול מהסף |
| x >= limit | גדול מהסף או שווה לו |
| x == limit | שווה לסף |
| x != limit | שונה מהסף |

אין צורך לשנן את כל הטבלה היום. הליבה היא `<=` וההבדל בין `<` ל־`<=`. `=` מציב ערך; `==` בודק שוויון.

## 2. if על ערך אחד
`if` פירושה: אם התנאי אמת, בצע את ההוראה המוזחת. אם התנאי שקר, דלג עליה. לא חייבים else.

```python
price_nis = 90000
if price_nis <= budget_nis:
    print("המחיר עומד בתקציב:", price_nis)
print("הבדיקה הסתיימה")
```

פירוק: `if` מתחילה בדיקה; אחריה תנאי; נקודתיים מסיימות את הכותרת. ארבעה רווחים מסמנים את גוף התנאי. הודעת הסיום מחוץ לתנאי ולכן מופיעה בכל מקרה.

ניסוי: שנו את המחיר ל־110000 והריצו תא זה בלבד. מה לא יודפס? לאחר הניסוי החזירו ל־90000 והריצו שוב.

## עצירת הבנה
למחיר 100000: האם גוף התנאי יתבצע? מה ישתנה אם נחליף ל־`<`? כתבו קודם תחזית ופירוש.

## 3. תנאי בתוך לולאה — רק לאחר שהשלב הקודם ברור
הלולאה בוחרת בכל מעבר מחיר אחד; התנאי בודק אותו. ההדפסה הפנימית שייכת גם ללולאה וגם לתנאי.

```python
for price_nis in prices_nis:
    if price_nis <= budget_nis:
        print("מחיר שעומד בתקציב:", price_nis)
print("סיימנו לבדוק את כל המחירים")
```

### טבלת מעקב
| מעבר | מחיר | מחיר <= 100000 | האם יודפס המחיר? |
|---|---|---|---|
| 1 | 90000 | True | כן |
| 2 | 110000 | False | לא |
| 3 | 100000 | True | כן |
| 4 | 80000 | True | כן |

יש ארבעה מעברים אך רק שלוש הדפסות של מחיר. הודעת הסיום מופיעה פעם אחת. הדפסת הערכים שנבחרו אינה יצירת רשימה חדשה ואינה מחיקה מהמקור.
