## פתרון מתועד — מחשבים

## למה לומדים את זה עכשיו?
בקניית מחשב יש מגבלת תקציב: מה ההבדל בין כלל סינון לבין תחזית מחיר?
**בתהליך הפרויקט:** בחירת נתונים.
**מסגרת פעילה:** 1 שעות לימוד — 45 דקות, כולל דוגמה ותרגול עצמאי. המורה בוחר את משימות הליבה שבמדריך; שאר הפעילויות לחזרה או לעיון, ללא תוספת שעות.

### 1. תוצאות תנאים
True, False, False, True. האופרטור `<` אינו כולל שוויון; `<=` כולל אותו.

### 2. ערך אחד
2600 עומד בתקציב. במחיר 3200 תופיע רק הודעת הסיום; לאחר בדיקה מחזירים ל־2600.

```python
price_nis = 2600
budget_nis = 3000
if price_nis <= budget_nis:
    print("המחיר עומד בתקציב:", price_nis)
print("הבדיקה הסתיימה")
```

### 3. רשימה ותנאי
הלולאה הראשונה מציגה את כל הערכים. השנייה מציגה רק ערכים שעבורם התנאי אמת.

```python
prices_nis = [1800, 3200, 3000, 4200, 2600]
for price_nis in prices_nis:
    print(price_nis)
```

```python
for price_nis in prices_nis:
    if price_nis <= budget_nis:
        print("עד התקציב:", price_nis)
```

### 4. מעקב
| מעבר | מחיר | תנאי | הוצג? |
|---|---|---|---|
| 1 | 1800 | True | כן |
| 2 | 3200 | False | לא |
| 3 | 3000 | True | כן |
| 4 | 4200 | False | לא |
| 5 | 2600 | True | כן |

חמישה מעברים ושלושה מחירים מוצגים. False אינו עוצר את הלולאה; עוברים לערך הבא.

### 5. גבול שאינו נכלל
יוצגו רק 1800 ו־2600; המחיר 3000 לא עומד בתנאי המחמיר.

```python
for price_nis in prices_nis:
    if price_nis < budget_nis:
        print("מתחת לתקציב:", price_nis)
```

### 6. מקור ופירוש
הרשימה לא השתנתה. זו בחירה להצגה בלבד. השאלה היא אילו מחירים בדוגמה אינם עולים על 3000 ₪. הנתונים מומצאים והבחירה מצמצמת את הקבוצה; אין להסיק ממנה על כלל השוק או על כדאיות מחשב. לא אומן מודל: הסף ניתן מראש ולא נלמד מדוגמאות.

```python
print("מקור:", prices_nis)
```

### 7. הזחה
בקוד השגוי הדפסת המספר נמצאת בתוך for אך מחוץ ל־if, ולכן מתבצעת בכל מעבר. מזיחים אותה לתוך התנאי.

```python
for price_nis in prices_nis:
    if price_nis <= budget_nis:
        print("מחיר מתאים")
        print(price_nis)
```

### הרחבה — מונה
כל תצפית שעומדת בתנאי מוסיפה 1, ולא את מחירה. המונה מתאר מספר מחשבים, ללא יחידת ₪. אין חובה ללמד זאת למי שלא למד צבירה.

```python
matching_count = 0
for price_nis in prices_nis:
    if price_nis <= budget_nis:
        matching_count = matching_count + 1
print("מספר מחירים שעומדים בתקציב:", matching_count)
```
