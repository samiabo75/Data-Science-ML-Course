**יחידה 3 — רשות / המשך מפגש 8**. מבוסס על המחברת Knn_Penguin_classification_exercise_3. עוגן ההשתלמות מציג Pipeline, GridSearchCV ו־cv_results_. כאן התרגול הוא קוד רץ שמונע התאמת Scaler מחוץ ל־CV.

```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split, cross_val_score, StratifiedKFold
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import make_pipeline
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score, classification_report, ConfusionMatrixDisplay
import matplotlib.pyplot as plt

df = pd.read_csv("https://raw.githubusercontent.com/samiabo75/Data-Science-ML-Course/main/Datasets/penguins.csv")
features = ["bill_length_mm","bill_depth_mm","flipper_length_mm","body_mass_g"]
clean = df.dropna(subset=features + ["species"]).copy()
X = clean[features]
y = clean["species"]
X_train,X_test,y_train,y_test = train_test_split(X,y,test_size=0.20,random_state=42,stratify=y)
print("Train:",X_train.shape,"Test:",X_test.shape)
```

```python
from sklearn.pipeline import Pipeline
from sklearn.model_selection import GridSearchCV
pipe=Pipeline([('scale',StandardScaler()),('knn',KNeighborsClassifier())])
grid=GridSearchCV(pipe,{'knn__n_neighbors':[1,3,5,7,9,15,25]},cv=5,scoring='accuracy')
grid.fit(X_train,y_train)
print('Best parameter:',grid.best_params_)
print('Best CV accuracy:',round(grid.best_score_,3))
print('Final Test:',round(accuracy_score(y_test,grid.predict(X_test)),3))
```

### חקר תוצאות
הציגו את `grid.cv_results_` בטבלה, מיינו לפי `mean_test_score`, וענו: האם ה־K שנבחר מבטיח תוצאה גבוהה יותר תמיד? מדוע Test שמור לסוף?

```python
cv_results=pd.DataFrame(grid.cv_results_)
display(cv_results[['param_knn__n_neighbors','mean_test_score','std_test_score']].sort_values('mean_test_score',ascending=False))
```
