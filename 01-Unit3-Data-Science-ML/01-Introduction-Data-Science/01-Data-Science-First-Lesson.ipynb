## מחשבים תשובה מנתונים
נחשב סך מכירות וממוצע על ארבעה ימים. אין צורך לשנן את פקודות Python כרגע — נסביר כל שורה.

```python
sales = [12, 18, 27, 14]  # רשימה של נתונים
print("מכירות בארבעת הימים:", sales)
print("סך המכירות:", sum(sales))
print("ממוצע מכירות ליום:", sum(sales) / len(sales))
```

## תכנות רגיל לעומת למידת מכונה
**תכנות רגיל:** אנחנו כותבים את הכלל; הנתונים נכנסים; המחשב מחשב תשובה.

**למידת מכונה:** מציגים למערכת דוגמאות של נתונים ותשובות נכונות; אלגוריתם מתאים מודל; אחר כך משתמשים במודל כדי לנבא תשובה לדוגמה חדשה.

ההדגמה הבאה היא כלל ידני — **לא מודל שנלמד מנתונים**.

```python
def rule_based_recommendation(temperature):
    if temperature >= 28:
        return "להציע שתייה קרה"
    return "להציע שתייה חמה"

print(rule_based_recommendation(31))
print(rule_based_recommendation(19))
```

## הדגמה של למידה מדוגמאות — 1-Nearest Neighbor
להמחשת הרעיון בלבד, נשתמש בדוגמאות מסומנות: טמפרטורה והמלצת מוצר. במקום לכתוב כלל של 28 מעלות, נחפש את הדוגמה הקרובה ביותר. זהו מודל שכנים פשוט, ללא אימון פרמטרים. נלמד את KNN לעומק בפרק 7.

```python
examples = [(15, "חמה"), (20, "חמה"), (27, "קרה"), (34, "קרה")]
new_temperature = 25
closest = examples[0]
for example in examples:
    if abs(example[0] - new_temperature) < abs(closest[0] - new_temperature):
        closest = example
print("טמפרטורה חדשה:", new_temperature)
print("המלצה לפי הדוגמה הקרובה:", closest[1])
```

**שאלות:** (1) איזו דוגמה נבחרה? (2) מה יקרה אם הטמפרטורה תהיה 22? (3) האם ארבע דוגמאות מספיקות להחלטה אמינה? (4) במה שתי הגישות שונות?
