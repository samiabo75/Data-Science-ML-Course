## 1. סיווג עיר או יער
לפי דף העבודה: x1 הוא ממוצע צבע אדום, x2 ממוצע צבע כחול, w1=0.6, w2=0.4, b=-100. מחשבים `z=w1*x1+w2*x2+b`. **אם z>0 — עיר (1); אחרת יער (0)**. קודם חשבו ידנית את ארבע הדוגמאות.

```python
import pandas as pd
images=pd.DataFrame({'x1_red':[153,93,81,90],'x2_blue':[173,83,124,18]})
w1,w2,b=0.6,0.4,-100
images['z']=w1*images['x1_red']+w2*images['x2_blue']+b
images['prediction']=(images['z']>0).astype(int)
display(images)
```

**בדיקת הבנה:** מה יקרה אם נחליף את b ב־0? האם קבלת החלטה בעזרת ערכי משקולות קבועים כבר נחשבת *אימון*? (לא; זו הפעלת מסווג.)

## 2. קו הפרדה
כאשר z=0 מתקבלת משוואת קו. בדוגמת המקור `x2=-1.5*x1+250`. נקודות מעל/מתחת לקו מסווגות לפי סימן z; היזהרו מהנחה ש־'מעל' פירושו תמיד מחלקה 1 ללא בדיקת סימן המקדם.

```python
import matplotlib.pyplot as plt
plt.scatter(images['x1_red'],images['x2_blue'],c=images['prediction'])
xs=[0,200]
ys=[(-w1*x-b)/w2 for x in xs]
plt.plot(xs,ys)
plt.xlabel('Mean red (x1)');plt.ylabel('Mean blue (x2)');plt.title('Perceptron boundary');plt.show()
```

## 3. אימון הפרספטרון — הדגמת עדכון יחיד
לפי המצגת מתחילים ב־w=(0,0), b=0. דוגמה חיובית שמזוהה כשלילית: `w=w+x`, `b=b+1`. דוגמה שלילית שמזוהה כחיובית: `w=w-x`, `b=b-1`. בדוגמה שבה z=0 החלטת הסיווג היא 0.

```python
x1,x2,y=94,129,1
w1=w2=b=0
z=w1*x1+w2*x2+b
prediction=int(z>0)
print('לפני עדכון:',z,prediction)
if y==1 and prediction==0:
    w1+=x1;w2+=x2;b+=1
print('אחרי עדכון:',w1,w2,b)
```

## 4. SVM — מסווג עם שוליים רחבים
פרספטרון מנסה להגיע להפרדה ליניארית; SVM ליניארי מחפש הפרדה בעלת שוליים גדולים תחת האילוצים המתאימים. SVM מאפשר גם Kernels לגבולות לא ליניאריים. התנסות מינימלית על Iris מתוך sklearn (ללא הורדת CSV).

```python
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.svm import SVC
from sklearn.metrics import accuracy_score
X,y=load_iris(return_X_y=True)
mask=y!=2
X,y=X[mask],y[mask]
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.25,random_state=42,stratify=y)
model=make_pipeline(StandardScaler(),SVC(kernel='linear',C=1.0))
model.fit(X_train,y_train)
print('Test Accuracy:',accuracy_score(y_test,model.predict(X_test)))
```
