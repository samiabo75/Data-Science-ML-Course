```
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.neighbors import KNeighborsClassifier
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import MinMaxScaler
from sklearn.metrics import classification_report, confusion_matrix, ConfusionMatrixDisplay
```

# Read and examine the Data

```
df=pd.read_csv('https://raw.githubusercontent.com/arielb30/datasets/main/diamonds.csv')
df.head()
```

```
X = df.drop(['price'],axis=1)
y = df['price']
print("start: ",X.shape, y.shape)
```

```
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.20, random_state = 42)
print("train: ", X_train.shape, y_train.shape)
print("test:  " , X_test.shape, y_test.shape)
```

```
dftrain = pd.concat([X_train,y_train],axis=1)
dftrain.head()
```

```
# Group by 'cut' and calculate the mean of 'price'
cut_price_mean = dftrain.groupby('cut')['price'].mean().sort_values(ascending=True)
cut_price_mean
```
