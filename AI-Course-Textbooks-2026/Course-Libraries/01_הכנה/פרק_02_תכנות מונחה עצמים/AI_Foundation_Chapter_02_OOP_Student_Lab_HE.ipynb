## הדגמה מודרכת 5 — fit / transform בלי מודל מאומן

```python
class SimpleMinMax:
    def __init__(self):
        self.min_value = None
        self.max_value = None

    def fit(self, values):
        self.min_value = min(values)
        self.max_value = max(values)
        return self

    def transform(self, values):
        gap = self.max_value - self.min_value
        return [0.0 if gap == 0 else (x - self.min_value)/gap
                for x in values]

scaler = SimpleMinMax().fit([30, 40, 50])
print(scaler.transform([45, 60]))
```
