# Matrix dot product

```python
mat_a = np.random.randint(0,10,(2,3))
mat_b = np.random.randint(0,10,(3,2))
print(mat_a)
print("\n",mat_b)
```

```python
def mat_mul_v1(mat_a, mat_b): #dot product
  m = mat_a.shape[0]
  n = mat_b.shape[1]
  k1 = mat_a.shape[1]
  k2 = mat_b.shape[0]
  if k1!=k2:
    raise ValueError(f"list/vectors are of different length: {k1} , {k2}")
  mat_c = np.zeros((m,n),dtype=int)
  for i in range(m):
    for j in range(n):
      sum=0
      for p in range(k1):
        sum += mat_a[i,p] * mat_b[p,j]
      mat_c[i,j] = sum
  return mat_c
```

```python
def mat_mul_v2(mat_a, mat_b): #dot product
  m = mat_a.shape[0]
  n = mat_b.shape[1]
  k1 = mat_a.shape[1]
  k2 = mat_b.shape[0]
  if k1!=k2:
    raise ValueError(f"list/vectors are of different length: {k1} , {k2}")
  mat_c = np.zeros((m,n),dtype=int)
  for i in range(m):
    for j in range(n):
      mat_c[i,j] = mat_a[i,:] @ mat_b[:,j]
  return mat_c
```

```python
print(mat_mul_v1(mat_a,mat_b))
print("\n",mat_mul_v2(mat_a,mat_b))
print("\n",mat_a@mat_b)
print("\n",np.dot(mat_a,mat_b))
print("\n",mat_a.dot(mat_b))
```
