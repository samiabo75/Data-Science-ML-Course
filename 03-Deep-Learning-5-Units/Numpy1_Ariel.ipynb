# Numpy row and column vectors

```python
a1 = np.array([1,5,10])
print(a1)
print(a1.shape)
```

```python
b1 = np.array([[1],[5],[10]]) #column vector
print(b1)
print(b1.shape)
```

```python
print(a1.reshape(-1,1))
print(a1.reshape(-1,1).shape)
```

```python
a1.shape
```

```python
print(np.expand_dims(a1,axis=1))
print(np.expand_dims(a1,axis=1).shape)
```

```python
print(b1.reshape(-1))
print(b1.reshape(-1).shape)
```

```python
print(np.squeeze(b1))
print(np.squeeze(b1).shape)
```

```python
c1 = np.array([[1,5,10]]) #row vector
print(c1)
print(c1.shape)
```

```python
print(a1.reshape(1,-1))
print(a1.reshape(1,-1).shape)
```

```python
print(np.expand_dims(a1,axis=0))
print(np.expand_dims(a1,axis=0).shape)
```

```python
print(c1.reshape(-1))
print(c1.reshape(-1).shape)
```

```python
print(np.squeeze(c1))
print(np.squeeze(c1).shape)
```
