```python
def  vectorcheck(u):
    u_valid = all(isinstance(x, (int, float)) for x in u)
    if u_valid:
            return True
    return False
def add(u, v):
    if vectorcheck(u):
        if len(u) == len(v):
            w=[]
            for index in range(0,len(u)):
                w.append(u[index] + v[index])
            return  w
    print('invalid input')
    return None
def multiply(u, v):
    if vectorcheck(u):
        if len(u) == len(v):
            w=[]
            for index in range(0,len(u)):
                w.append(u[index] * v[index])
            return  w
    print('invalid input')
    return None
def dot(u,v):
    if vectorcheck(u):
        if len(u) == len(v):
            w=0
            for index in range(0,len(u)):
                w += u[index] * v[index]
            return w
    print('invalid input')
    return None
def Scalar_mul(s,v):
    if vectorcheck(v):
        w=[]
        for index in range(0,len(v)):
            w.append(s * v[index])
        return w
    print('invalid input')
    return None

a=[1,2,3,4]
b=[3,2,1,0.0]
print(add(a,b))
print(dot(a,b))
print(Scalar_mul(2,a))
print(multiply(a,b))
```
