## 1. חיזוי ידני לפני ספרייה
המודל למציאת רוחב עלה לפי אורך: ŷ = wx + b. במצגת מוצגים w=0.416, b=-0.366. חשבו תחילה ביד עבור x=1.5, 2.5, 6.5. מה משמעות שיפוע חיובי?

```python
w = 0.416
b = -0.366
for x in [1.5, 2.5, 6.5]:
    prediction = w*x + b
    print('אורך עלה:', x, 'רוחב חזוי:', round(prediction, 3))
```

## 2. טעינת נתונים
**מאפיין (X):** אורך עלה כותרת. **משתנה מטרה (y):** רוחב עלה כותרת. ל־scikit-learn נדרש X דו־ממדי גם כשיש מאפיין אחד.

```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score

df = pd.read_csv("https://raw.githubusercontent.com/arielb30/datasets/main/Iris.csv")
display(df.head())
print("Shape:",df.shape)
```

## 3. בדיקה חזותית

```python
plt.scatter(df["PetalLengthCm"], df["PetalWidthCm"], alpha=0.55)
plt.xlabel("Petal length (cm)")
plt.ylabel("Petal width (cm)")
plt.title("Iris: petal length vs width")
plt.show()
```

## 4. אימון המודל
fit מחשב את המקדמים מתוך הנתונים. זו למידה מדוגמאות, בניגוד לערכים הידניים מהשלב הראשון.

```python
X = df[["PetalLengthCm"]]
y = df["PetalWidthCm"]
reg = LinearRegression()
reg.fit(X,y)
print("w:",reg.coef_[0])
print("b:",reg.intercept_)
```

## 5. חיזוי לדוגמאות חדשות

```python
new_flowers = pd.DataFrame({"PetalLengthCm":[3,5,7]})
print(reg.predict(new_flowers))
```

## 6. R² — איכות התאמה
הדגמה ראשונית על **אותם נתונים שעליהם אומן המודל**. זהו ציון אימון בלבד ואינו מוכיח יכולת הכללה. בהמשך נלמד בדיקה על Test נפרד.

```python
predictions = reg.predict(X)
print("R² training:",round(r2_score(y,predictions),4))
```

## 7. הצגת ישר הרגרסיה

```python
plt.scatter(X["PetalLengthCm"],y,alpha=0.45)
line_x = pd.DataFrame({"PetalLengthCm":[X["PetalLengthCm"].min(),X["PetalLengthCm"].max()]})
plt.plot(line_x["PetalLengthCm"],reg.predict(line_x),color="red")
plt.xlabel("Petal length (cm)")
plt.ylabel("Petal width (cm)")
plt.show()
```

## 8. יותר ממאפיין אחד — טעימה
במצגת מוסבר שהמודל הכללי הוא ŷ = w₁x₁ + w₂x₂ + ⋯ + b. רשמו למה נדרשת מטריצה X ולא וקטור בודד. בתרגול נרחיב לנתוני דירות.
