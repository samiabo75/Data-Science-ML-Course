# Calculating Q1 + Q3, n is odd

```python
arr2 = np.array([7.0,2.0,10.0,4.0,6.0])
```

```python
# def school_quartiles(data): #returns Q1,Q3
#     data = np.?(?)   # מיון המערך
#     n = ?(?) #מספר האיברים במערך

#     if ? % ?:  # odd
#         lower = ?[:?//?]
#         upper = ?[?//?+?:]
#     else:      # even
#         lower = data[:?//?]
#         upper = data[?//?:]

#     return np.?(?), np.?(?)


def school_quartiles(data): #returns Q1,Q3
    data = np.sort(data)   # מיון המערך
    n = len(data) #מספר האיברים במערך

    if n % 2:  # odd
        lower = data[:n//2]
        upper = data[n//2+1:]
    else:      # even
        lower = data[:n//2]
        upper = data[n//2:]

    return np.median(lower), np.median(upper)
```

```python
#computation
school_quartiles(arr2)
```

```python
arr3 = np.array([12.0,5.0,8.0,3.0])
```

```python
#computation
school_quartiles(arr3)
```

# Computing Mode

```python
arr4 = np.array([3.0,1.0,3.0,4.0,3.0,3.0,4.0,4.0,5.0,1.0])
```

```python
#automatic using scipy
# from scipy import stats
# mode_result = stats.?(?)
# print(mode_result)
# print(mode_result.?, mode_result.?)

from scipy import stats
mode_result = stats.mode(arr4)
print(mode_result)
print(mode_result.mode, mode_result.count)
```

```python
#computation part 1
# unique_elements, counts = np.?(?, return_counts=?)
# frequency_table = dict(zip(?, ?))
# frequency_table


unique_elements, counts = np.unique(arr4, return_counts=True)
frequency_table = dict(zip(unique_elements, counts))
frequency_table
```

```python
#computation part 2
# max, max_key = ?, ?
# for ?, ? in frequency_table.?():
#     if ? > ?:
#         ? = ?
#         ? = ?
# print(?, ?)

max, max_key = 0, None
for element, count in frequency_table.items():
    if count > max:
        max = count
        max_key = element
print(max_key, max)
```
