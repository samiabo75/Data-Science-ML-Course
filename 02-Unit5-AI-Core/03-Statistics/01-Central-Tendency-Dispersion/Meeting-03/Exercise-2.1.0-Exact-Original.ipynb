# Computing Frequency table

```python
arr1 = np.array([1.0,4.0,4.0,2.0,4.0,1.0,1.0,2.0,1.0])
```

```python
unique_elements, counts = np.unique(arr1, return_counts=True)
```

```python
unique_elements
```

```python
counts
```

```python
dic = dict(zip(unique_elements, counts))
dic
```

```python
for key,value in dic.items():
  print(f"{key} appears {value} times")
```

```python
import pandas as pd
df = pd.Series(dic).reset_index()
df.columns = ['element', 'count']
display(df)
```
