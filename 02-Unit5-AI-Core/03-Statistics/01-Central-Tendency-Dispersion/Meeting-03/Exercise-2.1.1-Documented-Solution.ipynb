# Calculating Q1 + Q3, n is odd

```python
arr2 = np.array([7.0,2.0,10.0,4.0,6.0])
```

**הסבר מתמטי ותכנותי:** הרבעונים מחושבים באמצעות חלוקת הנתונים לחצאים, ללא החציון כאשר n אי-זוגי — בהתאם למצגת.

```python
# def school_quartiles(data): #returns Q1,Q3

#     else:      # even



def school_quartiles(data): #returns Q1,Q3
    data = np.sort(data)   # מיון המערך
    n = len(data) #מספר האיברים במערך

    if n % 2:  # odd
        lower = data[:n//2]
        upper = data[n//2+1:]
    else:      # even
        lower = data[:n//2]
        upper = data[n//2:]

    return np.median(lower), np.median(upper)
```

```python
#computation
school_quartiles(arr2)
```

```python
arr3 = np.array([12.0,5.0,8.0,3.0])
```

```python
#computation
school_quartiles(arr3)
```

# Computing Mode

```python
arr4 = np.array([3.0,1.0,3.0,4.0,3.0,3.0,4.0,4.0,5.0,1.0])
```

**הסבר מתמטי ותכנותי:** stats.mode מחזירה שכיח וספירת מופעים; יש להבדיל בין מקרה חד-שכיח לרב-שכיח.

```python
#automatic using scipy
# from scipy import stats
# print(mode_result)

from scipy import stats
mode_result = stats.mode(arr4)
print(mode_result)
print(mode_result.mode, mode_result.count)
```

**הסבר מתמטי ותכנותי:** np.unique(...,return_counts=True) נותנת ערכים ייחודיים ומספר מופעים.

```python
#computation part 1
# frequency_table


unique_elements, counts = np.unique(arr4, return_counts=True)
frequency_table = dict(zip(unique_elements, counts))
frequency_table
```

**הסבר מתמטי ותכנותי:** עוברים על מילון השכיחויות ומעדכנים את השכיח הגבוה ביותר.

```python
#computation part 2

max, max_key = 0, None
for element, count in frequency_table.items():
    if count > max:
        max = count
        max_key = element
print(max_key, max)
```

# Computing population variance & std

```python
arr5 = np.array([2.0,7.0,4.0,11.0])
```

**הסבר מתמטי ותכנותי:** שונות אוכלוסייה מחלקת ב-N, ולכן ddof=0.

```python
#vectorized

np.var(arr5,ddof=0)
```

**הסבר מתמטי ותכנותי:** סכום ריבועי הסטיות מחולק ב-N.

```python
#loop
np.sum((arr5-np.mean(arr5))**2)/len(arr5)
```

**הסבר מתמטי ותכנותי:** במקור חושבה כאן שונות במקום סטיית תקן; תוקן ל-np.std עם ddof=0.

```python
# compute std direct — corrected original variance expression
np.std(arr5,ddof=0)
```

**הסבר מתמטי ותכנותי:** סטיית תקן היא שורש השונות; np.std מבצעת זאת ישירות.

```python
#compute std from var
np.std(arr5,ddof=0)
```

# Computing sample variance & std

```python
arr5 = np.array([2.0,7.0,4.0,11.0])
```

```python
np.mean(arr5)
```

**הסבר מתמטי ותכנותי:** שונות מדגם עם תיקון בסל מחלקת ב-n-1, ולכן ddof=1.

```python
#vectorized
np.var(arr5,ddof=1)
```

**הסבר מתמטי ותכנותי:** המכנה len(arr5)-1 הופך את החישוב לשונות מדגם.

```python
#loop
np.sum((arr5-np.mean(arr5))**2)/(len(arr5)-1)
```

**הסבר מתמטי ותכנותי:** סטיית תקן של מדגם מחושבת עם ddof=1.

```python
#compute std direct
np.std(arr5,ddof=1)
```

**הסבר מתמטי ותכנותי:** שורש שונות המדגם שווה לסטיית התקן שלו.

```python
#compute std from var
np.sqrt(np.var(arr5,ddof=1))
```
