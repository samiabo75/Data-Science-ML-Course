## אוכלוסייה ומדגם
N מסמן גודל אוכלוסייה ו־n גודל מדגם. מדגם הוא חלק מהקבוצה שמעניינת אותנו; בזמן אימון מודל יש להפריד נתוני אימון מנתוני בדיקה.

## ממוצע וחציון
ראשית מחשבים ידנית; אחר כך משתמשים ב־mean וב־median. בחציון חובה למיין.

```python
x=np.array([7,2,10,4,6], dtype=float)
print('ממוצע:',sum(x)/len(x),'בדיקת NumPy:',np.mean(x))
print('מיון:',sorted(x),'חציון:',np.median(x))
y=np.array([12,5,8,3],dtype=float)
print('חציון מספר זוגי:',np.median(y))
```

## רבעונים בשיטת המצגת (Tukey: ללא החציון במספר איברים אי־זוגי)
זה אינו בהכרח הפלט של np.percentile ברירת מחדל.

```python
def tukey_quartiles(values):
    a=sorted(values)
    n=len(a)
    middle=n//2
    low=a[:middle]
    high=a[middle+1:] if n%2 else a[middle:]
    return float(np.median(low)),float(np.median(a)),float(np.median(high))
for data in ([7,2,10,4,6],[12,5,8,3]):
    q1,q2,q3=tukey_quartiles(data)
    print('נתונים:',data,'Q1,Q2,Q3=',q1,q2,q3,'IQR=',q3-q1)
    print('np.percentile:',np.percentile(data,[25,50,75]))
```

## שכיח והתפלגות
השכיח הוא הערך או הערכים בעלי התדירות הגבוהה ביותר. בנתוני קטגוריות זו לעיתים הבחירה הטבעית ביותר.

```python
values=[3,1,3,4,3,3,4,4,5,1]
counts=Counter(values)
highest=max(counts.values())
modes=[x for x,count in counts.items() if count==highest]
print('שכיחים:',modes,'שכיחויות:',counts)
```

## שונות וסטיית תקן
ב־NumPy משתמשים ב־ddof=0 לאוכלוסייה וב־ddof=1 למדגם. מדובר באומד סטטיסטי אחר, לא בטעות ספרייה.

```python
data=np.array([2.,4.,6.,7.,10.])
mean=np.mean(data)
variance_pop=np.sum((data-mean)**2)/len(data)
variance_sample=np.sum((data-mean)**2)/(len(data)-1)
print('μ/x̄:',mean)
print('variance population:',variance_pop, 'np.var:',np.var(data,ddof=0))
print('variance sample:',variance_sample,'np.var ddof1:',np.var(data,ddof=1))
print('std population:',np.std(data,ddof=0),'std sample:',np.std(data,ddof=1))
```

## ערך חריג והשפעתו

```python
a=np.array([4,5,6,7,8])
b=np.array([4,5,6,7,100])
for name,z in [('רגיל',a),('עם חריג',b)]:
    print(name,'mean=',np.mean(z),'median=',np.median(z),'std=',np.std(z))
```
