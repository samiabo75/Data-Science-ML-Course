הסבר מתמטי והוראת Python מלאה לכל משימה. משימה 6 היא הרחבה מקצועית מעבר למצגת.

```python
import numpy as np
```

## 1. ממוצע וחציון עם ערך חריג
**משימה:** השוו [4,5,6,7,8] עם [4,5,6,7,100]. חשבו ממוצע וחציון ונמקו מה מושפע יותר.

**הסבר:** החציון עמיד יותר לחריגים מהממוצע.

```python
a=np.array([4,5,6,7,8]);b=np.array([4,5,6,7,100])
for x in [a,b]:print('mean',np.mean(x),'median',np.median(x))
```

## 2. חציון זוגי ואי־זוגי
**משימה:** מיינו [7,2,10,4,6] ו-[12,5,8,3] ומצאו חציון.

**הסבר:** למספר זוגי של איברים מחשבים ממוצע של שני האמצעיים.

```python
for x in [[7,2,10,4,6],[12,5,8,3]]:print(sorted(x),np.median(x))
```

## 3. רבעונים ו-IQR
**משימה:** חשבו בשיטת המצגת Q1,Q2,Q3,IQR עבור [7,2,10,4,6], והשוו ל-np.percentile.

**הסבר:** בשיטת המצגת: Q1=3,Q2=6,Q3=8.5,IQR=5.5. ברירת המחדל של NumPy משתמשת באינטרפולציה אחרת.

```python
x=np.array([7,2,10,4,6]);s=sorted(x);m=len(s)//2
q1=np.median(s[:m]);q2=np.median(s);q3=np.median(s[m+1:]);print(q1,q2,q3,q3-q1)
print(np.percentile(x,[25,50,75]))
```

## 4. שכיח
**משימה:** מצאו שכיח ל-[3,1,3,4,3,3,4,4,5,1] ותארו מצב עם יותר משכיח אחד.

**הסבר:** שכיח הוא הערך בתדירות מרבית; ייתכנו כמה שכיחים.

```python
from collections import Counter
x=[3,1,3,4,3,3,4,4,5,1];c=Counter(x);top=max(c.values());print([k for k,v in c.items() if v==top])
```

## 5. שונות וסטיית תקן
**משימה:** השוו שונות וסטיית תקן של אוכלוסייה ומדגם לנתונים [2,4,6,7,10].

**הסבר:** הבדל המכנה בין N לבין n−1; סטיית התקן היא השורש של השונות.

```python
x=np.array([2.,4.,6.,7.,10.]);avg=np.mean(x);sq=np.sum((x-avg)**2)
print('Population',sq/len(x),np.std(x,ddof=0))
print('Sample',sq/(len(x)-1),np.std(x,ddof=1))
```

## 6. AI — תקנון ללא דליפה
**משימה:** תקננו Train=[50,60,70,80] ו-Test=[90,100] בעזרת ממוצע וסטיית התקן של Train בלבד.

**הסבר:** הרחבה מקצועית: פרמטרי תקנון נלמדים על נתוני האימון בלבד.

```python
train=np.array([50.,60.,70.,80.]);test=np.array([90.,100.]);mu=np.mean(train);std=np.std(train)
print('train', (train-mu)/std)
print('test', (test-mu)/std)
```
