# יחידה 5 — פרק 2.2: התפלגות נורמלית
## התנסות דידקטית — פתרון מלא ומתועד
**חומר מקורי נוסף על מצגת ההשתלמות**, עם הסבר מתמטי, הקשר לפרויקטים ופתרון Python.

```python
import numpy as np
from scipy.stats import norm
```

## התנסות 1: ציוני תלמידים בקורס
**השאלה:** בקורס דוגמה ציונים מתפלגים נורמלית עם ממוצע 75 וסטיית תקן 10. תלמיד קיבל 90. חשבו Z והסיכוי לקבל ציון גבוה מ־90.

**ההסבר המתמטי והקשר לנתונים:** z=(90−75)/10=1.5. ההסתברות היא השטח מימין ל־90. זהו מודל הדגמה בלבד; ציוני אמת אינם בהכרח נורמליים.

```python
mu = 75
sigma = 10
score = 90
z = (score - mu) / sigma
prob = norm.sf(score, loc=mu, scale=sigma)
print('Z-score:', z)
print('P(score>90):', round(prob,4))
```

## התנסות 2: זמני משלוח בחנות
**השאלה:** נניח שזמן משלוח מתפלג נורמלית עם μ=30, σ=5 דקות. חשבו P(T>40) ואת P(25≤T≤35).

**ההסבר המתמטי והקשר לנתונים:** 40 דקות הן שתי סטיות תקן מעל הממוצע, ולכן כ־2.28% מעל הסף. התחום 25–35 הוא μ±σ ולכן כ־68.27%.

```python
mu=30
sigma=5
late=norm.sf(40,loc=mu,scale=sigma)
in_range=norm.cdf(35,loc=mu,scale=sigma)-norm.cdf(25,loc=mu,scale=sigma)
print('late',round(late,4),'between',round(in_range,4))
```

## התנסות 3: חישוב בדיוק לפי שקף 18
**השאלה:** בהתפלגות N(2.5,1.5²), חשבו P(X≤3), P(3≤X≤4), ו־P(X≥3) עם norm.cdf ו־norm.sf.

**ההסבר המתמטי והקשר לנתונים:** חישוב שטחים: CDF(3)≈0.63056; CDF(4)−CDF(3)≈0.21079; SF(3)≈0.36944.

```python
mu=2.5
sigma=1.5
a=3
b=4
left=norm.cdf(a,loc=mu,scale=sigma)
between=norm.cdf(b,loc=mu,scale=sigma)-norm.cdf(a,loc=mu,scale=sigma)
right=norm.sf(a,loc=mu,scale=sigma)
print('P(X<=3):',round(left,5))
print('P(3<=X<=4):',round(between,5))
print('P(X>=3):',round(right,5))
```

## התנסות 4: PDF מול CDF
**השאלה:** עבור N(170,10²), חשבו צפיפות ב־170 ואת P(X≤170). הסבירו למה התוצאות שונות.

**ההסבר המתמטי והקשר לנתונים:** PDF הוא גובה עקומת הצפיפות (כ־0.0399), לא הסתברות לנקודה. CDF(170)=0.5 הוא השטח המצטבר עד הממוצע.

```python
mu=170
sigma=10
print('PDF at mean:',norm.pdf(170,loc=mu,scale=sigma))
print('CDF at mean:',norm.cdf(170,loc=mu,scale=sigma))
```

## התנסות 5: המחשת כלל 68–95–99.7
**השאלה:** חשבו ב־Python את ההסתברות עבור |Z|≤1, |Z|≤2 ו־|Z|≤3.

**ההסבר המתמטי והקשר לנתונים:** מתקבלים בקירוב 68.27%,95.45%,99.73%. כלל זה מבוסס על התפלגות נורמלית.

```python
for k in [1,2,3]:
    p=norm.cdf(k)-norm.cdf(-k)
    print('Within +/-',k,'sigma:',round(100*p,3),'%')
```

## התנסות 6: ניתוח פרויקט מחירי דירות — מגבלות ההנחה
**השאלה:** צרו מערך מחירי דירות [800000,900000,950000,1000000,1100000,5000000]. חשבו ממוצע, חציון ו-Z של הדירה היקרה. האם ציון התקן מוכיח שמחירים נורמליים?

**ההסבר המתמטי והקשר לנתונים:** הממוצע קופץ בגלל דירה יקרה אחת; החציון עמיד יותר. מדד Z ניתן לחישוב גם בנתונים לא נורמליים, אך לא ניתן להסיק ממנו הסתברויות נורמליות ללא אימות הנחה.

```python
prices=np.array([800000,900000,950000,1000000,1100000,5000000],dtype=float)
mu=np.mean(prices)
sigma=np.std(prices,ddof=0)
print('mean',round(mu,2),'median',np.median(prices))
print('Z expensive',round((prices[-1]-mu)/sigma,3))
print('חשוב: Z אינו מוכיח התפלגות נורמלית')
```

## התנסות 7: תרגיל מצטיינים — סף 90%
**השאלה:** בנתוני משלוחים N(30,5²), מצאו את משך הזמן ש־90% מהמשלוחים קצרים ממנו.

**ההסבר המתמטי והקשר לנתונים:** PPF היא הפונקציה ההפוכה ל־CDF: מוצאת ערך מתוך הסתברות. התוצאה כ־36.41 דקות; הרחבה מעבר למצגת.

```python
threshold=norm.ppf(0.9,loc=30,scale=5)
print('90th percentile:',round(threshold,2),'minutes')
```
