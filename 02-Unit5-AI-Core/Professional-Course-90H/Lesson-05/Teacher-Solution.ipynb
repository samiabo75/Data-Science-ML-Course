# שיעור 5 — מערכות המלצה: יישום דמיון — פתרון מתועד למורה

**משימה לתלמיד:** שנה פרופיל משתמש ל־[1,5], חשב דמיון לשלושת הסרטים וקבע המלצה מנומקת.

**הכוונה:** זהו פתרון דוגמה לנתוני הדגמה, לא פתרון יחיד לכל בחירת נתונים של תלמיד. הערכים בתרגול העצמאי עשויים להיות שונים.

**הבסיס המתמטי:** דמיון גבוה יותר מעיד על כיוון וקטורים דומה, לא על איכות הסרט

```python
import numpy as np
user=np.array([4.,1.]);movies={'סרט א':np.array([5.,1.]),'סרט ב':np.array([1.,5.]),'סרט ג':np.array([3.,2.])}
for name,m in movies.items():
 print(name,np.dot(user,m)/(np.linalg.norm(user)*np.linalg.norm(m)))
```
