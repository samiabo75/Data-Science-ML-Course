**משימת המעבדה:** החלפת פרופיל משתמש ל־[1,5] ובחירת הסרט הדומה

זהו פתרון לדוגמה עם קוד מלא. פתרונות גלויים במאגר ציבורי — רצוי להפיץ לתלמידים רק את Student-Lab.ipynb.

```python
import numpy as np
user=np.array([1.,5.])
movies={'סרט א':np.array([5.,1.]),'סרט ב':np.array([1.,5.]),'סרט ג':np.array([3.,2.])}
scores={}
for name,m in movies.items():
    scores[name]=np.dot(user,m)/(np.linalg.norm(user)*np.linalg.norm(m))
    print(name,round(scores[name],3))
print('המלצה:',max(scores,key=scores.get))
assert max(scores,key=scores.get)=='סרט ב'
```

## הסבר תוצאות

סרט ב מקביל בדיוק לווקטור המשתמש ולכן הקוסינוס שלו 1; זו המלצה מתמטית, לא הבטחת שביעות רצון.
