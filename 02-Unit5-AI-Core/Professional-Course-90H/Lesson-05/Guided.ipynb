**הקשר:** איזה סרט מומלץ לצופה שדירג שני ז'אנרים?

**יעד:** לדרג מועמדים לפי דמיון קוסינוס תוך התייחסות לווקטור אפס

**הסבר מתמטי:** דמיון גבוה יותר מעיד על כיוון וקטורים דומה, לא על איכות הסרט

הפעל תאים לפי הסדר. חשב תחילה על נייר ונמק את הפלט. הערכים דידקטיים.

```python
import numpy as np
user=np.array([4.,1.]);movies={'סרט א':np.array([5.,1.]),'סרט ב':np.array([1.,5.]),'סרט ג':np.array([3.,2.])}
for name,m in movies.items():
    similarity=np.dot(user,m)/(np.linalg.norm(user)*np.linalg.norm(m))
    print(name,round(similarity,3))
```
