# מפגש 1 — פתרון מורה (אין לצרף למצגת או להפצה לתלמידים)
נתוני ספרייה סינתטיים.

```python
books = [[120, 3], [180, 4], [240, 5]]
labels = ["רגיל", "מומלץ", "מומלץ"]
rows = len(books)
features = len(books[0])
print(rows, features)
assert rows == 3 and features == 2
assert len(labels) == 3
```

```python
pages = [row[0] for row in books]
mean_pages = sum(pages) / len(pages)
print(mean_pages)
assert mean_pages == 180
```

```python
changed_pages = [120, 180, 300]
changed_mean = sum(changed_pages) / len(changed_pages)
print(changed_mean)
assert changed_mean == 200
assert changed_mean - mean_pages == 20
```

**הסבר למורה:** 120+180+240=540, חלקי 3 = 180 עמודים לספר; לאחר שינוי 240→300, הסכום 600, חלקי 3 = 200. התווית מומלץ/רגיל היא יעד הסיווג ואינה מאפיין. אם מנבאים דירוג, הדירוג אינו קלט שמותר לדעת מראש.
