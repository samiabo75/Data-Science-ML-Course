# מפגש 1 — מחברת הדגמה
נתוני דוגמה מומצאים, לא נתוני תלמידים אמיתיים. לפני כל הרצה ננבא פלט.

```python
X = [[1, 60], [2, 70], [4, 85], [5, 95]]
y = ["חיזוק", "חיזוק", "העשרה", "העשרה"]
print("rows =", len(X), "features =", len(X[0]))
```

## מחשבים ממוצע ביד: (60+70+85+95)/4 = 77.5

```python
scores = [row[1] for row in X]
mean_score = sum(scores) / len(scores)
print("Mean:", mean_score)
assert len(X) == len(y) == 4
assert all(len(row) == 2 for row in X)
assert mean_score == 77.5
```

## שינוי ערך: מה קורה אם הציון האחרון 100?

```python
new_scores = [60, 70, 85, 100]
print(sum(new_scores) / len(new_scores))
assert sum(new_scores)/len(new_scores) == 78.75
```
