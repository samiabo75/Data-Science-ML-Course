**משימה לתלמיד:** נתח ציוני שתי קבוצות משלך: מדדי מרכז, רבעונים, שונות וסטיית תקן. נמק איזו קבוצה יציבה יותר.

**הכוונה:** זהו פתרון דוגמה לנתוני הדגמה, לא פתרון יחיד לכל בחירת נתונים של תלמיד. הערכים בתרגול העצמאי עשויים להיות שונים.

**הבסיס המתמטי:** דווח גודל מדגם, מדדי מרכז ופיזור והסתייגות

```python
import numpy as np
scores=np.array([62,70,72,72,75,81,85,90,93,100])
print(len(scores),np.mean(scores),np.median(scores),np.percentile(scores,[25,75]),np.std(scores,ddof=1))
```
