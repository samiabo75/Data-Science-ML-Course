**משימת המעבדה:** השוואת מדדי סטטיסטיקה בשתי קבוצות

זהו פתרון לדוגמה עם קוד מלא. פתרונות גלויים במאגר ציבורי — רצוי להפיץ לתלמידים רק את Student-Lab.ipynb.

```python
import numpy as np
groups={'א':np.array([62,70,72,75,81]),'ב':np.array([70,70,71,72,90])}
for name,x in groups.items():
    q1,q3=np.percentile(x,[25,75])
    print('קבוצה',name,'n',len(x),'mean',np.mean(x),'median',np.median(x),'Q1/Q3',q1,q3,'std sample',np.std(x,ddof=1))
```

## הסבר תוצאות

זהו פתרון מדגים לשתי קבוצות שנבחרו בידי המורה; תלמיד רשאי להשתמש בנתונים אחרים, אך חייב לפרש פיזור ומרכז יחד.
