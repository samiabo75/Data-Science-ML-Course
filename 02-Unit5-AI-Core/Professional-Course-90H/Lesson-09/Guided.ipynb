# שיעור 9 — מעבדת סטטיסטיקה מסכמת — מחברת מודרכת

**הקשר:** איך מנתח נתונים מציג מסקנה מבוססת ולא רק מספר?

**יעד:** לנתח מאגר קטן ולנסח מסקנה מנומקת

**הסבר מתמטי:** דווח גודל מדגם, מדדי מרכז ופיזור והסתייגות

הפעל תאים לפי הסדר. חשב תחילה על נייר ונמק את הפלט. הערכים דידקטיים.

```python
import numpy as np
scores=np.array([62,70,72,72,75,81,85,90,93,100])
print('n',len(scores),'mean',np.mean(scores),'median',np.median(scores))
print('Q1 Q3',np.percentile(scores,[25,75]))
print('std',np.std(scores,ddof=1))
```
