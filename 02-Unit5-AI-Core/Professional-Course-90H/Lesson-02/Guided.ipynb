**הקשר:** איך מחשבים ממוצע של ציוני קבוצת תלמידים?

**יעד:** לייצג נתונים במערך NumPy ולהפעיל לולאה וסכימה

**הסבר מתמטי:** np.array מייצר מערך; np.mean מחשב ממוצע

הפעל תאים לפי הסדר. חשב תחילה על נייר ונמק את הפלט. הערכים דידקטיים.

```python
import numpy as np
scores = np.array([70, 85, 90, 95])
total = 0
for score in scores:
    total += score
print('ממוצע:', total / len(scores))
```
