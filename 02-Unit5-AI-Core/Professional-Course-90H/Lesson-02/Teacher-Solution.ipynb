**משימת המעבדה:** שישה ציונים, לולאת for ו־np.mean

זהו פתרון לדוגמה עם קוד מלא. פתרונות גלויים במאגר ציבורי — רצוי להפיץ לתלמידים רק את Student-Lab.ipynb.

```python
import numpy as np
scores=np.array([70,75,80,85,90,95])
total=0
for score in scores:
    total+=score
manual=total/len(scores)
print('סכום',total,'ממוצע בלולאה',manual,'ממוצע NumPy',np.mean(scores))
assert np.isclose(manual,np.mean(scores))
```
