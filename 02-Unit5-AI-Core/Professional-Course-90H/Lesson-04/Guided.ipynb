# שיעור 4 — מרחק ודמיון בין וקטורים — מחברת מודרכת

**הקשר:** אילו שתי דוגמאות במאגר דומות יותר?

**יעד:** להבחין בין Manhattan, Euclidean ודמיון קוסינוס

**הסבר מתמטי:** L1=Σ|uᵢ−vᵢ|; L2=√Σ(uᵢ−vᵢ)²; cos=(u·v)/(||u||||v||)

הפעל תאים לפי הסדר. חשב תחילה על נייר ונמק את הפלט. הערכים דידקטיים.

```python
import numpy as np
u=np.array([1.,2.]);v=np.array([4.,6.])
print('L1:',np.sum(np.abs(u-v)))
print('L2:',np.linalg.norm(u-v))
print('cos:',np.dot(u,v)/(np.linalg.norm(u)*np.linalg.norm(v)))
```
