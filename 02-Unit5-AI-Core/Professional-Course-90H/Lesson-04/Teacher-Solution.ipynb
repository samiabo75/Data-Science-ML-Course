# שיעור 4 — מרחק ודמיון בין וקטורים — פתרון מתועד למורה

**משימה לתלמיד:** חשב מרחקי L1/L2 ודמיון קוסינוס בין [2,1] לבין [5,5]; נמק איזה מדד מתאים למקרה נתון.

**הכוונה:** זהו פתרון דוגמה לנתוני הדגמה, לא פתרון יחיד לכל בחירת נתונים של תלמיד. הערכים בתרגול העצמאי עשויים להיות שונים.

**הבסיס המתמטי:** L1=Σ|uᵢ−vᵢ|; L2=√Σ(uᵢ−vᵢ)²; cos=(u·v)/(||u||||v||)

```python
import numpy as np
u=np.array([1.,2.]);v=np.array([4.,6.])
print(np.sum(np.abs(u-v)),np.linalg.norm(u-v))
print(np.dot(u,v)/(np.linalg.norm(u)*np.linalg.norm(v)))
```
