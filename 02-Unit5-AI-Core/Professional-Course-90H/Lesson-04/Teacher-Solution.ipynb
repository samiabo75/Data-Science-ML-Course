# פתרון למורה — שיעור 4

**משימת המעבדה:** u=[2,1],v=[5,5]: L1, L2 וקוסינוס

זהו פתרון לדוגמה עם קוד מלא. פתרונות גלויים במאגר ציבורי — רצוי להפיץ לתלמידים רק את Student-Lab.ipynb.

```python
import numpy as np
u=np.array([2.,1.]);v=np.array([5.,5.])
l1=np.sum(np.abs(u-v))
l2=np.linalg.norm(u-v)
cos=np.dot(u,v)/(np.linalg.norm(u)*np.linalg.norm(v))
print('L1',l1,'L2',l2,'cos',round(cos,4))
assert np.isclose(l1,7)
assert np.isclose(l2,5)
```

## הסבר תוצאות

L1=3+4=7; L2=sqrt(9+16)=5; הקוסינוס הוא 15/(sqrt(5)*sqrt(50))≈0.9487. יש לבדוק וקטור אפס לפני חישוב קוסינוס.
