**הקשר:** מהו ציון טיפוסי של קבוצת תלמידים?

**יעד:** לחשב ממוצע, חציון ושכיח ולהסביר יתרונות ומגבלות

**הסבר מתמטי:** ממוצע רגיש לערכים קיצוניים; החציון עמיד יותר

הפעל תאים לפי הסדר. חשב תחילה על נייר ונמק את הפלט. הערכים דידקטיים.

```python
import numpy as np
from collections import Counter
x=[60,70,70,80,100]
print('ממוצע',np.mean(x),'חציון',np.median(x),'שכיח',Counter(x).most_common(1)[0][0])
```
