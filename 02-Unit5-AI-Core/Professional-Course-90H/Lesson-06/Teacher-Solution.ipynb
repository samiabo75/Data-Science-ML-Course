# שיעור 6 — אוכלוסייה, מדגם ומדדי מרכז — פתרון מתועד למורה

**משימה לתלמיד:** חשב מדדי מרכז ל־[50,70,70,80,90,100,100] והוסף ערך קיצוני. הסבר איזה מדד השתנה יותר.

**הכוונה:** זהו פתרון דוגמה לנתוני הדגמה, לא פתרון יחיד לכל בחירת נתונים של תלמיד. הערכים בתרגול העצמאי עשויים להיות שונים.

**הבסיס המתמטי:** ממוצע רגיש לערכים קיצוניים; החציון עמיד יותר

```python
import numpy as np
from collections import Counter
x=[60,70,70,80,100]
print(np.mean(x),np.median(x),Counter(x).most_common(1)[0])
```
