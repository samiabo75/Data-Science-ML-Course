**משימת המעבדה:** מדדי מרכז עם ערך חריג

זהו פתרון לדוגמה עם קוד מלא. פתרונות גלויים במאגר ציבורי — רצוי להפיץ לתלמידים רק את Student-Lab.ipynb.

```python
import numpy as np
from collections import Counter
x=[50,70,70,80,90,100,100]
y=x+[300]
for label,a in [('מקור',x),('עם חריג',y)]:
    print(label,'mean',np.mean(a),'median',np.median(a),'modes',Counter(a).most_common())
print('שינוי ממוצע',np.mean(y)-np.mean(x))
```

## הסבר תוצאות

הממוצע רגיש לתוספת 300 יותר מן החציון. במקרה של תיקו שכיחויות יש לדווח על כמה שכיחים ולא על הראשון בלבד.
