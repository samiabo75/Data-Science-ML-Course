**משימת המעבדה:** משפט הגבול המרכזי — גדלים 4,16,64

זהו פתרון לדוגמה עם קוד מלא. פתרונות גלויים במאגר ציבורי — רצוי להפיץ לתלמידים רק את Student-Lab.ipynb.

```python
import numpy as np
import matplotlib.pyplot as plt
rng=np.random.default_rng(42)
source=rng.exponential(scale=5,size=5000)
print('סטיית תקן מקור',np.std(source,ddof=0))
for n in [4,16,64]:
    means=np.mean(rng.choice(source,size=(1000,n),replace=True),axis=1)
    print('n',n,'סטיית תקן אמפירית של ממוצעים',round(np.std(means),3),'תחזית',round(np.std(source)/np.sqrt(n),3))
    plt.figure(figsize=(5,3));plt.hist(means,bins=25);plt.title(f'ממוצעי מדגמים: n={n}');plt.show()
```

## הסבר תוצאות

ההתפלגות של ממוצעי המדגמים מתקרבת לנורמלית ומתכווצת סביב התוחלת כאשר n גדל; התפלגות המקור נשארת מוטה.
