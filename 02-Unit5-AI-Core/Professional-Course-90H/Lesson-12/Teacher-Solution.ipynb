# שיעור 12 — משפט הגבול המרכזי ודגימת ממוצעים — פתרון מתועד למורה

**משימה לתלמיד:** דמה 1000 ממוצעי מדגמים בגדלים 4,16,64 והשווה היסטוגרמות ושגיאות תקן. האם נתוני המקור עצמם הפכו לנורמליים?

**הכוונה:** זהו פתרון דוגמה לנתוני הדגמה, לא פתרון יחיד לכל בחירת נתונים של תלמיד. הערכים בתרגול העצמאי עשויים להיות שונים.

**הבסיס המתמטי:** במדגם i.i.d. עם שונות סופית: SE(ממוצע)=σ/√n; CLT על הממוצעים

```python
import numpy as np
rng=np.random.default_rng(42)
pop=rng.exponential(5,5000)
for n in [5,40]:
 means=np.mean(rng.choice(pop,size=(1000,n),replace=True),axis=1)
 print(n,np.mean(means),np.std(means))
```
