# שיעור 12 — משפט הגבול המרכזי ודגימת ממוצעים — מחברת מודרכת

**הקשר:** מדוע ממוצעים נוטים להציג צורת פעמון גם כשזמני ההמתנה מוטים?

**יעד:** להבחין בין התפלגות אוכלוסייה לבין התפלגות ממוצעי מדגמים

**הסבר מתמטי:** במדגם i.i.d. עם שונות סופית: SE(ממוצע)=σ/√n; CLT על הממוצעים

הפעל תאים לפי הסדר. חשב תחילה על נייר ונמק את הפלט. הערכים דידקטיים.

```python
import numpy as np
import matplotlib.pyplot as plt
rng=np.random.default_rng(42)
population=rng.exponential(scale=5,size=5000)
means5=np.mean(rng.choice(population,size=(1000,5),replace=True),axis=1)
means40=np.mean(rng.choice(population,size=(1000,40),replace=True),axis=1)
for data,label in [(population,'אוכלוסייה'),(means5,'ממוצעים n=5'),(means40,'ממוצעים n=40')]:
 plt.figure();plt.hist(data,bins=25);plt.title(label);plt.show()
```
