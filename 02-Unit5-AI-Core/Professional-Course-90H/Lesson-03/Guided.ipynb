# שיעור 3 — וקטורים ופעולות מתמטיות — מחברת מודרכת

**הקשר:** איך מתארים מוצר באמצעות מחיר, דירוג וזמן משלוח?

**יעד:** לחשב חיבור וקטורים, מכפלה פנימית ונורמה

**הסבר מתמטי:** u·v=Σuᵢvᵢ; ||u||₂=√Σuᵢ²

הפעל תאים לפי הסדר. חשב תחילה על נייר ונמק את הפלט. הערכים דידקטיים.

```python
import numpy as np
u = np.array([3, 4])
v = np.array([2, 1])
print('חיבור:', u + v)
print('מכפלה פנימית:', np.dot(u, v))
print('נורמה:', np.linalg.norm(u))
```
