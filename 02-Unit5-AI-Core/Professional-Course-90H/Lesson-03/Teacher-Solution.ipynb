# שיעור 3 — וקטורים ופעולות מתמטיות — פתרון מתועד למורה

**משימה לתלמיד:** נתונים u=[6,8], v=[1,2]. חשב ידנית וב־NumPy חיבור, מכפלה פנימית ונורמה.

**הכוונה:** זהו פתרון דוגמה לנתוני הדגמה, לא פתרון יחיד לכל בחירת נתונים של תלמיד. הערכים בתרגול העצמאי עשויים להיות שונים.

**הבסיס המתמטי:** u·v=Σuᵢvᵢ; ||u||₂=√Σuᵢ²

```python
import numpy as np
u=np.array([3,4]);v=np.array([2,1])
print(u+v, np.dot(u,v), np.linalg.norm(u))
```
