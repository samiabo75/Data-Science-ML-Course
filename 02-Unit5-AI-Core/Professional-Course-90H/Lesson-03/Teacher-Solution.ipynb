# פתרון למורה — שיעור 3

**משימת המעבדה:** u=[6,8], v=[1,2]: חיבור, מכפלה פנימית ונורמה

זהו פתרון לדוגמה עם קוד מלא. פתרונות גלויים במאגר ציבורי — רצוי להפיץ לתלמידים רק את Student-Lab.ipynb.

```python
import numpy as np
u=np.array([6,8]);v=np.array([1,2])
print('חיבור',u+v)
print('מכפלה פנימית',np.dot(u,v))
print('נורמה u',np.linalg.norm(u))
assert np.dot(u,v)==22
assert np.isclose(np.linalg.norm(u),10)
```

## הסבר תוצאות

u+v=[7,10], המכפלה 6·1+8·2=22, ונורמת u היא sqrt(36+64)=10.
