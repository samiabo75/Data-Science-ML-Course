# שיעור 7 — רבעונים וטווח בין־רבעוני — מחברת מודרכת

**הקשר:** האם זמן המתנה חריג ביחס לשאר הלקוחות?

**יעד:** לפרש Q1,Q2,Q3, IQR וגבולות חריגים

**הסבר מתמטי:** IQR=Q3−Q1; גבולות: Q1−1.5IQR, Q3+1.5IQR (כלל תיאורי)

הפעל תאים לפי הסדר. חשב תחילה על נייר ונמק את הפלט. הערכים דידקטיים.

```python
import numpy as np
x=np.array([2,3,4,5,6,7,8,20])
q1,q2,q3=np.percentile(x,[25,50,75])
iqr=q3-q1
print('רבעונים',q1,q2,q3,'גבולות',q1-1.5*iqr,q3+1.5*iqr)
```
