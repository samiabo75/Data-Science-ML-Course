# שיעור 7 — רבעונים וטווח בין־רבעוני — פתרון מתועד למורה

**משימה לתלמיד:** חשב רבעונים ו־IQR לערכים [3,4,4,5,6,8,10,30]. הדגש ששיטות חישוב הרבעונים עשויות להיות שונות.

**הכוונה:** זהו פתרון דוגמה לנתוני הדגמה, לא פתרון יחיד לכל בחירת נתונים של תלמיד. הערכים בתרגול העצמאי עשויים להיות שונים.

**הבסיס המתמטי:** IQR=Q3−Q1; גבולות: Q1−1.5IQR, Q3+1.5IQR (כלל תיאורי)

```python
import numpy as np
x=np.array([2,3,4,5,6,7,8,20])
q=np.percentile(x,[25,50,75]);print(q, q[2]-q[0])
```
