# פתרון למורה — שיעור 7

**משימת המעבדה:** Q1 Q2 Q3, IQR וחריגים

זהו פתרון לדוגמה עם קוד מלא. פתרונות גלויים במאגר ציבורי — רצוי להפיץ לתלמידים רק את Student-Lab.ipynb.

```python
import numpy as np
x=np.array([3,4,4,5,6,8,10,30])
q1,q2,q3=np.percentile(x,[25,50,75])
iqr=q3-q1
low=q1-1.5*iqr;high=q3+1.5*iqr
print('רבעונים',q1,q2,q3,'IQR',iqr,'גבולות',low,high)
print('חשודים כחריגים',x[(x<low)|(x>high)])
```

## הסבר תוצאות

np.percentile משתמש בשיטת אינטרפולציה ברירת מחדל; במחברת ובבחינה ידנית עשויה להידרש שיטת רבעונים אחרת. יש לציין את השיטה.
