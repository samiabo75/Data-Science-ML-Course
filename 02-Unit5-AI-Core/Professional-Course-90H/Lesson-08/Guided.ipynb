# שיעור 8 — שונות וסטיית תקן — מחברת מודרכת

**הקשר:** לשתי כיתות אותו ממוצע: האם פיזור הציונים דומה?

**יעד:** לחשב שונות וסטיית תקן ולהבחין בין אוכלוסייה למדגם

**הסבר מתמטי:** שונות אוכלוסייה מחלקים ב־N, אומדן שונות מדגם ב־n−1

הפעל תאים לפי הסדר. חשב תחילה על נייר ונמק את הפלט. הערכים דידקטיים.

```python
import numpy as np
x=np.array([60,70,80,90,100])
print('ממוצע',np.mean(x))
print('שונות אוכלוסייה',np.var(x,ddof=0))
print('שונות מדגם',np.var(x,ddof=1))
print('סטיית תקן',np.std(x,ddof=0))
```
