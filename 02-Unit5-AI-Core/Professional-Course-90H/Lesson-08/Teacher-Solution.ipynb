# שיעור 8 — שונות וסטיית תקן — פתרון מתועד למורה

**משימה לתלמיד:** השווה [70,75,80,85,90] ל־[50,65,80,95,110]. חשב ממוצע וסטיית תקן והסק מסקנה.

**הכוונה:** זהו פתרון דוגמה לנתוני הדגמה, לא פתרון יחיד לכל בחירת נתונים של תלמיד. הערכים בתרגול העצמאי עשויים להיות שונים.

**הבסיס המתמטי:** שונות אוכלוסייה מחלקים ב־N, אומדן שונות מדגם ב־n−1

```python
import numpy as np
x=np.array([60,70,80,90,100])
print(np.mean(x),np.var(x,ddof=0),np.std(x,ddof=0))
```
