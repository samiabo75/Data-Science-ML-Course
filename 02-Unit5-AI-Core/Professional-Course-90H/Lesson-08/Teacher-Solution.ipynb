**משימת המעבדה:** השוואת פיזור שתי קבוצות

זהו פתרון לדוגמה עם קוד מלא. פתרונות גלויים במאגר ציבורי — רצוי להפיץ לתלמידים רק את Student-Lab.ipynb.

```python
import numpy as np
a=np.array([70,75,80,85,90]);b=np.array([50,65,80,95,110])
for name,x in [('קבוצה א',a),('קבוצה ב',b)]:
    print(name,'ממוצע',np.mean(x),'שונות אוכלוסייה',np.var(x,ddof=0),'סטיית תקן',np.std(x,ddof=0))
assert np.mean(a)==np.mean(b)==80
assert np.std(a)<np.std(b)
```

## הסבר תוצאות

לשתי הקבוצות ממוצע 80, אך סטיית התקן של ב גדולה יותר ולכן ציוניה מפוזרים יותר.
