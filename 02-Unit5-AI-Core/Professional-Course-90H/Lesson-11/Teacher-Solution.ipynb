# פתרון למורה — שיעור 11

**משימת המעבדה:** X~N(70,10²), הסתברות קטע מול pdf בנקודה

זהו פתרון לדוגמה עם קוד מלא. פתרונות גלויים במאגר ציבורי — רצוי להפיץ לתלמידים רק את Student-Lab.ipynb.

```python
from scipy.stats import norm
mu,sigma=70,10
area=norm.cdf(80,loc=mu,scale=sigma)-norm.cdf(60,loc=mu,scale=sigma)
density=norm.pdf(70,loc=mu,scale=sigma)
print('P(60<X<80)=',round(area,5))
print('f(70)=',round(density,5))
print('P(X=70)=',0)
assert 0<area<1
```

## הסבר תוצאות

הסתברות בתחום היא כ־0.68269; f(70)≈0.03989 היא צפיפות (גובה), לא הסתברות; ההסתברות לנקודה בודדת היא 0.
