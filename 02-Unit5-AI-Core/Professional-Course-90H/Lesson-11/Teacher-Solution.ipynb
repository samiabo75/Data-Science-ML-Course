# שיעור 11 — Z-score, צפיפות והסתברות — פתרון מתועד למורה

**משימה לתלמיד:** עבור X~N(70,10²), מצא P(60<X<80), חשב pdf(70), והסבר מדוע pdf(70) אינו P(X=70).

**הכוונה:** זהו פתרון דוגמה לנתוני הדגמה, לא פתרון יחיד לכל בחירת נתונים של תלמיד. הערכים בתרגול העצמאי עשויים להיות שונים.

**הבסיס המתמטי:** Z=(x−μ)/σ; P(a<X<b)=F(b)−F(a); P(X=a)=0 ברציף בעל צפיפות

```python
from scipy.stats import norm
print(norm.cdf(115,100,15)-norm.cdf(85,100,15))
print(norm.pdf(100,100,15))
```
