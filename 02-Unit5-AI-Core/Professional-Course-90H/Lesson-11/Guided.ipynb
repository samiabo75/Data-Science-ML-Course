# שיעור 11 — Z-score, צפיפות והסתברות — מחברת מודרכת

**הקשר:** איך מחשבים הסתברות לציון בתחום ולא בנקודה?

**יעד:** לחשב Z, CDF והסתברות לקטע ולהבחין מצפיפות

**הסבר מתמטי:** Z=(x−μ)/σ; P(a<X<b)=F(b)−F(a); P(X=a)=0 ברציף בעל צפיפות

הפעל תאים לפי הסדר. חשב תחילה על נייר ונמק את הפלט. הערכים דידקטיים.

```python
from scipy.stats import norm
mu,sigma=100,15
a,b=85,115
print('ציוני Z',(a-mu)/sigma,(b-mu)/sigma)
print('הסתברות לקטע',norm.cdf(b,mu,sigma)-norm.cdf(a,mu,sigma))
print('צפיפות ב־100',norm.pdf(100,mu,sigma))
print('הסתברות לנקודה',0)
```
