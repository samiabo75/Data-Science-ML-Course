# שיעור 10 — התפלגות נורמלית ו־Plinko — פתרון מתועד למורה

**משימה לתלמיד:** שנה מספר כדורים מ־100 ל־3000 והשווה. מה קורה אם p=0.7? האם משתנה הנחיתה רציף?

**הכוונה:** זהו פתרון דוגמה לנתוני הדגמה, לא פתרון יחיד לכל בחירת נתונים של תלמיד. הערכים בתרגול העצמאי עשויים להיות שונים.

**הבסיס המתמטי:** ב־Plinko מספר תאים בדיד; ההתפלגות עשויה להתקרב לצורת פעמון

```python
import numpy as np
import matplotlib.pyplot as plt
rng=np.random.default_rng(42)
x=rng.binomial(12,0.5,3000)
plt.hist(x,bins=np.arange(-0.5,13.5,1));plt.show()
```
