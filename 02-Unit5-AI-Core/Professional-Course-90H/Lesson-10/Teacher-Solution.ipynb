# פתרון למורה — שיעור 10

**משימת המעבדה:** Plinko: 100 מול 3000 כדורים וגם p=0.7

זהו פתרון לדוגמה עם קוד מלא. פתרונות גלויים במאגר ציבורי — רצוי להפיץ לתלמידים רק את Student-Lab.ipynb.

```python
import numpy as np
import matplotlib.pyplot as plt
rng=np.random.default_rng(42)
for size,p in [(100,0.5),(3000,0.5),(3000,0.7)]:
    x=rng.binomial(n=12,p=p,size=size)
    plt.figure(figsize=(6,3));plt.hist(x,bins=np.arange(-0.5,13.5,1),edgecolor='black')
    plt.title(f'כדורים={size}, הסתברות ימינה={p}')
    plt.xlabel('מספר סטיות ימינה');plt.ylabel('שכיחות');plt.show()
    print('ממוצע',np.mean(x))
```

## הסבר תוצאות

ב־p=0.5 ההתפלגות סימטרית בקירוב; ב־p=0.7 מרכז ההתפלגות זז ימינה. המדידה בדידה (13 תוצאות אפשריות עבור 12 הסתעפויות).
