**הקשר:** מדוע כדורי Plinko רבים מתרכזים במרכז?

**יעד:** להבדיל בין ניסוי בינומי בדיד לעקומת צפיפות נורמלית רציפה

**הסבר מתמטי:** ב־Plinko מספר תאים בדיד; ההתפלגות עשויה להתקרב לצורת פעמון

הפעל תאים לפי הסדר. חשב תחילה על נייר ונמק את הפלט. הערכים דידקטיים.

```python
import numpy as np
import matplotlib.pyplot as plt
rng=np.random.default_rng(42)
landings=rng.binomial(12,0.5,size=3000)
plt.hist(landings,bins=np.arange(-0.5,13.5,1),edgecolor='black')
plt.xlabel('מספר סטיות ימינה');plt.ylabel('שכיחות');plt.show()
```
