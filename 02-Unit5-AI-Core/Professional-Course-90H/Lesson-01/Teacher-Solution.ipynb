**משימת המעבדה:** ארבע רשומות עם שלושה מאפיינים, מספר שורות ועמודות

זהו פתרון לדוגמה עם קוד מלא. פתרונות גלויים במאגר ציבורי — רצוי להפיץ לתלמידים רק את Student-Lab.ipynb.

```python
data = [[2, 4, 1], [3, 5, 0], [4, 3, 1], [1, 2, 0]]
print('מספר רשומות:', len(data))
print('מספר מאפיינים:', len(data[0]))
for row in data:
    print(row)
```

## הסבר תוצאות

יש 4 רשומות ו־3 מאפיינים בכל אחת. כל שורה היא דוגמה, וכל עמודה מייצגת מאפיין. התווית/יעד אינה מוגדרת כאן בנפרד.
